# Voľný čas a spotreba

V predchádzajúcej kapitole sme analyzovali spotrebiteľovo rozhodovanie
medzi rôznymi spotrebnými statkami. V makroekonómii sa pozornosť často
sústreďuje na rozhodovanie medzi spotrebou produktov a spotrebou
voľného času.

Budeme uvažovať agregovaný, reálny a statický model spotrebiteľa.

Agregované hľadisko znamená, že všetky spotrebné statky nahradíme
jediným kompozitným statkom $C$, ktorý predstavuje celkovú spotrebu
tovarov a služieb.

Kompozitný statok $C$ plní v modeli aj úlohu peňazí (média výmeny).
Príjmy, výdavky a ostatné transakcie spotrebiteľa preto vyjadrujeme
v jednotkách spotreby, nie v peňažných jednotkách. Ak je premenná
vyjadrená v jednotkách spotreby, ide o jej reálnu hodnotu, ktorá nie je
ovplyvnená trhovou cenou spotreby.

Ak by sme pracovali s nominálnymi premennými, hodnotu spotreby by sme zapisovali ako $PC$,

kde $P$ označuje cenu jednotky spotreby.

V tejto kapitole budeme používať reálne premenné, a preto cenu jednotky
spotreby považujeme za rovnú jednej,

$$
P=1.
$$

Statické hľadisko znamená, že spotrebiteľa analyzujeme, akoby
žil iba v jednom období. Neuvažujeme preto sporenie, zadlžovanie ani
presuny spotreby medzi rôznymi obdobiami. Dynamické hľadisko správania spotrebiteľa predstavíme v nasledovnej kapitole.

Spotrebiteľ rozdeľuje svoj exogénny disponibilný čas $h$ medzi prácu $N^s$ a
voľný čas $\ell$

```{math}
:label: r_casove_ohranicenie
h=N^s+\ell.
```

Voľný čas a spotreba sú v modeli interpretované ako spotrebné statky.

## Preferencie spotrebiteľa

Predpokladáme, že voľný čas je pre spotrebiteľa statok ako spotreba a preferencie voči týmto statkom vyjadríme dvakrát diferencovateľnou funkciou užitočnosti

```{math}
:label: r_uzitocnost_lC
u=u(\ell,C).
```

Predpokladáme, že preferencie spotrebiteľa majú rovnaké vlastnosti ako
v predchádzajúcej kapitole. Funkcia užitočnosti je rastúca vo svojich argumentoch a reprezentuje
spojité a konvexné preferencie v rozhodovacom priestore $(\ell,C)$.
Indiferenčné krivky sú preto klesajúce a konvexné.

Predpokladáme, že voľný čas aj spotreba sú normálne ({numref}`tab_prijmovy_efekt`) a štandardné ({numref}`tab_zmena_ceny`) statky.

Hraničná miera spotrebiteľskej substitúcie medzi voľným časom a spotrebou je

```{math}
:label: r_mrs_lC
MRS(\ell,C)
=
\frac{\partial u(\ell,C)/\partial \ell}
     {\partial u(\ell,C)/\partial C}.

Hraničná miera spotrebiteľskej substitúcie udáva približné množstvo spotreby,
ktorého je spotrebiteľ ochotný vzdať sa za dodatočnú jednotku voľného
času pri nezmenenej úrovni užitočnosti.

## Rozpočtové ohraničenie

Spotrebiteľ získa príjem z ponuky práce a z vlastníctva výrobných
faktorov.

Ak spotrebiteľ pracuje $N^s$ jednotiek času za reálnu mzdovú sazdbu $w$,
dosahuje pracovný príjem $wN^s$.



Okrem pracovného príjmu môže spotrebiteľ získať aj zisk firmy, ktorú vlastní. Tento príjem označíme $\pi$.

Ak spotrebiteľ platí paušálnu daň $T$, jeho rozpočtové ohraničenie je

```{math}
:label: r_rozpocet_spotrebitela_praca
C \le wN^s+\pi-T.
```


```{admonition} Skrátené pojmy
Na mzdovú sadzbu $w$ sa v texte budeme skrátene odvolávať aj ako na mzdu. Mzdová sadzba a mzda sú v ďalšom texte synonyma.

Príjem $\pi-T$ je nepracovný príjem po zaplatení paušálnej dane alebo skrátene iba nepracovný príjem.
```

(def_pausal_dan)=
Paušálna daň 
: je daň, ktorej výška nezávisí od rozhodnutí
spotrebiteľa. Spotrebiteľ ju platí bez ohľadu na výšku svojej
spotreby, množstvo odpracovaného času alebo dosiahnutý príjem.

Spotreba nemôže byť vyššia ako disponibilný príjem spotrebiteľa.
Disponibilný príjem pozostáva z pracovného príjmu, zisku firmy a je znížený o zaplatené dane.

Po dosadení časového ohraničenia {eq}`r_casove_ohranicenie`

$$
N^s=h-\ell
$$

dostaneme rozpočtové ohraničenie v tvare

```{math}
:label: r_rozpocet_spotrebitela_praca2
C \le w(h-\ell)+\pi-T.
```

Dodatočná jednotka voľného času znižuje pracovný príjem o $w$
jednotiek spotreby. Reálna mzdová sadzba preto predstavuje alternatívne náklady
voľného času.

## Hranica spotrebiteľských možností

Rozpočtové ohraničenie {eq}`r_rozpocet_spotrebitela_praca` vymedzuje množinu spotrebiteľských možností,
teda všetky kombinácie voľného času a spotreby, ktoré si môže
spotrebiteľ dovoliť.

Ak v rozpočtovom ohraničení nahradíme znak nerovnosti znakom rovnosti,
získame hranicu spotrebiteľských možností

```{math}
:label: r_hsm
C=w(h-\ell)+\pi-T.
```

Hranica spotrebiteľských možností oddeľuje dosiahnuteľné kombinácie
voľného času a spotreby od nedosiahnuteľných kombinácií.

Hraničná miera rozpočtovej substitúcie medzi voľným časom a spotrebou je

```{math}
:label: r_mrb_lC
MRB(\ell,C)=w.
```

Hraničná miera rozpočtovej substitúcie udáva množstvo spotreby,
ktoré musí spotrebiteľ obetovať za dodatočnú jednotku voľného času.
Je určená sklonom rozpočtového ohraničenia

$$
-\frac{\partial{C}}{\partial{\ell}} = w
$$

a jej hodnota sa rovná
reálnej mzdovej sadzbe.

```{admonition} Všimnite si

Doteraz sme rozpočtové ohraničenie zapisovali v reálnych jednotkách
spotreby. V nominálnych jednotkách by po dosadení časového ohraničenia malo rozpočtové ohraničenie tvar

$$
PC \le \omega(h-\ell)+\Pi-\mathcal{T}.
$$

kde $P$ je cena jednotky spotreby, $\omega = wP$ nominálna mzdová sadzba,
$\Pi = P \pi$ nominálny dôchodok z vlastníctva výrobných faktorov a $\mathcal{T} = PT$
nominálna paušálna daň.

Tento zápis je analógiou rozpočtového ohraničenia spotrebiteľa

$$
p_1x_1+p_2x_2\le M,
$$

pričom

$$
p_1=\omega,
\qquad
p_2=P,
\qquad
M=\omega h+\Pi-\mathcal{T}.
$$

Cena voľného času je preto nominálna mzdová sadzba $\omega$. Ak sa
spotrebiteľ rozhodne spotrebovať o jednotku voľného času viac,
vzdáva sa pracovného príjmu vo výške $\omega$.

Hraničná miera rozpočtovej substitúcie medzi voľným časom a spotrebou je

$$
MRB(\ell,C)
=
\frac{p_1}{p_2}
=
\frac{\omega}{P}
=
w,
$$

kde $w$ označuje reálnu mzovú sadzbu.
```

````python
import numpy as np
import matplotlib.pyplot as plt

# parametre
h = 10
w = 1

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# =====================================================
# π - T > 0
# =====================================================

A = 3  # A = π - T

lmax = h + A / w

# hranica rozpočtových možností
l1 = np.linspace(0, h, 200)
C1 = w * (h - l1) + A

# množina spotrebiteľských možností
poly_x = [0, lmax, h, h, 0]
poly_y = [0, 0, 0, A, w * h + A]

ax[0].fill(poly_x, poly_y, color="lightgreen", alpha=0.6)

# zelená hranica
ax[0].plot(l1, C1, color="green", lw=3)
ax[0].plot([h, h], [0, A], color="green", lw=3)

# pomocné čiary
ax[0].axvline(h, ls="--", color="gray")
ax[0].axhline(A, ls="--", color="gray")

# bod bez práce
ax[0].scatter([h], [A], color="black", zorder=10)
ax[0].annotate(r'$[h,\pi-T]$',
               xy=(h, A),
               xytext=(10, 10),
               textcoords='offset points')

# sklon
x_arrow = 5
y_arrow = w * (h - x_arrow) + A

ax[0].annotate(r'sklon $=-w$',
               xy=(x_arrow, y_arrow),
               xytext=(4.0, y_arrow + 2),
               arrowprops=dict(arrowstyle='->'))

ax[0].set_title(r'$\pi-T>0$')
ax[0].set_xlabel(r'$\ell$')
ax[0].set_ylabel(r'$C$')

ax[0].set_xticks([h])
ax[0].set_xticklabels([r'$h$'])

ax[0].set_xlim(0, 14)
ax[0].set_ylim(0, 15)

# prerušované predĺženie rozpočtovej priamky

l_ext = np.linspace(h, lmax, 100)
C_ext = w*(h - l_ext) + A

ax[0].plot(
    l_ext,
    C_ext,
    color="green",
    lw=2,
    ls="--"
)

# označenie extrémov

ax[0].scatter([0, h, lmax],
              [w*h + A, A, 0],
              color="black")

ax[0].set_xticks([h, lmax])
ax[0].set_xticklabels([
    r'$h$',
    r'$h+\frac{\pi-T}{w}$'
])

ax[0].set_yticks([A, w*h + A])
ax[0].set_yticklabels([
    r'$\pi-T$',
    r'$wh+\pi-T$'
])

# =====================================================
# π - T ≤ 0
# =====================================================

A = -1.5

l0 = h + A/w

l2 = np.linspace(0, l0, 200)
C2 = w*(h - l2) + A

# trojuholník spotrebiteľských možností
ax[1].fill_between(
    l2,
    0,
    C2,
    color="lightgreen",
    alpha=0.6
)

# hranica možností
ax[1].plot(
    l2,
    C2,
    color="green",
    lw=3
)

# zvislé časové ohraničenie
ax[1].axvline(h,
              ls="--",
              color="gray")

# extrémy
ax[1].scatter(
    [l0, 0],
    [0, w*(h)+A],
    color="black"
)

# sklon
ax[1].annotate(
    r'sklon $=-w$',
    xy=(5.5, w*(h-5.5)+A),
    xytext=(4.0, y_arrow - 2),
    arrowprops=dict(arrowstyle='->')
)

ax[1].set_title(r'$\pi-T\leq0$')

ax[1].set_xlabel(r'$\ell$')
# ax[1].set_ylabel(r'$C$')

ax[1].set_xticks([l0, h])
ax[1].set_xticklabels([
    r'$h+\frac{\pi-T}{w}$',
    r'$h$'
])

ax[1].set_yticks([w*(h)+A])
ax[1].set_yticklabels([
    r'$wh+\pi-T$'
])

ax[1].set_xlim(0, 14)
ax[1].set_ylim(0, 15)

plt.tight_layout()
plt.show()
````

Obrázok zobrazuje hranice spotrebiteľských možností pre dva prípady. V oboch prípadoch je sklon hranice rovný $-w$, pričom reálna mzdová sadzba $w$ predstavuje alternatívne náklady voľného času.

Ak platí $\pi-T>0$, spotrebiteľ disponuje kladným nepracovným príjmom. Aj bez ponuky práce preto môže financovať spotrebu vo výške $\pi-T$. Hranica spotrebiteľských možností pretína obe osi a spotrebiteľské možnosti sú ohraničené rozpočtovým aj časovým ohraničením. Hranica spotrebiteľských možností má v tomto prípade konkávny tvar.

Ak platí $\pi-T\le 0$, spotrebiteľ nedokáže financovať spotrebu iba z nepracovného príjmu. Na dosiahnutie kladnej spotreby musí ponúkať prácu. Hranica spotrebiteľských možností preto pretína os voľného času v bode $h+(\pi-T)/w$, ktorý leží naľavo od bodu $h$.

Ďalšie grafické interpretácie v tejto kapitole budú predpokladať
$\pi - T > 0$.

## Úloha spotrebiteľa

Úloha spotrebiteľova je maximalizácia funckie užitočnosti {eq}`r_uzitocnost_lC` vzhľadom na rozpočtové ohraničenie {eq}`r_rozpocet_spotrebitela_praca2`, časové ohraničenie {eq}`r_casove_ohranicenie` a nezápornosť rozhodovacích premenných. Formálne ide o úlohu

```{math}
:label: r_max_ulc
u_{max} = \max_{\ell,C} u(\ell,C),
```

vzhľadom na {eq}`r_rozpocet_spotrebitela_praca2`

$$
C \le w(h-\ell)+\pi-T
$$

a

```{math}
:label: r_nezapornost_ulc
0 \le \ell \le h \quad C \ge 0.
```

## Podmienky prvého rádu

Podmienky prvého rádu sú 

- hraničná miera spotrebiteľskej substitúcie {eq}`r_mrs_lC` a hraničná miera rozpočtovej substitúcie medzi voľným časom a spotrebou {eq}`r_mrb_lC` sa rovnajú a
- výber voľného času a spotreby sa realizuje na hranici spotrebiteľských možností. 

```{math}
:label: r_optimum_lC
MRS(\ell,C)=MRB(\ell,C).
```

a {eq}`r_hsm`:

$$
C=w(h-\ell)+\pi-T.
$$

Po dosadení vzťahov {eq}`r_mrs_lC` a {eq}`r_mrb_lC`  do {eq}`r_optimum_lC` dostaneme

```{math}
:label: r_optimum_lC_w
MRS(\ell,C)=w.
```

Spotrebiteľ v optime oceňuje dodatočnú jednotku voľného času rovnako,
ako sú jej alternatívne náklady vyjadrené reálnou mzdovou sadzbou.


{ref}`redukovany_tvar` úlohy {eq}`r_max_ulc`, {eq}`r_rozpocet_spotrebitela_praca2` a {eq}`r_nezapornost_ulc` možno zapísať funkciami

```{math}
:label: r_volny_cas
l^* = l(\Pi-T,w),
```

a

```{math}
:label: r_spotreba_lC
C^* = C(\Pi-T,w).
```

Kombináciou {eq}`r_volny_cas` a {eq}`r_casove_ohranicenie` získame ponuku práce v redukovanom tvare:

```{math}
:label: r_ponuka_prace
N^{s*} = N^s(\Pi-T,w).
```

Redukovaný tvar za predpkladu Cobbovho Douglasovho tvaru funkcie užitočnosti je odvodený v príklade {ref}`priklad_CD_lC`: {eq}`r_volny_cas_CD`, {eq}`spotreba_lC_CD` a {eq}`ponuka_prace_CD`.

````python
# parametre
h = 24
w = 4

# optimum
l_star = 10
C_star = 80

fig, ax = plt.subplots(figsize=(12, 7))

# =====================================================
# π - T > 0
# =====================================================

A = 24  # A = π - T

lmax = h + A / w

# hranica rozpočtových možností
l = np.linspace(0, h, 200)
C = w * (h - l) + A

# zelená hranica
ax.plot(l, C, color="green", lw=3)
ax.plot([h, h], [0, A], color="green", lw=3)

# pomocné čiary
ax.axvline(h, ls="--", color="gray")
ax.axhline(A, ls="--", color="gray")

# bod bez práce
ax.scatter([h], [A], color="black", zorder=10)
ax.annotate(r'$[h,\pi-T]$',
               xy=(h, A),
               xytext=(10, 10),
               textcoords='offset points')

# sklon
x_arrow = 5
y_arrow = w * (h - x_arrow) + A

ax.text(
    10.1,
    125,
    r'Podmienka optimálnosti: $MRS(\ell^*,C^*)=w$'
)

ax.set_xlabel(r'$\ell$')
ax.set_ylabel(r'$C$')

ax.set_xticks([h])
ax.set_xticklabels([r'$h$'])

ax.set_xlim(0, 33)
ax.set_ylim(0, 130)

# prerušované predĺženie rozpočtovej priamky

l_ext = np.linspace(h, lmax, 100)
C_ext = w*(h - l_ext) + A

ax.plot(
    l_ext,
    C_ext,
    color="green",
    lw=2,
    ls="--"
)

# označenie extrémov

ax.scatter([0, h, lmax],
              [w*h + A, A, 0],
              color="black")

ax.set_xticks([l_star, h, lmax])

ax.set_xticklabels([
    r'$\ell^*$',
    r'$h$',
    r'$h+\frac{\pi-T}{w}$'
])

ax.set_yticks([C_star, A, w*h + A])

ax.set_yticklabels([
    r'$C^*$',
    r'$\pi-T$',
    r'$wh+\pi-T$'
])

# indiferenčná krivka prechádzajúca optimom

l_ic = np.linspace(1, 26, 500)
C_ic = 80*np.sqrt(10/l_ic)

ax.plot(
    l_ic,
    C_ic,
    color="blue",
    lw=2,
    label="indiferenčná krivka"
)

ax.scatter(
    [l_star],
    [C_star],
    color="red",
    s=70,
    zorder=10
)

ax.axvline(
    l_star,
    color="black",
    ls="--",
    lw=1
)

ax.axhline(
    C_star,
    color="black",
    ls="--",
    lw=1
)

ax.annotate(
    '',
    xy=(h, 10),
    xytext=(l_star, 10),
    arrowprops=dict(
        arrowstyle='<->',
        color='darkorange',
        lw=2
    )
)

ax.text(
    (l_star + h)/2,
    12,
    r'$N^{s*}$',
    color='darkorange',
    ha='center'
)

plt.tight_layout()
plt.show()
````

```

Grafická interpretácia riešenia úlohy {eq}`r_max_ulc`, {eq}`r_rozpocet_spotrebitela_praca2` a {eq}`r_nezapornost_ulc` je na {numref}`graf_optimum_lC` Zelená čiara predstavuje hranicu
spotrebiteľských možností a modrá krivka indiferenčnú krivku
prechádzajúcu optimálnym bodom.

Optimálna kombinácia voľného času a spotreby $(\ell^*,C^*)$ je bodom dotyku indiferenčnej krivky a hranice spotrebiteľských možností.
V tomto bode sa hraničná miera spotrebiteľskej substitúcie medzi
voľným časom a spotrebou rovná hraničnej miere rozpočtovej substitúcie.

Tretia endogénna premenná modelu je ponuka práce, ktorú možno dopočítať z časového ohraničenia {eq}`r_casove_ohranicenie`. Optimálna ponuka práce zodpovedá časti disponibilného času, ktorú
spotrebiteľ nevenuje voľnému času,

$$
N^{s*}=h-\ell^*.
$$

Graficky je optimálna ponuka práce vyznačená ako vzdialenosť medzi
optimálnym voľným časom $\ell^*$ a celkovým disponibilným časom $h$.

Ak optimálne riešenie neleží v extrémnych bodoch
$\ell=0$ alebo $\ell=h$, konvexnosť preferencií a tvar hranice
spotrebiteľských možností zabezpečujú, že podmienky prvého rádu
{eq}`r_optimum_lC_w` a {eq}`r_hsm` sú postačujúce pre existenciu optima.

Na {numref}`graf_h_l` je zobrazené rozhodnutie spotrebiteľa v bode $[\ell^*,C^*] = [h,\pi-T]$. Spotrebiteľove rozhodnutie je nepracovať ($N^{s*}=0$) a spotrebu financovať výlučne z nepracovného príjmu.
V tomto bode je hraničná miera spotrebiteľskej substitúcie vyššia než
reálna mzdová sadzba,

$$
MRS(h,\pi-T)>w.
$$

Rozhodnutie spotrebiteľa nepracovať je priblížené v príklade {ref}`priklad_CD_lC`. Príklad predpkkladá Cobbov-Douglasov tvar funkcie užitočnosti.

````python
import numpy as np
import matplotlib.pyplot as plt

def indifference_curve(l, l0, C0, alpha):
    return C0 * (l0 / l)**(alpha / (1 - alpha))

alpha = 3/4

# parametre
h = 24
w = 3
A = 48      # π - T
l0 = h
C0 = A

fig, ax = plt.subplots(figsize=(12, 7))

# hranica spotrebiteľských možností
l = np.linspace(0, h, 200)
C = w * (h - l) + A

# hlavná časť hranice
ax.plot(l, C, color="green", lw=3)

# zvislá časť hranice
ax.plot([h, h], [0, A], color="green", lw=3)

# pomocné čiary
ax.axvline(h, ls="--", color="gray")
ax.axhline(A, ls="--", color="gray")

# predĺženie rozpočtovej priamky
lmax = h + A / w

l_ext = np.linspace(h, lmax, 100)
C_ext = w * (h - l_ext) + A

ax.plot(
    l_ext,
    C_ext,
    color="green",
    lw=2,
    ls="--"
)

# extrémy
ax.scatter(
    [0, h, lmax],
    [w*h + A, A, 0],
    color="black",
    zorder=10
)

# označenie bodu bez práce
ax.annotate(
    r'$[h,\pi-T]$',
    xy=(h, A),
    xytext=(10, 10),
    textcoords='offset points'
)

# osi
ax.set_xlabel(r'$\ell$')
ax.set_ylabel(r'$C$')

ax.set_xlim(0, 42)
ax.set_ylim(0, 130)

# označenie osí
ax.set_xticks([h, lmax])
ax.set_xticklabels([
    r'$\ell^* = h$',
    r'$h+\frac{\pi-T}{w}$'
])
ax.set_yticks([A, w*h + A])

ax.set_yticklabels([
    r'$C^*=\pi-T$',
    r'$wh+\pi-T$'
])

l_ic = np.linspace(4, 40, 500)

C_ic = indifference_curve(l_ic, l0, C0, alpha)

ax.plot(
    l_ic,
    C_ic,
    color="blue",
    lw=2
)

plt.tight_layout()
plt.show()
````

## Vplyv zníženia zisku alebo zvýšenia paušálnej dane

{numref}`graf_zmena_pi_T` znázorňuje vplyv poklesu nepracovného príjmu $(\pi-T)$, ktorý môže byť spôsobený napríklad poklesom zisku firmy $\pi$ alebo zvýšením paušálnej dane $T$.

Pokles nepracovného príjmu má **čisto príjmový efekt**. Posúva hranicu spotrebiteľských možností smerom nadol. Keďže voľný čas aj spotreba sú normálne statky, spotrebiteľ v dôsledku príjmového efektu obmedzuje spotrebu aj voľný čas. Optimálny voľný čas sa znižuje z $\ell_1^*$ na $\ell_2^*$ a optimálna spotreba z $C_1^*$ na $C_2^*$.

Pokles voľného času zároveň znamená nárast ponuky práce. Zníženie nepracovného príjmu preto vedie k zvýšeniu ponuky práce domácnosti.

````python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# model
# --------------------------------------------------

h = 24

def optimum(alpha, w, A):
    M = w*h + A
    l_star = alpha * M / w
    C_star = (1 - alpha) * M
    return l_star, C_star

def indifference_curve(l, alpha, l_star, C_star):
    return C_star * (l_star/l)**(alpha/(1-alpha))

# --------------------------------------------------
# scenáre
# --------------------------------------------------

alpha = 1/3

scenarios = [
    {"A": 30, "w": 4, "color": "blue"},
    {"A": 10, "w": 4, "color": "red"},
]

# --------------------------------------------------
# graf
# --------------------------------------------------

fig, ax = plt.subplots(figsize=(12,7))

for s in scenarios:

    A = s["A"]
    w = s["w"]
    color = s["color"]

    # optimum
    l_star, C_star = optimum(alpha, w, A)

    # hranica možností
    l = np.linspace(0, h, 300)
    C = w*(h-l) + A

    ax.plot(l, C, color=color, lw=2)

    # vodorovná časť pri π-T
    ax.plot([h, h], [0, A], color=color, lw=2)

    # vodorovná pomocná čiara π-T
    ax.axhline(A, color=color, ls="--", alpha=0.5)

    # indiferenčná krivka
    l_ic = np.linspace(1, h, 500)
    C_ic = indifference_curve(
        l_ic,
        alpha,
        l_star,
        C_star
    )

    ax.plot(
        l_ic,
        C_ic,
        color=color,
        ls=":"
    )

    # optimum
    ax.scatter(
        [l_star],
        [C_star],
        color=color,
        s=70,
        zorder=10
    )
    # vodiace čiary k optimu

    ax.plot(
        [0, l_star],
        [C_star, C_star],
        color=color,
        ls="--",
        lw=1
    )

    ax.plot(
        [l_star, l_star],
        [0, C_star],
        color=color,
        ls="--",
        lw=1
    )

# spoločná zvislá čiara h
ax.axvline(h, ls="--", color="black")

ax.set_xticks([])
ax.set_yticks([])

# modré optimum
l_star_blue, C_star_blue = optimum(alpha, 4, 30)

ax.text(
    l_star_blue,
    -4,
    r'$\ell^*_1$',
    color='blue',
    ha='center'
)

ax.text(
    -2,
    C_star_blue,
    r'$C^*_1$',
    color='blue',
    va='center'
)

# červené optimum
l_star_red, C_star_red = optimum(alpha, 4, 10)

ax.text(
    l_star_red,
    -4,
    r'$\ell^*_2$',
    color='red',
    ha='center'
)

ax.text(
    -2,
    C_star_red,
    r'$C^*_2$',
    color='red',
    va='center'
)

# π - T
ax.text(
    -2,
    30,
    r'$(\pi-T)_1$',
    color='blue',
    va='center'
)

ax.text(
    -2,
    10,
    r'$(\pi-T)_2$',
    color='red',
    va='center'
)

ax.text(
    h,
    -4,
    r'$h$',
    ha='center'
)

ax.set_xlabel(r'$\ell$')
ax.set_ylabel(r'$C$')

ax.set_xlim(0, 26)
ax.set_ylim(0, 130)

plt.tight_layout()
plt.show()
````

## Vplyv zvýšenia mzdovej sadzby

Na {numref}`graf_zmena_w_lC` je znázornený vplyv zvýšenia reálnej mzdovej sadzby $w$ na optimálnu kombináciu spotreby a voľného času.

Zvýšenie reálnej mzdovej sadzby zvyšuje sklon hranice spotrebných možností.
Pri nezmenenom nepracovnom príjme $(\pi-T)$ sa hranica spotrebných
možností otáča okolo bodu $[h,\pi-T]$.

Tento bod leží na každej hranici spotrebných možností bez ohľadu na
výšku mzdovej sadzby. 

Po zvýšení $w$ môže spotrebiteľ pri nižšom množstve voľného času dosiahnuť vyššiu spotrebu ako pred zvýšením mzdy.

Zo zobrazeného optimálneho riešenia je zrejmé, že optimálna spotreba $C$ sa zvýši. Spotrebiteľ využíva vyššie zhodnotenie práce na dosiahnutie vyššej úrovne spotreby.

Vplyv zvýšenia mzdy na optimálne množstvo voľného času $\ell$ však nie je bezprostredne zrejmý. Na {numref}`graf_zmena_w_lC` síce pozorujeme pokles optimálneho množstva voľného času, tento výsledok však nie je všeobecný.

Ak by sa optimálne množstvo voľného času znížilo, zvýšila by sa ponuka práce

$$
N^s=h-\ell.
$$

Naopak, ak by sa optimálne množstvo voľného času zvýšilo, ponuka práce by sa znížila.

Celkový vplyv zmeny mzdy na voľný čas a ponuku práce preto nemožno určiť iba z geometrie hranice spotrebných možností. Vyššia mzda totiž súčasne mení relatívnu cenu voľného času aj dôchodok spotrebiteľa. V ďalšom texte preto rozložíme celkový efekt zmeny mzdy na substitučný a príjmový efekt.

````python
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# Parametre modelu
# ==================================================

h = 24
alpha = 1/3

scenarios = [
    {"A": 24, "w": 3, "color": "blue"},
    {"A": 24, "w": 10, "color": "red"},
]

# ==================================================
# Funkcie
# ==================================================

def budget(l, w, A):
    """
    Hranica spotrebných možností

    C = w(h-l) + A
    """
    return w * (h - l) + A


def optimum(alpha, w, A):
    """
    Cobb-Douglas

    U(C,l)=C^(1-alpha) l^alpha
    """

    M = w * h + A

    l_star = alpha * M / w
    C_star = (1 - alpha) * M

    return l_star, C_star


def indifference_curve(l, alpha, l_star, C_star):
    """
    Indiferenčná krivka pre úroveň užitočnosti
    prechádzajúcu optimom.
    """

    U = C_star**(1-alpha) * l_star**alpha

    return (U / (l**alpha))**(1/(1-alpha))


# ==================================================
# Graf
# ==================================================

fig, ax = plt.subplots(figsize=(10, 6))

l_grid = np.linspace(1e-16, h, 500)

for i, s in enumerate(scenarios, start=1):

    A = s["A"]
    w = s["w"]
    color = s["color"]

    # optimum
    l_star, C_star = optimum(alpha, w, A)

    # hranica možností
    ax.plot(
        l_grid,
        budget(l_grid, w, A),
        color=color,
        lw=2
    )
    
    l_ext = np.linspace(h, h+8, 100)

    ax.plot(l_ext,
        budget(l_ext,w,A),
        color=color,
        lw=1.5,
        ls="--")

    # indiferenčná krivka
    ax.plot(
        l_grid,
        indifference_curve(
            l_grid,
            alpha,
            l_star,
            C_star
        ),
        color=color,
        ls=":"
    )

    # optimum
    ax.scatter(
        l_star,
        C_star,
        color=color,
        s=60,
        zorder=5
    )

    # pomocné čiary
    ax.plot(
        [0, l_star],
        [C_star, C_star],
        ls="--",
        lw=1,
        color=color,
        alpha=0.7
    )

    ax.plot(
        [l_star, l_star],
        [0, C_star],
        ls="--",
        lw=1,
        color=color,
        alpha=0.7
    )

    # označenia
    ax.text(
        -3,
        C_star,
        rf"$C_{i}$",
        color=color,
        va="center"
    )

    ax.text(
        l_star,
        -8,
        rf"$\ell_{i}$",
        color=color,
        ha="center"
    )

# ==================================================
# π-T
# ==================================================

A = scenarios[0]["A"]

ax.axhline(
    A,
    linestyle="--",
    color="black",
    alpha=0.7
)

ax.text(
    -3,
    A,
    r"$\pi-T$",
    va="center"
)

# ==================================================
# h
# ==================================================

ax.axvline(
    h,
    linestyle="--",
    color="black",
    alpha=0.7
)

ax.plot(
    [h, h],
    [0, A],
    color="blue",
    lw=2
)

ax.text(
    h,
    -8,
    r"$h$",
    ha="center"
)

# ==================================================
# Úprava osí
# ==================================================

C_max = max(
    budget(0, s["w"], s["A"])
    for s in scenarios
)

ax.set_xlim(0, h + 8)
ax.set_ylim(0, 300)

ax.set_xticks([])
ax.set_yticks([])

ax.set_xlabel(r"$\ell$")
ax.set_ylabel(r"$C$")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.tight_layout()
plt.show()
````

## Rozklad vplyvu zmeny mzdovej sadzby na substitučný a príjmový efekt. 

Vplyv zvýšenia mzdy na optimálne množstvo voľného času nie je
bezprostredne zrejmý. Vyššia mzda totiž pôsobí na rozhodovanie
spotrebiteľa prostredníctvom dvoch protichodných efektov.

Substitučný efekt vzniká preto, že zvýšenie mzdy zvyšuje
oportunitné náklady voľného času. Každá hodina voľného času znamená
stratu väčšieho množstva spotreby ako pred zvýšením mzdy. Spotrebiteľ
preto nahrádza časť voľného času spotrebou a prácou.

Substitučným efektom klesne voľný čas a zvýši sa spotreba. Keďže voľný čas klesne, ponuka práce narastie. V modeli rozhodovania medzi voľným časom a spotrebou sa tento efekt nazýva **substitučný efekt medzi voľným časom a spotrebou**.

Príjmový efekt vzniká preto, že vyššia mzda zvyšuje príjem
spotrebiteľa. Ak sú voľný čas a spotreba normálne statky, spotrebiteľ využije
časť dodatočného príjmu na zvýšenie voľného času a spotreby.

Príjmovým efektom narastú voľný čas aj spotreba a ponuka práce klesne.

Substitučný a príjmový efekt pôsobia na voľný čas opačným smerom. Zatiaľ čo
substitučný efekt vedie k poklesu voľného času, príjmový efekt vedie
k jeho rastu.

:name: tab_rozklad_lC

| Efekt | Voľný čas $\ell$ | Spotreba $C$ | Ponuka práce $N=h-\ell$ |
|--------|--------|--------|--------|
| Substitučný efekt (SE) | klesne | narastie | narastie |
| Príjmový efekt (PE) | narastie | narastie | klesne |
| Celkový efekt (CE) | nejednoznačný | narastie | nejednoznačný |
```

Na určenie výsledného vplyvu zvýšenia mzdy na voľný čas a ponuku práce
možno celkový efekt zmeny mzdy na substitučný a príjmový
efekt rozložiť pomocou Hicksovho rozkladu na {numref}`graf_rozklad_lC`.

````python
import numpy as np
import matplotlib.pyplot as plt

# ==================================================
# Parametre modelu
# ==================================================

h = 24
alpha = 1/3

scenarios = [
    {"A": 24, "w": 3, "color": "blue"},
    {"A": 24, "w": 10, "color": "red"},
]

# ==================================================
# Funkcie
# ==================================================

def budget(l, w, A):
    """
    Hranica spotrebných možností

    C = w(h-l) + A
    """
    return w * (h - l) + A


def optimum(alpha, w, A):
    """
    Cobb-Douglas

    U(C,l)=C^(1-alpha) l^alpha
    """

    M = w * h + A

    l_star = alpha * M / w
    C_star = (1 - alpha) * M

    return l_star, C_star


def indifference_curve(l, alpha, l_star, C_star):
    """
    Indiferenčná krivka pre úroveň užitočnosti
    prechádzajúcu optimom.
    """

    U = C_star**(1-alpha) * l_star**alpha

    return (U / (l**alpha))**(1/(1-alpha))


# ==================================================
# Graf
# ==================================================

fig, ax = plt.subplots(figsize=(10, 6))

l_grid = np.linspace(1e-16, h, 500)

for i, s in enumerate(scenarios, start=1):

    A = s["A"]
    w = s["w"]
    color = s["color"]

    # optimum
    l_star, C_star = optimum(alpha, w, A)

    # hranica možností
    ax.plot(
        l_grid,
        budget(l_grid, w, A),
        color=color,
        lw=2
    )
    
    l_ext = np.linspace(h, h+8, 100)

    ax.plot(l_ext,
        budget(l_ext,w,A),
        color=color,
        lw=1.5,
        ls="--")

    # indiferenčná krivka
    ax.plot(
        l_grid,
        indifference_curve(
            l_grid,
            alpha,
            l_star,
            C_star
        ),
        color=color,
        ls=":"
    )

    # optimum
    ax.scatter(
        l_star,
        C_star,
        color=color,
        s=60,
        zorder=5
    )

    # pomocné čiary
    ax.plot(
        [0, l_star],
        [C_star, C_star],
        ls="--",
        lw=1,
        color=color,
        alpha=0.7
    )

    ax.plot(
        [l_star, l_star],
        [0, C_star],
        ls="--",
        lw=1,
        color=color,
        alpha=0.7
    )

    # označenia
    ax.text(
        -3,
        C_star,
        rf"$C_{i}$",
        color=color,
        va="center"
    )

    ax.text(
        l_star,
        -8,
        rf"$\ell_{i}$",
        color=color,
        ha="center"
    )

# ==================================================
# π-T
# ==================================================

A = scenarios[0]["A"]

ax.axhline(
    A,
    linestyle="--",
    color="black",
    alpha=0.7
)

ax.text(
    -3,
    A,
    r"$\pi-T$",
    va="center"
)

# ==================================================
# h
# ==================================================

ax.axvline(
    h,
    linestyle="--",
    color="black",
    alpha=0.7
)

ax.plot(
    [h, h],
    [0, A],
    color="blue",
    lw=2
)

ax.text(
    h,
    -8,
    r"$h$",
    ha="center"
)

# ==================================================
# Úprava osí
# ==================================================

C_max = max(
    budget(0, s["w"], s["A"])
    for s in scenarios
)

ax.set_xlim(0, h + 8)
ax.set_ylim(0, 300)

ax.set_xticks([])
ax.set_yticks([])

ax.set_xlabel(r"$\ell$")
ax.set_ylabel(r"$C$")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

# ==================================================
# Hicksova kompenzácia
# ==================================================

# pôvodný scenár
A0 = scenarios[0]["A"]
w0 = scenarios[0]["w"]

l0, C0 = optimum(alpha, w0, A0)

# pôvodná užitočnosť
U0 = C0**(1-alpha) * l0**alpha

# nová mzda
w1 = scenarios[1]["w"]

# Hicksovo optimum

ls = U0 / (
    (w1 * (1-alpha)/alpha)**(1-alpha)
)

Cs = (
    w1
    * (1-alpha)/alpha
    * ls
)

# kompenzovaný nepracovný príjem

As = Cs - w1*(h-ls)

# ==================================================
# Hicksova hranica
# ==================================================

ax.plot(
    l_grid,
    budget(l_grid, w1, As),
    color="green",
    lw=2
)

ax.scatter(
    ls,
    Cs,
    color="green",
    s=60,
    zorder=6
)

ax.plot(
    [0, ls],
    [Cs, Cs],
    color="green",
    ls="--",
    lw=1
)

ax.plot(
    [ls, ls],
    [0, Cs],
    color="green",
    ls="--",
    lw=1
)

ax.text(
    ls,
    -8,
    r'$\ell_s$',
    color='green',
    ha='center'
)

ax.text(
    -3,
    Cs,
    r'$C_s$',
    color='green',
    va='center'
) 

plt.tight_layout()
plt.show()
````

## Ponuka práce

Teoretický aj empirický výskum spravidla predpokladá, že ponuka práce {eq}`r_ponuka_prace` je rastúcou funkciou mzdovej sadzby.

Ponuka práce sa po zvýšení mzdovej sadzby *{ref}`cp-label`* zvýši, iba ak substitučný efekt medzi voľným časom a spotrebou je vyšší ako príjmový. 

Preto teoretický a empirický výskum predpokladá, že pri bežných
hodnotách mzdy prevláda substitučný efekt nad príjmovým efektom.
Zvýšenie mzdy preto vedie k rastu ponuky práce a k poklesu spotreby
voľného času.

Predpoklad, že substitučný efekt medzi voľným časom a spotrebou je vyšší ako príjmový bude dôležitý aj v nasledujúcich kapitolách, pretože ho
využívajú mnohé kompaktné modely všeobecnej rovnováhy a moderné
makroekonomické modely trhu práce.

Graf ponuky práce je na {numref}`graf_ponuka_prace`. Jej priebeh je rastúci.

Na ponuku práce má vplyv aj nepracovný príjem $\pi-T$. Zvýšením zisku alebo znížením paušálnej dane sa zvýšia spotreba a voľný čas a ponuka práca klesne. Dôsledkom toho sa ponuka práce posunie smerom dole z modrej čiary do červenej.

Tvar ponuky práce je dôsledkom zvolenej funkcie užitočnosti. V našom modeli je ponuka práce rastúca a konkávna. Teória spotrebiteľa však vo všeobecnosti neurčuje, či má byť ponuka práce konkávna alebo konvexná. Tvar ponuky práce závisí od preferencií spotrebiteľa a od špecifikácie modelu.

## Riziko-averzné preferencie

V {ref}`priklad_CD_lC` sú spotrebiteľove preferencie opísané Cobbovou-Douglasovou
funkciou užitočnosti s konštantným výnosomo z rozsahu

```{math}
:label: r_CD_lC
u(\ell,C)
=
\ell^\alpha C^{1-\alpha}.
```

V príklade je preukázané, že substitučný efekt medzi voľným časom a spotrebou na voľný čas závisí od nepracovného príjmu $\pi-T$. Motiváciou tejto časti je predstaviť preferencie, pri ktorých možno
rastúcu ponuku práce odvodiť pri slabších predpokladoch než v prípade
Cobbovej-Douglasovej funkcie užitočnosti. Takéto preferencie by zodpobvedali súčasnému ekonomickému poznaniu.

V mnohých makroekonomických modeloch sa používa zápis

```{math}
:label: CRRA_lC
u(\ell,C,)=\ell + \frac{C^\gamma}{\gamma},\qquad\gamma \neq 0.
```

```{Admonition} Poznámka

Keďže logaritmus je rastúca funkcia, maximalizácia Cobbovej-Douglasovej
funkcie užitočnosti

$$
u(\ell,C,)
=
\ell^\alpha C^{1-\alpha}
$$

vedie k rovnakému optimálnemu rozhodnutiu ako maximalizácia jej
logaritmu

$$
\ln u(\ell,C,)
=
\alpha\ln\ell
+
(1-\alpha)\ln C.
$$

Preferencie možno preto ekvivalentne opísať funkciou

$$
u(\ell,C,)
=
\alpha\ln\ell
+
(1-\alpha)\ln C.
$$

Logaritmická funkcia užitočnosti je špeciálnym prípadom všeobecnejšej
triedy preferencií

$$
u(\ell,C,)
=
\alpha\ln\ell
+
(1-\alpha)\frac{C^\gamma-1}{\gamma},
\qquad \gamma\neq 0,
$$

pre ktorú pri

$$
\gamma\rightarrow 0
$$

získame

$$
u(\ell,C,)
=
\alpha\ln\ell
+
(1-\alpha)\ln C.
$$
```

Ak sa táto funkcia používa v prostredí neistoty, zodpovedá konkávnej
funkcii užitočnosti a riziko-averznému spotrebiteľovi.

Riziko-averzný spotrebiteľ 
: uprednostňuje istú spotrebu pred neistou spotrebou s
rovnakou očakávanou hodnotou.

Rizikovú averziu možno v {eq}`CRRA_lC` vyjadriť parametrom $\gamma$. Porovnajme spotrebiteľovu reakciu na extrémne nízku hodnotu pri rôznych hodnotách $\gamma$.

Pri preferenciach {eq}`r_CD_lC`, 

$$
\gamma\rightarrow 0
$$

ak sa spotreba blíži k nule,

$$
C \rightarrow 0,
$$

platí

$$
u(\ell,C,)\rightarrow 0.
$$

Extrémne nízka spotreba je preto spojená s veľmi nízkou úrovňou
užitočnosti.

Pri preferenciach {eq}`CRRA_lC`, ak

$$
\gamma<0,
$$

potom pri spotrebe blížiacej sa k nule platí

$$
\frac{C^\gamma}{\gamma}
\rightarrow
-\infty.
$$

Extrémne nízka spotreba je preto penalizovaná extrémne nízkou
zápornou hodnotou funkcie užitočnosti.

Takéto preferencie vyjadrujú, že spotrebiteľ prikladá veľký význam
vyhnutiu sa veľmi nízkym úrovniam spotreby. Pokles spotreby z nízkej
úrovne spôsobuje oveľa väčší pokles užitočnosti než rovnaký pokles
spotreby z vysokej úrovne.

Platí, že čím nižšia je $\gamma$, tým vyššia je averzia k riziku.

Pre naše účely uvažujme preferencie

```{math}
:label: CRRA_N_rastie_w
u(\ell,C)
=
\ell
+
\frac{C^\gamma}{\gamma},
\qquad
0<\gamma<1.
```

Pri

$$
0<\gamma<1
$$

je funkcia užitočnosti zo spotreby stále konkávna, a teda zodpovedá
riziko-averzným preferenciám. Miera rizikovej averzie je však nižšia
ako pri Cobbových Douglasových preferenciach.

Pre tieto preferencie {ref}`priklad_CRRA` odvodzuje redukovaný tvar
ponuky práce ako rastúcej funkcie mzdovej sadzby pri slabších
predpokladoch než v prípade Cobbovej-Douglasovej funkcie užitočnosti.

Takéto preferencie sa často používajú v modernej makroekonómii,
pretože umožňujú zachovať konkávnosť funkcie užitočnosti zo spotreby,
a teda aj rizikovú averziu spotrebiteľa, pričom súčasne vedú k
jednoduchej a jednoznačne rastúcej ponuke práce.

### Príklad: Ponuka práce pri Cobbovej-Douglasovej funkcii užitočnosti

Predpokladajme, že spotrebiteľove preferencie sú dané Cobbovou-Douglasovou funkciou užitočnosti {eq}`r_CD_lC`

$$
u(\ell,C)
=
\ell^\alpha C^{1-\alpha}.
$$

1. Odvoďte optimálny voľný čas, spotrebu a ponuku práce.
2. Preskúmajte vplyv zmeny nepracovného príjmu $(\pi-T)$ na voľný čas, spotrebu a ponuku práce.
3. Preskúmajte vplyv zmeny mzdy $w$ na voľný čas, spotrebu a ponuku práce.
4. Preukážte, že vplyv mzdovej sadzby závisí od $(\pi-T)$.

#### Riešenie

**Redukovaný tvar**

Podmienka prvého rádu je

$$
MRS(\ell,C)=
w.
$$

Pre Cobbovu-Douglasovu funkciu užitočnosti platí

$$
MU_\ell=
\alpha
\ell^{\alpha-1}
C^{1-\alpha},
$$

$$
MU_C=
(1-\alpha)
\ell^\alpha
C^{-\alpha}.
$$

Preto

$$
MRS(\ell,C)=
\frac{MU_\ell}{MU_C}=
\frac{\alpha}{1-\alpha}
\frac{C}{\ell}.
$$

Podmienku optima možno zapísať v tvare

$$
\frac{\alpha}{1-\alpha}
\frac{C}{\ell}=
w,
$$

odkiaľ získame

$$
C=
\frac{1-\alpha}{\alpha}
w\ell.
$$

Po dosadení do rozpočtového obmedzenia

$$
\frac{1-\alpha}{\alpha}
w\ell=
w(h-\ell)+\pi-T
$$

a po úprave

$$
w\ell=
\alpha wh+\alpha(\pi-T).
$$

Optimálny voľný čas je

```{math}
:label: r_volny_cas_CD
\ell^*=\alpha h+\alpha\frac{\pi-T}{w}.
```

Optimálna spotreba je

```{math}
:label: spotreba_lC_CD
C^*=
(1-\alpha)
\bigl(
wh+\pi-T
\bigr).
```

Ponuka práce je

$$
N^{s*}=
h-\ell^*.
$$

Po dosadení za $\ell^*$ dostaneme

```{math}
:label: ponuka_prace_CD
N^{s*}
=
(1-\alpha)h
-
\alpha\frac{\pi-T}{w}.
```

---

**Komparatívna statika**

*Vplyv zmeny nepracovného príjmu*

Pre optimálny voľný čas platí

$$
\frac{\partial \ell^*}
{\partial(\pi-T)}
=
\frac{\alpha}{w}
>
0.
$$

Zvýšenie nepracovného príjmu zvyšuje spotrebu voľného času, *ceteris paribus*.

Pre optimálnu spotrebu platí

$$
\frac{\partial C^*}
{\partial(\pi-T)}
=
1-\alpha
>
0.
$$

Zvýšenie nepracovného príjmu zvyšuje spotrebu, *ceteris paribus*.

Pre ponuku práce platí

$$
\frac{\partial N^{s*}}
{\partial(\pi-T)}
=
-\frac{\alpha}{w}
<
0.
$$

Zvýšenie nepracovného príjmu znižuje ponuku práce, *ceteris paribus*.

---

**Vplyv zmeny mzdovej sadzby**

Pre optimálnu spotrebu platí

$$
\frac{\partial C^*}
{\partial w}
=
(1-\alpha)h
>
0.
$$

Zvýšenie mzdy vedie k rastu spotreby, *ceteris paribus*.

Pre optimálny voľný čas platí

$$
\frac{\partial \ell^*}
{\partial w}
=
-\alpha
\frac{\pi-T}{w^2}.
$$

Ak

$$
\pi-T>0,
$$

potom

$$
\frac{\partial \ell^*}
{\partial w}
<
0.
$$

Zvýšenie mzdy vedie k poklesu voľného času, *ceteris paribus*.

Pre ponuku práce platí

$$
\frac{\partial N^{s*}}
{\partial w}
=
\alpha
\frac{\pi-T}{w^2}
>
0.
$$

Ak

$$
\pi-T>0,
$$

potom

$$
\frac{\partial N^{s*}}
{\partial w}
>
0.
$$

Zvýšenie mzdy vedie k rastu ponuky práce, *ceteris paribus*, ak spotrebiteľov nepracovný príjem je kladný.

Voľný čas sa po zvýšení mzdovej sadzby zvýši, iba ak nepracovný príjem je kladný.

Ak je nepracovný príjem kladný, substitučný efekt vplyvu $w$ na voľný čas je vyšší ako príjmový. V takom prípade je ponuka práce rastúcou funkciou mzdovej sadzby.

Ak je nepracovný príjem záporný, substitučný efekt vplyvu $w$ na voľný čas je nižší ako príjmový.

Ak sa nepracovný príjem rovná 0, substitučný efekt vplyvu $w$ na voľný čas sa rovná príjmovému.

#### Python

````python
# Use python to express marginal rate of consumer's substitution and express the allocation.

l,C,alpha,pi,T,w,Ns,h = symbols('l,C,alpha,pi,T,w,Ns,h',positive=True)

U = l**alpha * C**(1-alpha)

print('MRS(l,C) = ', simplify(diff(U,l)/diff(U,C)))


kos = solve([diff(U,l)/diff(U,C) - w, #first FOC MRS = w
           w*(h-l) + pi -T - C, # second FOC, constraint is fulfilled
            Ns + l - h], 
          [l,C,Ns], # endogenous 
         dict = True)

kos = kos[0] if type(kos) is list else kos

print("spotrebiteľský kôš")
display(kos)

# komparatívna statika
print()
print("Vplyv nepracovného príjmu")
print()
print("Na voľný čas")
display( simplify( diff(kos[l],pi) ) )
print()
print("Na spotrebu")
display( -simplify( diff(kos[C],T) ) )
print()
print("Na ponuku práce")
display( simplify( diff(kos[Ns],pi) ) )
print()
print("Vplyv mzdovej sadzby")
print()
print("Na voľný čas")
display( simplify( diff(kos[l],w) ) )
print()
print("Na spotrebu")
display( simplify( diff(kos[C],w) ) )
print()
print("Na ponuku práce")
display( simplify( diff(kos[Ns],w) ) )
````
