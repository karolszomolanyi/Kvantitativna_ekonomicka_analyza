# Dynamický model zatvorenej výrobnej ekonomiky

V kapitole ({doc}`StaticModel`) sme predstavili statický model výrobnej ekonomiky. Spotrebiteľ sa rozhodoval medzi voľným časom a spotrebou, zatiaľ čo firma využívala prácu na výrobu spotrebného statku.

V kapitole ({doc}`Finan`) sme uviedli dynamický model s dvoma obdobiami. Spotrebiteľ sa v ňom rozhodoval medzi súčasnou a budúcou spotrebou.

Obidva modely bolo možné analyzovať v dvojrozmernom rozhodovacom priestore. Spotrebiteľ sa rozhodoval buď medzi voľným časom a spotrebou, alebo medzi súčasnou a budúcou spotrebou.

V realistickom prostredí však jednotlivci robia komplexnejšie rozhodnutia. Spotrebiteľ sa rozhoduje nielen o rozdelení časového fondu medzi voľný čas a prácu, ale aj o tom, akú časť svojho dôchodku spotrebuje a akú časť odloží na budúcu spotrebu. Firma využíva úspory domácností na tvorbu kapitálu, ktorý spolu s prácou vstupuje do výrobného procesu ako výrobný faktor.

Model využíva koncept reprezentatívneho spotrebiteľa a reprezentatívnej firmy. Spotrebiteľ volí spotrebu, úspory a voľný čas v oboch obdobiach tak, aby maximalizoval svoj blahobyt. Firma volí prácu v oboch obdobiach a investície tak, aby maximalizovala svoj zisk. Vláda vyberá paušálnu daň na financovanie vlastnej spotreby v oboch obdobiach.

Aby sme zachytili vzájomné väzby medzi rozhodnutiami spotrebiteľa a firmy, vytvoríme grafický rámec, v ktorom budeme analyzovať agregátnu ponuku a dopyt po práci a agregátnu ponuku a dopyt po výstupe. Produkčná funkcia pritom prepája trh práce s trhom produktov.

## Reprezentatívny spotrebiteľ

Spotrebiteľove preferencie charakterizuje funkcia užitočnosti s rozhodovacími premennými spotreba a voľný čas v oboch obdobiach

```{math}
:label: U_DGE

U(\ell,\ell',C,C')=u(\ell,C)+\beta u(\ell',C').
```

Označenie nadväzuje na kapitoly {doc}`StaticModel` a {doc}`Finan`, teda spotreba reprezentatívneho spotrebiteľa je $C$, jeho voľný čas je $\ell$. Apostrofom označujeme hodnoty v druhom období. Preferencie budúcnosti sú vyjadrené diskontným faktorom $0<\beta<1$. Funkcia $u$ je dvakrát diferencovateľná, rastúca a konkávna v $\ell,C$. 

Spotrebiteľov rozpočet v prvom období je

```{math}
:label: rozpocet1_DGE

C + S \le w(h-\ell)+\pi-T.
```

Spotrebiteľov príjem v prvom období tvorí príjem z práce $w(h-\ell)$ a zisk $\pi$, pričom $h$ je disponibilný čas a $w$ je mzdová sadzba. Zo svojho príjmu platí paušálnu daň $T$. Spotrebiteľ rozdelí svoj príjem medzi spotrebu a úspory $S$.

Spotrebiteľov rozpočet v druhom období je

```{math}
:label: rozpocet2_DGE

C' \le w'(h-\ell')+\pi'-T'+(1+r)S.
```

Kombináciou {eq}`rozpocet1_DGE` a {eq}`rozpocet2_DGE` získame životný rozpočet reprezentatívneho spotrebiteľa:

```{math}
:label: zivotny_rozpocet_DGE

C+\frac{C'}{1+r} \le w(h-\ell)+\frac{w'(h-\ell')}{1+r}+\pi+\frac{\pi'}{1+r}-T-\frac{T'}{1+r}.
```

Na pravej strane životného rozpočtu {eq}`zivotny_rozpocet_DGE` je súčasná hodnota životného bohatstva spotrebiteľa.

Reprezentatívny spotrebiteľ rieši úlohu výberom $\{\ell,\ell',C,C'\}$ maximalizovať funkciu užitočnosti {eq}`U_DGE` vzhľadom na životný rozpočet {eq}`zivotny_rozpocet_DGE`.

Podmienky prvého rádu sú:

```{math}
:label: MRS_w

MRS(\ell,C)=w,
```

```{math}
:label: MRSp_wp

MRS(\ell',C')=w',
```

```{math}
:label: MRS_1r

\frac{\frac{du(\ell,C)}{dC}} {\beta \frac{du(\ell',C')}{dC'}}=1+r,
```

a

```{math}
:label: zivotny_rozpocet_DGE_PPR

C+\frac{C'}{1+r} = w(h-\ell)+\frac{w'(h-\ell')}{1+r}+\pi+\frac{\pi'}{1+r}-T-\frac{T'}{1+r}.
```

Podmienka prvého rádu {eq}`MRS_w` je totožná s podmienkou prvého rádu správania spotrebiteľa v statickom prostredí {eq}`r_optimum_lC_w`.

Keďže reprezentatívny spotrebiteľ robí to isté rozhodnutie aj v druhom období, druhá podmienka prvého rádu {eq}`MRSp_wp` je totožná s prvou, iba s budúcimi hodnotami spotreby, voľného času a mzdovej sadzby.

Tretia podmienka prvého rádu {eq}`MRS_1r` opisuje vzťah medzi súčasnou a budúcou spotrebou a zodpovedá Eulerovej rovnici {eq}`r_Euler`.

Podľa poslednej podmienky {eq}`zivotny_rozpocet_DGE_PPR` je spotrebiteľove rozhodnutie na hranici spotrebiteľských možností. Spotrebiteľ v priebehu života využije všetky dostupné zdroje na financovanie spotreby v oboch obdobiach.

Predpokladáme, že podmienky prvého rádu {eq}`MRS_w`--{eq}`zivotny_rozpocet_DGE_PPR` sú nutnými a postačujúcimi podmienkami optimality.

## Ponuka práce

Spotrebiteľova ponuka práce je $(h-l)$. Podľa podmienok prvého rády správania spotrebiteľa {eq}`MRS_w` až {eq}`zivotny_rozpocet_DGE_PPR`, na ponuku práce vplývajú nasledovné premenné:

- **Mzdová sadzba** $w$: Podľa {eq}`MRS_w`, ak substitučný efekt zmeny mzdovej sadzby prevláda nad príjmovým efektom, ponuka práce je rastúcou funkciou mzdovej sadzby (pozri {ref}`ponuka_prace`).
- **Úroková miera** $r$: V kapitole ({doc}`Dynam`) sme ukázali, aký je {ref}`kap_r_dynam` na súčasnú a budúcu spotrebu. Voľný čas tiež interpretujeme ako  spotrebný statok. Dynamický vplyv úrokovej miery je rovnaký na voľný čas ako na spotrebu. Vplyv úrokovej miery možno na spotrebu rozložiť na dynamický substitučný efekt a príjmový efekt. Dynamickým substitučným efektom sa zvýšením úrokovej miery súčasná spotreba spotrebiteľa zníži. Príjmový efekt sa líši podľa toho, či spotrebiteľ je veriteľ alebo dlžník. Po zvýšení $r$ veriteľ príjmovým efektom súčasnú spotrebu zvýši, dlžník zníži. Predpokladajme, že dynamický substitučný efekt je väčší ako príjmový. Potom sa zvýšením úrokovej miery zníži súčasná spotreba. Analogicky sa dynamickým substitučným efektom zníži voľný čas. Ak sa zvýšením $r$ zníži voľný čas, ponuka práce sa pri pevnom časovom fonde spotrebiteľa zvýši.
- **Životné bohatstvo** $we$: Ak sa spotrebiteľovi zvýši životné bohatstvo, spotrebiteľ príjmovým efektom zvýši spotrebu aj voľný čas. Ak sa zvýši voľný čas, ponuka práce klesne.

```{admonition} Všimite si
V modeloch výrobnej ekonomiky predpokladáme, že substitučný efekt medzi voľným časom a spotrebou vplyvu mzdovej sadzby na spotrebu a voľný čas prevyšuje príjmový efekt a dynamický substitučný efekt vplyvu úrokovej miery na súčasnú spotrebu a súčasný voľný čas prevyšuje príjmový efekt.
```

Ponuku práce možno zapísať diferencovateľnou funkciou

```{math}
:label: r_Ns_DGE

N^s=N^s(w,r,we),
```

pričom funkcia $N^s$ je rastúca v $w$ a $r$ a klesajúca v $we$.

Ponuka práce je zobrazená na {numref}`graf_NS_w_DGE` a {numref}`graf_NS_we_DGE`. Ponuka práce rastie s $w$. Ak sa zvýši úroková miera, čiara ponky práce na sa {numref}`graf_NS_w_DGE` posunie doprava. Naopak, ak sa spotrebiteľove životné bohatstvo zvýši, čiara ponuky práca sa na {numref}`graf_NS_we_DGE` posunie doľava.

````python
import matplotlib.pyplot as plt
import grafy
a=0 # intercept Ns
a2=-0.5
b=1 # sklon Ns
rozpatie=1.8
xmin=1-rozpatie/2
xmax=1+rozpatie/2
def Ns(a,b,w):
    return a+b*w
grafy.set_theory_style()
fig,ax=plt.subplots()
grafy.nakresli_funkciu(ax,lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$")
grafy.nakresli_funkciu(ax,lambda w:Ns(a2,b,w),xmin-a2,xmax,nazov="$N^s(w,r_2,we)$",color="red")
ax.legend()
grafy.ekonomicke_osi(ax,"$N^s$","$w$")
plt.show()
````

## Spotrebiteľov dopyt po výstupe

Podľa podmienok prvého rádu správania spotrebiteľa {eq}`MRS_w` až {eq}`zivotny_rozpocet_DGE_PPR` ovplyvňujú spotrebiteľov dopyt po súčasných spotrebných statkoch najmä úroková miera a životné bohatstvo.

- **Úroková miera** $r$: Ak sa zvýši úroková miera a dynamický substitučný efekt preváži nad príjmovým efektom, spotrebiteľ obmedzí súčasnú spotrebu v prospech budúcej spotreby. Dopyt po súčasných spotrebných statkoch preto klesá.

- **Životné bohatstvo** $we$: Ak sa zvýši životné bohatstvo, spotrebiteľ príjmovým efektom zvýši súčasnú aj budúcu spotrebu. Dopyt po súčasných spotrebných statkoch preto rastie.

Spotrebiteľov dopyt po výstupe zapíšeme funkciou súčasného výstupu a úrokovej miery

```{math}
:label: r_Cd_DGE

C^d=C^d(r,we),
```

pričom funkcia $C^d$ je klesajúca v $r$ a rastúca v $we$.

Spotrebiteľov dopyt po výstupe je zobrazený na {numref}`graf_Cd_DGE`. Zvýšenie životného bohatstva posúva na krivku dopytu po spotrebných statkoch doprava.

## Reprezentatívna firma

Reprezentatívna firma vyrába statok pomocou práce a kapitálu. V prvom období je zásoba kapitálu $K$ predeterminovaná. Firma rozhoduje o práci v oboch obdobiach $N$ a $N'$ a o investíciách $I$ v prvom období, ktorými sa zvýši zásoba kapitálu v druhom období $K'$.

Produkčná funkcia v prvom období je

```{math}
:label: Y1_DGE

Y=AF(K,N).
```
Produkčná funkcia v druhom období je

```{math}
:label: Y2_DGE

Y'=A'F(K',N').
```

Pri vyjadrení zásoby kapitálu v ďalšom období využívame rovnaký princíp ako v Solowovom modeli rastu ({doc}`UvodRast`). Zásoba kapitálu sa zvyšuje investíciami, pričom časť existujúceho kapitálu sa v dôsledku opotrebenia znehodnotí. Parameter $\delta$ vyjadruje mieru znehodnotenia kapitálu. Hrubé investície $I$ zahŕňajú čisté investície aj investície do obnovy existujúceho kapitálu. Kapitál v ďalšom období je

```{math}
:label: K2_DGE
K'=(1-\delta)K+I
```

Zisk firmy v prvom období je rozdiel medzi príjmom $Y$ a nákladmi. Firma v prvom období má náklady z práce a náklady z investícií, platí

```{math}
:label: pi1_DGE

\pi = Y - wN - I.
```

V druhom období firma neuskutočňuje ďalšie investície. Na konci obdobia jej však zostáva neznehodnotená časť kapitálu \((1-\delta)K'\), ktorá predstavuje zostatkovú hodnotu kapitálu. Zisk firmy v druhom období je

```{math}
:label: pi2_DGE

\pi' = Y' - w'N' + (1-\delta)K'.
```

Reprezentatívna firma vyberie takú prácu v prvom a druhom období a investície, aby maximalizovala súčasnú hodnotu zisku

```{math}
:label: max_firma_DGE

\pi_\max = \max_{N,N',K'}\left(\pi+\frac{\pi'}{1+r}\right)
```


za podmienok {eq}`Y1_DGE`, {eq}`Y2_DGE`, {eq}`K2_DGE`.
Podmienky prvého rádu sú

```{math}
:label: MPN_w_DGE

MPN=w,
```

```{math}
:label: MPNp_wp_DGE

MPN'=w'
```

a

```{math}
:label: MPK_r_DGE

MPK'-\delta=r.
```

Podľa prvých dvoch podmienok firma zamestnáva pracovníkov dovtedy, kým sa mzdová sadzba rovná hraničnému produktu práce. Posledná podmienka určuje optimálnu zásobu kapitálu. Firma investuje dovtedy, kým sa budúci čistý hraničný produkt kapitálu, teda hraničný produkt kapitálu znížený o znehodnotenie kapitálu, rovná úrokovej miere.

## Dopyt po práci

Dopyt po práci je vyjadrený podmienkou {eq}`MPN_w_DGE`, podľa ktorej sa hraničný produkt práce rovná mzdovej sadzbe. Na dopyt po práci vplývajú nasledovné premenné:

- **Mzdová sadzba** $w$: Hraničný produkt práce je klesajúcou funkciou množstva práce. Ak sa zvýši mzdová sadzba, firma môže podmienku {eq}`MPN_w_DGE` splniť iba znížením pracovného dopytu. Dopyt po práci preto klesá.

- **Celková produktivita faktorov v prvom období** $A$: Zvýšenie celkovej produktivity faktorov zvyšuje hraničný produkt práce pri každej úrovni pracovného dopytu. Pri nezmenenej mzdovej sadzbe preto firma zvýši dopyt po práci.

- **Zásoba kapitálu** $K$: Vyššia zásoba kapitálu zvyšuje hraničný produkt práce. Pri nezmenenej mzdovej sadzbe preto firma zvýši dopyt po práci. Naopak, ak sa zásoba kapitálu zníži, dopyt po práci klesne.

Funkciu dopytu po práci môžeme zapísať v tvare

```{math}
:label: r_Nd_DGE

N^d=N^d(w,A,K),
```

pričom funkcia $N^d$ je klesajúca v $w$ a rastúca v $A$ a $K$.

Dopyt po práci je zobrazený na {numref}`graf_Nd_w_DGE` a {numref}`graf_Nd_K_DGE`. Zvýšenie celkovej produktivity faktorov posúva krivku dopytu po práci doprava. Naopak, pokles zásoby kapitálu posúva krivku dopytu po práci doľava.

Keďže v prvom období je zásoba kapitálu $K$ predeterminovaná, firma môže na zmeny mzdovej sadzby reagovať iba úpravou pracovného dopytu.

````python
c=2. 
d=-1.
c2=2.5
rozpatie=1.8
xmin=1-rozpatie/2
xmax=1+rozpatie/2
def Nd(c,d,w):
    return c+d*w
grafy.set_theory_style()
fig,ax=plt.subplots()
grafy.nakresli_funkciu(ax,lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A_1,K)$")
grafy.nakresli_funkciu(ax,lambda w:Nd(c2,d,w),xmin,xmax+0.5,nazov="$N^d(w,A_2,K)$",color="red")
ax.legend()
grafy.ekonomicke_osi(ax,"$N^d$","$w$")
plt.show()
````

## Investičný dopyt

Investičný dopyt je vyjadrený podmienkou {eq}`MPK_r_DGE`, podľa ktorej sa budúci čistý hraničný produkt kapitálu rovná úrokovej miere. Na investičný dopyt vplývajú nasledovné premenné:

- **Úroková miera** $r$: Hraničný produkt kapitálu je klesajúcou funkciou zásoby kapitálu. Ak sa zvýši úroková miera, firma môže podmienku {eq}`MPK_r_DGE` splniť znížením stavu kapitálu v druhom období. Investičný dopyt klesá.

- **Celková produktivita faktorov v druhom období** $A'$: Zvýšenie budúcej celkovej produktivity faktorov zvyšuje budúci hraničný produkt kapitálu pri každej úrovni budúcej zásoby kapitálu. Pri nezmenenej úrokovej miera sa preto firma usiluje o vyššiu zásobu kapitálu v druhom období a zvýši investičný dopyt.

- **Zásoba kapitálu** $K$: Vzťah medzi investíciami a súčasnou zásobou kapitálu je {eq}`K2_DGE`. Pri danej požadovanej budúcej zásobe kapitálu vyššia súčasná zásoba kapitálu znižuje potrebu nových investícií. Naopak, pokles súčasnej zásoby kapitálu zvyšuje investičný dopyt.

Funkciu investičného dopytu môžeme zapísať v tvare

```{math}
:label: r_Id_DGE

I^d=I^d(r,A',K),
```

pričom funkcia $I^d$ je klesajúca v $r$ a $K$ a rastúca v $A'$.

Investičný dopyt je zobrazený na {numref}`graf_Id_DGE`. Zvýšenie budúcej celkovej produktivity faktorov a pokles stavu kapitálu posúva krivku investičného dopytu doprava.

### Investície, asymetrické informácie a finančná kríza

Pri odvodení investičného dopytu sme predpokladali, že firmy si môžu bez problémov požičiavať za trhovú úrokovú mieru $r$. V praxi však finančné trhy často nefungujú dokonale. Dôležitým zdrojom nedokonalostí sú asymetrické informácie, ktorým sme sa venovali v kapitole ({doc}`Finan`), konkrétne v časti {ref}`kap_urokova_prirazka`.

Asymetrická informácia vzniká vtedy, keď jedna strana finančnej transakcie disponuje lepšími informáciami než druhá. V prípade investícií firmy spravidla poznajú kvalitu svojich projektov lepšie než banky alebo ostatní veritelia. Finančné inštitúcie preto nedokážu s istotou odlíšiť perspektívne firmy od firiem s vysokým rizikom zlyhania.

Predpokladajme, že domácnosti ukladajú svoje úspory v bankách za trhovú úrokovú mieru $r$. Banky následne poskytujú úvery firmám. Keďže však časť dlžníkov nemusí svoje záväzky splatiť, banky požadujú od dlžníkov vyššiu úrokovú mieru

```{math}
:label: r_rx

r^L=r+r_p,
```

kde $r_p$ predstavuje úrokovú prirážku.

Pre firmu, ktorá financuje investície pomocou úveru, už nie sú nákladom investícií iba trhové úroky $r$, ale úroky vrátane úrokovej prirážky $r+r_p$. Podmienku prvého rádu {eq}`MPK_r_DGE` prepíšeme do tvau

$$
MPK'-\delta=r+r_p.
$$

Po úprave získame

```{math}
:label: r_MPKx_DGE

MPK'-\delta-r_p=r.
```

Úroková prirážka pôsobí podobne ako dodatočné zdanenie investícií. Pri danej úrokovej miere znižuje čistý výnos z kapitálu a vedie firmy k nižším investíciám.

Zvýšenie úrokovej prirážky posúva krivku investičného dopytu doľava. Takáto situácia môže nastať počas finančnej krízy, keď banky očakávajú vyšší podiel nesplatených úverov a sprísňujú úverové podmienky. Pokles investičného dopytu následne znižuje agregátny dopyt a hospodársku aktivitu.

Mechanizmus predstavuje jeden z kanálov, ktorým sa problémy finančného sektora prenášajú do reálnej ekonomiky. Rast neistoty a rizikových prirážok vedie k poklesu investícií, produkcie a zamestnanosti aj v prípade, ak sa základná trhová úroková miera výrazne nezmení.

Vplyv zvýšenia úrokovej prirážky na investičný dopyt je zobrazený na {numref}`graf_Id_x_DGE`.

````python
c=2. 
d=-1.
c2=1.5
rozpatie=1.8
xmin=1-rozpatie/2
xmax=1+rozpatie/2
grafy.set_theory_style()
fig,ax=plt.subplots()
grafy.nakresli_funkciu(ax,lambda r:c+d*r,xmin,xmax,nazov="$I^d(r+r_{p1},A,K)$")
grafy.nakresli_funkciu(ax,lambda r:c2+d*r,xmin,xmax-0.5,nazov="$I^d(r+r_{p2},A,K)$",color="red")
ax.legend()
grafy.ekonomicke_osi(ax,"$I^d$","$r$")
plt.show()
````

## Vláda

Vláda v oboch obdobiach vyberá paušálne dane $T,T'$ na financovanie vlastnej spotreby $G,G'$. V prvom období môže emitovať vládne dlhopisy $B$ na financovanie rozpočtového deficitu $G-T$. V kapitole {doc}`Finan` sme odvodili životný rozpočet vlády {eq}`r_SHrozpocet_vlady`:

```{math}
:label: TGz_DGE

G+\frac{G'}{1+r}=T+\frac{T'}{1+r}.
```

## Konkurenčná rovnováha

(def_konkurencna_DGE_decentr)=

Konkurenčná rovnováha
: Pri danej spotrebe vlády $\{G,G'\}$ a technológii firmy $\{A,A'\}$ je konkurenčnou rovnováhou taká alokácia spotrebiteľa $\{\ell,\ell',C,C'\}$, alokácia firmy $\{N,N',Y,Y',K'\}$ a množina cien $\{w,w',r\}$, že

- spotrebiteľ rieši úlohu maximalizovať funkciu užitočnosti {eq}`U_DGE` vzhľadom na životný rozpočet {eq}`zivotny_rozpocet_DGE`,
- firma rieši úlohu {eq}`max_firma_DGE` za podmienok {eq}`Y1_DGE`, {eq}`Y2_DGE`,
- vláda spĺňa životný rozpočet {eq}`TGz_DGE` a 
- trh práce je v rovnováhe

$$
N=h-\ell,
$$

$$
N'=h-\ell',
$$

- trh produktov je v rovnováhe

$$
Y=C+I+G.
$$


Konkurenčnú rovnováhu možno charakterizovať sústavou podmienok prvého rádu spotrebiteľa {eq}`MRS_w` až {eq}`zivotny_rozpocet_DGE_PPR`, podmienok prvého rádu firmy {eq}`MPN_w_DGE` až {eq}`MPK_r_DGE` životného rozpočtu vlády {eq}`TGz_DGE`, produkčnou funkciou a výpočtom budúceho kapitálu {eq}`Y1_DGE` až {eq}`K2_DGE`.

## Dopyt po výstupe v súčasnom období

Dopyt po výstupe pozostáva z dopytu spotrebiteľov {eq}`r_Cd_DGE`, investícií {eq}`r_Id_DGE` a dopytu vlády $G$. Môžeme ho zapísať funkciou

```{math}
:label: r_Yd_DGE

Y^d(r,we,A',K,G)=C^d(r,we)+I^d(r,A',K)+G.
```

Keďže spotrebný aj investičný dopyt sú klesajúcimi funkciami úrokovej miery, dopyt po výstupe je tiež klesajúcou funkciou úrokovej miery.

Krivka funkcie dopytu po výstupe {eq}`r_Yd_DGE` sa na {numref}`graf_Yd_DGE` posunie doprava ak sa zvýši príjem $Y$, životné bohatstvo $we$ a budúca celková produktivita faktorov $A'$, alebo ako sa zníži predetermnovaný kapitál $K$.

- Ak sa zvýši príjem $Y$ alebo životné bohatvsto $we$, spotrebiteľ príjmovým efektom zvýši svoj dopyt. Životné bohatstvo sa môže zvýšiť
  - znížením daní v súčasnom alebo budúcom období $T,T'$ alebo
  - zvýšením budúceho výstupu $Y'$.

## Vplyv zvýšenia súčasných vládnych nákupov na konkurenčnú rovnováhu

Predpokladajme, že ekonomika je pred zmenou vládnych nákupov v rovnováhe. Na {numref}`graf_DGE_G` je rovnováha charakterizovaná priesečníkom kriviek dopytu a ponkuky na pracovnom trhu a trhu výstupu. Trh práce je v rovnováhe a rovnovážne hodnoty práce a mzdovej sadzby sú $[N^*_1,w^*_1]$. Rovnováha na trhu výstupu je $[Y^*_1,r^*_1]$.

Vládne nákupy majú priamy efekt na dopyt po výstupe $Y^d(r,we,A',K,G)$ a na životné bohatstvo spotrebiteľa $we$. Zmena životného bohatsva ďalej ovplyvní dopyt po výstupe $Y^d(r,we,A',K,G)$, ponuku práce $N^s(w,r,we)$ a ponuku výstupu $Y^s(r,A,we,K)$. Pohyby dopytu a ponuky výstupu ovplyvnia úrokovú mieru $r$ a cez ňu ďalej ponuku práce $N^s(w,r,we)$.

Aký je vplyv vládnych nákupov na životné bohatsvo spotrebiteľa?

Vláda financuje zvýšené nákupy zvýšením súčasných alebo budúcich daní $T,T'$. Platí, že ak sa súčasná hodnota daní zvýši o jednotku, zníži sa o jednotku životné bohatstvo spotrebiteľa;

```{math}
:label: r_we_G

\frac{\partial{we}}{\partial{G}}=-1.
``` 

Preskúmajme teraz, ako sa posunie dopyt po výstupe $Y^d(r,we,A',K,G)$ pri nezmenej úrokovej miere $r$.

Celkový vplyv vládnych nákupov na dopyt po výstupe $Y^d(r,we,A',K,G)$ je:

$$
\frac{\partial{Y^d(r^*_1)}}{\partial{G}}=\frac{\partial{C^d(r^*_1)}}{\partial{G}}+\frac{\partial{I^d(r^*_1)}}{\partial{G}}+\frac{\partial{G}}{\partial{G}}.
$$

Investičný dopyt $I^d(r,A',K)$ je nezávislý od vládnych nákupov $G$ aj od životného bohatstva $we$. Rovnicu možno prepísať do tvaru

$$
\frac{\partial{Y^d(r^*_1)}}{\partial{G}}=\frac{\partial{C^d(r^*_1)}}{\partial{we}}\frac{\partial{we}}{\partial{G}}+1.
$$

Zohľadnením zmeny životného bohatsva spôsobenej jednotkovou zmenou vládnych nákupov máme

$$
\frac{\partial{Y^d}(r^*_1)}{\partial{G}}=-\frac{\partial{C^d(r^*_1)}}{\partial{we}}+1.
$$

Ak sa spotrebiteľovi zvýši daň o jednotku, zníži sa mu o jednotku životné bohatstvo a disponibilný príjem. Reakcia spotrebného dopytu na zmenu príjmu je {ref}`def_mpc`. Preto prvý člen na pravej strane rovnice so znamienkom *mínus* môžeme zapísať v tvare

$$
\frac{\partial{C^d}(r^*_1)}{\partial{we}}=mpc,
$$

kde $mpc$ je hraničný sklon k spotrebe. Za predpokladu, že zmena vládnych nákupov je dočasná, podľa Friedmanovej hypotézy trvalého príjmu ({ref}`kap_friedman_trvaly`) je hraničný sklon k spotrebe kladný a menší ako jedna: $0<mpc<1$. Preto platí

```{math}
:label: r_DGE_G


\frac{\partial{Y^d}(r^*_1)}{\partial{G}}=1-mpc > 0;
```

zvýšením vládnych nákupov sa v pravom paneli {numref}`graf_DGE_G` dopyt po výstupe posunie doprava zo zelenej čiary do tmavčervenej.

Ak sa zníži spotrebiteľovo životné bohatstvo, spotrebiteľ príjmovým efektom zníži nielen spotrebný dopyt aj voľný čas -- zvýši ponuku práce. V ľavom paneli {numref}`graf_DGE_G` sa krivka ponuky práce posunie doprava z modrej do oranžovej.

Zvýšením ponuky práce sa zvýši ponuka výstupu. V pravom paneli {numref}`graf_DGE_G` sa krivka ponuky práce posunie doprava z modrej do červenej. 

Predpokladáme, že príjmový efekt zníženia životného bohatstva nie je dostatočne veľký, aby zvýšenie ponuky výstupu pokrylo zvýšenie dopytu po výstupe. Rovnováha na trhu výstupu sa zmení z $[Y^*_1,r^*_1]$ na $[Y^*_2,r^*_2]$. Rovnovážny výstup aj rovnovážna úroková miera vzrastú.

Aké sú fiškálne multiplikátory {ref}`def_multiplikatorG`?.

Výstup sa vplyvom zvýšenia vládnej spotreby zvýši. Pri nezmenenej úrokovej miere by sa dopyt po výstupe zvýšil o $(1-MPC)\Delta G$. Keďže sa v rovnováhe zvýši úroková miera, časť súkromného spotrebného a investičného dopytu je vytlačená. Rovnovážny výstup preto vzrastie o menej ako $(1-MPC)\Delta G$.

V pravom paneli {numref}`graf_DGE_G` sa dopyt po výstupe zvýši o $(1-MPC)\Delta G$. Bod označený ako $Y^d(r^*_1)$ predstavuje posun dopytu pri nezmenenej úrokovej miere $r^*_1$. Výstup sa zvýši menej, lebo úroková miera v skutočnosti narastie. Platí

$$
0<\frac{\partial Y}{\partial G}<1-MPC.
$$

Spotreba sa zníži dvoma efektmi:

- príjmovým efektom zníženia životného bohatstva a
- dynamickým substitučným efektom zvýšenia úrokovej miery.

Spotreba sa zvýšením vládnych nákupov zníži, *ceteris paribus*.

Ak sa zvýšia vládne nákupy, zvýši sa úroková miera a zníži sa investičný dopyt.

Investície sa zvýšením vládnych nákupov, *ceteris paribus*, znížia.

Rovnako ako v statickom modeli ({doc}`StaticModel`) aj v tomto modeli sa súkromný dopyt po výstupe zníži. Vládne nákupy **vytláčajú** súkromnú spotrebu $C$ a súkromné investície $I$.

## Vplyv sektorových šokov (simultánny pokles ponuky a dopytu po práci)

V časti {doc}`HladPrac` sme predstavili efekt vyhľadávania na trhu práce. Ukázali sme, že zvýšenim efektu vyhľdávania sa zníži miera nezamestnanosti, zatiaľ čo miera voľných miest sa nemení.

Veľký vplyv na efekt vyhľadávania má štruktúra dopytu a ponuky práce. Ak v ekonomike existuje kvalifikovaná pracovná sila v určitom výrobnom sektore, v ekonomike bude nízka miera nezamestnanosti, ak v nej má daný priemyselný sektor zastúpenie. Ak sa v ekonomike mení štruktúra priemyslu, mení sa efekt vyhľadávania na trhu práce. Hovoríme o **sektorovom šoku**.

Po páde komunistické režimu na Slovensku v 1989 sa podstatne zúžil zbrojársky priemysel. Táto transformácia predstavovala pokles efektu vyhľadávania, lebo pracujúci kvalifikovaní v strojárskom priemysle si ťažšie hľadali prácu. Na prelome tisícorčí bola miera nezamestnanosti na Slovensku takmer 20 %.

S príchodom automobilového priemyslu sa efekt vyhľadávania začal zväčšovať a v čase písania tejto kapitoly (jeseň 2026) je miera nezamestnanosti nižšia ako 6 %.

Sektorový šok možno v modeli výrobnej ekonomiky formulovať simutlánnym posunom ponuky a dopytu práce.

V ľavom paneli {numref}`graf_DGE_e` je zobrazený vplyv poklesu efektu vyhľadávania. Krivky ponuky a dopytu práce sa posúvajú doľava; z modrej do oranžovej a zo zelenej do tmavočervenej čiary.

Model možno opísať sústavou podmienok prvého rádu spotrebiteľa {eq}`MRS_w`--{eq}`MRS_1r` a podmienok prvého rádu firmy {eq}`MPN_w_DGE`--{eq}`MPK_r_DGE` spolu s ďalšími rovnicami špecifikujúcimi ostatné endogénne premenné.

Druhé obdobie je posledné obdobie modelu. Firmy už neinvestujú do rozširovania kapitálu pre ďalšie obdobie, pretože ďalšie obdobie neexistuje. Investície preto slúžia iba na obnovu opotrebovaného kapitálu v objeme $\delta K'$. Rovnováha na trhu výstupu v druhom období má preto tvar

```{math}
:label: r_Y2_DGE

Y'=C'+I'+G',
```

pričom

$$
I'=\delta K'
$$

Časové ohraničenie spotrebiteľa v oboch obdobiach je 

```{math}

h=N+\ell
```

```{math}

h=N'+\ell'
```

### Funkčné tvary a všetky rovnice modelu

Preferencie reprezentatívneho spotrebiteľa opisuje funkcia užitočnosti v jednoduchom tvare:

$$
U(C,l,C',l') = \ln{C} + \ln{\ell} + \beta(\ln{C'} + \ln{\ell'}),
$$

kde $\beta$ je diskontný faktor.

Produkčná funkcia {eq}`Y1_DGE`,{eq}`Y2_DGE` je v Cobbovom-Douglasovom tvare

```{math}
:label: CD1_DGE

Y = AK^{\alpha} N^{1-\alpha},
```

```{math}
:label: CD2_DGE

Y' = A'K'^{\alpha} N'^{1-\alpha}.
```

Dosaďme funkčné tvary a zapíšme sústavu všetkých rovníc. Podmienky prvého rádu spotrebiteľa {eq}`MRS_w`--{eq}`MRS_1r` sú

```{math}
:label: C_l_w_DGE

\frac{C}{\ell}=w,
```

```{math}
:label: Cp_lp_wp_DGE

\frac{C'}{\ell'}=w'
```

a

```{math}
:label: C_Cp_1r_DGE

\frac{C'}{\beta C}=1+r.
```

Podmienky prvého rádu firmy {eq}`MPN_w_DGE`--{eq}`MPK_r_DGE` sú

```{math}
:label: CD_MPN_DGE

(1-\alpha)AK^\alpha N^{-\alpha} = w,
```

```{math}
:label: CD_MPNp_DGE

(1-\alpha)A'K'^\alpha N'^{-\alpha} = w'
```

a

```{math}
:label: CD_MPKp_DGE

\alpha A'K'^{\alpha-1} N'^{1-\alpha} - \delta = r.
```

Investície v oboch obdobiach sú

```{math}
:label: r_I_DGE

I = K' - (1-\delta)K
```

a

```{math}
:label: r_Ip_DGE

I' = \delta K'.
```

Výstup {eq}`CD1_DGE` a {eq}`CD2_DGE`

$$
Y = AK^{\alpha} N^{1-\alpha}
$$

a

$$
Y' = A'K'^{\alpha} N'^{1-\alpha}
$$

Rovnováha na trhoch produktov {eq}`r_Y1_DGE` a {eq}`r_Y2_DGE` je

$$
C+I+G=Y
$$

a

$$
C'+I'+G'=Y'.
$$

a časové ohraničenia {eq}`r_cas_DGE` a {eq}`r_casp_DGE` sú

$$
\ell+N=h
$$

a

$$
\ell'+N'=h.
$$

Označme množinu endogénnych premenných symbolom

$$
\mathcal{Y}=
\{
C,C',\ell,\ell',
N,N',
K',
Y,Y',
I,I',
w,w',
r
\}.
$$

Model pozostáva zo 14 rovníc pre 14 neznámych z množiny $\mathcal{Y}$.

Konkurenčná rovnováha
: Pri danej spotrebe vlády $\{G,G'\}$ a technológii firmy $\{A,A'\}$ je konkurenčnou rovnováhou taká alokácia spotrebiteľa $\{\ell,\ell',C,C'\}$, alokácia firmy $\{N,N',K'\}$, zodpvedajúca množina endogénnych premenných $\{Y,Y',I,I'\}$, a množina cien $\{w,w',r\}$, ktoré zodpovedajú sústave rovníc pozostávajúcich z podmienok prvého rádu spotrebiteľa {eq}`C_l_w_DGE`--{eq}`C_Cp_1r_DGE` podmienok prvého rádu firmy {eq}`CD_MPN_DGE`--{eq}`CD_MPKp_DGE` investičnej funkcii {eq}`r_I_DGE` a {eq}`r_Ip_DGE`, produkčnej funkcii {eq}`CD1_DGE` a {eq}`CD2_DGE` rovnováhe na trhu výstupu {eq}`r_Y1_DGE` a {eq}`r_Y2_DGE` a časového ohraničeniu {eq}`r_cas_DGE` a {eq}`r_casp_DGE`.

Sústavu rovníc a endogénne premenné možno v jazyku Python špecifikovať nasledovnou funkciou

````python
def ppr(endog):

    C, Cp, l, lp, N, Np, Y, Yp, I, Ip, Kp, w, wp, r = endog

    return (

        C/l - w,                                            # MRS(l,C)=w
        Cp/lp - wp,                                         # MRS(l',C')=w'
        Cp/(beta*C) - (1+r),                                # MRS(C,C')=1+r

        (1-alpha)*A*K**alpha*N**(-alpha) - w,               # MPN=w
        (1-alpha)*Ap*Kp**alpha*Np**(-alpha) - wp,           # MPN'=w'
        alpha*Ap*Kp**(alpha-1)*Np**(1-alpha) - delta - r,   # MPK'-delta=r

        Y - A*K**alpha*N**(1-alpha),                        # Y=AF(K,N)
        Yp - Ap*Kp**alpha*Np**(1-alpha),                    # Y'=A'F(K',N')

        I - (Kp - (1-delta)*K),                             # I=K'-(1-delta)K
        Ip - delta*Kp,                                      # I'=delta*K'

        C + I + G - Y,                                      # C+I+G=Y
        Cp + Ip + Gp - Yp,                                  # C'+I'+G'=Y'

        N + l - h,                                          # N+l=h
        Np + lp - h                                         # N'+l'=h

    )
````

Pri analýze hospodárskych cyklov nás bude zaujímať aj súčasný priemerný produkt práce

```{math}
:label: APN_DGE

APN=\frac{Y}{N}.
```

### Kalibrácia modelu

Množina exogénnych premenných a parametrov je 

$$
\mathcal{X}=\{G,G',A,A',h,\alpha,\beta,\delta \}
$$

Model možno riešiť iba numericky, preto za exogénne premenné a parametre dosadíme hodnoty, ktoré najviac zodpovedajú skutočnej ekonomike. Tomuto procesu hovoríme {ref}`kalibracia`.

Pri kalibrácii budeme postupovať rovnako v príklade ({ref}`Numericky_priklad_lC`).

Hodnoty parametrov a exogénnych premenných $\mathcal{X}$ rozdelíme do dvoch skupín. V prvej skupine sú hodnoty, ktoré zvolíme tak, aby výsledná konkurenčná rovnováha mala želané výsledky. V druhej sú hodnoty, ktoré zodpovedajú priemerom pozorovaných dát v ekonomike.

**Normalizácia**

Pri voľbe parametrov v prvej skupine zvolíme normalizácu v ustálenom stave, podľa ktorej si želáme, aby sa vstupy a výstup rovnali $1$ v oboch obdobiach. Získame tak hodnoty $\{G,G',A,A',h,\beta\}$.

Definujme normalizovanú konkurenčnú rovnováhu ako hodnoty endogénnych premenných, ktoré zodpovedajú normalizácii.

Ustálený stav
: Definujme ustálený stav, v ktorom sa premenné dynamicky nemenia, platí:

$$
Y=Y',
$$

$$
K=K'
$$

a

$$
N=N'.
$$

Vzhľadom na charakter modelu v dvoch obdobiach ide o hypotetický ustálený stav, v ktorom sú veličiny v oboch obdobiach rovnaké.

Vzhľadom na požiadavku jednotkových vstupov a výstupu platí:

```{math}
:label: PFss

Y=Y'=K=K'=N=N'=1.
```

Ak predpokladáme, že v ustálenom stave sú vstupy a výstup jednotkové, nie je zložité vypočítať konkurenčnú rovnováhu v ustálenom stave. Ukážeme, že daný ustálený stav zodpovedá jednoznačným hodnotám prvej skupiny parametrov a exogénnych premenných.

Produkčná funkcia {eq}`CD1_DGE` a {eq}`CD2_DGE` vyhovuje požiadavke {eq}`PFss`, iba ak

```{math}
:label: Ass_DGE

A=A'=1.
```

Dosadením {eq}`PFss` a {eq}`Ass_DGE` do podmienok prvého rádu firmy vyjadrujúcich dopyt po práci {eq}`CD_MPN_DGE` a {eq}`CD_MPNp_DGE` získame mzdovú sadzbu v oboch obdobiach v ustálenom stave

```{math}
:label: wss_DGE

w=w'=(1-\alpha).
```

Dosaďme do časového ohraničenia {eq}`r_cas_DGE` a {eq}`r_casp_DGE` za prácu {eq}`PFss`. Získame voľný čas v ustálenom stave

```{math}
:label: lss_DGE

\ell=\ell'=h-1.
```

Dosaďme voľný čas {eq}`lss_DGE` a mzdovú sadzbu {eq}`wss_DGE` do podmienok prvého rádu spotrebiteľa {eq}`C_l_w_DGE` a {eq}`Cp_lp_wp_DGE`. Získame

$$
\frac{C}{h-1}=1-\alpha
$$

a

$$
\frac{C'}{h-1}=1-\alpha
$$

Po úprave platí


```{math}
:label: Css_DGE

C=C'=(1-\alpha)(h-1).
```

Za predpokladu jednotkového kapitálu v oboch obdobiach investície {eq}`r_I_DGE` a {eq}`r_Ip_DGE` sú

```{math}
:label: Iss_DGE

I=I'=\delta
```

Keďže $K=K'=1$, investície v oboch obdobiach slúžia iba na obnovu opotrebovaného kapitálu.

Spotrebu možno vyjadriť v podmienkách rovnováhy na trhu produktov {eq}`r_Y1_DGE` a {eq}`r_Y2_DGE`, keď z rovníc rovnováhy

$$
C+\delta+G=1
$$

a

$$
C'+\delta+G'=1.
$$

Platí

```{math}
:label: C_ss_DGE

C=C'=1-\delta-G.
```

Z Eulerovej rovnice {eq}`C_Cp_1r_DGE` vzhľadom na konštantnú spotrebu v ustálenom stave platí

```{math}
:label: Euler_DGE

\frac{1}{\beta}=1+r
```

Úrokovú miera v ustálenom stave tiež môžeme vyjadriť z podmienky prvého rádu firmy vyjadrujúce investičný dopyt {eq}`CD_MPKp_DGE`. Po dosadení jednotkových vstupov {eq}`PFss` platí

```{math}
:label: rss_DGE

r = \alpha - \delta.
```

Existujú dva spôsoby vyjadrenia spotreby {eq}`Css_DGE` a {eq}`C_ss_DGE` a úrokovej miery {eq}`Euler_DGE` a {eq}`rss_DGE`. Vďaka tomu možno nájsť jedinú hodnotu disponibilného času $h$ a diskontného faktora $\beta$, ktoré zodpovedajú normalizovanému ustálenému stavu.

Kombináciou {eq}`Euler_DGE` a {eq}`rss_DGE` získame

```{math}
:label: beta_DGE

\beta = \frac{1}{1+\alpha - \delta}.
```

Kombináciou {eq}`Css_DGE` a {eq}`C_ss_DGE` získame

$$
(1-\alpha)(h-1) = 1 - G - \delta.
$$

Odvoďme $h$

```{math}
:label: h_DGE

h = \frac{1 - G - \delta}{1-\alpha} + 1.
```

Normalizovanému ustálenému stavu vyhovujú exogénne premenné a parametre {eq}`Ass_DGE`, {eq}`beta_DGE` a {eq}`h_DGE`.

**Skutočné pozorovania**

Z podmienok rovnováhy na trhu produktov {eq}`r_Y1_DGE` a {eq}`r_Y2_DGE` vyplýva, že ak $Y=Y'=1$, potom $G$, $G'$, $C$ a $C'$ predstavujú podiely vládnych nákupov a spotreby na výstupe.

Za podiel vládnych nákupov na výstupe možno zvoliť jeho dlhodobý priemer $g$. Realistická hodnota je

$$
g=0.25.
$$

V rozvinutých ekonomikách zodpovedá podiel kapitálu na výstupe približne hodnote

$$
\alpha=\frac{1}{3},
$$

pričom priemerná miera znehodnotenia kapitálu je približne

$$
\delta=0.1.
$$

Po dosadení týchto hodnôt získame z rovníc {eq}`beta_DGE` a {eq}`h_DGE` hodnoty $\beta,h$.

Nasleduje kalibrácia modelu v jazyku Python.

````python
A = 1.      # celková produktivita faktorov
Ap = 1.     # budúca celková produktivita faktorov
alpha = 1/3 # podiel kapitálu na HDP
delta = 0.1 # miera znehodnotenia kapitálu
G = 1/4     # vládne nákupy (podiel n HDP)
Gp = 1/4
K = 1.      # predeterminovaný kapitál

beta = 1 / (1 + alpha - delta)  # diskontný faktor kalibrovaný tak, aby v ustálenom stave platilo Y=K=N=1
h = (1-G-delta)/(1-alpha) + 1   # disponibilný čas
````

Dosadením kalibrovaných hodnôt parametrov do sústavy rovníc a jej numerickým vyriešením získame konkurenčnú rovnováhu. Výsledok bude zodpovedať analyticky odvodenému ustálenému stavu {eq}`wss_DGE`--{eq}`Iss_DGE` a {eq}`rss_DGE`.

````python
C, Cp, l, lp, N, Np, Y, Yp, I, Ip, Kp, w, wp, r = fsolve(
    ppr,
    (1-delta-G, 1-delta-G, h-1, h-1, 1., 1.,1., 1.,delta, delta,1.,1-alpha,1-alpha,alpha-delta)
)

APN = Y / N # priemerný produkt práce vyjde 1

#zobraz výsledky
print("Všeobecná rovnováha")
print("Spotreba v prvom období", round(C,2), "v druhom", round(Cp,2))
print("Práca v prvom období", round(N,2), "v druhom", round(Np,2))
print("HDP v prvom období", round(Y,2), "v druhom", round(Yp,2))
print("Investície v prvom období",round(I,2), "v druhom", round(Ip,2))
print("Mzdová sadzba v prvom období", round(w,2), "v druhom", round(wp,2))
print("Úroková miera", round(r,2))
print("Priemerný produkt práce", round(APN,2))
````

Výsledné hodnoty nemusia zodpovedať reálnym úrovniam ekonomických veličín, pretože model je kalibrovaný tak, aby konkurenčná rovnováha zodpovedala normalizovanému ustálenému stavu. Spotreba, investície a vládne nákupy možno interpretovať ako podiely na HDP, keďže v ustálenom stave platí $Y=1$.

Kapitálový podiel na HDP sa v Cobbovej-Douglasovej produkčnej funkcii rovná parametru $\alpha$. V konkurenčnej rovnováhe sa tento podiel rozdeľuje medzi čistý výnos kapitálu $r$ a náklady na obnovu opotrebovaného kapitálu $\delta$, preto platí

$$
r+\delta=\alpha=0.33.
$$

To znamená, že približne tretina výstupu pripadá vlastníkovi kapitálu, pričom časť kompenzuje znehodnotenie kapitálu a zvyšok predstavuje čistý výnos kapitálu.

V nasledujúcej časti simulujeme dosahy rôznych šokov na hospodárstvo. V súlade s výkladom z predchádzajúcich častí pomocou modelu preskúmame, ako sa zmení hospodárstvo, ak sa

- zníži stav predeterminovaného kapitálu (katastrofa),

- zvýši úroková prirážka (finančný šok),

### Vplyv úrokovej prirážky v investičnej funkcii

V modeli upravíme podmienku prvého rádu firmy pre investičný dopyt {eq}`CD_MPKp_DGE`. Predpokladajme, že firmy financujú investície úverom a platia úrokovú mieru zvýšenú o úrokovú prirážku $r_p$. Podmienka optimálneho investičného rozhodnutia má potom tvar

$$
\alpha A'K'^{\alpha-1} N'^{1-\alpha} - \delta = r+r_p.
$$

Úroková prirážka zvyšuje náklady financovania investícií a pôsobí podobne ako zvýšenie trhovej úrokovej miery. Pri danej hodnote $r$ preto vedie k poklesu optimálnej zásoby kapitálu a investícií.

Zmenený model je potrebné opätovne špecifikovať v jazyku Python.

````python
def ppr_rp(endog):

    C, Cp, l, lp, N, Np, Y, Yp, I, Ip, Kp, w, wp, r = endog

    return (

        C/l - w,                                           
        Cp/lp - wp,                                       
        Cp/(beta*C) - (1+r),                              

        (1-alpha)*A*K**alpha*N**(-alpha) - w,               
        (1-alpha)*Ap*Kp**alpha*Np**(-alpha) - wp,           
        alpha*Ap*Kp**(alpha-1)*Np**(1-alpha) - delta - r - r_p,  # ZMENA, MPK'-delta=r+r_p

        Y - A*K**alpha*N**(1-alpha),                        
        Yp - Ap*Kp**alpha*Np**(1-alpha),                   

        I - (Kp - (1-delta)*K),                             
        Ip - delta*Kp,                                      

        C + I + G - Y,                                      
        Cp + Ip + Gp - Yp,                                  

        N + l - h,                                          
        Np + lp - h                                         

    )
````

Predpokladáme, že v ustálenom stave platí $r_p=0$. Zavedenie úrokovej prirážky preto nemení kalibráciu ani pôvodný ustálený stav modelu.

Podľa teórie {ref}`kap_DGE_rp` zvýšenie úrokovej prirážky znižuje rovnovážnu úrokovú mieru. Vplyv na ostatné endogénne premenné nie je jednoznačný, pretože výsledok závisí od relatívnej veľkosti príjmového efektu poklesu životného bohatstva a dynamického substitučného efektu vyvolaného poklesom úrokovej miery.

Preskúmajme tento vplyv pomocou numerickej simulácie.

````python
r_p=0.01

# nová konkurenčná rovnováha
C_r, Cp_r, l_r, lp_r, N_r, Np_r, Y_r, Yp_r, I_r, Ip_r, Kp_r, w_r, wp_r, r_r = fsolve(
    ppr_rp,
    (C, Cp, l, lp, N, Np, Y, Yp, I, Ip, Kp, w, wp, r)
)

APN_r = Y_r/N_r # nový priemerný produkt práce

print("Percentuálne zmeny po zvýšení úrokovej prirážky o 1 percentuálny bod.")
print("Spotreba", round(percent_zmena(C,C_r),4))
print("Práca", round(percent_zmena(N,N_r),4))
print("HDP", round(percent_zmena(Y,Y_r),4))
print("Mzdová sadzba", round(percent_zmena(w,w_r),4))
print("Úroková miera", round(percent_zmena(r,r_r),4))
print("Priemerný produkt práce", round(percent_zmena(APN,APN_r),4))
print("Investície", round(percent_zmena(I,I_r),4))

r_p = 0. # pôvodná hodnota 
````

Grafická analýza predpokladala, že zvýšenie úrokovej prirážky pôsobí nielen na investičný dopyt firiem, ale aj na spotrebiteľa prostredníctvom zníženia životného bohatstva a horšieho prístupu k úverom. V numerickom modeli je úroková prirážka zavedená iba ako dodatočný náklad investícií v podmienke prvého rádu firmy. Simulácia preto zachytáva investičný kanál finančného šoku, ale abstrahuje od jeho priameho vplyvu na životné bohatstvo domácností.

### Vplyv zníženia dopytu a ponuky práce (sektorový šok)

V modeli potrebujeme formulovať exogénne zmeny ponuky a dopytu práce. Upravme časové ohraničenie spotrebiteľa a podmienku prvého rádu firmy pre dopyt po práci do tvaru

$$
\ell+\nu N=h,
$$

a

$$
(1-\alpha)AK^\alpha(\epsilon N)^{-\alpha}=w,
$$

kde $\nu=\epsilon=1$ v pôvodnom modeli.

Ak sa parameter $\nu$ zvýši, ponuka práce sa zníži. Na získanie jednotky práce je potrebné obetovať väčšie množstvo disponibilného času.

Ak sa parameter $\epsilon$ zvýši, zníži sa dopyt po práci, pretože pri danom množstve práce klesne hraničný produkt práce.

````python
def ppr_e(endog):

    C, Cp, l, lp, N, Np, Y, Yp, I, Ip, Kp, w, wp, r = endog

    return (

        C/l - w,
        Cp/lp - wp,
        Cp/(beta*C) - (1+r),

        (1-alpha)*A*K**alpha*(epsilon*N)**(-alpha) - w,     # ZMENA
        (1-alpha)*Ap*Kp**alpha*Np**(-alpha) - wp,

        alpha*Ap*Kp**(alpha-1)*Np**(1-alpha) - delta - r,

        Y - A*K**alpha*N**(1-alpha),
        Yp - Ap*Kp**alpha*Np**(1-alpha),

        I - (Kp - (1-delta)*K),
        Ip - delta*Kp,

        C + I + G - Y,
        Cp + Ip + Gp - Yp,

        l + nu*N - h,                                        # ZMENA 
        Np + lp - h
    )
````

Podľa teórie {ref}`kap_DGE_e`, ak sa simultánne znížia dopyt aj ponuka práce,

- zníži sa práca,
- zníži sa HDP,
- zvýši sa úroková miera,
- zníži sa spotreba,
- znížia sa investície,
- zvýši sa priemerný produkt práce.

Reakcia reálnej mzdovej sadzby nie je jednoznačná. Závisí od relatívnej veľkosti poklesu dopytu a ponuky práce a od sily dynamického substitučného efektu vyvolaného rastom úrokovej miery.

````python
nu=1.01
epsilon=1.01

# nová konkurenčná rovnováha
C_e, Cp_e, l_e, lp_e, N_e, Np_e, Y_e, Yp_e, I_e, Ip_e, Kp_e, w_e, wp_e, r_e = fsolve(
    ppr_e,
    (C, Cp, l, lp, N, Np, Y, Yp, I, Ip, Kp, w, wp, r)
)

APN_e = Y_e/N_e # nový priemerný produkt práce

print("Percentuálne zmeny po simultánnom poklese ponuky a dopytu po práci.")
print("Spotreba", round(percent_zmena(C,C_e),4))
print("Práca", round(percent_zmena(N,N_e),4))
print("HDP", round(percent_zmena(Y,Y_e),4))
print("Mzdová sadzba", round(percent_zmena(w,w_e),4))
print("Úroková miera", round(percent_zmena(r,r_e),4))
print("Priemerný produkt práce", round(percent_zmena(APN,APN_e),4))
print("Investície", round(percent_zmena(I,I_e),4))

nu=1
epsilon=1
````

Numerická simulácia potvrdzuje teoretické predikcie modelu. Pokles efektu vyhľadávania znižuje prácu, výstup a spotrebu. Keďže sa znižuje ponuka výstupu pri nezmenenom dopyte, úroková miera rastie a investície klesajú.

Priemerný produkt práce rastie, pretože ekonomika sa pohybuje po produkčnej funkcii smerom k nižšiemu využitiu práce. Dynamický substitučný efekt rastu úrokovej miery je dostatočne silný na to, aby mierne znížil reálnu mzdovú sadzbu.
