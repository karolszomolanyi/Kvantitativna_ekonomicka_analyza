```python
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

# skontroluj, či v pracovnom priečinku existuje podpriečinok data, ak nie vytvor
Path("data").mkdir(exist_ok=True)

def eurostat_to_datetime(s):
    s = s.astype(str)

    if s.str.match(r"^\d{4}$").all():          # ročné: 1995
        return pd.to_datetime(s, format="%Y")

    elif s.str.match(r"^\d{4}-Q[1-4]$").all(): # štvrťročné: 1995-Q1
        return pd.PeriodIndex(s, freq="Q").to_timestamp()

    elif s.str.match(r"^\d{4}-\d{2}$").all():  # mesačné: 1996-01
        return pd.to_datetime(s, format="%Y-%m")

    else:
        return pd.to_datetime(s)

def eurostat_to_parquet(
    url,
    filename,
    index,
    columns,
    values="OBS_VALUE",
    rename=None
):
    df = pd.read_csv(url)

    df[index] = eurostat_to_datetime(df[index]) # prevod do dátumu

    
    df = df.pivot(
        index=index,
        columns=columns,
        values=values
    )

    if rename:
        df = df.rename(columns=rename)

    df.to_parquet(filename)

    return df

def eurostat_single_series_to_parquet(
    url,
    filename,
    name,
    index="TIME_PERIOD"
):
    df = pd.read_csv(url)

    df[index] = eurostat_to_datetime(df[index]) # prevod do dátumu

    df = df[[index, "OBS_VALUE"]]

    df = df.rename(columns={
        "OBS_VALUE": name
    })

    df = df.set_index(index)

    df.to_parquet(filename)

    return df

def eurostat_multiindex_to_parquet(
    url,
    filename,
    columns,
    rename=None
):
    df = pd.read_csv(url)

    df["TIME_PERIOD"] = eurostat_to_datetime(df["TIME_PERIOD"])

    df = df.pivot(
        index="TIME_PERIOD",
        columns=columns,
        values="OBS_VALUE"
    )
    
    df.columns = [
        "_".join(map(str, col))
        for col in df.columns
    ]

    if rename:
        df = df.rename(columns=rename)

    df.to_parquet(filename)

    return df
```

## Údaje z ECB {cite}`ecb`

```python
import pandas as pd

def ecb_to_parquet(
    dataflow,
    series_key,
    colname,
    filename
):

    url = (
        "https://data-api.ecb.europa.eu/service/data/"
        f"{dataflow}/{series_key}"
        "?format=csvdata"
    )

    df = pd.read_csv(url)

    df["TIME_PERIOD"] = pd.to_datetime(df["TIME_PERIOD"])

    df = (
        df[["TIME_PERIOD", "OBS_VALUE"]]
        .rename(columns={"OBS_VALUE": colname})
        .set_index("TIME_PERIOD")
    )

    df.to_parquet(filename)

    return df
```

#### Všeobecné funkcie a zobrazenie časových radov

```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

def set_theory_style():

    plt.rcParams.update({
        "figure.figsize": (8,6),
        "figure.dpi": 120,
        "font.size": 12,
        "lines.linewidth": 2,
        "axes.grid": False,
        "axes.spines.top": False,
        "axes.spines.right": False,
    })


def set_data_style():

    plt.rcParams.update({
        "figure.figsize": (8,4.5),
        "figure.dpi": 120,
        "font.size": 11,
        "lines.linewidth": 2,
        "axes.grid": True,
        "grid.alpha": 0.3,
        "axes.spines.top": False,
        "axes.spines.right": False,
        "legend.frameon": False,
    })

def plot_ts(
    data,
    filename=None,
    labels=None,
    title=None,
    ylabel=None,
    common_sample=True,
    normalize=False,
    zero_line=False,
    recessions=None,
    figsize=None,
    YearLocator=5
):
    """
    Graf priebehu jedného alebo viacerých časových radov.

    Parameters
    ----------
    data : pd.Series, pd.DataFrame alebo list[pd.Series]
        Vstupné časové rady.

    labels : list[str], optional
        Názvy do legendy.

    title : str, optional
        Názov grafu.

    ylabel : str, optional
        Popis osi y.

    common_sample : bool, default=True
        Oreže rady na spoločné obdobie.

    normalize : bool, default=False
        Normalizuje prvé pozorovanie na 100.

    zero_line : bool, default=False
        Nakreslí horizontálnu čiaru v nule.

    recessions : bool, default=False
        Zobrazí vybrané recesie.

    figsize : tuple, optional
        Prepíše predvolenú veľkosť grafu.
    
    YearLocator : interval rokov zobrazovaných na osi x 

    Returns
    -------
    fig, ax, df
    """

    # --------------------------------------------------
    # Vytvorenie DataFrame
    # --------------------------------------------------

    if isinstance(data, pd.Series):
        df = data.to_frame()

    elif isinstance(data, pd.DataFrame):
        df = data.copy()

    elif isinstance(data, (list, tuple)):
        df = pd.concat(data, axis=1)

    else:
        raise TypeError(
            "data musí byť Series, DataFrame alebo list Series"
        )

    # --------------------------------------------------
    # Datetime index
    # --------------------------------------------------

    if not pd.api.types.is_datetime64_any_dtype(df.index):
        try:
            df.index = pd.to_datetime(df.index)
        except Exception:
            raise ValueError(
                "Index nie je datetime a nepodarilo sa ho previesť."
            )

    df = df.sort_index()

    # --------------------------------------------------
    # Legendy
    # --------------------------------------------------

    if labels is not None:

        if len(labels) != len(df.columns):
            raise ValueError(
                "Počet labels sa nerovná počtu radov."
            )

        df.columns = labels

    # --------------------------------------------------
    # Spoločná vzorka
    # --------------------------------------------------

    if common_sample:

        start = max(
            df[c].first_valid_index()
            for c in df.columns
        )

        end = min(
            df[c].last_valid_index()
            for c in df.columns
        )

        df = df.loc[start:end]

    # --------------------------------------------------
    # Normalizácia
    # --------------------------------------------------

    if normalize:

        df = (
            df
            .div(df.iloc[0])
            .mul(100)
        )

    # --------------------------------------------------
    # Graf
    # --------------------------------------------------

    fig, ax = plt.subplots(figsize=figsize)

    for col in df.columns:
        ax.plot(df.index, df[col], label=col)

    # --------------------------------------------------
    # Voľby
    # --------------------------------------------------

    if title:
        ax.set_title(title)

    if ylabel:
        ax.set_ylabel(ylabel)

    if zero_line:
        ax.axhline(
            0,
            color="black",
            lw=0.8
        )

    if recessions:

        for peak, trough in recessions:

            ax.axvspan(
                pd.Timestamp(peak),
                pd.Timestamp(trough),
                color="lightgray",
                alpha=0.3
            )

    # --------------------------------------------------
    # Os x
    # --------------------------------------------------

    
    ax.xaxis.set_major_locator(
        mdates.YearLocator(YearLocator)
    )

    ax.xaxis.set_major_formatter(
        mdates.DateFormatter("%Y")
    )

    fig.autofmt_xdate()

    if len(df.columns) > 1:
        ax.legend()
    
    plt.tight_layout()

    if filename is not None:
        plt.savefig(
            f"Data/{filename}.png",
            dpi=300,
            bbox_inches="tight"
        )
        plt.close()

    return fig, ax, df
```

#### Log-kvadratický trend

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm


def decompose_log_quadratic(series, log_quadratic=True):
    """
    Rozklad časového radu pomocou kvadratického trendu.

    Parameters
    ----------
    series : pandas.Series

    log_quadratic : bool
        True  -> log-kvadratický trend
        False -> obyčajný kvadratický trend

    Returns
    -------
    out : pandas.DataFrame

        y
        trend
        cycle
        cycle_pct (iba pri logaritmickom trende)

    model : statsmodels RegressionResults
    """

    series = (
        series
        .dropna()
        .astype(float)
        .sort_index()
        .copy()
    )

    t = np.arange(len(series))

    X = pd.DataFrame(
        {
            "const": 1.0,
            "t": t,
            "t2": t**2
        },
        index=series.index
    )

    # premenná použitá v regresii
    if log_quadratic:
        y_reg = np.log(series)
    else:
        y_reg = series

    # odhad trendu
    model = sm.OLS(
        y_reg,
        X
    ).fit()

    trend_reg = pd.Series(
        model.predict(X),
        index=series.index
    )

    # prevod trendu späť do pôvodných jednotiek
    if log_quadratic:

        trend_level = np.exp(trend_reg)

        cycle = (
            np.log(series)
            - trend_reg
        )

        out = pd.DataFrame(
            {
                "y": series,
                "trend": trend_level,
                "cycle": cycle,
                "cycle_pct": 100 * cycle
            },
            index=series.index
        )

    else:

        cycle = series - trend_reg

        out = pd.DataFrame(
            {
                "y": series,
                "trend": trend_reg,
                "cycle": cycle
            },
            index=series.index
        )

    return out, model
```

#### Cyckické charakteristiky, prevod dataframeu do obrázku

```python
def cycle_characteristics(
    x,
    y,
    name=None,
    max_lag=1,
    acyclic_threshold=0.2
):
    """
    Cyklické charakteristiky premennej x
    vzhľadom na cyklickú zložku HDP y.

    Ak log_quadratic=False:
        cyklická zložka = y - trend
    """

    df = pd.concat([x, y], axis=1).dropna()

    x = df.iloc[:, 0]
    y = df.iloc[:, 1]

    # volatilita
    sd_x = x.std()
    sd_y = y.std()

    rel_vol = sd_x / sd_y

    # súčasná korelácia
    corr0 = x.corr(y)

    # korelácie s posunom
    corrs = {}

    for k in range(-max_lag, max_lag + 1):
        corrs[k] = x.shift(k).corr(y)

    best_lag = max(
        corrs,
        key=lambda k: abs(corrs[k])
    )

    best_corr = corrs[best_lag]

    # cyklickosť
    if abs(best_corr) < acyclic_threshold:
        cyclicality = "acyklická"
        timing = "neurčená"
    
    elif best_corr > 0:
        cyclicality = "procyklická"
    
    else:
        cyclicality = "kontracyklická"
    
    # časová charakteristika
    if cyclicality != "acyklická":
    
        if best_lag < 0:
            timing = "oneskorená"
    
        elif best_lag > 0:
            timing = "predbiehajúca"
    
        else:
            timing = "súbežná"

    out = pd.DataFrame(
        {
            "Premenná": [name],
            "σ": [sd_x],
            "σ/σ(Y)": [rel_vol],
            "corr(Y)": [corr0],
            "Typ": [cyclicality],
            "Posun": [best_lag],
            "corr(max)": [round(float(best_corr), 2)],
            "Vzťah k cyklu": [timing]
        }
    )

    return out, pd.Series(corrs)


def df_to_png0(df, filename, title=None):

    fig, ax = plt.subplots(
        figsize=(12, 2 + 0.4 * len(df))
    )

    ax.axis("off")

    tbl = ax.table(
        cellText=df.map(
            lambda x: f"{x:.2f}".replace(".", ",")
            if isinstance(x, (int, float))
            else x
        ).values,
        rowLabels=df.index,
        colLabels=df.columns,
        cellLoc="center",
        loc="center"
    )

    tbl.auto_set_font_size(False)
    tbl.set_fontsize(8)

    tbl.auto_set_column_width(
        col=list(range(len(df.columns)))
    )

    tbl.scale(1.1, 1.4)

    if title:
        ax.set_title(title)

    plt.savefig(
        f"Data/{filename}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
```

## Výpočty

```python
import matplotlib.pyplot as plt
import pandas as pd

# výpočet priemerných rastov
def avg_growth(series):

    x0 = series.iloc[0]
    xt = series.iloc[-1]

    n = len(series) - 1

    return (np.log(xt) - np.log(x0)) / n

# dataframe do obrázka
def df_to_png(df, filename, title=None):

    fig, ax = plt.subplots(
        figsize=(6, 2 + 0.4 * len(df))
    )

    ax.axis("off")

    tbl = ax.table(
        cellText=df.map(
            lambda x: f"{x:.2f}".replace(".", ",")
            if isinstance(x, (int, float))
            else x
        ).values,
        rowLabels=df.index,
        colLabels=df.columns,
        cellLoc="center",
        loc="center"
    )

    tbl.auto_set_font_size(False)
    tbl.set_fontsize(10)
    tbl.scale(1.1, 1.4)

    if title:
        ax.set_title(title)

    plt.savefig(
        f"Data/{filename}.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
```
