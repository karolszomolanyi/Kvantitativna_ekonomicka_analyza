# Úvod do teórie rastu

Pri pohľade na dlhodobý hospodársky vývoj krajín možno rozlíšiť tri charakteristické rastové príbehy.

1. **Vyspelé ekonomiky**, medzi ktoré patria Spojené štáty a krajiny západnej Európy, zaznamenávali počas veľmi dlhého obdobia relatívne stabilný rast reálneho HDP na obyvateľa. Priemerná ročná miera rastu HDP na obyvateľa dosahovala v USA od konca občianskej vojny v 19. storočí až po globálnu finančnú krízu v roku 2008 približne 2 %. Aj relatívne nízke tempo rastu, ak pretrváva mnoho desaťročí, vedie k obrovskému kumulatívnemu zvýšeniu životnej úrovne. Dnešné veľké rozdiely v HDP na obyvateľa sú preto do značnej miery výsledkom rozdielneho hospodárskeho vývoja počas dlhého obdobia. Ekonomický rast pritom zohráva významnú úlohu pri zlepšovaní materiálnych životných podmienok.

2. **Konvergujúce ekonomiky** začínali s podstatne nižšou úrovňou HDP na obyvateľa, ale počas určitého obdobia rástli výrazne rýchlejšie ako vyspelé ekonomiky. Medzi najvýraznejšie historické príklady patria ekonomiky východnej Ázie. Japonsko, Singapur, Hongkong či Južná Kórea dokázali počas povojnových desaťročí výrazne zmenšiť rozdiel v životnej úrovni voči najvyspelejším ekonomikám. Dnes patria medzi ekonomiky s vysokou úrovňou HDP na obyvateľa.

3. **Ekonomiky s pomalou alebo neúspešnou konvergenciou** zostávajú napriek hospodárskemu rastu relatívne chudobné. Ich HDP na obyvateľa môže rásť, jeho rast však nestačí na výraznejšie približovanie sa k vyspelým ekonomikám. V jednotlivých obdobiach môžu niektoré z týchto ekonomík zaznamenať aj dlhšie obdobia stagnácie alebo poklesu životnej úrovne.

Konvergencia
: Konvergenciou budeme označovať proces, pri ktorom chudobnejšia ekonomika rastie rýchlejšie ako bohatšia ekonomika, takže sa rozdiel v ich HDP na obyvateľa postupne zmenšuje.

Slovenská ekonomika predstavuje ďalší zaujímavý príklad konvergencie. Podobne ako ostatné postkomunistické ekonomiky strednej a východnej Európy vstupovalo Slovensko do obdobia transformácie s podstatne nižšou úrovňou HDP na obyvateľa než západoeurópske ekonomiky. HDP na obyvateľa prepočítaný paritou kúpnej sily dosahoval v polovici 90. rokov minulého storočia približne 51 % priemernej úrovne EÚ.

Ako vyplýva z {numref}`graf_PPP`, slovenská ekonomika rástla mimoriadne rýchlo najmä v období pred globálnou finančnou krízou, keď sa HDP na obyvateľa prepočítaný pomocou parity kúpnej sily zvýšil približne na tri štvrtiny európskeho priemeru. Slovenský HDP na obyvateľa sa následne pohyboval približne medzi 70 % a 80 % priemeru EÚ a do roku 2025 nedošlo k ďalšiemu významnému približovaniu.

Slovenská skúsenosť tak obsahuje prvky dvoch odlišných rastových príbehov. Rýchly rast a približovanie sa k vyspelejším ekonomikám pripomínajú konvergenčný rast, aký v minulosti zaznamenali úspešné ekonomiky východnej Ázie. Následné spomalenie konvergencie však ukazuje, že rýchly rast nemusí automaticky pokračovať navždy. Ako ukazuje {numref}`tab_rast_udalosti`, priemerné tempo rastu HDP na obyvateľa bolo pred globálnou finančnou krízou výrazne vyššie ako v nasledujúcich obdobiach. V období pred globálnou finančnou krízou bol priemerný rast HDP na obyvateľa vysoko nad 2 %, v období medzi globálnou finančnou krízou a pandémiou COVID-19 mierne nad 2 % a v súčasnom období dokonca pod 2 %.

Tieto rozdielne skúsenosti predstavujú základné otázky teórie ekonomického rastu. Prečo niektoré ekonomiky dokážu dlhodobo zvyšovať životnú úroveň? Prečo chudobnejšie ekonomiky niekedy rýchlo dobiehajú bohatšie krajiny, zatiaľ čo iné zaostávajú? Akú úlohu pri ekonomickom raste zohrávajú akumulácia kapitálu, miera sporenia, populačný vývoj a rast produktivity? A do akej miery môže dlhodobý ekonomický rast ovplyvniť hospodárska politika?

Konštrukcia teórie, ktorá nám umožní na tieto otázky odpovedať, je predmetom tejto a nasledujúcej kapitoly. Teória ekonomického rastu vysvetľuje vývoj ekonomiky z dlhodobého hľadiska. Neskôr ju doplníme teóriou hospodárskych cyklov, ktorá sa sústreďuje na krátkodobé výkyvy ekonomickej aktivity.

## Solowov model rastu

Na vysvetlenie dlhodobých trendov v ekonomike sa využíva model, ktorý nezávisle od seba prezentovali Solow {cite}`Solow1956` a Swan {cite}`Swan1956`. Model budeme ďalej označovať ako **Solowov model rastu**. 

Od kapitoly {doc}`Stati` sme predpokladali, že výstup ekonomiky je kompozitný statok, ktorý predstavuje celkovú produkciu tovarov a služieb. Statok je vyrábaný firmami a spotrebovávaný spotrebiteľmi. Predpokladáme zatvorenú ekonomiku bez vlády.

### Produkčná funkcia

Technologické možnosti firiem sú opísané makroekonomickou produkčnou funkciou, ktorú sme predstavili v časti {ref}`kap_makro_pf`.

Zopakujme, že vstupmi produkčnej funkcie sú kapitál $K$ a práca $N$. Efektívnosť využívania vstupov vo výrobe vyjadruje celková produktivita faktorov $A$.

Z hľadiska dlhodobého ekonomického rastu celkovú produktivitu faktorov $A$ interpretujeme ako **úroveň technológie**.

O produkčnej funkcii {eq}`r_makro_pf` predpokladáme tieto vlastnosti:

1. je rastúca v oboch vstupoch, kapitáli a práci,
2. je konkávna,
3. elasticita výstupu podľa celkovej produktivity faktorov sa rovná 1,
4. prechádza začiatkom súradnicového systému,
5. má konštantné výnosy z rozsahu.

Vďaka konštantným výnosom z rozsahu možno produkčnú funkciu vyjadriť v intenzívnom tvare {eq}`intenzivna_pf`.

Intenzívny kapitál $k$ a intenzívny výstup yyy sú vyjadrené na jednotku práce. Ak predpokladáme konštantný počet jednotiek práce na obyvateľa, môžeme intenzívny výstup $y$ interpretovať aj ako HDP na obyvateľa.

### Miera rastu práce

Predpokladáme, že populácia spotrebiteľov rastie konštantnou mierou $n$. Keďže predpokladáme konštantný počet jednotiek práce na obyvateľa, rovnakou mierou rastie aj práca:

```{math}
:label: r_rast_prace

\frac{N'}{N}=1+n.
```

Podľa konvencie tejto učebnice premenná označená apostrofom predstavuje hodnotu v nasledujúcom období; $N'$ je práca v ďalšom období.

### Úspory

V predchádzajúcich modeloch sme ukázali, že rovnováha na jednotlivých trhoch zabezpečuje, aby sa zároveň spĺňali identity systému národných účtov. Aplikovaním Walrasovho zákona ({ref}`kap_walras`) sme napríklad v {doc}`StaticModel` ukázali, že rovnováha na trhu práce vedie k rovnosti medzi vytvoreným a použitým výstupom. Podobne sme v {doc}`Finan` ukázali súvislosť medzi rovnováhou na finančnom trhu, úsporami a použitím príjmu.

V Solowovom modeli nebudeme konkurenčnú rovnováhu jednotlivých trhov explicitne odvodzovať. Správanie spotrebiteľov zjednodušíme predpokladom, že v každom období usporia konštantný podiel $s$ svojho príjmu:

```{math}
:label: r_miera_sporenia

S=sY.
```

Podiel výstupu $s$, ktorý spotrebitelia usporia, nazývame **mierou sporenia**. Predpokladáme, že:

$$
0 < s < 1.
$$

Predpokladáme pritom, že trhy sú v rovnováhe, a preto sa musia spĺňať aj príslušné identity systému národných účtov. Keďže ekonomika je zatvorená a neuvažujeme vládu, jej bežný účet sa rovná nule a agregátne úspory sa musia rovnať agregátnym investíciám:

```{math}
:label: identita_S_I_zatvorena

S=I,
```

čo zodpovedá identite úspor a investícií {eq}`identita_S_I`.

Z identity rozdelenia agregovaného príjmu {eq}`r_agregovany_prijem` vyjadríme reálnu spotrebu. Ekonomika je uzavretá, a preto sa čisté príjmy faktorov a čisté jednostranné transfery zo zahraničia rovnajú nule. V modeli neuvažujeme vládu, takže vládne nákupy sa rovnajú nule. Identitu rozdelenia agregovaného príjmu môžeme preto vyjadriť v tvare:

```{math}
:label: r_rozdelenie_mala_zatvorena

Y=C+S.
```

Kombináciou {eq}`identita_S_I_zatvorena` a {eq}`r_rozdelenie_mala_zatvorena` získame

```{math}
:label: r_solow_spotreba

C=(1-s)Y.
```

Z každej jednotky výstupu spotrebitelia usporia podiel $s$ a spotrebujú podiel $1-s$.

Zopakujme, že hrubé súkromné investície ({ref}`def_I`) pozostávajú z čistých investícií ({ref}`def_ciste_I`) a odpisov ({ref}`def_odpisy`).

Čisté investície zvyšujú kapitál $K$; rovnajú sa zmene stavu kapitálu $K'-K$.

Predpokladajme, že v každom období sa znehodnotí konštantný podiel $\delta$ existujúceho kapitálu. Odpisy sa preto rovnajú $\delta K$. Podiel existujúceho kapitálu, ktorý sa počas jedného obdobia opotrebuje, $\delta$, nazývame **mierou znehodnotenia kapitálu**.

Hrubé súkromné investície sú súčtom čistých investícií a odpisov:

```{math}
:label: r_solow_investice

I=K'-K+\delta K.
```

V základnej verzii Solowovho modelu predpokladáme, že úroveň technológie $A$ je konštantná. Tento predpoklad zmeníme v ďalšej kapitole.

## Základná rovnica Solowovho modelu

Z rovnice hrubých súkromných investícií {eq}`r_solow_investice` vyjadrime kapitál v nasledujúcom období

$$
K'=(1-\delta)K+I
$$

Za hrubé súkromné investície dosadíme hrubé súkromné úspory z identity {eq}`identita_S_I_zatvorena`

$$
K'=(1-\delta)K+S,
$$

ktoré ďalej zapíšeme  podľa {eq}`r_miera_sporenia`:

$$
K'=(1-\delta)K+sY.
$$

Za reálny HDP dosaďme produkčnú funkciu {eq}`r_makro_pf` a prehoďme poradie členov na pravej strane:

$$
K'= sA F(K,N) + (1-\delta)K.
$$

Ak chceme skonštruovať model, ktorý má vysvetliť rast v rôzne veľkých ekonomikách, je výhodné vyjadriť dynamiku premenných v intenzívnom tvare podelením prácou $N$:

$$
\frac{K'}{N}=sA\frac{F(K,N)}{N}+(1-\delta)\frac{K}{N}.
$$

Za predpokladu konštantných výnosov z rozsahu, produkčná funkcia v intenzívnom tvare je {eq}`intenzivna_pf`, kde $\frac{K}{N}=k$. Preto rovnicu môžeme zapísať v tvare

$$
\frac{K'}{N}=sAf(k)+(1-\delta)k.
$$

Z formulácie rastu práce {eq}`r_rast_prace` vieme vyjadriť prácu:

$$
N=\frac{N'}{1+n}.
$$

Po dosadení

$$
(1+n)\frac{K'}{N'}=sAf(k)+(1-\delta)k.
$$

Rovnako ako v prvom období, aj v nasledovnom období je intenzívny kapitál

$$
k'=\frac{K'}{N'}
$$

Po dosadení získame **základnú rovnicu Solowovho modelu**:

```{math}
:label: r_solow_zakladna

k'=\frac{sAf(k)}{1+n}+\frac{(1-\delta)}{1+n}k.
```

Základná rovnica Solowovho modelu vyjadruje dynamiku intenzívneho kapitálu $k$. Intenzívny kapitál v nasledujúcom období vypočítame využitím kapitálu v prítomnom období $k$ a parametrov miery rastu práce $n$, úrovne technológie $A$, miery sporenia $s$ a miery znehodnotenia kapitálu $\delta$.


## Grafická interpretácia základnej rovnice Solowovho modelu

Označme funkciu pravej strany základnej rovnice Solowovho modelu {eq}`r_solow_zakladna` symbolom $\phi(k)$,

$$
k'=\phi(k)=\frac{sAf(k)}{1+n}+\frac{(1-\delta)}{1+n}k.
$$

Na {numref}`graf_45` je zobrazená funkcia $\phi(k)$ modrou čiarou. Funkcia je rastúca a keďže produkčná funkcia $Af(k)$ prvom člene je konkávna, funkcia $\phi(k)$ je konkávna. Funkcia vychádza zo začiatku súradnicového systému.

O uvedenom môžeme presvedčiť za predpokladu, že $\phi(k)$ je dvakrát diferencovateľaná. Platí:

$$
\frac{d^2 \phi(k)}{d k^2} = \frac{sA}{1+n} \frac{d^2 f(k)}{d k^2}<0
$$

a

$$
\phi(0)=\frac{sAf(0)}{1+n}=0.
$$

Využitím základnej rovnice Solowovho modelu môžeme definovať stav, keď sa intenzívny kapitál nemení, t.j. intenzívny kapitál v nasledujúcom období sa rovná kapitálu v prítomnom období, $k′ = k$. Hovoríme o **ustálenom stave**.

Aby sme vyjadrili hodnotu intenzívneho kapitálu v ustálenom stave, zobrazíme priamku, ktorá prislúcha rovnosti $k′ = k$. Priamka zobrazená čiernou farbou vychádza zo začiatku súradnicového systému a zviera s osami $45^\circ$ uhol. V bodoch kde sa priamka pretne s funkciou $\phi(k)$ sa realizuje rovnica {eq}`r_solow_zakladna` a zároveň $k′ = k$. V tomto bode je hodnota intenzívneho kapitálu v ustálenom stave.

Na {numref}`graf_45` sa $\phi(k)$ pretne s priamkou $k'=k$ v dvoch bodoch. Prvý je v začiatku súradnicového systému, $k=0$. Druhý je označený hviezdičkou $k*$.

Dynamika intenzívneho kapitálu závisí od hodnoty v prítomnom období. Nech $k_0$ na {numref}`graf_45` je množstvo prítomného intenzívneho kapitálu. Podľa základnej rovnice Solowovho modelu {eq}`r_solow_zakladna` vieme vypočítať hodnotu intenzívneho kapitálu pracujúceho v ďalšom období. Ak ju označíme $k_1$, platí:

$$
k_1=\phi(k_0)=\frac{sAf(k_0)}{1+n}+\frac{(1-\delta)}{1+n}k_0.
$$

Graficky určíme $k_1$ tak, že z hodnoty $k_0$ na horizontálnej osi postupujeme vertikálne k funkcii $\phi(k)$. Funkčná hodnota $\phi(k_0)$ na vertikálnej osi predstavuje kapitál $k_1$ v nasledujúcom období.

Aby sme mohli hodnotu $k_1$ použiť ako kapitál na horizontálnej osi pri výpočte ďalšieho obdobia, využijeme priamku $k'=k$. Keďže táto priamka zviera s oboma osami uhol $45^\circ$, rovnaká hodnota má na oboch osiach rovnakú vzdialenosť od začiatku súradnicového systému. Horizontálnym posunom od funkcie $\phi(k)$ k priamke $k'=k$ preto prenesieme hodnotu $k_1$ z vertikálnej osi na horizontálnu os. Ak $0<k_0<k^*$ hodnota $k1$ je väčšia ako $k0$. Intenzívny kapitál rastie. 

Rovnaký postup môžeme zopakovať. Dosadením hodnoty $k_1$ do funkcie $\phi(k)$ získame hodnotu

$$
k_2=\phi(k_1),
$$
 
dosadením $k_2$
 
$$
k_3=\phi(k_2),
$$
 
a všeobecne
 
$$
k_{t+1}=\phi(k_t).
$$

Opakovaním tohto postupu základná rovnica Solowovho modelu generuje postupnosť $k_0,k_1,k_2,k_3,\ldots$

Intenzívny kapitál rastie, ak štartovacia hodnota $0<k_0$ je nižšia ako $k^*$ a naopak klesá, ak $k_0>k^*$ a približuje sa k ustálenému stavu $k^*$. 

Ak štartovacia hodnota $k_0=k^*$, intenzívny kapitál je od začiatku v ustálenom stave $k^*$ a v čase sa nemení. Rovnaké platí, ak $k_0=0$: ak ekonomika nemá žiadny kapitál, pri našich predpokladoch nevyrába, nevytvára úspory ani investície a kapitál zostáva nulový.

Rozdiel medzi ustálenými stavmi $k=0$ a $k=k^*$ je, že $k^*$ je **stabilný**. Ak aj na začiatku intenzívny kapitál nebol v ustálenom stave $0<k_0 \ne k^*$, časom sa začne približovať k ustálenému stavu $k^*$. 

Ustálený stav $k=0$ túto vlastnosť nemá. Ak ekonomika začína s ľubovoľnou kladnou zásobou kapitálu, jej intenzívny kapitál sa od nulového ustáleného stavu vzďaľuje. V ďalšom výklade sa preto budeme zaoberať ustáleným stavom $k^*$ a budeme predpokladať $k>0$.

````python
import grafy

grafy.set_theory_style()
k0=0.3
A = 2
s = 0.3
alpha = 0.3
delta = 0.4
n = 0
k_us, y_us, i_us, c_us = grafy.us(A, alpha, n, s, delta)
fig,ax=grafy.plot_cobweb(
    lambda k: grafy.phi(k,A,alpha,n,s,delta),
    0,
    2.3,
    k0,
    var="k",
    steady_state=k_us,
    figsize=(6,6)
)
````

## Časové rady

Základná rovnica Solowovho modelu {eq}`r_solow_zakladna` generuje postupnosť $k_0,k_1,k_2,k_3,\ldots$, ktorá má vlastnosti časového radu. Časový rad intenzívneho kapitálu vygenerovaný rovnicou {eq}`r_solow_zakladna` je zobrazený na {numref}`graf_k_solow`. Pri štartovacej hodnote $0<k_0<k^*$ sa intenzívny kapitál každým ďalším obdobím zväčšuje a približuje sa k ustálenému stavu $k^*$ zobrazenému prerušovanou čiernou čiarou. Platí, že čím bližšie je $k$ k ustálenému stavu $k^*$ tým menšie sú jeho prírastky v každom ďalšom období.

Intenzívny kapitál preto rastie čoraz pomalšie, až jeho prírastky pri približovaní sa k ustálenému stavu smerujú k nule.

Opačný proces nastáva, ak je štartovacia hodnota $k_0>k^*$. Intenzívny kapitál v každom ďalšom období klesá a postupne sa približuje k ustálenému stavu $k^*$, pričom jeho pokles je čoraz menší.

Ak je štartovacia hodnota $k_0=k^*$, ekonomika sa už nachádza v ustálenom stave. Intenzívny kapitál sa v čase nemení, $k_t=k^*$ pre každé obdobie, a jeho časový rad má preto tvar vodorovnej čiary (prerušovaná čiara na {numref}`graf_k_solow`).

````python
N0=1
k,index,*_=grafy.rady(k0,N0,A,alpha=alpha,n=n,s=s,delta=delta,T=10)
fig,ax,df=grafy.plot_model_path(
    k,
    labels=[r"$k_t$"],
    ylabel=r"$k_t$",
    steady_states={
        r"$k^*$": k_us
    }
)
````

Od kapitálu v intenzívnom tvare $k$ závisia aj ďalšie endogénne premenné: intenzívny výstup

```{math}
:label: r_int_y

y=Af(k),
```

intenzívne investície

```{math}
:label: r_int_i

i=sy=sAf(k),
```

a intenzívna spotreba

```{math}
:label: r_int_c

c=(1-s)y=(1-s)Af(k).
```

Keďže v zatvorenej ekonomike platí $S=I$, intenzívne investície $i$ sa rovnajú intenzívnym úsporám $sy$.

Všetky tri intenzívne premenné rastú s $k$. Ak je intenzívny kapitál $k$ vyjadrený časovým radom, potom aj intenzívny výstup $y$, investície $i$ a spotreba $c$ vytvárajú časové rady.

Platí:

- Ak kapitál $k$ v čase rastie, rastú aj $y$, $i$ a $c$.
- Ak kapitál $k$ v čase klesá, klesajú aj $y$, $i$ a $c$.
- Ak sa kapitál $k$ v čase nemení, nemenia sa ani $y$, $i$ a $c$.

Z posledného tvrdenia vyplýva, že ak je kapitál v ustálenom stave $k=k^*$, potom sú v ustálenom stave aj výstup, investície a spotreba. Ich ustálené hodnoty označujeme hviezdičkou: $k^*,y^*,i^*=sy^*,c^*$.

Keď sa približovanie intenzívneho kapitálu $k$ k ustálenému stavu $k^*$ spomaľuje, spomaľuje sa aj približovanie ostatných uvažovaných premenných k ich ustáleným hodnotám.

## Ustálený stav

V dynamických ekonomických modeloch závisí hodnota endogénnych premenných v jednom období od ich hodnôt v predchádzajúcom období.

(def_us)=
Ustálený stav 
: Ustálený stav je stav ekonomiky, v ktorom sa vybrané endogénne premenné v čase nemenia. 

Ak danú endogénnu premennú označíme $x$, v ustálenom stave platí

$$
x'=x=x^*,
$$

kde hviezdička označuje hodnotu premennej v ustálenom stave.

Ak dynamiku premennej opisuje rovnica

$$
x'=\phi(x),
$$

jej ustálený stav $x^*$ spĺňa podmienku

$$
x^*=\phi(x^*).
$$

Ustálený stav je teda **pevný bod dynamického systému**: ak sa ekonomika nachádza v ustálenom stave a $x=x^*$, dynamická rovnica v nasledujúcom období vygeneruje opäť rovnakú hodnotu $x'=x^*$.

Samotná existencia ustáleného stavu ešte neznamená, že k nemu ekonomika bude smerovať. Ak sa ekonomika po odchýlení od ustáleného stavu postupne vracia späť, ustálený stav nazývame **stabilným**. Ak sa od neho naopak ďalej vzďaľuje, ustálený stav je **nestabilný**.

V základnom Solowovom modeli sú v ustálenom stave konštantné intenzívne premenné $k$, $y$, $i$ a $c$. Existujú dva ustálené stavy. Kladný ustálený stav $k^*,y^*,i^*=sy^*,c^*$ je stabilný, zatiaľ čo nulový ustálený stav $k=y=i=c=0$ je nestabilný.

(def_ps)=
Prechodný stav
: Ak sa ekonomika nenachádza v ustálenom stave, nachádza sa v prechodnom stave. V prechodnom stave sa vybrané endogénne premenné v čase menia. V základnom Solowovom modeli sa ekonomika v prechodnom stave postupne približuje ku kladnému stabilnému ustálenému stavu.

(def_konvergencia_us)=
Konvergencia k ustálenému stavu
: Ak sa hodnota endogénnej premennej $x_t$ v čase približuje k jej hodnote v ustálenom stave $x^*$, hovoríme, že premenná $x$ konverguje k ustálenému stavu.

Konvergenciu k ustálenému stavu možno zapísať limitou

$$
\lim_{t\to \infty} (x_t) = x^*.
$$

Časové rady na {numref}`graf_k_solow`--{numref}`graf_c_solow` zodpovedajú ekonomike v prechodnom stave. Všetky štyri premenné konvergujú k svojim hodnotám v ustálenom stave.

## Grafická identifikácia ustáleného stavu

Pre ustálený stav platí 

$$
k^*=\phi(k^*)=\frac{sAf(k^*)}{1+n}+\frac{(1-\delta)}{1+n}k^*.
$$

Po úprave získame rovnicu vhodnú pre grafickú identifikáciu ustáleného stavu v tvare

```{math}
:label: r_us

sAf(k^*)=(\delta+n)k^*
```

Na ľavej strane {eq}`r_us` sú intenzívne úspory a teda aj investície v ustálenom stave. V ustálenom stave sa investície a úspory rovnajú:

```{math}
:label: r_i_us

i^*=sy^*=(\delta+n)k^*.
```

Na oboch stranách {eq}`r_us` sú funkcie kapitálu. Funkcia $sAf(k)$ je rastúca a konkávna a na {numref}`graf_us` je zobrazená oranžovou čiarou. Funkcia $(\delta+n)k$ zobrazená zelenou je lineárna rastúca. Obe funkcie vychádzajú zo začiatku súradnicového systému. Na {numref}`graf_us` je tiež modrou farbou zobrazená produkčná funkcia $Af(k)$.

Rovnica {eq}`r_us` sa spĺňa v bodoch, v ktorých sa pretíňajú krivky $sAf(k)$ a $(\delta+n)k$. Priesečníky identifikujú ustálený stav. Zaujíma nás kladný $k^*$, ktorý identifikujeme na horizontálnej osi. V bode priesečníka dosahujú investície aj úspory ustálenú hodnotu $i^*=sy^*$.

Pomocou produkčnej funkcie možno pri kapitáli $k^*$ získať výstup v ustálenom stave $y^*=Af(k^*)$. Spotreba v ustálenom stave $c^*$ je rozdiel medzi výstupom a úsporami, zobrazená vertikálnou dvojšípkou ako vzdialenosť medzi $y^*$ a $i^*$.

````python
import matplotlib.pyplot as plt
import numpy as np

grafy.set_theory_style()

A = 1.
alpha = 1/2
s = 0.4
n = 0.02
delta = 0.2

k_us,y_us,i_us,*_=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)

fig, ax = plt.subplots()

grafy.nakresli_funkciu(
    ax,
    lambda k: A * grafy.f(k, alpha),
    0,
    5,
    nazov=r"$y=Af(k)$"
)

grafy.nakresli_funkciu(
    ax,
    lambda k: s * A * grafy.f(k, alpha),
    0,
    5,
    nazov=r"$i=sAf(k)$"
)

grafy.nakresli_funkciu(
    ax,
    lambda k: (n + delta) * k,
    0,
    5,
    nazov=r"$(n+\delta)k$"
)

grafy.vyznac_na_osi(ax, k_us, y_us, r"$k^*$", popis_y=r"$y^*$")
grafy.vyznac_na_osi(ax, k_us, i_us, popis_y=r"$i^*=sy^*$")
grafy.nakresli_bod(ax, k_us, y_us, nazov=None, vodiace_ciary=True)
grafy.nakresli_bod(ax, k_us, i_us, nazov=None, vodiace_ciary=True)

x_arrow = 0.07 * ax.get_xlim()[1]

grafy.nakresli_dvojsipku(
    ax,
    0.1,
    i_us,
    0.1,
    y_us,
    nazov=r"$c^*$"
)

grafy.ekonomicke_osi(ax,xlabel=f"$k$",ylabel=None)
ax.legend()

plt.show()
````

## Extenzívne premenné v ustálenom stave

Extenzívnu premennú získame prenásobením intenzívnej premennej množstvom práce. V ustálenom stave platí.

$$
X^*=Nx^*, \qquad x^* \in \{k^*,y^*,i^*,c^*\}, \qquad X^* \in \{K^*,Y^*,I^*,C^*\}.
$$

```{admonition} Upozornenie
Hviezdička pri extenzívnej premennej $X^*$ označuje jej hodnotu, keď sa ekonomika nachádza v ustálenom stave. Neznamená, že táto hodnota je v čase konštantná.
```

Aká je dynamika extenzívnych premennýh v ustálenom stave?

Zapíšme extenzívne premenné v nasledujúcom období, pričom zohľadníme definíciu ustáleného stavu, $x'^*=x^*,

$$
X'^*=N'x'^*=N'x^*, \qquad x^* \in \{k^*,y^*,i^*,c^*\}, \qquad X'^* \in \{K'^*,Y'^*,I'^*,C'^*\}.
$$

Miera rastu extenzívnej premennej je

$$
\frac{X'^*}{X^*}=\frac{N'x^*}{Nx^*}=\frac{N'}{N}=1+n, \qquad X^* \in \{K^*,Y^*,I^*,C^*\}, \qquad X'^* \in \{K'^*,Y'^*,I'^*,C'^*\}.
$$

Ustálený stav neznamená, že ekonomika prestane rásť. Konštantné zostávajú intenzívne premenné, zatiaľ čo extenzívne rastú spolu s množstvom práce mierou $n$.

Na {numref}`graf_Ke_solow`--{numref}`graf_Ce_solow` sú zobrazené časové rady extenzívnych premenných.

## Zlaté pravidlo

Americký ekonóm Edmund Phelps položil otázku: **Ak je ekonomika v ustálenom stave, za akých podmienok je intenzívna spotreba maximálna?** {cite}`Phelps1961`

Podľa {numref}`graf_zlate` hľadáme takú mieru sporenia, pri ktorej je intenzívna spotreba v ustálenom stave maximálna. Rôzne miery sporenia $s$ generujú rôzne ustálené stavy, a teda aj rôzne hodnoty spotreby $c^*$. Na {numref}`graf_zlate` je pri miere sporenia $s_1$ spotreba $c_1^*$, pri miere sporenia $s_2$ spotreba $c_2^*$ a pri miere sporenia $s^*$ dosahuje spotreba svoju maximálnu hodnotu $c^{**}$.

Miera sporenia ovplyvňuje ustálený stav ekonomiky. Keďže všetky intenzívne endogénne premenné možno charakterizovať pomocou intenzívneho kapitálu, každý kladný ustálený stav môžeme charakterizovať príslušnou hodnotou $k^*$.

Spotreba v ustálenom stave je rozdielom medzi výstupom a úsporami, resp. vzhľadom na rovnosť úspor a investícií medzi výstupom a investíciami:

$$
c^*=y^**i^*.
$$

Za $y^*$ dosaďme produkčnú funkciu a za $i^*$ vzťah {eq}`r_i_us`. Získame spotrebu v ustálenom stave:

```{math}
:label: r_c_us

c^*=Af(k^*)-(\delta+n)k^*.
```

Získali sme tak spotrebu v ustálenom stave $c^*$ ako funkciu kapitálu v ustálenom stave $k^*$.

Phelpsov problém môžeme riešiť maximalizáciou {eq}`r_c_us` vzhľadom na $k^*>0$. Podmienka prvého rádu je

$$
\frac{d c^*}{d k^*}=A\frac{d f(k^*)}{d k^*}-(n+\delta)=0.
$$


Prvý člen predstavuje intenzívny hraničný produkt kapitálu:

$$
mpk(k)=A\frac{df(k)}{dk}.
$$

Označme hodnotu kapitálu, pri ktorej je spotreba v ustálenom stave maximálna, $k^{**}$. Podmienku prvého rádu môžeme zapísať v tvare

```{math}
:label: r_zlate

mpk(k^{**})-\delta=n.
```

Hraničný produkt kapitálu $mpk$ vyjadruje prírastok výstupu vytvorený dodatočnou jednotkou kapitálu. Časť tohto prírastku vo výške $\delta$ je potrebná na náhradu znehodnotenia kapitálu. Rozdiel

$$
mpk-\delta
$$

preto predstavuje čistý výnos z dodatočnej jednotky kapitálu.

Trh dlhopisov a trh kapitálu sú v rovnováhe, ak čistý príjem z dlhopisov sa rovná čistému príjmu z kapitálu. Čistý príjem z dlhopisov je úroková miera.

Intenzívna spotreba v ustálenom stave je teda maximálna vtedy, keď sa čistý výnos z kapitálu, resp. reálna úroková miera, rovná miere rastu práce


$$
r=n.
$$


Túto podmienku nazývame **zlatým pravidlom akumulácie kapitálu**.

Hraničný produkt kapitálu je na {numref}`graf_zlate` prezentovaný dotyčnicou produkčnej funkcie zobrazenej modrou farbou. Pri maximálnej spotrebe v ustálenom stave je sklon dotyčnice rovnaký ako sklon oranžovej priamky $(n+\delta)k$.

````python
import matplotlib.pyplot as plt
import numpy as np

grafy.set_theory_style()

A = 1.
alpha = 1/2
n = 0.02
delta = 0.2

#množina parametrov miery sporenia  
ska=[0.4,0.5,0.6]
slabels=[r"$i=s_1Af(k)$",r"$i=s^*Af(k)$",r"$i=s_2Af(k)$"]
clabels=[r"$c^*_1$",r"$c^{**}$",r"$c^*_2$"]

k_zlate,y_zlate,i_zlate,*_=grafy.us(A=A,alpha=alpha,n=n,s=ska[1],delta=delta)

fig, ax = plt.subplots()

grafy.nakresli_funkciu(
    ax,
    lambda k: A * grafy.f(k, alpha),
    0,
    xmax=k_zlate*2,
    nazov=r"$y=Af(k)$",linewidth=1.
)

grafy.nakresli_funkciu(
    ax,
    lambda k: (n + delta) * k,
    0,
    xmax=k_zlate*2,
    nazov=r"$(n+\delta)k$",linewidth=1.
)


for s, slabel, clabel in zip(ska, slabels, clabels):
    us=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)
    
    grafy.nakresli_funkciu(
    ax,
    lambda k: s * A * grafy.f(k, alpha),
    0,
    xmax=k_zlate*2,
    nazov=slabel,
    linewidth=1.
    )
    grafy.nakresli_bod(ax, us[0], us[2], nazov=None, vodiace_ciary=False)
    grafy.nakresli_bod(ax, us[0], us[1], nazov=None, vodiace_ciary=False)

    grafy.nakresli_dvojsipku(
    ax,
    us[0],
    us[2],
    us[0],
    us[1],
    nazov=clabel,
    )

grafy.ekonomicke_osi(ax,xlabel=f"$k$",ylabel=None)
grafy.nakresli_dotycnicu(ax,lambda k: A * grafy.f(k, alpha),lambda k: A*alpha*k**(alpha-1),k_zlate,2
                         ,label=f"Sklon = $n+\delta$",linewidth=1.)

grafy.vyznac_na_osi(ax, k_zlate, i_zlate, popis_x=r"$k^{**}$", popis_y=r"$i^{**}=sy^{**}$")
grafy.nakresli_bod(ax, k_zlate, i_zlate, nazov=None, vodiace_ciary=True)

ax.legend()

plt.show()
````

$$

#### Python

Pre aplikovanie základnej rovnice Solowovho modelu využijeme funkciu `grafy.phi`.

V kóde budeme rozlišovať číselné hodnoty premenných a ich symbolický zápis. Číselné hodnoty označíme príponou h a symbolické premenné príponou s.

````python
# Exogénne premenné, hodnoty
Kh=300
Nh=1000
Ah=2
sh=0.3
alphah=0.3
deltah=0.38
nh=0.02

kh=Kh/Nh
khp=grafy.phi(k=kh,A=Ah,alpha=alphah,n=nh,s=sh,delta=deltah
            )
print("Intenzívny kapitál v súčasnom období", kh)
print("Intenzívny kapitál v nasledujúcom období", round(khp,2))
````

Nasledujúci kód odvodí intenzívny kapitál v ustálenom stave z {eq}`r_us`.

````python
# Exogénne premenné, symboly
As,alphas,ns,ss,deltas,k_star=symbols("A,alpha,n,s,delta,k_star",positive=True)

k_us=solve(ss*As*k_star**alphas-(deltas+ns)*k_star,k_star,dict=True)
k_us=k_us[0]
display(k_us[k_star])
````

Dosadením hodnôt za symboly získame $k^*$.

````python
# Za symboly dosadíme hodnoty
exogenne=({As:Ah,ss:sh,alphas:alphah,deltas:deltah,ns:nh})

k_us[k_star].subs(exogenne)
````

````python
display(Math("$[k^*,y^*,i^*,c^*]=$"))
np.round(grafy.us(A=Ah,alpha=alphah,n=nh,s=sh,delta=deltah
            ),2)
````

V balíčku `grafy`, funkcia `grafy.rovnice` slúži na aplikáciu rovníc {eq}`r_int_y`--{eq}`r_int_c`.

````python
display(Math("$[y,i,c]=$"))
np.round(grafy.rovnice(k=kh,A=Ah,alpha=alphah,s=sh),2)
````

### Príklad: Extenzívne premenné v ustálenom stave

O ekonomike s agregátnou produkčnou funkciou v Cobbovom-Douglasovom tvare poznáte: úroveň technológie (1), mieru sporenia (1/3), podiel kapitálu (1/3), mieru znehodnotenia kapitálu (0,1) a mieru rastu práce (0,02). V súčasnom období žije v ekonomike 1000 obyvateľov. Predpokladáme, že ekonomika je v ustálenom stave.

1. Vypočítajte hodnoty intenzívnych premenných v ustálenom stave.
2. Vypočítajte hodnoty extenzívnych premenných v súčasnom období a v nasledovnom období.
3. Vypočítajte mieru rastu extenzívnych premenných medzi obidvoma obdobiami.

#### Riešenie

Najprv vypočítame **intenzívny kapitál v ustálenom stave**. Pre Cobbovu-Douglasovu produkčnú funkciu platí podľa {eq}`r_CD_us`

$$
k^*=\left[\frac{sA}{n+\delta}\right]^{\frac{1}{1-\alpha}}.
$$

Po dosadení

$$
k^*=\left[\frac{\frac{1}{3}\times1}{0,02+0,1}\right]^{\frac{1}{1-\frac{1}{3}}}\approx4,63.
$$

**Intenzívny výstup v ustálenom stave** vypočítame z produkčnej funkcie:

$$
y^*=A{k^*}^{\alpha}.
$$

Po dosadení

$$
y^*=1\times4,63^{\frac{1}{3}}\approx1,67.
$$

**Intenzívne investície v ustálenom stave** sú

$$
i^*=sy^*,
$$

a teda

$$
i^*=\frac{1}{3}\times1,67\approx0,56.
$$

Rovnakú hodnotu môžeme overiť pomocou podmienky ustáleného stavu:

$$
i^*=(n+\delta)k^*=(0,02+0,1)\times4,63\approx0,56.
$$

**Intenzívna spotreba v ustálenom stave** je

$$
c^*=(1-s)y^*
$$

a teda

$$
c^*=\left(1-\frac{1}{3}\right)\times1,67\approx1,11.
$$

Hodnoty intenzívnych premenných v ustálenom stave sú preto približne

| Premenná | Hodnota |
|:---|---:|
| Intenzívny kapitál $k^*$ | 4,63 |
| Intenzívny výstup $y^*$ | 1,67 |
| Intenzívne investície $i^*$ | 0,56 |
| Intenzívna spotreba $c^*$ | 1,11 |

**Extenzívne premenné v súčasnom období**

Extenzívnu premennú získame prenásobením intenzívnej premennej množstvom práce. V súčasnom období je

$$
N=1000.
$$

Extenzívny kapitál je

$$
K^*=Nk^*=1000\times4,63\approx4629,63.
$$

Výstup je

$$
Y^*=Ny^*=1000\times1,67\approx1666,67.
$$

Investície sú

$$
I^*=Ni^*=1000\times0,56\approx555,56.
$$

Spotreba je

$$
C^*=Nc^*=1000\times1,11\approx1111,11.
$$

**Extenzívne premenné v nasledujúcom období**

Ekonomika je v ustálenom stave, preto sa intenzívne premenné v nasledujúcom období nemenia:

$$
k'^*=k^*, \qquad
y'^*=y^*, \qquad
i'^*=i^*, \qquad
c'^*=c^*.
$$

Množstvo práce však rastie mierou $n=0,02$. V nasledujúcom období preto bude

$$
N'=(1+n)N=1,02\times1000=1020.
$$

Extenzívny kapitál v nasledujúcom období je

$$
K'^*=N'k^*=1020\times4,63\approx4722,22.
$$

Výstup je

$$
Y'^*=N'y^*=1020\times1,67\approx1700,00.
$$

Investície sú

$$
I'^*=N'i^*=1020\times0,56\approx566,67.
$$

Spotreba je

$$
C'^*=N'c^*=1020\times1,11\approx1133,33.
$$

### Príklad: Zlaté pravidlo

Ak je agregovaná produkčná funkcia v Cobbovom-Douglasovom tvare s konštantnými výnosmi z rozsahu, odvoďte, pri akej hodnote miery sporenia je intenzívna spotreba v ustálenom stave maximálna.

#### Riešenie

Podmienka prvého rádu zlatého pravidla je {eq}`r_zlate`:

$$
mpk(k^{**})-\delta=n.
$$

Pri Cobbovej-Douglasovej produkčnej funkcii

$$
y=Ak^\alpha
$$

je hraničný produkt kapitálu

$$
mpk(k)=\alpha A k^{\alpha-1}.
$$

Podmienku zlatého pravidla preto môžeme zapísať ako

$$
\alpha A(k^{**})^{\alpha-1}=n+\delta.
$$

Intenzívny kapitál v ustálenom stave pri Cobbovej-Douglasovej produkčnej funkcii je podľa {eq}`r_CD_us`

$$
k^*=\left[\frac{sA}{n+\delta}\right]^{\frac{1}{1-\alpha}}.
$$

Ak má byť ustálený stav zároveň stavom, v ktorom je intenzívna spotreba maximálna, platí

$$
k^{**}=\left[\frac{s^*A}{n+\delta}\right]^{\frac{1}{1-\alpha}}.
$$

Dosadením do podmienky zlatého pravidla získame

$$
\alpha A\left[\left(\frac{s^*A}{n+\delta}\right)^{\frac{1}{1-\alpha}}\right]^{\alpha-1}=n+\delta.
$$

Keďže

$$
\frac{\alpha-1}{1-\alpha}=-1
$$

dostaneme

$$
\alpha A\left( \frac{s^*A}{n+\delta}\right)^{-1}=n+\delta.
$$

Preto

$$
\alpha A\frac{n+\delta}{s^*A}=n+\delta.
$$

Po vykrátení $A$ a $n+\delta$ získame

$$
\frac{\alpha}{s^*}=1,
$$

a teda

```{math}
:label: r_s_zlate

s^*=\alpha.
```

Pri Cobbovej-Douglasovej produkčnej funkcii je preto intenzívna spotreba v ustálenom stave maximálna vtedy, keď sa **miera sporenia rovná podielu kapitálu na výstupe**.

#### Python

Na odvodenie využijeme symboly, ktoré sme deklarovali v príklade {ref}`priklad_solow_zaklad`. Keďže sme si prepísali kapitál v ustálenom stave `k_us`, vyjadríme ho ešte raz.

````python
# Odvodenie kapitálu v ustálenom stave ešte raz
k_us=solve(ss*As*k_star**alphas-(deltas+ns)*k_star,k_star,dict=True)
k_us=k_us[0]

s_golden=solve(alphas*As*k_us[k_star]**(alphas-1)-ns-deltas,ss,dict=True)
s_golden=s_golden[0]
print("Miera sporenia v zlatom pravidle")
simplify(s_golden[ss])
````

Keďže sme v kóde deklarovali iba pozitívnosť parametra $\alpha$, SymPy neuvažuje predpoklad $0<\alpha<1$, ktorý používame v Cobbovej-Douglasovej produkčnej funkcii. Preto vo výstupe nie je ďalšie zjednodušenie výrazu. Pri $0<\alpha<1$ však z odvodenia vyplýva {eq}`r_s_zlate`.
