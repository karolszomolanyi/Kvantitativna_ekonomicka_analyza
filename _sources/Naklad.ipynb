# Náklady firmy

- Krátkodobé a dlhodobé hľadisko firmy:
    - z dlhodobého hľadiska sú všetky vstupy *variabilné*,
    - z krátkodobého existujú *fixné* vstupy.

Variabilné vstupy
: Výrobné faktory, ktorých objem môže firma meniť aj v krátkom období.

Fixné vstupy
: Výrobné faktory, ktorých objem je v krátkom období daný a firma ho nemôže meniť (výrobné haly, stroje a pod.)

Aby sme zjednodušili analýzu, predpokladajme, že firma využíva iba dva vstupy. Kapitál $K$ je fixný vstup a práca $N$ je variabilný vstup,

$$
\mathbf{x}
=
\begin{pmatrix}
K\\
N
\end{pmatrix}.
$$

Ceny vstupov sú cena jednotky kapitálu a mzdová sadzba

$$
\mathbf{c}
=
\begin{pmatrix}
q\\
\omega
\end{pmatrix}.
$$

## Krátkodobé náklady

Celkové náklady firmy sú súčtom nákladov na spotrebu fixného a variabilného vstupu:

$$
C=qK+\omega N,
$$

kde $q$ je cena jednotky kapitálu a $\omega$ mzdová sadzba.

Do nákladovej funkcie možno za vstupy dosadiť podmienené dopytové funkcie {eq}`podmieneny_dopyt` a získať nákladovú funkciu {eq}`nakladova_funkcia`.

Z krátkodobého hľadiska je variabilný iba vstup $N$, ktorého dopyt je $N(y,q,\omega)$.

Vstup $K$ je fixný, preto nezávisí od objemu výroby ani od cien vstupov.

**Krátkodobá nákladová funkcia** je v tvare

```{math}
:label: SC
SC(y,K,q,\omega)
=
qK
+
\omega N(y,q,\omega).
```

Prvý člen $qK$ predstavuje **fixné náklady**.

Definujme funkcie krátkodobých variabilných nákladov, krátkodobých priemerných nákladov, krátkodobých priemerných variabilných nákladov a krátkodobých hraničných nákladov.

Krátkodobé variabilné náklady
: Krátkodobé náklady po odpočítaní fixných nákladov,

```{math}
:label: SVC
SVC(y,q,\omega)
=
\omega N(y,q,\omega).
```

Krátkodobé priemerné náklady
: Podiel krátkodobých nákladov na objeme výroby,

```{math}
:label: SAC
SAC(y)
=
\frac{SC(y,K,q,\omega)}{y}.
```

Krátkodobé priemerné variabilné náklady
: Podiel krátkodobých variabilných nákladov na objeme výroby,

```{math}
:label: SAVC

SAVC(y)
=
\frac{SVC(y,q,\omega)}{y}.
```

Krátkodobé hraničné náklady
: Výstupný {ref}`multiplikator-label` krátkodobých nákladov,

```{math}
:label: SMC

SMC(y)
=
\frac{\partial SC(y,K,q,\omega)}
{\partial y}.
```

## Dlhodobé náklady

Z dlhodobého hľadiska sú všetky vstupy variabilné. Firma preto hľadá optimálny objem spotreby nielen variabilných, ale aj krátkodobo fixných vstupov.

Popri spotrebe práce plánuje aj optimálnu veľkosť výrobných hál, strojového vybavenia a ďalších výrobných kapacít.

Dlhodobá nákladová funkcia má tvar

$$
C(y,q,\omega)
=
qK(y,q,\omega)
+
\omega N(y,q,\omega).
$$

Definujme funkcie dlhodobých priemerných nákladov a dlhodobých hraničných nákladov.

Dlhodobé priemerné náklady
: podiel dlhodobej nákladovej funkcie na výstupe,

```{math}
:label: LAC
LAC = \frac{C(y,q,w)}{y}.
```

Dlhodobé hraničné náklady
: výstupný multiplikátor dlhodobých nákladov,

```{math}
:label: LMC
LMC = \frac{\partial{C(y,q,w)}}{\partial{y}}.
```

## Vlastnosti krátkodobých nákladov

Na {Numref}`kratkodobe_naklady` sú zobrazené priebehy definovaných krátkodobých funkcií.  

````python
# Priemerné, priemerné varibilné a hraničné náklady

# potrebné knižnice
import numpy as np
import matplotlib.pyplot as plt #kreslenie grafov

def hn(y):
    MC = 0.126*y**2 - 1.8*y + 11 # hraničné náklady
    return MC

def pvn(y):
    AVC = 0.042*y**2 - 0.9*y + 11 # priemerné variabilné náklady
    return AVC

def pn(y):
    AC = 0.042*y**2 - 0.9*y + 11 +7/y # priemerné náklady
    return AC

y = np.linspace(1e-12,20,100) # definičný obor funkcií

naklady = (hn(y), pvn(y), pn(y)) # do listu troch vektorov vlož funkčné hodnoty hraničných, priemerných variabilných 
                                # a priemerných nákladov
popisky = ("SMC", "SAVC", "SAC")

fig,ax = plt.subplots(figsize = [10,8]) # ideme kresliť nákladové funkcie

# v každom cykle vyber z listu naklady a z listu popisky jednu hodnotu do n (naklady) a do p (popisky) v danom poradí
for n,p in zip(naklady,popisky): 
    ax.plot(y,n, label = p)

ax.set_xlabel("$y$") # názov osi x
ax.set_ylabel("náklady na jednotku výstupu")

ax.set_ylim([4, 12]) # rozmer osi y

ax.legend() # zobraz legendu
plt.show() # zobraz graf
````

$$
SMC=SAVC.

### Vzťah medzi hraničnými a priemernými nákladmi

Podľa prvej vlastnosti priebeh krátkodobých priemerných nákladov $SAC$ závisí od ich porovnania s krátkodobými hraničnými nákladmi $SMC$.

Priebeh krátkodobých priemerných nákladov možno vyjadriť prvou deriváciou:

$$
\frac{\partial SAC}{\partial y}
=
\frac{\partial}{\partial y}
\left(
\frac{SC(y,q,\omega)}{y}
\right).
$$

Ak je táto derivácia kladná, krátkodobé priemerné náklady rastú, ak je záporná, krátkodobé priemerné náklady klesajú. V bode minima sa rovná nule.

Po rozpísaní priemerných nákladov a použití pravidla pre deriváciu podielu platí

$$
\frac{\partial}{\partial y}
\left(
\frac{SC(y,q,\omega)}{y}
\right)
=
\frac{
\frac{\partial SC(y,q,\omega)}{\partial y}\,y
-
SC(y,q,\omega)
}
{y^2}.
$$

Po dosadení

$$
SMC
=
\frac{\partial SC(y,q,\omega)}{\partial y}
$$

dostaneme

$$
\frac{
SMC\cdot y
-
SC(y,q,\omega)
}
{y^2}.
$$

Keďže

$$
SAC
=
\frac{SC(y,q,\omega)}{y},
$$

platí

$$
\frac{
SMC\cdot y
-
SC(y,q,\omega)
}
{y^2}
=
\frac{
SMC
-
\frac{SC(y,q,\omega)}{y}
}
{y}
=
\frac{SMC-SAC}{y}.
$$

Preto

$$
\frac{\partial SAC}{\partial y}
=
\frac{SMC-SAC}{y}.
$$

Znamienko prvej derivácie priemerných nákladov závisí od znamienka čitateľa zlomku na pravej strane $SMC - SAC$.

```{table} Vzťah medzi priemernými a hraničnými nákladmi
:name: tab-SMC_SAC

|--------------------------|---------------|
| $SMC < SAC$ | $SAC$ klesajú |

| $SMC > SAC$ | $SAC$ rastú |
```

### Vzťah medzi hraničnými a priemernými variabilnými nákladmi

Podľa druhej vlastnosti priebeh krátkodobých priemerných variabilných nákladov $SAVC$ závisí od ich porovnania s krátkodobými hraničnými nákladmi $SMC$.

Priebeh krátkodobých priemerných variabilných nákladov možno vyjadriť prvou deriváciou:

$$
\frac{\partial SAVC}{\partial y}
=
\frac{\partial}{\partial y}
\left(
\frac{SVC(y,q,\omega)}{y}
\right).
$$

Ak je táto derivácia kladná, krátkodobé priemerné variabilné náklady rastú, ak je záporná, krátkodobé priemerné variabilné náklady klesajú. V bode minima sa rovná nule.

Po rozpísaní priemerných variabilných nákladov a použití pravidla pre deriváciu podielu platí

$$
\frac{\partial}{\partial y}
\left(
\frac{SVC(y,q,\omega)}{y}
\right)
=
\frac{
\frac{\partial SVC(y,q,\omega)}{\partial y}\,y
-
SVC(y,q,\omega)
}
{y^2}.
$$

Keďže fixné náklady nezávisia od objemu výroby,

$$
SC(y,q,\omega)
=
SFC
+
SVC(y,q,\omega),
$$

platí

$$
\frac{\partial SVC(y,q,\omega)}{\partial y}
=
\frac{\partial SC(y,q,\omega)}{\partial y}
=
SMC.
$$

Po dosadení dostaneme

$$
\frac{
SMC\cdot y
-
SVC(y,q,\omega)
}
{y^2}.
$$

Keďže

$$
SAVC
=
\frac{SVC(y,q,\omega)}{y},
$$

platí

$$
\frac{
SMC\cdot y
-
SVC(y,q,\omega)
}
{y^2}
=
\frac{
SMC
-
\frac{SVC(y,q,\omega)}{y}
}
{y}
=
\frac{SMC-SAVC}{y}.
$$

Preto

$$
\frac{\partial SAVC}{\partial y}
=
\frac{SMC-SAVC}{y}.
$$

Znamienko prvej derivácie priemerných variabilných nákladov závisí od znamienka čitateľa zlomku na pravej strane $SMC-SAVC$.

Výsledok možno zhrnúť do nasledujúcej tabuľky.

```{table} Vzťah medzi priemernými variabilnými a hraničnými nákladmi
:name: tab-SMC_SAVC

| Vzťah medzi $SMC$ a $SAVC$ | Priebeh $SAVC$ |
|---------------------------|----------------|
| $SMC < SAVC$ | $SAVC$ klesajú |
| $SMC = SAVC$ | $SAVC$ dosahujú minimum |
| $SMC > SAVC$ | $SAVC$ rastú |
```

Pri tretej vlastnosti využijeme L'Hôpitalovo pravidlo. Ak

$$
\lim_{x\rightarrow 0}f(x)=0,
$$

potom platí

$$
\lim_{x\rightarrow 0}
\frac{f(x)}{x}
=
\lim_{x\rightarrow 0}
\frac{\partial f(x)}{\partial x}.
$$

Skúmajme správanie krátkodobých priemerných variabilných nákladov pri veľmi nízkom výstupe:

$$
\lim_{y\rightarrow 0}
SAVC
=
\lim_{y\rightarrow 0}
\frac{SVC(y,q,\omega)}{y}.
$$

Použitím L'Hôpitalovho pravidla získame

$$
\lim_{y\rightarrow 0}
\frac{SVC(y,q,\omega)}{y}
=
\lim_{y\rightarrow 0}
\frac{\partial SVC(y,q,\omega)}
{\partial y}.
$$

Keďže

$$
\frac{\partial SVC(y,q,\omega)}
{\partial y}
=
SMC,
$$

platí

$$
\lim_{y\rightarrow 0}
SAVC
=
\lim_{y\rightarrow 0}
SMC.
$$

Pri veľmi nízkom objeme výroby sú preto krátkodobé priemerné variabilné náklady a krátkodobé hraničné náklady takmer rovnaké.

### Priemerné náklady

S rastom výstupu $y$ sa krátkodobé priemerné náklady a krátkodobé priemerné variabilné náklady k sebe približujú.

Platí totiž

$$
SAC
=
SAVC
+
\frac{qK}{y}.
$$

Preto

$$
\lim_{y \rightarrow \infty} SAC
=
\lim_{y \rightarrow \infty}
\left(
SAVC+\frac{qK}{y}
\right).
$$

Použitím pravidiel pre limity dostaneme

$$
\lim_{y \rightarrow \infty} SAC
=
\lim_{y \rightarrow \infty} SAVC
+
\lim_{y \rightarrow \infty} \frac{qK}{y}.
$$

Keďže fixné náklady $qK$ sú konštantné,

$$
\lim_{y \rightarrow \infty}
\frac{qK}{y}
=
0.
$$

Preto

$$
\lim_{y \rightarrow \infty} SAC
=
\lim_{y \rightarrow \infty} SAVC.
$$

Pri vysokom objeme výroby je príspevok fixných nákladov na jednotku produkcie zanedbateľný a krátkodobé priemerné náklady sa približujú ku krátkodobým priemerným variabilným nákladom.

Podľa poslednej vlastnosti, pri veľmi nízkom objeme výroby sú krátkodobé priemerné náklady veľmi vysoké.

Platí

$$
SAC
=
SAVC
+
\frac{qK}{y}.
$$

Pri limite pre $y\rightarrow 0$ dostaneme

$$
\lim_{y\rightarrow 0}
SAC
=
\lim_{y\rightarrow 0}
\left(
SAVC
+
\frac{qK}{y}
\right).
$$

Keďže fixné náklady $qK$ sú kladná konštanta,

$$
\lim_{y\rightarrow 0}
\frac{qK}{y}
=
+\infty.
$$

Preto

$$
\lim_{y\rightarrow 0}
SAC
=
+\infty.
$$

Pri veľmi nízkom objeme výroby sa fixné náklady rozpočítavajú na malý počet jednotiek produkcie, čo spôsobuje prudký rast krátkodobých priemerných nákladov.

### Vzťah medzi produkčnou funkciou a krátkodobými nákladmi

Krátkodobé variabilné náklady {eq}`SVC` možno zapísať aj v tvare vyjadrujúcom približné zvýšenie variabilných nákladov spôsobené zvýšením množstva práce:

$$
dSVC=\omega\, dN.
$$

Hraničný produkt práce {eq}`MPN` vyjadruje približné zvýšenie výstupu spôsobené jednotkovým zvýšením množstva práce.

Obrátená hodnota hraničného produktu práce preto vyjadruje približné množstvo práce potrebné na výrobu dodatočnej jednotky produkcie:

$$
dN=\frac{1}{MPN}dy.
$$

Dosadením do predchádzajúceho vzťahu dostaneme

$$
dSVC=
\frac{\omega}{MPN}dy.
$$

Keďže prvá derivácia krátkodobých variabilných nákladov podľa produkcie sa rovná krátkodobým hraničným nákladom, platí

```{math}
:label: r_SMC_MPN
SMC
=
\frac{dSVC}{dy}
=
\frac{\omega}{MPN}.
```

Vzťah ukazuje, že hraničné náklady závisia od ceny práce a jej hraničného produktu. Čím vyšší je hraničný produkt práce, tým menšie množstvo práce je potrebné na výrobu dodatočnej jednotky produkcie a tým nižšie sú hraničné náklady.

Ak sa pri raste produkcie hraničný produkt práce znižuje, na výrobu dodatočnej jednotky produkcie je potrebné stále väčšie množstvo práce. Rast krátkodobých hraničných nákladov je preto dôsledkom zákona klesajúcich hraničných výnosov práce.

```{Table} Vzťah medzi produkčnou funkciou a krátkodobými nákladmi
:name: produkcna_funkcia_k_naklady

| Hraničný produkt práce $MPN$ | Krátkodobé hraničné náklady $SMC$ | Krátkodobé celkové náklady $SC$ |
|------------------------|------------------|-----------------|
| $MPN$ rastie | $SMC$ klesajú | $SC$ sú konkávne |
| $MPN$ dosahuje maximum | $SMC$ dosahujú minimum | $SC$ majú inflexný bod |
| $MPN$ klesá | $SMC$ rastú | $SC$ sú konvexné |
```

## Vlastnosti dlhodobých nákladov

Analogicky ako pri krátkodobých nákladoch platia nasledovné vlastnosti.

1. Dlhodobé priemerné náklady klesajú, ak $LMC<LAC$, rastú, ak $LMC>LAC$, a v bode minima platí

$$
LMC=LAC.
$$

2. Pri veľmi nízkom objeme výroby sú dlhodobé hraničné náklady približne rovnaké ako dlhodobé priemerné náklady.

### Vzťah medzi hraničnými a priemernými nákladmi

Dlhodobé priemerné náklady sú

$$
LAC=\frac{C(y,q,\omega)}{y}.
$$

Ich prvá derivácia je

$$
\frac{\partial LAC}{\partial y}
=
\frac{\partial}{\partial y}
\left(
\frac{C(y,q,\omega)}{y}
\right).
$$

Po použití pravidla pre deriváciu podielu dostaneme

$$
\frac{\partial LAC}{\partial y}
=
\frac{
LMC\cdot y
-
C(y,q,\omega)
}
{y^2}.
$$

Keďže

$$
LAC
=
\frac{C(y,q,\omega)}{y},
$$

platí

$$
\frac{\partial LAC}{\partial y}
=
\frac{LMC-LAC}{y}.
$$

Preto:

```{table} Vzťah medzi priemernými a hraničnými dlhodobými nákladmi
:name: tab-LMC_LAC

| Vzťah medzi $LMC$ a $LAC$ | Priebeh $LAC$ |
|--------------------------|---------------|
| $LMC < LAC$ | $LAC$ klesajú |
| $LMC = LAC$ | $LAC$ dosahujú minimum |
| $LMC > LAC$ | $LAC$ rastú |
```

### Správanie pri nízkom objeme výroby

Keďže

$$
LAC
=
\frac{C(y,q,\omega)}{y},
$$

a pri nulovom výstupe platí

$$
C(0,q,\omega)=0,
$$

môžeme použiť L'Hôpitalovo pravidlo:

$$
\lim_{y\rightarrow 0}LAC
=
\lim_{y\rightarrow 0}
\frac{\partial C(y,q,\omega)}
{\partial y}
=
\lim_{y\rightarrow 0}LMC.
$$

Pri veľmi nízkom objeme výroby sú preto dlhodobé priemerné a dlhodobé hraničné náklady približne rovnaké.

## Vzťah medzi krátkodobými a dlhodobými nákladmi

Krátkodobé náklady závisia od objemu fixných vstupov. Každá hodnota fixného vstupu preto určuje inú krátkodobú nákladovú funkciu.

Z dlhodobého hľadiska môže firma meniť všetky vstupy vrátane tých, ktoré sú v krátkom období fixné. Pre každý objem výroby si preto vyberie takú kombináciu vstupov, ktorá minimalizuje náklady.

Dlhodobá nákladová funkcia preto predstavuje najnižšie náklady, ktoré možno pri danom objeme výroby dosiahnuť.

Grafická interpretácia vzťahov medzi dlhodobými a krátkodobými nákladmi je na {ref}`priemerne_naklady` a {ref}`vztahy_naklady`.

Dlhodobá krivka priemerných nákladov je dolným obalom krátkodobých kriviek priemerných nákladov. V bodoch dotyku platí

$$
LAC=SAC(y,K_i).
$$

V rovnakých bodoch sa zároveň rovnajú aj dlhodobé a krátkodobé hraničné náklady:

$$
LMC=SMC(y,K_i).
$$

Bod $y(K_1)$ leží v oblasti rastúcich výnosov z rozsahu. Pri zvýšení objemu fixného vstupu môže firma v dlhom období znížiť priemerné náklady.

Bod $y(K_2)$ zodpovedá minimu krivky dlhodobých priemerných nákladov. Firma disponuje optimálnym objemom fixného vstupu.

Bod $y(K_3)$ leží v oblasti klesajúcich výnosov z rozsahu. Pri znížení objemu fixného vstupu môže firma v dlhom období znížiť priemerné náklady.

Z grafu vyplývajú nasledovné vlastnosti.

1. Dlhodobé priemerné náklady nikdy nie sú vyššie ako krátkodobé priemerné náklady

$$
LAC(y)\le SAC(y,K).
$$

2. Dlhodobá krivka priemerných nákladov je dolným obalom krátkodobých kriviek priemerných nákladov.

3. Pre každý objem výroby existuje krátkodobá krivka priemerných nákladov, ktorá sa dotýka dlhodobej krivky priemerných nákladov.

4. V bode dotyku platí

$$
LAC=SAC.
$$

5. V bode dotyku majú krátkodobé a dlhodobé hraničné náklady rovnakú hodnotu,

$$
LMC=SMC.
$$



Dlhodobé náklady vznikajú minimalizáciou krátkodobých nákladov podľa objemu fixných vstupov:

$$
C(y,q,\omega)
=
\min_K SC(y,K,q,\omega).
$$

Preto pre ľubovoľný objem fixného vstupu platí

$$
C(y,q,\omega)
\le
SC(y,K,q,\omega).
$$

Po vydelení objemom výroby dostaneme

$$
LAC(y)
\le
SAC(y,K).
$$

Dlhodobá krivka priemerných nákladov preto leží pod všetkými krátkodobými krivkami priemerných nákladov a tvorí ich dolný obal.

Pre každý objem výroby existuje optimálny objem fixného vstupu $K^*(y)$, pre ktorý platí

$$
C(y,q,\omega)
=
SC(y,K^*(y),q,\omega).
$$

Preto

$$
LAC(y)
=
SAC(y,K^*(y)).
$$

V každom bode dlhodobej krivky sa teda nachádza krátkodobá krivka, ktorá sa jej dotýka.

Dlhodobá krivka priemerných nákladov je preto dolným obalom krátkodobých kriviek priemerných nákladov.

V bode dotyku krátkodobej a dlhodobej nákladovej funkcie sa krátkodobé a dlhodobé hraničné náklady rovnajú.

Dôkaz vyplýva z derivácie rovnosti

$$
C(y,q,\omega)
=
SC(y,K^*(y),q,\omega)
$$

podľa objemu výroby a zo skutočnosti, že optimálny objem fixného vstupu $K^*(y)$ minimalizuje krátkodobé náklady.

Platí

$$
LMC
=
SMC(K^*(y)).
$$

## Ponuka firmy

Ponuková funkcia firmy vyplýva z úlohy maximalizácie zisku výberom výstupu. Podmienka prvého rádu úlohy je {eq}`PPR_y`.

### Krátkodobá ponuková funkcia

Dosadením funkcie krátkodobých hraničných nákladov {eq}`SMC` do podmienky prvého rádu {eq}`PPR_y` získame

$$
p
=
SMC(y,K,q,\omega).
$$

V podmienkach dokonalej konkurencie považuje firma cenu výstupu $p$ za danú. O objeme výroby rozhoduje porovnaním ceny s nákladmi na jednotku výstupu.

Ak platí

$$
p>SAC,
$$

firma dosahuje kladný ekonomický zisk. Tržba z jednej jednotky výstupu prevyšuje priemerné náklady na jej výrobu.

Ak platí

$$
p=SAC=SMC,
$$

firma dosahuje nulový ekonomický zisk. Tržby presne pokrývajú všetky ekonomické náklady vrátane implicitných nákladov vlastných výrobných faktorov.

Ak platí

$$
SAVC \le p < SAC,
$$

firma dosahuje stratu. Napriek tomu pokračuje vo výrobe, pretože tržby pokrývajú všetky variabilné náklady a zároveň aspoň časť fixných nákladov. Strata pri pokračovaní vo výrobe je preto menšia ako strata pri zastavení výroby.

Ak platí

$$
p<SAVC,
$$

firma výrobu zastaví. Tržby už nestačia ani na pokrytie variabilných nákladov. Pokračovanie vo výrobe by znamenalo, že firma musí uhrádzať nielen fixné náklady, ale aj časť variabilných nákladov z vlastných zdrojov. Výhodnejšie je preto výrobu zastaviť a znášať iba fixné náklady.

Geometrická interpretácia zisku firmy je na {numref}`graf_zisk`. Zisk z jednotky výroby je $p - SAC(y^*)$. Vynásobením realizovaným objemom výroby získame zisk $ [ p - SAC(y^* ) ] y^*$. Možno sa o tom presvedčiť aj zo ziskovej funkcie

$$
\pi = p y^* - SC(y^*) =
p y^* - SAC(y^*) y^* =
[ p - SAC(y^* ) ] y^*
$$

````python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

# Hraničné náklady
def hn(y):
    return 0.126 * y**2 - 1.8 * y + 11

# Priemerné variabilné náklady
def pvn(y):
    return 0.042 * y**2 - 0.9 * y + 11

# Priemerné náklady
def pn(y):
    return 0.042 * y**2 - 0.9 * y + 11 + 7 / y


# Definičný obor
y = np.linspace(0.5, 20, 5000)

# Funkčné hodnoty
SMC = hn(y)
SAVC = pvn(y)
SAC = pn(y)

# Minimum SAC (nulový ekonomický zisk)
idx_min = np.argmin(SAC)
y_zero = y[idx_min]
p_zero = SAC[idx_min]

# Tri cenové hladiny
p_profit = p_zero + 1.2
p_loss = p_zero - 1.2

prices = [p_profit, p_zero, p_loss]
titles = ["Kladný zisk", "Nulový zisk", "Strata"]

# Obrázok s troma panelmi
fig, axes = plt.subplots(1, 3, figsize=(18, 6), sharey=True)

for ax, p, title in zip(axes, prices, titles):

    # Nákladové krivky
    ax.plot(y, SMC, label="SMC", linewidth=2)
    ax.plot(y, SAVC, label="SAVC", linewidth=2)
    ax.plot(y, SAC, label="SAC", linewidth=2)

    # Nulový zisk
    if title == "Nulový zisk":

        y_star = y_zero
        AC_star = p_zero

    else:

        # Pravý priesečník MC = p
        crossings = np.where(np.diff(np.sign(SMC - p)))[0]
        idx = crossings[-1]

        x1 = y[idx]
        x2 = y[idx + 1]

        mc1 = SMC[idx]
        mc2 = SMC[idx + 1]

        # Presnejší priesečník pomocou interpolácie
        y_star = x1 + (p - mc1) * (x2 - x1) / (mc2 - mc1)

        AC_star = pn(y_star)

    # Cenová čiara
    ax.axhline(
        p,
        color="black",
        linestyle="--",
        linewidth=1.8,
        label="p"
    )

    # y*
    ax.vlines(
        y_star,
        ymin=0,
        ymax=max(p, AC_star),
        colors="gray",
        linestyles=":",
        linewidth=1.5
    )

    # Pomocná vodorovná čiara AC(y*)
    ax.hlines(
        AC_star,
        xmin=0,
        xmax=y_star,
        colors="gray",
        linestyles=":",
        linewidth=1.2
    )

    # Ekonomický zisk / strata
    profit = p - AC_star

    if profit > 0.02:

        rect = Rectangle(
            (0, AC_star),
            y_star,
            profit,
            facecolor="green",
            edgecolor="green",
            alpha=0.30
        )

        ax.add_patch(rect)

        ax.text(
            y_star / 2,
            (p + AC_star) / 2,
            "zisk",
            ha="center",
            va="center"
        )

    elif profit < -0.02:

        rect = Rectangle(
            (0, p),
            y_star,
            AC_star - p,
            facecolor="red",
            edgecolor="red",
            alpha=0.30
        )

        ax.add_patch(rect)

        ax.text(
            y_star / 2,
            (p + AC_star) / 2,
            "strata",
            ha="center",
            va="center"
        )

    # Bod AC(y*)
    ax.plot(y_star, AC_star, "ko", markersize=5)

    ax.annotate(
        r"$y^*$",
        xy=(y_star, 4.1),
        xytext=(6, 0),
        textcoords="offset points",
        fontsize=11
    )

    # Označenie ceny
    ax.text(
        0.3,
        p + 0.05,
        r"$p$",
        fontsize=11
    )

    ax.set_title(title)
    ax.set_xlabel(r"$y$")
    ax.set_xlim(0, 20)
    ax.set_ylim(4, 12)

axes[0].set_ylabel("náklady na jednotku výstupu")
axes[0].legend(loc="upper right")

plt.tight_layout()
plt.show()
````

### Dlhodobá ponuková funkcia

Dosadením funkcie dlhodobých hraničných nákladov {eq}`LMC` do podmienky prvého rádu {eq}`PPR_y` získame

$$
p
=
LMC(y,K,q,\omega).
$$

Keďže v dlhom období firma optimálne prispôsobuje aj množstvo kapitálu, dlhodobé hraničné náklady závisia iba od výstupu a cien výrobných faktorov. Preto platí

$$
p
=
LMC(y,q,\omega).
$$

Optimálny výstup firmy je určený priesečníkom ceny a krivky dlhodobých hraničných nákladov. Firma však vyrába iba vtedy, ak cena pokrýva minimálne dlhodobé priemerné náklady. Dlhodobú ponukovú funkciu preto tvorí rastúca časť krivky dlhodobých hraničných nákladov ležiaca nad minimom krivky dlhodobých priemerných nákladov {eq}`LAC`.

````python
# potrebné knižnice
import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(
    1, 2,
    figsize=(14, 6)
)

# ==================================================
# KRÁTKODOBÁ PONUKA
# ==================================================

def hn(y):
    return 0.126*y**2 - 1.8*y + 11

def pvn(y):
    return 0.042*y**2 - 0.9*y + 11

def pn(y):
    return 0.042*y**2 - 0.9*y + 11 + 7/y

y = np.linspace(1e-12, 20, 500)

SMC = hn(y)
SAVC = pvn(y)
SAC = pn(y)

ax[0].plot(y, SMC, label="SMC")
ax[0].plot(y, SAVC, label="SAVC")
ax[0].plot(y, SAC, label="SAC")

# minimum SAVC

idx_savc = np.argmin(SAVC)

y_s = y[idx_savc]
savc_min = SAVC[idx_savc]

# krátkodobá ponuka

mask = y >= y_s

ax[0].plot(
    y[mask],
    SMC[mask],
    color="red",
    linewidth=4,
    label="krátkodobá ponuka"
)

ax[0].scatter(
    y_s,
    savc_min,
    color="black",
    zorder=10
)

ax[0].text(
    y_s + 0.4,
    savc_min - 0.2,
    r"$\min (SAVC)$"
)

ax[0].axvline(
    y_s,
    linestyle=":",
    color="gray"
)

ax[0].set_title("Krátkodobá ponuka firmy")

ax[0].set(
    xlabel="$y$",
    ylabel="náklady na jednotku výstupu",
    ylim=(4,12)
)

ax[0].legend()

# ==================================================
# DLHODOBÁ PONUKA
# ==================================================

y = np.linspace(1,20,500)

LAC = 1 + 12/y + 0.15*y

LC = y * LAC

LMC = np.gradient(LC,y)

ax[1].plot(
    y,
    LAC,
    color="black",
    linewidth=3,
    label="LAC"
)

ax[1].plot(
    y,
    LMC,
    color="tab:blue",
    linewidth=2.5,
    label="LMC"
)

# minimum LAC

idx_lac = np.argmin(LAC)

y_l = y[idx_lac]
lac_min = LAC[idx_lac]

# dlhodobá ponuka

mask = y >= y_l

ax[1].plot(
    y[mask],
    LMC[mask],
    color="red",
    linewidth=4,
    label="dlhodobá ponuka"
)

ax[1].scatter(
    y_l,
    lac_min,
    color="black",
    zorder=10
)

ax[1].scatter(
    y_l,
    lac_min,
    color="black",
    zorder=10
)

ax[1].text(
    y_l + 0.4,
    lac_min - 0.2,
    r"$\min LAC$"
)

ax[1].axvline(
    y_l,
    linestyle=":",
    color="gray"
)

ax[1].set_title("Dlhodobá ponuka firmy")

ax[1].set(
    xlabel="$y$",
    ylabel="náklady na jednotku výstupu",
    ylim=(1,8)
)

ax[1].legend()

plt.tight_layout()
plt.show()
````

## Agregácia

V aplikovanej ekonómii pochádzajú údaje často z pozorovaní agregovaných premenných, ako je celkový výstup odvetvia, dopyt odvetvia po určitom výrobnom faktore alebo celkový zisk v sektore. Doteraz sme diskutovali len o produkcii, vstupe a zisku jednej firmy. Je potrebné rozšíriť tieto diskusie aj o agregované premenné. Klasickým prístupom k tomuto problému je predpokladať, že agregované premenné sú generované jedinou firmou — fiktívnou firmou. Ak je to tak, všetky doterajšie, sú okamžite aplikovateľné aj na agregované premenné.
Konkrétne, ak máme $n$ nezávislých firiem, môžeme agregované ich ponukové a ziskové funkcie považovať za ponukovú a ziskovú funkciu jednej firmy? Odpoveď je áno za určitých podmienok.
Predpokladajme, že v ekonomike je $m$ firiem alebo závodov. Každá firma $i$ ponúka výstup podľa ponukovej funkcie $s_i(p)$ pre všetky $ i = \{1,2, \cdots m \}$. Agregovaná ponuková funkcia je

$$
s(p) = \sum_{i=1}^m {s_i(p)}.
$$

$$
\boxed{LMC(y)=\frac{4}{2^{2/3}}y^{1/3}}.
$$

**Krivka dlhodobých priemerných nákladov tvorí spodný obal rodiny krátkodobých priemerných nákladov.**

Krátkodobé priemerné náklady majú tvar

$$
SAC(y,K)
=
\frac{K^2}{y}
+
\frac{y}{K}.
$$

Pre každú pevnú hodnotu kapitálu $K$ dostaneme jednu krivku krátkodobých priemerných nákladov.

V predchádzajúcej úlohe sme ukázali, že náklady sú minimálne pri zásobe kapitálu

$$
K^*(y)
=
\left(
\frac{y^2}{2}
\right)^{1/3}.
$$

Dosadením optimálneho kapitálu do funkcie $SAC$ získame

$$
SAC(y,K^*(y))
=
\frac{\left(\frac{y^2}{2}\right)^{2/3}}{y}
+
\frac{y}
{\left(\frac{y^2}{2}\right)^{1/3}}.
$$

Po úprave

$$
SAC(y,K^*(y))
=
\frac{3}{2^{2/3}}
y^{1/3}.
$$

To je presne funkcia dlhodobých priemerných nákladov

$$
LAC(y)
=
\frac{3}{2^{2/3}}
y^{1/3}.
$$

Platí teda

$$
LAC(y)
=
SAC(y,K^*(y)).
$$

Keďže firma pre každý objem výroby vyberá takú zásobu kapitálu, ktorá minimalizuje náklady, platí pre ľubovoľné $K$

$$
LAC(y)
\le
SAC(y,K).
$$

Rovnosť nastáva iba pri optimálnej zásobe kapitálu

$$
K=K^*(y).
$$

Krivka $LAC$ sa preto v každom bode dotýka jednej z kriviek $SAC$ a súčasne leží pod všetkými ostatnými krivkami $SAC$.

Preto krivka dlhodobých priemerných nákladov tvorí spodný obal rodiny krátkodobých priemerných nákladov.

$$
\boxed{
LAC(y)
=
\min_K SAC(y,K)
}
$$

**Krátkodobá ponuková funkcia**

Krátkodobé hraničné náklady majú tvar

$$
SMC(y,K)
=
\frac{2y}{K}.
$$

V krátkom období je kapitál fixný, preto optimálny objem výroby určuje podmienka

$$
p
=
SMC(y,K).
$$

Po dosadení dostaneme

$$
p
=
\frac{2y}{K}.
$$

Po vyjadrení výstupu:

$$
y
=
\frac{K}{2}p.
$$

Krátkodobá ponuková funkcia firmy je teda

$$
\boxed{
y_S(p,K)
=
\frac{K}{2}p.
}
$$

Keďže

$$
SAC(y,K)
=
\frac{K^2}{y}
+
\frac{y}{K},
$$

v tomto modeli neexistujú fixné náklady nezávislé od objemu výroby a platí

$$
SAVC(y,K)=SAC(y,K).
$$

Minimum priemerných nákladov nastáva v bode

$$
y=\frac{K^{3/2}}{\sqrt2},
$$

pričom

$$
p_{\min}
=
\sqrt{2K}.
$$

Firma preto vyrába iba vtedy, ak

$$
p\ge\sqrt{2K}.
$$

Krátkodobú ponukovú funkciu možno zapísať v tvare

$$
y_S(p,K)
=
\begin{cases}
0,
&
p<\sqrt{2K},
\\[6pt]
\dfrac{K}{2}p,
&
p\ge\sqrt{2K}.
\end{cases}
$$

**Dlhodobá ponuková funkcia**

V dlhom období môže firma meniť aj zásobu kapitálu. Dlhodobé hraničné náklady majú tvar

$$
LMC(y)
=
\frac{4}{2^{2/3}}
y^{1/3}.
$$

Optimálny výstup určuje podmienka

$$
p
=
LMC(y).
$$

Po dosadení:

$$
p
=
\frac{4}{2^{2/3}}
y^{1/3}.
$$

Vyjadríme výstup:

$$
y
=
\left(
\frac{2^{2/3}}{4}p
\right)^3.
$$

Po úprave dostaneme

$$
\boxed{
y_L(p)
=
\frac{p^3}{32}.
}
$$

Keďže

$$
LAC(y)
=
\frac{3}{2^{2/3}}
y^{1/3}
$$

je rastúca funkcia, neexistuje konečná cena, pri ktorej by firma dosahovala nulový ekonomický zisk. Minimum funkcie $LAC$ sa nachádza v bode

$$
y=0.
$$

Dlhodobá ponuková funkcia je preto celá krivka $LMC$.

**Porovnanie**

Krátkodobá ponuka firmy je

$$
y_S(p,K)
=
\frac{K}{2}p,
$$

teda lineárna funkcia ceny.

Dlhodobá ponuka firmy je

$$
y_L(p)
=
\frac{p^3}{32},
$$

pričom firma pri každom objeme výroby optimálne prispôsobuje zásobu kapitálu.

V dlhom období preto reaguje na zmenu ceny výraznejšie než v krátkom období, čo je prejav Le Chatelierovho princípu.

Na {numref}`graf_priklad2` sú zobrazené funkcie krátkodobých priemerných nákladov pre rôzne hodnoty kapitálu. Krivka dlhodobých priemerných nákladov tvorí ich spodný obal.

````python
import numpy as np
import matplotlib.pyplot as plt

# interval produkcie

y = np.linspace(0.2, 25, 2000)

# dlhodobé priemerné náklady

LAC = 3 / (2**(2/3)) * y**(1/3)

# rôzne úrovne kapitálu

K_values = [1, 2, 3, 4, 6, 8, 10]

plt.figure(figsize=(9, 6))

# rodina SAC

for K in K_values:

    SAC = K**2 / y + y / K

    plt.plot(
        y,
        SAC,
        lw=1.5,
        alpha=0.8,
        label=f"SAC, K={K}"
    )

    # bod dotyku SAC a LAC

    y_star = np.sqrt(2 * K**3)
    sac_star = K**2 / y_star + y_star / K

    plt.plot(
        y_star,
        sac_star,
        "ko",
        ms=4
    )

# LAC

plt.plot(
    y,
    LAC,
    color="navy",
    lw=3,
    label="LAC"
)

plt.xlabel("Výstup $y$")
plt.ylabel("Priemerné náklady")
plt.title("LAC ako spodný obal rodiny SAC")

plt.xlim(0, 25)
plt.ylim(0, 15)

plt.grid(alpha=0.3)

# legenda mimo grafu

plt.legend(
    loc="center left",
    bbox_to_anchor=(1.02, 0.5)
)

plt.tight_layout()
plt.show()
````

Krivka dlhodobých priemerných nákladov $LAC$ tvorí dolný obal rodiny krátkodobých priemerných nákladov $SAC$. Pre každý objem výroby si firma môže zvoliť takú zásobu kapitálu, ktorá minimalizuje priemerné náklady. Krivka $LAC$ preto v každom bode predstavuje najnižšie dosiahnuteľné priemerné náklady pri danom objeme produkcie.

#### Python

````python
Krátkodobé nákladové funkcie
````

````python
# symboly

y, K = symbols("y K", positive=True)

# krátkodobá nákladová funkcia

SC = K**2 + y**2/K

print("SC =", SC)

# krátkodobé priemerné náklady

SAC = simplify(SC/y)

print("SAC =", SAC)

# krátkodobé hraničné náklady

SMC = simplify(diff(SC, y))

print("SMC =", SMC)
````

````python
Optimálny kapitál pri danom objeme výroby.
````

````python
# optimálny kapitál

dSC_dK = simplify(diff(SC, K))

print("dSC/dK =", dSC_dK)

K_star = solve(dSC_dK, K,dict=True)
K_star = K_star[0] if type(K_star) is list else K_star

print(K_star)
````

````python
Dlhodobá nákladová funkcia.
````

````python
# dlhodobá nákladová funkcia

C = simplify(SC.subs(K_star))

print("C(y) =", C)
````

````python
# dlhodobé priemerné náklady

LAC = simplify(C/y)

print("LAC =", LAC)

# dlhodobé hraničné náklady

LMC = simplify(diff(C, y))

print("LMC =", LMC)
````

````python
SAC_star = simplify(SAC.subs(K_star))

print("SAC(K*) =", SAC_star)
print("LAC      =", LAC)

print("Rozdiel medzi SAC(K*) a LAC =", simplify(SAC_star - LAC))
````

````python
# krátkodobá ponuka

s_SR = solve(p - SMC, y, dict = True)
s_SR = s_SR[0] if type(s_SR) is list else s_SR

print("kratkodoba ponuka =", s_SR)
````
