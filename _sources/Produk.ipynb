# Produkcia firmy

(vst-label)=
Vstupy a výstupy
: Firma spotrebúva statky a služby, ktoré vstupujú do výrobného procesu. Tieto veličiny nazývame *vstupy*. Výsledkom výrobného procesu sú *výstupy*. 


```{sidebar} Príklad
Mlyn (*firma*) produkuje múku a na jej výrobu spotrebuje pšenicu. *Vstup* firmy je pšenica, *výstup* je múka. 
```

Predpokladajme, že firma vyrába $y \ge 0$ jednotiek finálneho výstupu. Na výrobu spotrebuje $x \ge 0$ jednotiek vstupu.

(mnozina_vyrob_moznosti)=
## Množina výrobných možností

```{sidebar} Príklad
Ak firma spotrebuje 100 ton pšenice, technológia jej umožňuje vyrobiť najviac 10 ton múky. Každá produkcia od 0 do 10 ton je preto výrobnou možnosťou firmy.
```

Výrobné možnosti firmy
: *Výrobné možnosti* firmy sú obmedzené. Pri danom množstve vstupu $x$, firma môže vyrobiť iba obmedzené množstvo výstupu $y$.

Množina výrobných možností
: *Množina výrobných možností* firmy je množina všetkých prípustných možností výroby firmy. Množina výrobných možností firmy je zobrazená na {numref}`mnozina_vyr_moznosti`. Ak je množstvo vstupu aj výstupu nekonečne deliteľné, množina výrobných možností je nekonečná; každý bod v zelene vyznačenom priestore je prípustný, predstavuje výrobnú možnosť.

````python
# Obrázok Množica výrobných možností

import matplotlib.pyplot as plt
import numpy as np

#Definícia veľkosti všetkých obrázkov
plt.rcParams["figure.figsize"] = (10, 6) #set default figure size

#produkčná funkcia
def f(x):
    y = x**α
    return y

#definičný obor produkčnej funkcie a podiel vstupu
x = np.linspace(0,150,200)
α = 0.5


y = f(x)

#zobraz množinu výrobných možností
fig, ax = plt.subplots()
ax.fill_between(x,y, color = 'g')
ax.set_title("Množina výrobných možností")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.show()
````

:name: mnozina_vyr_moznosti

Množina výrobných možností
```

Zelená plocha na obrázku predstavuje množinu výrobných možností firmy. Každý bod v tejto oblasti predstavuje prípustnú kombináciu vstupu a výstupu.

Body nachádzajúce sa nad zelenou plochou nie sú technologicky uskutočniteľné. Firma ich nedokáže vyrobiť pri dostupnej technológii.

Body nachádzajúce sa vo vnútri množiny sú prípustné, ale firma pri nich nevyužíva svoj výrobný potenciál naplno.

Na {numref}`vyrobne_moznosti` je z množiny výrobných možností zvýraznená možnosť $[100,5]$ červenou farbou. Táto výrobná možnosť nie je efektívna; 
- pri využití 100 jednotiek vstupu firma môže vyrobiť viac ako 5 jednotiek výstupu, maximálne 10 (bod zvýraznený žltou farbou); 
- na výrobu 5 jednotiek výstupu postačuje aj menšie množstvo vstupu; pri spotrebe 25 jednotiek vstupu dokáže firma stále vyrobiť 5 jednotiek výstupu (modrý bod).
 
Pre firmu nie je prípustná možnosť $[50,8]$ (fialový bod); tento bod nepatrí do množiny výrobných možností.

````python
#Obrázok Porovnanie niektorých výrobných možností
#Predchádzajúci obrázok rozšírime o zobrazenia niektorých konkrétnych výrobných možností a tiež o zvýraznenie hlavných osí 
#(pre lepšiu grafickú interpretovarľnosť)


#zobrazíme 4 body, ktoré sú charakterizované súradnicami x, y a farbou
body_x = [100,25,100,50]
body_y = [5,5,10,8]
farba = ['r','b','y','m']

fig, ax = plt.subplots()

#množina výrobných možností
ax.fill_between(x,y, color = 'g')

#body
for xx,yy,fa in zip(body_x, body_y, farba):
    ax.scatter(xx,yy,marker='x', color = fa)
    ax.plot([xx,xx], [yy,0], ':', color = fa)
    ax.plot([xx,0], [yy,yy], ':', color = fa)

xd, xh = ax.get_xlim()
yd, yh = ax.get_ylim()

#osi
ax.arrow(xd,0,xh-xd,0)
ax.arrow(0,yd,0,yh-yd)

ax.set_title("Efektívne, neefektívne a neprípustné výrobné možnosti")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.show()
````

## Hranica výrobných možností

Efektívne výrobné možnosti sú na hranici ich množiny, sú to výrobné možnosti, ktoré zodpovedajú maximálnej možnej výrobe pre danú hodnotu vstupu. 

Hranica výrobných možností
: Hranica výrobných možností je množina efektívnych výrobných možností.

Výrobná možnosť je efektívna, ak pri danom množstve vstupu poskytuje maximálny možný výstup. Efektívne výrobné možnosti ležia na hranici množiny výrobných možností.

Ak hodnotíme výrobné možnosti podľa kritérií minimalizácie vstupu a maximalizácie výstupu, žiadna výrobná možnosť na hranici nie je dominovaná inou výrobnou možnosťou.

Na obr. {numref}`hranica_vyrobnych_moznosti` je hranica výrobných možností zobrazená čiernou čiarou. Ohraničuje množinu výrobných možností zhora a predstavuje efektívnu hranicu výroby firmy.

````python
#Obrázok: Množina a hranica výrobných možností
#Prvý obrázok rozšírime o hranicu výrobných možností

fig, ax = plt.subplots()

#množina výrobných možností
ax.fill_between(x,y, color = 'g')

#hranica výrobných možností
ax.plot(x,y, color = 'k', linewidth = 4)

ax.arrow(xd,0,xh-xd,0)
ax.arrow(0,yd,0,yh-yd)

ax.set_title("Množina a hranica výrobných možností")
plt.xlabel("$x$")
plt.ylabel("$y$")
plt.show()
````

## Produkčná funkcia

Racionálna firma sa usiluje vyrábať efektívne.

Zo všetkých prípustných výrobných možností uprednostňuje efektívne výrobné možnosti, ktoré ležia na hranici výrobných možností.  

Pre analýzu správania racionálnej firmy nám preto postačuje poznať hranicu jej výrobných možností.

Bude užitočné, ak hranicu výrobných možností dokážeme zapísať matematickou funkciou {ref}`funkcia-label`.

$$
y=f(x).
$$

Táto funkcia sa nazýva **produkčná funkcia**. 

Produkčná funkcia
: Produkčná funkcia vyjadruje maximálne množstvo výstupu, ktoré možno vyrobiť pri danom množstve vstupu.

Pre ďalšie účely predpokladajme, že produkčná funkcia $f(x)$ je dvakrát diferencovateľná.

### Príklady produkčných funkcií s jedným vstupom a jedným výstupom ###

Príklady matematických tvarov produkčnej funkcie môžu byť:

```{math}
:label: odmocnina_pf

y=f(x) = A\sqrt{x},
```


```{math}
:label: CobbD_pf

y=f(x) = Ax^{\alpha},
```

```{math}
:label: linearna_pf

y=f(x) = ax.
```

Tvar produkčnej funkcie {eq}`linearna_pf` je lineárny, {eq}`CobbD_pf` Cobbov-Douglasov. Parameter $a$ v {eq}`linearna_pf` je vstupný {ref}`multiplikator-label` výstupu, pretože udáva, koľko jednotiek výstupu pripadá na jednotku vstupu. Parameter $\alpha$ v  je vstupná {ref}`elasticita-label` výstupu. Vyjadruje percentuálnu zmenu výstupu pri jednopercentnej zmene vstupu. Keďže možno predpokladať, že zvyšovaním vstupu produkcia výstupu rastie, predpokladáme $A,a,\alpha>0$. Ak v produkčnej funkcii v Cobbovom-Douglasovom tvare {eq}`CobbD_pf` dosadíme za elasticitu $\alpha = 1/2$, získame produkčnú funkciu v tvare {eq}`odmocnina_pf`.

Produkčná funkcia opisuje technologické obmedzenia firmy. Sama osebe nehovorí nič o tom, koľko firma skutočne vyrába. Skutočný objem výroby závisí aj od cien vstupov a výstupov.

(mp-label)=

Hraničný produkt vstupu
: Hraničný produkt vstupu je vstupný multiplikátor výstupu. Vyjadruje zmenu výstupu spôsobenú malou zmenou vstupu.

$$
MPx = \frac{\partial{f(x)}}{\partial{x}} > 0.
$$

Hraničný produkt vstupu vyjadruje, ako sa zvýši výstup, ak sa vstup zvýši o jednotku, *ceteris paribus*.

Vo všeobecnosti možno predpokladať, že hraničný produkt vstupu je kladný; s rastom vstupu rastie výstup. Produkčná funkcia je preto rastúcou funkciou vstupu.

(ap-label)=

Priemerný produkt vstupu 
: je podiel výstupu a vstupu,

$$
APx = \frac{{f(x)}}{{x}} > 0.
$$

Geometrická interpretácia priemerného a hraničného produktu je na {numref}`produkty`.

Priemerný produkt vstupu zodpovedá sklonu úsečky, ktorá spája začiatok súradnicového systému s bodom produkčnej funkcie. V bode A je táto úsečka strmšia než v bode B, preto je priemerný produkt vstupu v bode A vyšší.

Hraničný produkt vstupu zodpovedá sklonu dotyčnice k produkčnej funkcii. Na obr. {numref}`produkty` sú hraničné produkty v bodoch A a B reprezentované dotyčnicami.

````python
#Obrázok: Priemerný produkt vstupu vyjadruje sklon úsečky spájajúce začiatok súradnicového systému a danej dvojice vstupu a výstupu
#Z predchádzajúceho obrázka zobrazíme iba produkčnú funkciu a pridáme spojnicu

# dotyčnica
def dotycnica(g,gder,x):
    b = gder(x)
    a = g(x) - gder(x)*x
    return a,b

# funkcia prvej derivácie
def fder(x):
    y = α*x**(α-1)
    return y

x_bod = (40,90)

fig, ax = plt.subplots()

x_prespojnicu = (np.linspace(0,x_bod[0],50),np.linspace(0,x_bod[1],50))

#produkčná funkcia
ax.plot(x,y, color = 'k', linewidth = 4)

# zobraz spojnice
for bod,d in zip(x_bod,x_prespojnicu):
    AP=f(bod)/bod
    y_prespojnicu = AP*d
    ax.plot(d,y_prespojnicu)

# zobraz dotyčnice
for bod in x_bod:
    x_predotycnicu = np.linspace(bod-30,bod+30,10)
    a, b = dotycnica(f,fder,bod)
    y_predotycnicu = a + b*x_predotycnicu
    ax.plot(x_predotycnicu,y_predotycnicu)

ax.text(58,5.5,"sklon = $APx$")
ax.text(38,6.6,"$A$")
ax.text(90,9.9,"$B$")

ax.arrow(xd,0,xh-xd,0)
ax.arrow(0,yd,0,yh-yd)

plt.xlabel("$x$")
plt.ylabel("$y$")
plt.show()
````

Ktorá výrobná možnosť je optimálna?

Nasledujúca úvaha je len predbežnou ilustráciou. Podrobnú analýzu optimálneho rozhodovania firmy vykonáme v nasledujúcej kapitole. 

Z našich doterajších znalostí sa však dokážeme prepracovať k odpovedi (pozri tiež {ref}`priklad_max_zisk_mat`. Pýtame sa, ktorá z množiny efektívných výrobných možností je tá, pre ktorú sa rozhodne firma usilujúca o maximálny zisk. 

Predpokladajme, že cena vstupu je $c$ a cena výstupu je $p$.

Ak firma zvýši vstup o jednotku, produkcia sa približne zvýši o $MPx$ jednotiek výstupu. Hodnota tejto dodatočnej produkcie je $p\,MPx$.

Racionálna firma zvyšuje vstup dovtedy, kým sa hodnota dodatočného výstupu rovná nákladom na dodatočnú jednotku vstupu:

$$
p\,MPx=c.
$$

Po vydelení cenou výstupu dostaneme

$$
MPx=\frac{c}{p}.
$$

Na {numref}`pf_optimal` optimálny výber firmy zodpovedá dotyčnice produkčnej funkcie so sklonom $c/p$

````python
#Obrázok: Optimálny výber firmy maximalizujúcej zisk
#Produkčná funkcia a pridáme dotyčnicu 

fig, ax = plt.subplots()

a,b = dotycnica(f,fder,40) 
x_predotycnicu = np.linspace(10,70,50)

y_predotycnicu = a + b*x_predotycnicu

#produkčná funkcia
ax.plot(x,y, color = 'k', linewidth = 4)
ax.plot(x_predotycnicu,y_predotycnicu)

# optimálny bod
x_opt = 40
y_opt = f(x_opt)

ax.scatter(x_opt, y_opt,
           color='red',
           s=80,
           zorder=5)

ax.plot([x_opt,x_opt],
        [0,y_opt],
        ':',
        color='red')

ax.plot([0,x_opt],
        [y_opt,y_opt],
        ':',
        color='red')

ax.text(x_opt + 3,
        y_opt - 0.3,
        r"$[x^*,y^*]$")

ax.text(22,7,"sklon = $c/p$")

ax.arrow(xd,0,xh-xd,0)
ax.arrow(0,yd,0,yh-yd)

plt.xlabel("$x$")
plt.ylabel("$y$")
plt.show()
````

```{figure} images/produk/pf_optimal.png
:name: pf_optimal

Optimálny výber firmy maximalizujúcej zisk 
```

Formálne zapísané, firma rieši {ref}`volny_extrem` v tvare:

$$
\pi_{max}=\max_x \; \left[ p f(x)-cx \right],\\
x \ge 0.
$$

kde zisk firmy počítame ako rozdiel medzi príjmom $pf(x)$ a nákladmi $cx$.

```{admonition} Poznámka
Presne vzaté, nejde o úlohu na voľný extrém, pretože rozhodovanie firmy je obmedzené podmienkou

$$
x \ge 0.
$$

Ak však optimálne riešenie leží vo vnútri prípustnej množiny ($x>0$), možno ho nájsť rovnakým postupom ako voľný extrém, teda pomocou podmienok prvého a druhého rádu.

V našom prípade predpokladáme riešenie vo vnútri prípustnej množiny, preto postačuje riešiť úlohu rovnakým spôsobom ako voľný extrém.
```

Podmienka prvého rádu je

$$
p MPx = p\frac{\partial{f(x)}}{\partial{x}} = c.
$$

Podmienka druhého rádu je konkávnosť produkčnej funkcie

$$
\frac{\partial{MPx}}{\partial{x}} = \frac{\partial^2{f(x)}}{\partial{x^2}} < 0,
$$

ak predpokadáme, že cena výstupu je kladná $p>0$.

```{Table} Produkcia firmy ako optimalizačný problém
:name: tab_produkcia_optimalizacia

| Produkcia firmy                                | Teória optimalizácie        |
| ---------------------------------------------- | --------------------------- |
| Množina výrobných možností                     | Množina prípustných riešení |
| Hranica výrobných možností (efektívna hranica) | Množina efektívnych riešení |
| Výrobná možnosť maximalizujúca zisk            | Optimálne riešenie          |


Množina výrobných možností firmy zohráva v produkčnej teórii rovnakú úlohu ako {ref}`mnozina_pripustnych_rieseni` v matematickej optimalizácii. Hranica výrobných možností predstavuje množinu efektívnych riešení ({ref}`efektivne`) a firma maximalizujúca zisk z nej vyberá optimálne riešenie ({ref}`volny_extrem`).

## Produkčná funkcia s dvoma vstupmi

Doteraz sme predpokladali, že firma využíva iba jeden vstup. V praxi však výroba spravidla vyžaduje viacero vstupov, napríklad prácu a kapitál.

Ak firma využíva dva vstupy, produkčnú funkciu zapisujeme v tvare

$$
y=f(x_1,x_2),
$$

kde

- $x_1$ je prvý vstup,
- $x_2$ je druhý vstup,
- $y$ je výstup,
- predpokladáme, že $f(x_1,x_2)$ je dvakrát diferencovateľná.

Hraničné produkty vstupov sú

$$
MP_1 = \frac{\partial{Af(x_1,x_2)}}{\partial{x_1}},\\
MP_2 = \frac{\partial{Af(x_1,x_2)}}{\partial{x_2}}.
$$

### Vrstevnice produkčnej funkcie

Produkčná funkcia s dvoma vstupmi je zobrazená v trojrozmernom priestore. Na jednoduchšiu analýzu sa často využíva jej vrstevnicové zobrazenie.

Vrstevnica produkčnej funkcie je množina všetkých kombinácií vstupov, ktoré vedú k rovnakej úrovni výstupu. Formálne ju definujeme rovnicou

$$
f(x_1,x_2)=\bar y,
$$

kde $\bar y$ je pevne daná úroveň produkcie.

### Izokvanty

Vrstevnice produkčnej funkcie sa v ekonómii nazývajú **izokvanty**.

(izokvanta)=
Izokvanta
: Izokvanta je množina kombinácií vstupov, ktoré umožňujú vyrobiť rovnaké množstvo výstupu.

Pri pohybe po izokvante zostáva výstup nezmenený. Zvýšenie jedného vstupu preto musí byť kompenzované znížením druhého vstupu.

(MRTS)=
### Hraničná miera technickej substitúcie

Sklon izokvanty vyjadruje, v akom pomere možno nahrádzať jeden vstup druhým bez zmeny objemu produkcie.

Hraničná miera technickej substitúcie
: Hraničná miera technickej substitúcie je hraničná miera substitúcie {ref}`MRS_mat` aplikovaná na produkčnú funkciu.

Na {numref}`izokvanty` sú zobrazené izokvanty produkčnej funkcie. Hraničnú mieru technickej substitúcie vyjadruje sklon dotyčnice. V bode $B$ je hraničná miera technickej substitúcie relatívne nízka, v bode $A$ vysoká.

Hraničná miera technickej substitúcie vyjadruje, o koľko jednotiek vstupu $x_2$ možno znížiť jej spotrebu pri zvýšení vstupu $x_1$ o jednotku tak, aby sa produkcia nezmenila.

Podobne ako pri hraničnej miere substitúcie ju vyjadríme ako podiel prvých derivácií produkčnej funkcie

$$
MRTS(x_1,x_2)
=
-\frac{dx_2}{dx_1}
=
\frac{\frac{\partial f}{\partial x_1}}
     {\frac{\partial f}{\partial x_2}}
=
\frac{MP_1}{MP_2}.
$$

Hraničná miera technickej substitúcie je teda podielom hraničných produktov oboch vstupov.

````python
#Obrázok zobrazuje izokvanty.

def h(yi,x1):
    x2 = yi**(1.5)*x1**(-.5)
    return x2

def h_der(yi,x1):
    x2 = -0.5*yi**(1.5)*x1**(-1.5)
    return x2

# Keďže uvažujeme funkcie 2 premenných potrebujeme predefinovať funkciu dotyčnice.
def dotycnica2(g,gder,y,x):
    b = gder(y,x)
    a = g(y,x) - gder(y,x)*x
    return a,b

body_x1 = (20,50,80) #body dotyčníc (x-ova súradnica)

x_1 = np.linspace(0.001,100) #definičný obor funkcií

y = np.array([8,10,12]) #3 izočiary

#zobraz priebehy funkcií
fig, ax = plt.subplots(figsize = (8,8))
i = 1
for y_i in y:
    x_2 = h(y_i, x_1)
    ax.plot(x_1,x_2,linewidth = 2, label=f'$y_{i}$')
    ax.legend()
    ax.set(xlabel = "$x_1$", ylabel = "$x_2$",ylim = (0,11), xlim = (1,100))
    i = i + 1

#Zobraz dotyčnice
yi = 10

for bod in body_x1:
    x1 = np.linspace(bod-5,bod+5,10)
    a, b = dotycnica2(h,h_der,yi,bod)
    y = a + b*x1
    ax.plot(x1,y,color='k')
        
#komenty
ax.text(17,6.5,"$A$", color = 'k')
ax.text(78,3.1,"$B$", color = 'k')
ax.text(48,4.1,"$C$", color = 'k')

plt.show()
````

## Produkčná funkcia s viac vstupmi

Firma môže na výrobu potrebovať viac vstupov.

```{sidebar} Príklad
Na výrobu múky sa využíva nielen pšenica, ale napríklad aj energia (vietor, vodný tok, palivo) a ľudská práca. 
```

Produkčná funkcia s $n$ vstupmi je v tvare:

$$
y=f(x_1, x_2 \dots ,x_n).
$$

Hodnoty vstupov možno zapísať vektorom:

$$\mathbf{x} = \begin{pmatrix} x_1 \\ x_2 \\ \vdots \\ x_n \end{pmatrix}.$$

Potom možno produkčnú funkciu s viac vstupmi zapísať v tvare:

$$
y=f(\mathbf{x}).
$$

Nech produkčná funkcia $f(\mathbf{x})$ je dvakrát diferencovateľná. Hraničný produkt vstupu $i$ je:

```{math}
:label: mp_i
MP_i = \frac{\partial f(\mathbf{x})}{\partial x_i}>0,
\qquad i=1,\dots,n.
```

Priemerný produkt vstupu $i$ je:

$$
AP_i = \frac{f(\mathbf{x})}{x_i}>0,
\qquad i=1,\dots,n.
$$

Hraničná miera technickej substitúcie medzi vstupmi $i$ a $j$ je

```{math}
:label: r_MRTS

MRTS(x_i,x_j)
=
\frac{\frac{\partial f}{\partial x_i}}
     {\frac{\partial f}{\partial x_j}}
=
\frac{MP_i}{MP_j}.
```

Produkčná funkcia s jedným vstupom je špeciálnym prípadom produkčnej funkcie s viacerými vstupmi, keď $n=1$.


```{admonition} Poznámka
Produkčná funkcia s jedným alebo dvoma vstupmi sa používa najmä na ilustráciu základných ekonomických princípov. Reálne firmy spravidla využívajú veľký počet vstupov a často produkujú viac výstupov súčasne.

Vo všeobecnosti výrobu firmy opisuje vektorová funkcia

$$
\mathbf{y}=\mathbf{f}(\mathbf{x}).
$$
```

## Vlastnosti produkčnej funkcie

Nech produkčná funkcia $f(\mathbf{x})$ je dvakrát diferencovateľná. Potom môžeme uviesť nasledovné vlastnosti produkčnej funkcie.

### Rastúcosť produkčnej funkcie

Rastúcosť produkčnej funkcie
: Produkčná funkcia je rastúca v každom vstupe.

Ak sa zvýši množstvo niektorého vstupu a ostatné vstupy zostanú nezmenené, produkcia sa nezníži.

$$
MP_i=
\frac{\partial f(\mathbf{x})}{\partial x_i}
\ge 0,
\qquad i=1,\dots,n.
$$

Rastúcosť vyjadruje predpoklad, že väčšie množstvo vstupov umožňuje vyrobiť aspoň rovnaké množstvo výstupu.

(hranicne_produkty_vstupov)=

### Hraničné produkty vstupov

Hraničný produkt vstupu vyjadruje prírastok produkcie spôsobený zvýšením vstupu.

Priebeh hraničného produktu závisí od druhej derivácie produkčnej funkcie.

Ak platí

$$
\frac{\partial^2 f(\mathbf{x})}{\partial x_i^2}\le 0,
$$

produkčná funkcia je vo vstupe $x_i$ konkávna. Hraničný produkt vstupu je potom klesajúci vo vstupe $x_i$.

Ak platí

$$
\frac{\partial^2 f(\mathbf{x})}{\partial x_i^2}\ge 0,
$$

produkčná funkcia je vo vstupe $x_i$ konvexná. Hraničný produkt vstupu je potom rastúci vo vstupe $x_i$.

```{Table}
:name: Vlastnosti produkčnej funkcie podľa priebehu hraničných produktov

| Druhá derivácia                         | Tvar produkčnej funkcie vo vstupe $x_i$ | Hraničný produkt vstupu   |
| --------------------------------------- | --------------------------------------- | --------------------------|
| $\frac{\partial^2 f}{\partial x_i^2}<0$ | konkávna                                | klesajúci vo vstupe $x_i$ |
| $\frac{\partial^2 f}{\partial x_i^2}=0$ | lineárna                                | konštantný                |
| $\frac{\partial^2 f}{\partial x_i^2}>0$ | konvexná                                | rastúci vo vstupe $x_i$   |
```

Sklon izokvanty je určený hraničnou mierou technickej substitúcie {eq}`r_MRTS`

Priebeh hraničných produktov preto ovplyvňuje tvar izokvánt.

Ak hraničné produkty vstupov klesajú, spravidla klesá aj hraničná miera technickej substitúcie. Izokvanty sú potom konvexné voči počiatku súradnicového systému. {numref}`izokvanty` zobrazuje konvexné izokvanty.

### Výnosy z rozsahu

Výnosy z rozsahu
: Výnosy z rozsahu vyjadrujú, ako sa zmení výstup pri rovnomernom zvýšení všetkých vstupov.

Nech sa všetky vstupy zvýšia $t$-násobne. Potom porovnávame pôvodný výstup

$$
f(\mathbf{x})
$$

a nový výstup

$$
f(t\mathbf{x}).
$$

#### Konštantné výnosy z rozsahu

Ak platí

$$
f(t\mathbf{x})
=
t f(\mathbf{x}),
$$

produkčná funkcia má konštantné výnosy z rozsahu.

#### Rastúce výnosy z rozsahu

Ak platí

$$
f(t\mathbf{x})
>
t f(\mathbf{x}),
$$

produkčná funkcia má rastúce výnosy z rozsahu.

#### Klesajúce výnosy z rozsahu

Ak platí

$$
f(t\mathbf{x})
<
t f(\mathbf{x}),
$$

produkčná funkcia má klesajúce výnosy z rozsahu.

Výnosy z rozsahu možno interpretovať aj pomocou elasticity výstupu vzhľadom na súčasnú zmenu všetkých vstupov.

Predpokladajme, že všetky vstupy zvýšime o 1 %. Potom

$$
t = 1.01.
$$

Nové množstvo vstupov je

$$
t\mathbf{x}=1.01\mathbf{x}.
$$

Výnosy z rozsahu porovnávajú, o koľko percent sa zvýši výstup v reakcii na toto jednopercentné zvýšenie všetkých vstupov.

Ak sa výstup zvýši o viac než 1 %, produkčná funkcia má rastúce výnosy z rozsahu.

Ak sa výstup zvýši presne o 1 %, produkčná funkcia má konštantné výnosy z rozsahu.

Ak sa výstup zvýši o menej než 1 %, produkčná funkcia má klesajúce výnosy z rozsahu.

Elasticita výstupu vzhľadom na súčasnú proporcionálnu zmenu všetkých vstupov je rovná súčtu všetkých vstupných elasticít výstupu

$$
\sum_{i=1}^{n}\varepsilon_i.
$$

Preto:

- ak

$$
\sum_{i=1}^{n}\varepsilon_i > 1,
$$

výstup rastie rýchlejšie než všetky vstupy a produkčná funkcia má rastúce výnosy z rozsahu,

- ak

$$
\sum_{i=1}^{n}\varepsilon_i = 1,
$$

výstup rastie rovnakým tempom ako všetky vstupy a produkčná funkcia má konštantné výnosy z rozsahu,

- ak

$$
\sum_{i=1}^{n}\varepsilon_i < 1,
$$

výstup rastie pomalšie než všetky vstupy a produkčná funkcia má klesajúce výnosy z rozsahu.

Súčet elasticít výstupu podľa všetkých vstupov možno interpretovať ako elasticitu výstupu podľa rozsahu výroby.

```{Table} Výnosy z rozsahu
:name: vynosy_rozsahu
| Výnosy z rozsahu | Produkčná funkcia | Súčet vstupných elasticít výstupu |
|---|---|---|
| Rastúce | $f(t\mathbf{x}) > t f(\mathbf{x})$ | $\sum_i \varepsilon_i > 1$ |
| Konštantné | $f(t\mathbf{x}) = t f(\mathbf{x})$ | $\sum_i \varepsilon_i = 1$ |
| Klesajúce | $f(t\mathbf{x}) < t f(\mathbf{x})$ | $\sum_i \varepsilon_i < 1$ |
```

## Makroekonomické vlastnosti produkčnej funkcie

Predpokladáme, že technológia je verejne dostupná pre všetky firmy. Agregovanú (makroekonomickú) produkciu ekonomiky preto môžeme aproximovať jedinou produkčnou funkciou.

V makroekonómii sa najčastejšie predpokladá, že výstup ekonomiky závisí od kapitálu, práce a technológie.

Makroekonomická produkčná funkcia má tvar

```{math}
:label: r_makro_pf

Y = A F(K,N),
```

kde

- $Y \ge 0$ je výstup ekonomiky,
- $K \ge 0$ je zásoba kapitálu,
- $N \ge 0$ je množstvo práce,
- $A > 0$ je celková produktivita faktorov.

Zásoba kapitálu a práca sú vstupy. Zásoba kapitálu je kumulované množstvo všetkých budov, strojov a zariadení používaných vo výrobe. 

Práca môže byť vyjadrená množstvom pracujúcich (z dlhodobého hľadiska) alebo úhrnným odpracovaným časom (z krátkodobého hľadiska). 

Celková produktivita faktorov vyjadruje efekt využitia vstupov vo výrobe. Čím je vyššia, tým lepšie sa vstupy využívajú.

Makroekonomická produkčná funkcia má tieto vlastnosti:

1. je rastúca vo vstupoch kapitál a práca,
2. je konkávna,
3. elasticita výstupu podľa celkovej produktivity faktorov sa rovná 1,
4. prechádza začiatkom súradnicového systému,
5. má konštantné výnosy z rozsahu.

Predpokladajme, že produkčná funkcia je dvakrát diferencovateľná.

```{Table} Makroekonomické vlastnosti produkčnej funkcie
:name: makro_vlastnosti_pf
| Vlastnosť | Matematické vyjadrenie |
|---|---|
| Rastúcosť | $MPK \ge 0,\ MPN \ge 0$ |
| Konkávnosť | $\frac{\partial^2Y}{\partial K^2}\le0,\ \frac{\partial^2Y}{\partial N^2}\le0$ |
| Elasticita podľa $A$ | $\varepsilon_A = 1$ |
| Začiatok súradnicového systému | $F(0,0)=F(K,0)=F(0,N)=0$ |
| Konštantné výnosy z rozsahu | $F(tK,tN)=tF(K,N)$ |

### Rastúca produkčná funkcia

Produkcia rastie so zvyšovaním kapitálu aj práce:

$$
\frac{\partial Y}{\partial K}
=
A\frac{\partial F(K,N)}{\partial K}
\ge 0,
$$

$$
\frac{\partial Y}{\partial N}
=
A\frac{\partial F(K,N)}{\partial N}
\ge 0.
$$

Hraničné produkty vstupov sú:

(def_MPK)=
Hraničný produkt kapitálu
: kapitálový {ref}`multiplikator-label` výstupu.

$$
MPK = A\frac{\partial F(K,N)}{\partial K}
\ge 0,
$$

(def_MPN)=
Hraničný produkt práce
: pracovný {ref}`multiplikator-label` výstupu.

$$
MPN = A\frac{\partial F(K,N)}{\partial N}
\ge 0.
$$

### Konkávnosť

Produkčná funkcia je konkávna vo vstupoch kapitál a práca.

$$
\frac{\partial^2 Y}{\partial K^2}
\le 0,
$$

$$
\frac{\partial^2 Y}{\partial N^2}
\le 0.
$$

Hraničné produkty kapitálu a práce sú preto klesajúce vo vlastných vstupoch.

### Celková produktivita faktorov

Elasticita výstupu podľa celkovej produktivity faktorov sa rovná 1:

$$
\varepsilon_A
=
\frac{\partial Y}{\partial A}\frac{A}{Y}.
$$

Keďže

$$
Y=AF(K,N),
$$

platí

$$
\frac{\partial Y}{\partial A}
=
F(K,N).
$$

Preto

```{math}
:label: elasticita_a

\varepsilon_A
=
F(K,N)
\frac{A}{AF(K,N)}
=
1.
```

Zvýšenie celkovej produktivity faktorov o 1 % vedie k zvýšeniu výstupu o 1 %. Preto sa z dlhodobého hľadiska parameter $A$ interpretuje ako technologická úroveň ekonomiky.

### Začiatok súradnicového systému

Ak ekonomika nedisponuje kapitálom alebo prácou, predpokladáme nulovú produkciu:

$$
F(0,0)=F(K,0)=F(0,N)=0.
$$

Preto

$$
Y=A\,F(0,0)=A\,F(K,0)=A\,F(0,N)=0.
$$

(KVR)=

### Konštantné výnosy z rozsahu

Ak sa kapitál aj práca zvýšia rovnakým násobkom $t$, výstup sa zvýši v rovnakom pomere:

```{math}
:label: r_konst_vynos_rozsah

A\,F(tK,tN)
=
A\,tF(K,N)=t\,Y.
```

Makroekonomická produkčná funkcia má preto konštantné výnosy z rozsahu.

Konštantné výnosy z rozsahu znamenajú, že zvýšenie kapitálu aj práce o 1 % vedie k zvýšeniu výstupu o 1 %.

Pri konštantných výnosoch z rozsahu platí, že súčet elasticít výstupu podľa kapitálu a práce je rovný jednej:

```{math}
:label: r_KVR
\varepsilon_K+\varepsilon_N=1.
```

Priemerné produkty majú podobný priebeh ako hraničné produkty a pri rastúcej konkávnej produkčnej funkcii sú tiež klesajúcimi funkciami vstupov.

````python
def nakresli_osi(ax):
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()

    ax.arrow(xd,0,xh-xd,0,
             length_includes_head=True)

    ax.arrow(0,yd,0,yh-yd,
             length_includes_head=True)


# --------------------------------------------------
# Produkčná funkcia
# --------------------------------------------------

α = 0.5

def f(x):
    return x**α

def fder(x):
    return α*x**(α-1)

def dotycnica(g, gder, x):
    b = gder(x)
    a = g(x) - b*x
    return a, b

# --------------------------------------------------
# Definičný obor
# --------------------------------------------------

x = np.linspace(1e-15, 150, 500)

# body pre ilustráciu
K1 = 40
K2 = 90

# --------------------------------------------------
# hodnoty produkcie
# --------------------------------------------------

Y = f(x)

# hraničné produkty
MP = fder(x)

# priemerné produkty
AP = f(x)/x

# --------------------------------------------------
# obrázok
# --------------------------------------------------

fig, ax = plt.subplots(
    3, 2,
    figsize=(14, 12),
    sharex='col'
)

# ==================================================
# 1. riadok: produkčné funkcie
# ==================================================

for j, (nazov, symbol) in enumerate([
    ("Kapitál", "K"),
    ("Práca", "N")]):

    g = ax[0,j]

    g.plot(x, Y, color='black', linewidth=3)

    for bod, meno in zip([K1, K2], ["1","2"]):

        # bod na produkčnej funkcii
        g.scatter(
            bod,
            f(bod),
            color='red',
            zorder=10
        )

        g.text(
            bod+2,
            f(bod)+0.8,
            rf"${symbol}_{meno}$"
        )

        # spojnice (AP)
        g.plot(
            [0,bod],
            [0,f(bod)],
            color='blue',
            linestyle='--'
        )

        # dotyčnice (MP)
        a,b = dotycnica(f,fder,bod)

        xx = np.linspace(
            max(1,bod-25),
            bod+25,
            50
        )

        yy = a + b*xx

        g.plot(
            xx,
            yy,
            color='green'
        )

        # pomocná zvislá čiara
        g.axvline(
            bod,
            linestyle=':',
            color='gray'
        )

    g.set_title(
        rf"$Y({symbol})$"
    )

    g.set_ylabel("$Y$")

    nakresli_osi(g)


# ==================================================
# 2. riadok: hraničné produkty
# ==================================================

for j, (nazov, symbol) in enumerate([
    ("Kapitál", "K"),
    ("Práca", "N")]):

    g = ax[1,j]

    g.plot(
        x,
        MP,
        color='green',
        linewidth=3
    )

    for bod, meno in zip([K1,K2],["1","2"]):

        g.scatter(
            bod,
            fder(bod),
            color='green',
            zorder=10
        )

        g.text(
            bod+2,
            fder(bod)+0.005,
            rf"$MP{symbol}_{meno}$"
        )

        g.axvline(
            bod,
            linestyle=':',
            color='gray'
        )

    g.set_title(
        rf"$MP{symbol}$"
    )

    g.set_ylabel("MP")



# ==================================================
# 3. riadok: priemerné produkty
# ==================================================

for j, nazov in enumerate(["K","N"]):

    g = ax[2,j]

    g.plot(
        x,
        AP,
        color='blue',
        linewidth=3
    )

    for bod, meno in zip([K1,K2],["1","2"]):

        ap_bod = f(bod)/bod

        g.scatter(
            bod,
            ap_bod,
            color='blue',
            zorder=10
        )

        g.text(
            bod+2,
            ap_bod+0.005,
            rf"$AP{nazov}_{meno}$"
        )

        g.axvline(
            bod,
            linestyle=':',
            color='gray'
        )

    g.set_title(
        rf"$AP{nazov}$"
    )

    g.set_ylabel("AP")
    g.set_xlabel(f"${nazov}$")

# --------------------------------------------------

ax[1,0].set_ylim(0,0.2)
ax[1,1].set_ylim(0,0.2)
    
ax[2,0].set_ylim(0,0.35)
ax[2,1].set_ylim(0,0.35)

plt.tight_layout()
plt.show()
````

## Dekompozícia rastu

Ako sa celková produktivita faktorov {A} a vstupy kapitál {K} a práca {N} podieľajú na raste?

Najskôr potrebujeme zapísať rast výstupu. Rast ekonomických premenných meriame relatívne v %. Za predpokladu diferencovateľnosti produkčnej funkcie {eq}`r_makro_pf`, relatívny rast výstupu je $dY/Y$. 

Analogicky relatívny rast celkovej produktivity faktorov zapíšeme ako $dA/A$, kapitálu $dK/K$ a práce $dN/N$.

Ak v {eq}`r_makro_pf` rastie $Y$, možno to pripísať rastu $A$, $N$ alebo $K$.

Percentuálny rast jednej premennej spôsobený rastom inej premennej je {ref}`elasticita-label`.

Elasticita výstupu vzhľadom na celkovú produktivitu faktorv {eq}`elasticita_a` sa rovná 1. Označme kapitálovú elasticitu výstupu symbolom $\alpha$ a pracovnú symbolom $\beta$. Platí:

```{math}
:label: r_alpha
\varepsilon_K = \alpha = \frac{\partial Y}{\partial K}\frac{K}{Y},
```

```{math}
:label: r_beta
\varepsilon_N=\beta=
\frac{\partial Y}{\partial N}\frac{N}{Y}.
```

Potom produkčnú funkciu {eq}`r_makro_pf` možno zapíasať v tvare

```{math}
:label: r_dekompozicia_rastu
\frac{dY}{Y}
=
\frac{dA}{A}
+
\alpha \frac{dK}{K}
+
\beta \frac{dN}{N}.
```

Rast výstupu možno preto rozložiť na príspevok celkovej produktivity faktorov, kapitálu a práce.

### Vstupné elasticity výstupu a konštantné výnosy z rozsahu

Predpoklad {ref}`KVR` v {eq}`r_makro_pf` sa spĺňa, ak {eq}`r_KVR`, teda:

```{math}
:label: elasticity_KVR
\alpha + \beta = 1
```

Ak predpokladáme konštantné výnosy z rozsahu, stačí nám poznať kapitálová elasticita výstupu, keďže pracovnú elasticitu výstupu možno vypočítať z {eq}`elasticity_KVR`:

```{math}
:label: beta_z_KVR
\beta = 1 - \alpha
```

### Ekonomická interpretácia vstupných elasticít výstupu

Vstupné elasticity výstupu {eq}`r_alpha` {eq}`r_beta` možno zapísať tiež v tvare:


```{math}
:label: r_podiel_kapitalu
\alpha
=
\frac{MPK\cdot K}{Y},
```

V {eq}`r_podiel_kapitalu` hraničný produkt kapitálu $MPK$ vyjadruje objem výstupu zodpovedajúci jednotke kapitálu. Ak ho vynásobíme objemom kapitálu, získame celkový výstup vyprodukovaný kapitálom. Podelením celkovým objemom výstupu dostaneme podiel kapitálu na výstupe.

(podiel_kapitalu)=

Podiel kapitálu na výstupe 
: je kapitálová elasticita výstupu $\alpha$.

Analogicky

```{math}
:label: r_podiel_prace
\beta
=
\frac{MPN\cdot N}{Y},
```
(podiel_prace)=

Podiel práce na výstupe 
: je pracovná elasticita výstupu $\beta$.

Ak hovoríme o agregovanom hospodárstve, jeho výstup možno vyjadriť hrubým domácim produktom. Potom tiež hovoríme o **podieloch vstupov na hrubom domácom produkte**.

Ak výnosy z rozsahu sú konštantné, súčet podielov vstupov na výstupe sa rovná 1 {eq}`elasticity_KVR`.

Do {eq}`elasticity_KVR` dosaďme podiely za elasticity {eq}`r_podiel_kapitalu` a {eq}`r_podiel_prace`:

$$
\frac{MPK\cdot K}{Y} + \frac{MPN\cdot N}{Y} = 1.
$$

Vynásobením výstupm $Y$, získame, ako výstup možno rozložiť na kapitálový a pracovný príspevok:

```{math}
:label: prispevky_vstupov
MPK\cdot K + MPN\cdot N = Y.
```

{eq}`prispevky_vstupov` je špeciálnym prípadom Eulerovej vety pre homogénne funkcie prvého stupňa. Produkčná funkcia s konštantnými výnosmi z rozsahu rozdeľuje celý výstup medzi výrobné faktory kapitál a prácu.

## Intenzívny tvar produkčnej funkcie

Konštantné výnosy z rozsahu umožňujú previesť makroekonomickú produkčnú funkciu na jednotku pracujúce (na pracujúceho, na odpracovanú hodinu).

Pripomeňme, že pri produkčnej funkcii {eq}`r_makro_pf` z makroekonomického hľadiska predpokladáme konštantné výnosy z rozsahu {eq}`r_konst_vynos_rozsah`.

Do rovnice {eq}`r_konst_vynos_rozsah` dosadíme

$$
t=\frac{1}{N}.
$$

Potom

$$
F(K,N)
=
N\,F\left(\frac{K}{N},1\right).
$$

Po dosadení do produkčnej funkcie získame

$$
Y
=
AN\,F\left(\frac{K}{N},1\right).
$$

Po vydelení prácou

$$
\frac{Y}{N}
=
A\,F\left(\frac{K}{N},1\right).
$$

Definujme intenzívny kapitál a intenzívny výstup.

Intenzívny kapitál
: Kapitál na jednotku práce,

$$
k=\frac{K}{N}.
$$

Intenzívny výstup
: Výstup na jednotku práce,

$$
y=\frac{Y}{N}.
$$

Potom produkčnú funkciu môžeme zapísať v intenzívnom tvare

```{math}
:label: intenzivna_pf

y=A f(k).
```

Produkčná funkcia v intenzívnom tvare
: Produkčná funkcia v intenzívnome tvare vyjadruje vzťah medzi intenzívnym výstupom a intenzívnym kapitálom.

Výhodou intenzívneho tvaru je, že produkčnú funkciu dvoch vstupov možno analyzovať pomocou funkcie jednej premennej.

V nasledujúcom texte preukážeme, že:

- produkčná funkcia prechádza začiatkom súradnicového systému,
- intenzívny a extenzívny hraničný produkt kapitálu sa rovnajú
- intenzívny a extenzívny priemerný produkt kapitálu sa rovnajú
- elasticita výstupu podľa celkovej produktivity faktorov sa rovná 1.

Z uvedeného možno odvodiť vlastnosti produkčnej funkcie v intenzívnom tvare

### Produkčná funkcia v intenzívom tvare prechádza začiatkom súradnicového systému ####

Extenzívny tvar produkčnej funkcie prechádza začiatkom súradnícového systému. ukážeme, že rovnaké musí platiť pre intenzívny tvar.

Ak

$$
K=0,
$$

potom

$$
Y=0.
$$

Z definície

$$
k=\frac{K}{N}
$$

a

$$
y=\frac{Y}{N}
$$

dostávame

$$
k=0,
\qquad
y=0.
$$

Intenzívnay tvar produkčnej funkcie preto prechádza začiatkom súradnicového systému.

### Hraničné a priemerné produkty v intenzívnom tvare

Hraničný produkt intenzívneho kapitálu je

$$
mpk
=
\frac{\partial y}{\partial k}.
$$

Keďže

$$
y=\frac{Y}{N},
\qquad
k=\frac{K}{N},
$$

a produkčná funkcia má konštantné výnosy z rozsahu, platí

```{math}
:label: intenzivny_mpk
mpk = MPK.
```

Hraničný produkt kapitálu je teda rovnaký v extenzívnom aj intenzívnom tvare produkčnej funkcie.

Priemerný produkt intenzívneho kapitálu je

$$
apk=\frac{y}{k}.
$$

Po dosadení definícií intenzívneho výstupu a intenzívneho kapitálu dostávame

```{math}
:label: intenzivny_apk
apk
=
\frac{Y/N}{K/N}
=
\frac{Y}{K}
=
APK.
```

Priemerný produkt kapitálu je preto tiež rovnaký v agregovanom aj intenzívnom tvare produkčnej funkcie.

Z rovností

$$
mpk = MPK,
\qquad
apk = APK
$$

vyplýva, že produkčná funkcia v intenzívnom tvare má rovnaké niektoré vlastnosti extenzívnej produkčnej funkcie.

Keďže agregovaná produkčná funkcia je rastúca a konkávna vo vstupe kapitál, rovnaké vlastnosti má aj produkčná funkcia v intenzívnom tvare. Jej hraničný produkt kapitálu je kladný a klesajúci vo vstupe kapitál.

### Elasticita výstupu podľa celkovej produktivity faktorov v intenzívnoom tvare produkčnej funkcie

Podobne ako extenzívny tvar produkčnej funkcie, aj produkčná funkcia v intenzívnom tvare má jednotkovú elasticitu výstupu podľa celkovej produktivity faktorov.

Keďže

$$
y=A f(k),
$$

platí

```{math}
:label: elasticita_a_intenzivna
\varepsilon_A
=
\frac{\partial y}{\partial A}\frac{A}{y}
=
1.
```

Zvýšenie celkovej produktivity faktorov o 1 % preto vedie k zvýšeniu intenzívneho výstupu o 1 %.

#### Vlastnosti produkčnej funkcie v intenzívnom tvare

Produkčná funkcia v intenzívnom tvare

$$
y=Af(k)
$$

má tieto vlastnosti:

1. je rastúca vo vstupe intenzívny kapitál,
2. je konkávna,
3. prechádza začiatkom súradnicového systému,
4. elasticita výstupu podľa celkovej produktivity faktorov sa rovná 1.

#### Intenzívna produkčná funkcia je rastúca 

$$
mpk
=
\frac{\partial y}{\partial k}
\ge 0.
$$

#### Konkávnosť

$$
\frac{\partial^2 y}{\partial k^2}
=
\frac{\partial mpk}{\partial k}
\le 0.
$$

#### Začiatok súradnicového systému

Ak

$$
k=0,
$$

potom

$$
y=0.
$$

#### Elasticita podľa celkovej produktivity faktorov

$$
\varepsilon_A
=
\frac{\partial y}{\partial A}\frac{A}{y}
=
1.
$$

Zvýšenie celkovej produktivity faktorov o 1 % vedie k zvýšeniu intenzívneho výstupu o 1 %.

Intenzívny tvar produkčnej funkcie spolu s hraničným a priemerným produktom sú zobrazené na {numref}`graf_intenzivna`

````python
import numpy as np
import matplotlib.pyplot as plt

# --------------------------------------------------
# Produkčná funkcia pracujúceho
# --------------------------------------------------

A = 1
α = 0.5

def f(k):
    return A*k**α

# --------------------------------------------------
# Definičný obor
# --------------------------------------------------

k = np.linspace(1e-15,150,500)

k1 = 40
k2 = 90

# --------------------------------------------------
# hodnoty funkcií
# --------------------------------------------------

y = f(k)

mpk = fder(k)

apk = f(k)/k

# --------------------------------------------------
# obrázok
# --------------------------------------------------

fig, ax = plt.subplots(
    3,1,
    figsize=(10,12),
    sharex=True
)

# ==================================================
# Produkčná funkcia
# ==================================================

g = ax[0]

g.plot(
    k,
    y,
    color='black',
    linewidth=3
)

for bod, meno in zip([k1,k2],["1","2"]):

    # bod
    g.scatter(
        bod,
        f(bod),
        color='red',
        zorder=10
    )

    g.text(
        bod+2,
        f(bod)+0.25,
        rf"$k_{meno}$"
    )

    # spojnica (APK)
    g.plot(
        [0,bod],
        [0,f(bod)],
        color='blue',
        linestyle='--'
    )

    # dotyčnica (MPK)
    a,b = dotycnica(f,fder,bod)

    xx = np.linspace(
        max(1,bod-25),
        bod+25,
        50
    )

    yy = a + b*xx

    g.plot(
        xx,
        yy,
        color='green'
    )

    g.axvline(
        bod,
        linestyle=':',
        color='gray'
    )

nakresli_osi(g)
g.set_title(r"$y=A f(k)$")
g.set_ylabel("$y$")

# ==================================================
# MPK
# ==================================================

g = ax[1]

g.plot(
    k,
    mpk,
    color='green',
    linewidth=3
)

for bod, meno in zip([k1,k2],["1","2"]):

    g.scatter(
        bod,
        fder(bod),
        color='green'
    )

    g.text(
        bod+2,
        fder(bod),
        rf"$mpk_{meno}$"
    )

    g.axvline(
        bod,
        linestyle=':',
        color='gray'
    )

g.set_title(r"$mpk$")
g.set_ylabel("$mpk$")
g.set_ylim(0,0.2)

# ==================================================
# APK
# ==================================================

g = ax[2]

g.plot(
    k,
    apk,
    color='blue',
    linewidth=3
)

for bod, meno in zip([k1,k2],["1","2"]):

    ap = f(bod)/bod

    g.scatter(
        bod,
        ap,
        color='blue'
    )

    g.text(
        bod+2,
        ap,
        rf"$apk_{meno}$"
    )

    g.axvline(
        bod,
        linestyle=':',
        color='gray'
    )

g.set_title(r"$apk$")
g.set_ylabel("$apk$")
g.set_xlabel("$k$")
g.set_ylim(0,0.35)

plt.tight_layout()
plt.show()
````

## Vplyv zvýšenia celkovej produktivity faktorov na produkčnú funkciu

Predpokladajme, že sa zvýši celková produktivita faktorov z $A_1$ na $A_2$, pričom

$$
A_2>A_1.
$$

Produkčná funkcia sa posunie nahor:

$$
Y=A_1F(K,N)
\quad \rightarrow \quad
Y=A_2F(K,N).
$$

Pri každej úrovni kapitálu a práce ekonomika vyrobí viac výstupu.

Keďže celková produktivita faktorov vstupuje do produkčnej funkcie ako multiplikátor, zvýšenie $A$ vedie nielen k zvýšeniu výstupu, ale aj hraničných a priemerných produktov vstupov.

Zvýšením celkovej produktivity faktorov sa preto zvyšujú:

- výstup $Y$,
- intenzívny výstup $y$,
- hraničný produkt kapitálu $MPK$ a $mpk$,
- hraničný produkt práce $MPN$,
- priemerný produkt kapitálu $APK$ a $apk$,
- priemerný produkt práce $APN$.

Grafická interpretácia zvýšenia celkovej produktivity faktorov je na {numref}`extenzivna_a` a {numref}`intenzivna_a`.

````python
# --------------------------------------------------
# Parametre
# --------------------------------------------------

A1 = 1
A2 = 1.5

α = 0.5

# --------------------------------------------------
# Funkcie
# --------------------------------------------------

def f(x,A):
    return A*x**α

def mp(x,A):
    return A*α*x**(α-1)

def ap(x,A):
    return f(x,A)/x

# --------------------------------------------------
# Dáta
# --------------------------------------------------

x = np.linspace(1e-15,80,500)

K1 = 40

# --------------------------------------------------
# Obrázok
# --------------------------------------------------

fig, ax = plt.subplots(
    3,2,
    figsize=(14,12),
    sharex='col'
)

# ==================================================
# Produkčné funkcie
# ==================================================

for j,nazov in enumerate(["K","N"]):

    g = ax[0,j]

    g.plot(x,f(x,A1),color="black",linewidth=3)
    g.plot(x,f(x,A2),color="black",linewidth=3)

    g.axvline(K1,linestyle=":",color="gray")

    g.scatter(K1,f(K1,A1),color="red")
    g.scatter(K1,f(K1,A2),color="red")

    if nazov=="K":
        g.text(K1+2,f(K1,A1)-0.8,r"$Y_1$")
        g.text(K1+2,f(K1,A2)-0.8,r"$Y_2$")
    else:
        g.text(K1+2,f(K1,A1)-0.8,r"$Y_1$")
        g.text(K1+2,f(K1,A2)-0.8,r"$Y_2$")

    # spojnice

    g.plot(
        [0,K1],
        [0,f(K1,A1)],
        "--",
        color="tab:blue",
        linewidth=2
    )

    g.plot(
        [0,K1],
        [0,f(K1,A2)],
        "--",
        color="navy",
        linewidth=2
    )

    # dotyčnice

    xx = np.linspace(20,60,50)

    a,b = dotycnica(
        lambda z: f(z,A1),
        lambda z: mp(z,A1),
        K1
    )

    g.plot(
        xx,
        a+b*xx,
        color="limegreen",
        linewidth=2
    )

    a,b = dotycnica(
        lambda z: f(z,A2),
        lambda z: mp(z,A2),
        K1
    )

    g.plot(
        xx,
        a+b*xx,
        color="darkgreen",
        linewidth=2
    )

    # šípka

    x_s = 60

    g.annotate(
        "",
        xy=(x_s,f(x_s,A2)),
        xytext=(x_s,f(x_s,A1)),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=2
        )
    )

    g.text(
        x_s+2,
        (f(x_s,A1)+f(x_s,A2))/2,
        r"$A_2>A_1$"
    )

    nakresli_osi(g)

    g.set_title(rf"$Y({nazov})$")
    g.set_ylabel("$Y$")

# ==================================================
# MPK a MPN
# ==================================================

for j,nazov in enumerate(["K","N"]):

    g = ax[1,j]

    g.plot(
        x,
        mp(x,A1),
        color="limegreen",
        linewidth=3
    )

    g.plot(
        x,
        mp(x,A2),
        color="darkgreen",
        linewidth=3
    )

    g.axvline(
        K1,
        linestyle=":",
        color="gray"
    )

    g.scatter(
        K1,
        mp(K1,A1),
        color="limegreen"
    )

    g.scatter(
        K1,
        mp(K1,A2),
        color="darkgreen"
    )

    g.text(
        K1+2,
        mp(K1,A1)-0.017,
        rf"$MP{nazov}_1$"
    )

    g.text(
        K1+2,
        mp(K1,A2)+0.005,
        rf"$MP{nazov}_2$"
    )

    g.set_title(rf"$MP{nazov}$")
    g.set_ylabel("MP")
    g.set_ylim(0,0.2)

# ==================================================
# APK a APN
# ==================================================

for j,nazov in enumerate(["K","N"]):

    g = ax[2,j]

    g.plot(
        x,
        ap(x,A1),
        color="tab:blue",
        linewidth=3
    )

    g.plot(
        x,
        ap(x,A2),
        color="navy",
        linewidth=3
    )

    g.axvline(
        K1,
        linestyle=":",
        color="gray"
    )

    g.scatter(
        K1,
        ap(K1,A1),
        color="tab:blue"
    )

    g.scatter(
        K1,
        ap(K1,A2),
        color="navy"
    )

    g.text(
        K1+2,
        ap(K1,A1)-0.027,
        rf"$AP{nazov}_1$"
    )

    g.text(
        K1+2,
        ap(K1,A2)+0.005,
        rf"$AP{nazov}_2$"
    )

    g.set_title(rf"$AP{nazov}$")
    g.set_ylabel("AP")
    g.set_xlabel(f"${nazov}$")
    g.set_ylim(0,0.35)

plt.tight_layout()
plt.show()
````

## Vplyv zvýšenia kapitálu na produkčnú funkciu

Predpokladajme, že sa zásoba kapitálu zvýši z $K_1$ na $K_2$, pričom

$$
K_2>K_1.
$$

V dôsledku rastúcej produkčnej funkcie sa výstup zvýši.

Keďže produkčná funkcia je konkávna, hraničný produkt kapitálu je klesajúci. Pri vyššej zásobe kapitálu preto platí

$$
MPK(K_2)<MPK(K_1).
$$

Rovnaký záver platí aj pre priemerný produkt kapitálu:

$$
APK(K_2)<APK(K_1).
$$

Naopak, kapitál zvyšuje produktivitu práce. Zvýšením kapitálu sa preto zvyšujú

$$
MPN
\quad\text{a}\quad
APN.
$$

Zvýšenie kapitálu teda pôsobí na produkty výrobných faktorov opačným spôsobom:

- znižuje hraničný a priemerný produkt kapitálu,

### Príklad: Produkčná funkcia s jedným vstupom a optimálna produkcia

Firma vyrába podľa produkčnej funkcie

$$ 
y=x^{1/2}, \quad x>0. 
$$

Preukážte, že produkčná funkcia je rastúca a konkávna.

Ak je cena produkcie

$$
p=10
$$

a cena vstupu

$$
c=1,
$$

vypočítajte optimálnu spotrebu vstupu a optimálnu produkciu firmy.

#### Riešenie

Najskôr vyjadríme prvú deriváciu produkčnej funkcie

$$
f(x)=x^{1/2}.
$$

Platí

$$
f'(x)=\frac{1}{2}x^{-1/2}
=\frac{1}{2\sqrt{x}}.
$$

Keďže

$$
f'(x)>0
$$

pre všetky

$$
x<0,
$$

produkčná funkcia je rastúca.

Vyjadríme druhú deriváciu

$$
f''(x)
=
-\frac{1}{4}x^{-3/2}
=
-\frac{1}{4x^{3/2}}.
$$

Keďže

$$
f''(x)\le0
$$

pre všetky

$$
x\ge0,
$$

produkčná funkcia je konkávna.

Zisk firmy je

$$
\pi = pf(x)-cx.
$$

Po dosadení dostaneme

$$
\pi = 10\sqrt{x}-x.
$$

Podmienka prvého rádu je

$$
\frac{d\pi}{dx}
=
\frac{10}{2\sqrt{x}}-1
=
0.
$$

Po úprave dostaneme

$$
\frac{5}{\sqrt{x}}
=
1.
$$

Odtiaľ

$$
\sqrt{x}=5
$$

a teda

$$
x^*=25.
$$

Optimálna produkcia je

$$
y^*
=
\sqrt{25}
=
5.
$$

Optimálna spotreba vstupu je

$$
x^*=25
$$

a optimálna produkcia je

$$
y^*=5.
$$

#### Python

Prvú a druhú deriváciu môžeme vypočítať aj pomocou balíka SymPy.

````python
x = symbols("x", positive=True)

f = x**(1/2)

print("Prvá derivácia:")
print(diff(f, x))

print()

print("Je prvá derivácia nezáporná?")
print(ask(Q.nonnegative(diff(f, x))))

print()

print("Druhá derivácia:")
print(diff(f, x, 2))

print()

print("Je druhá derivácia nekladná?")
print(ask(Q.nonpositive(diff(f, x, 2))))
````

Optimálnu spotrebu vstupu a produkciu vypočítame nasledovne.

````python
x = symbols("x", nonnegative=True)

p = 10
c = 1

zisk = p*x**(1/2) - c*x

podmienka = diff(zisk, x)

x_riesenie = solve(podmienka, x, dict=True)
x_riesenie = x_riesenie[0] if type(x_riesenie) is list else x_riesenie
x_opt = x_riesenie[x]

y_opt = simplify(x_opt**(1/2))

print("Optimálna spotreba vstupu =", x_opt)
print("Optimálna produkcia =", y_opt)
````

### Príklad: Produkčná funkcia s dvoma vstupmi

Firma vyrába podľa produkčnej funkcie

$$
y=x_1^{1/3}x_2^{2/3}, \quad x_1>0,\; x_2>0.
$$

1. Preukážte, že produkčná funkcia je rastúca vo vstupoch $x_1$ a $x_2$.
2. Preukážte, že produkčná funkcia je konkávna vo vstupoch $x_1$ a $x_2$.
3. Odvoďte izokvantu produkčnej funkcie.
4. Preukážte, že izokvanta je klesajúca a konvexná.
5. Vyjadrite a interpretujte hraničnú mieru technickej substitúcie.

#### Riešenie

Parciálne derivácie produkčnej funkcie sú

$$
\frac{\partial y}{\partial x_1}
=
\frac{1}{3}x_1^{-2/3}x_2^{2/3}
$$

a

$$
\frac{\partial y}{\partial x_2}
=
\frac{2}{3}x_1^{1/3}x_2^{-1/3}.
$$

Keďže sú obidve parciálne derivácie kladné pre všetky prípustné hodnoty vstupov, produkčná funkcia je rastúca vo vstupoch $x_1$ aj $x_2$.

Druhé parciálne derivácie sú

$$
\frac{\partial^2 y}{\partial x_1^2}
=
-\frac{2}{9}x_1^{-5/3}x_2^{2/3}
$$

a

$$
\frac{\partial^2 y}{\partial x_2^2}
=
-\frac{2}{9}x_1^{1/3}x_2^{-4/3}.
$$

Keďže sú záporné pre všetky prípustné hodnoty vstupov, produkčná funkcia vykazuje klesajúce hraničné produkty vstupov.

Vrstevnicu produkčnej funkcie získame z podmienky

$$
\bar y=x_1^{1/3}x_2^{2/3}.
$$

Po umocnení na tretiu dostaneme

$$
\bar y^3=x_1x_2^2.
$$

Po vyjadrení vstupu $x_2$

$$
x_2=\sqrt{\frac{\bar y^3}{x_1}}.
$$

To je rovnica vrstevnice produkčnej funkcie.

Prvá derivácia vrstevnice je

$$
\frac{dx_2}{dx_1}
=
-\frac{\sqrt{\bar y^3}}{2x_1^{3/2}}.
$$

Keďže je záporná, vrstevnica je klesajúca.

Druhá derivácia vrstevnice je

$$
\frac{d^2x_2}{dx_1^2}
=
\frac{3\sqrt{\bar y^3}}{4x_1^{5/2}}.
$$

Keďže je kladná, vrstevnica je konvexná.

Hraničná miera technickej substitúcie je

$$
MRTS=
\frac{MP_1}{MP_2}.
$$

Po dosadení

$$
MRTS=
\frac{
\frac13x_1^{-2/3}x_2^{2/3}
}{
\frac23x_1^{1/3}x_2^{-1/3}
}.
$$

Po úprave dostaneme

$$
MRTS=
\frac12\frac{x_2}{x_1}.
$$

Hraničná miera technickej substitúcie vyjadruje množstvo vstupu $x_2$, ktoré možno ušetriť pri zvýšení vstupu $x_1$ o jednotku tak, aby produkcia zostala nezmenená.

Keďže

$$
MRTS=
\frac12\frac{x_2}{x_1},
$$

jej hodnota klesá s rastom vstupu $x_1$. Produkčná funkcia preto vykazuje klesajúcu hraničnú mieru technickej substitúcie.

#### Python

Parciálne derivácie vyjadrítame nasledovne.

````python
x1, x2 = symbols("x1 x2", positive=True)

y = x1**(1/3) * x2**(2/3)

mp1 = diff(y, x1)
mp2 = diff(y, x2)

print("MP1 =", mp1)
print("MP2 =", mp2)

print()
print("MP1 >= 0:", ask(Q.nonnegative(mp1)))
print("MP2 >= 0:", ask(Q.nonnegative(mp2)))
````

### Príklad Cobbov-Douglasov tvar produkčnej funkcie

Najznámenjší a najčastejšie používaný tvar produkčnej funkcie {eq}`r_makro_pf` je Cobbov-Douglasov tvar:

```{math}
:label: CDouglas
Y=A K^\alpha N^\beta.
```

Za akých podmienok Cobbov-Douglasov tvar produkčnej funkcie {eq}`CDouglas` spĺňa štandardné makroekonomické vlastnosti? Zistite podmienky, za ktorých produkčná funkcia:

1. prechádza začiatkom súradnicovej sústavy,
2. je rastúca vo vstupoch $K$ a $N$,
3. je konkávna vo vstupoch $K$ a $N$,
4. má elasticitu výstupu vzhľadom na parameter $A$ rovnú jednej,
5. vykazuje konštantné výnosy z rozsahu.

#### Riešenie

Produkčná funkcia má tvar

$$
Y=A K^\alpha N^\beta.
$$

Predpokladajme

$$
A>0.
$$

Produkčná funkcia prechádza začiatkom súradnicovej sústavy, ak platí

$$
Y(0,N)=0
$$

a

$$
Y(K,0)=0.
$$

To nastane práve v prípade

$$
\alpha \ne 0
$$

a

$$
\beta \ne 0.
$$

Parciálna derivácia podľa kapitálu je

$$
MPK=
\frac{\partial Y}{\partial K}
=
A\alpha K^{\alpha-1}N^\beta.
$$

Kombinujme deriváciu s produkčnou funkciou {eq}`CDouglas`:

```{math}
:label: MPK
MPK=
\alpha A \frac{K^\alpha N^\beta}{K}=
\alpha \frac{Y}{K}.
```

Parciálna derivácia podľa práce je

```{math}
:label: MPN
MPN=
\frac{\partial Y}{\partial N}
=
A\beta K^\alpha N^{\beta-1}=
\beta \frac{Y}{N}.
```

Produkčná funkcia je rastúca vo vstupoch, ak platí

$$
MPK>0
$$

a

$$
MPN>0.
$$

Keďže

$$
A>0,
\qquad
K>0,
\qquad
N>0,
$$

platí to práve vtedy, ak

$$
\alpha>0
$$

a

$$
\beta>0.
$$

Druhá parciálna derivácia podľa kapitálu je

$$
\frac{\partial^2Y}{\partial K^2}
=
A\alpha(\alpha-1)K^{\alpha-2}N^\beta.
$$

Kombináciou s {eq}`CDouglas` platí:

$$
\frac{\partial^2Y}{\partial K^2}
=
\alpha \frac{Y}{K^2}.
$$

Druhá parciálna derivácia podľa práce je

$$
\frac{\partial^2Y}{\partial N^2}
=
A\beta(\beta-1)K^\alpha N^{\beta-2}=
\alpha \frac{Y}{N^2}.
$$

Produkčná funkcia je konkávna vo vstupoch, ak sú tieto derivácie nekladné.

Keďže ostatné členy sú kladné, musí platiť

$$
\alpha(\alpha-1)\le 0
$$

a

$$
\beta(\beta-1)\le 0.
$$

Odtiaľ dostaneme podmienky

$$
0<\alpha\le 1
$$

a

$$
0<\beta\le 1.
$$

Elasticita výstupu vzhľadom na parameter $A$ je

$$
\varepsilon_A
=
\frac{\partial Y}{\partial A}
\frac{A}{Y}.
$$

Keďže

$$
\frac{\partial Y}{\partial A}
=
K^\alpha N^\beta,
$$

dostaneme

$$
\varepsilon_A
=
K^\alpha N^\beta
\frac{A}
     {A K^\alpha N^\beta}
=
1.
$$

Elasticita výstupu vzhľadom na parameter $A$ sa teda konštantne rovná jednej.

Pre výnosy z rozsahu zvýšime oba vstupy $t$-násobne:

$$
F(tK,tN)
=
A(tK)^\alpha (tN)^\beta.
$$

Po úprave dostaneme

$$
F(tK,tN)
=
t^{\alpha+\beta}
A K^\alpha N^\beta.
$$

Ak má produkčná funkcia konštantné výnosy z rozsahu, musí platiť

$$
F(tK,tN)
=
tF(K,N).
$$

Teda

$$
t^{\alpha+\beta}=t.
$$

Odtiaľ vyplýva podmienka

$$
\alpha+\beta=1.
$$

Cobbova-Douglasova produkčná funkcia preto spĺňa makroekonomické predpoklady, ak platí

$$
A>0,
$$

$$
0<\alpha\le 1,
$$

$$
0<\beta\le 1,
$$

a pri konštantných výnosoch z rozsahu navyše

$$
\alpha+\beta=1.
$$

#### Python

Najskôr vypočítame parciálne derivácie prvého a druhého rádu.

````python
A, K, N, alpha, beta = symbols(
    "A K N alpha beta",
    positive=True
)

Y = A * K**alpha * N**beta

MPK = diff(Y, K)
MPN = diff(Y, N)

print("MPK =")
print(MPK)

print()

print("MPN =")
print(MPN)

print()

print("Druhá derivácia podľa K =")
print(diff(Y, K, 2))

print()

print("Druhá derivácia podľa N =")
print(diff(Y, N, 2))
````

Elasticitu výstupu vzhľadom na parameter $A$ vypočítame podľa definície.

````python
EA = simplify(
    diff(Y, A) * A / Y
)

print("Elasticita vzhľadom na A =")
print(EA)
````

Výnosy z rozsahu overíme zväčšením vstupov $t$-násobne.

````python
t = symbols("t", positive=True)

Y_t = Y.subs({
    K: t*K,
    N: t*N
})

print("F(tK,tN) =")
print(simplify(Y_t))
````

Faktor pri pôvodnej produkčnej funkcii získame nasledovne.

````python
print(
    simplify(
        Y_t / Y
    )
)
````

Z výsledku 

```python
t**(alpha + beta)
```

vyplýva, že Cobbova-Douglasova produkčná funkcia má konštantné výnosy z rozsahu práve vtedy, ak platí

```python
alpha + beta = 1.
```

### Príklad: Intenzívny tvar produkčnej funkcie

Za predpokladu konštantných výnosov z rozsahu možno Cobbov-Douglasov tvar produkčnej funkcie {eq}`CDouglas` napísať v tvare

```{math}
:label: CD
Y=A K^\alpha N^{(1-\alpha)}.
```

1. Odvoďte intenzívny tvar produkčnej funkcie.
2. Preukážte, že intenzívna produkčná funkcia prechádza začiatkom súradnicovej sústavy.
3. Preukážte, že intenzívna produkčná funkcia je rastúca.
4. Preukážte, že intenzívna produkčná funkcia je konkávna.
5. Vyjadrite elasticitu produkcie vzhľadom na technologický parameter.

#### Riešenie

Vydelme produkčnú funkciu {eq}`CD` prácou $N$.

Získame

$$
\frac{Y}{N}
=
A K^\alpha N^{1-\alpha} N^{-1}.
$$

Po úprave

$$
\frac{Y}{N}
=
A K^\alpha N^{-\alpha}.
$$


$$
\frac{Y}{N}
=
A\left(\frac{K}{N}\right)^\alpha.
$$


Výstup vyjadrený na jednotku práce (intenzívny výstup) je

$$
y=\frac{Y}{N}.
$$

Analogicky intenzívny kapitál je

$$
k=\frac{K}{N}
$$

Intenzívny tvar Cobbovej-Douglasovej produkčnej funkcie je

```{math}
:label: intenzivny_CD
y=A k^\alpha.
```

Pre

$$
k=0
$$

platí

$$
y=A0^\alpha=0.
$$

Intenzívna produkčná funkcia prechádza začiatkom súradnicovej sústavy.

Prvá derivácia intenzívnej produkčnej funkcie je

$$
\frac{dy}{dk}
=
A\alpha k^{\alpha-1}=
\alpha \frac{y}{k} .
$$

Keďže

$$
A>0,
$$

$$
\alpha>0,
$$

a

$$
k>0,
$$

platí

$$
\frac{dy}{dk}>0.
$$

Intenzívna produkčná funkcia je rastúca.

Druhá derivácia intenzívnej produkčnej funkcie je

$$
\frac{d^2y}{dk^2}
=
A\alpha(\alpha-1)k^{\alpha-2}=
\alpha \frac{y}{k^2} .
$$

Pre

$$
0<\alpha\le1
$$

platí

$$
\alpha(\alpha-1)\le0.
$$

Preto

$$
\frac{d^2y}{dk^2}\le0.
$$

Intenzívna produkčná funkcia je konkávna.

Elasticita výstupu vzhľadom na celkovú produktivitu faktorov $A$

je

$$
\varepsilon_A=
\frac{\partial y}{\partial A}\frac{A}{y}=.
$$

Keďže

$$
\frac{\partial y}{\partial A}
=
k^\alpha,
$$

získame

$$
\varepsilon_A=
k^\alpha
\frac{A}
     {A k^\alpha}
=
1.
$$

Elasticita výstupu vzhľadom na celkovú produktivitu faktorov sa rovná 1.

#### Python

Vyjadrime intenzívny tvar funkcie.

````python
A, K, N, alpha = symbols(
    "A K N alpha",
    positive=True
)

Y = A * K**alpha * N**(1-alpha)

y = simplify(Y / N)

print(y)
````

Produkčná funkcia prechádza začiatkom súradnicového systému.

````python
k = symbols(
    "k",
    positive=True
)

y = A * k**alpha

print(y.subs(k, 0))
````

Produkčná funkcia je rastúca v $k$.

````python
print(diff(y, k))
````

Produkčná funkcia je konkávna v $k$.

````python
print(diff(y, k, 2))
````

Elasticita výstupu podľa celkovej produktivity faktorov sa rovná 1.

````python
EA = simplify(
    diff(y, A) * A / y
)

print(EA)
````

### Príklad: Vstupné elasticity výstupu

Z produkčnej funkcie v Cobbovom-Douglasovom tvare {eq}`CDouglas` odvoďte vstupné elasticity výstupu.

### Riešenie

Kapitálová elasticita výstupu je 

$$
\varepsilon_K=
\frac{\partial Y}{\partial K}
\frac{K}{Y}.
$$

Prvý zlomok je hraničný produkt kapitálu {eq}`MPK`. Po dasedení a postupnými úpravami

$$
\varepsilon_K=
\alpha \frac{Y}{K}
\frac{K}{Y} = \alpha.
$$

Elasticita výstupu vzhľadom na kapitál sa rovná parametru $\alpha$

Analogicky, pracovná elasticita výstupu je

$$
\varepsilon_N=
\frac{\partial Y}{\partial N}
\frac{N}{Y}.
$$

Dosaďme {eq}`MPK` za hraničný produkt 

$$
\varepsilon_N=
\beta\frac{Y}{N} \frac{N}{Y} = \beta.
$$

Elasticita výstupu vzhľadom na kapitál sa rovná parametru $\beta$

Parametre Cobbovej-Douglasovej produkčnej funkcie $\alpha$ a $\beta$ priamo vyjadrujú vstupné elasticity výstupu.

Predpoklad konštantných výnosov z rozsahu je {eq}`elasticity_KVR`. Vstupné elasticity interpretujeme ako {ref}`podiel_kapitalu` {eq}`r_podiel_kapitalu` a {ref}`podiel_prace` {eq}`r_podiel_prace`.
