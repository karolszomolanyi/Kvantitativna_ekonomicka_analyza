# Správanie firmy

V analýze správania firmy potrebujeme poznať zisk firmy.

Zisk firmy
: Rozdiel medzi príjmami a ekonomickými nákladmi firmy.

Príjmy firmy 
: tržby z predaja produkcie.

Náklady firmy 
: ekonomické náklady, respektíve príležitostné náklady.

Náklady pozostávajú z
- kapitálu,
- práce,
- materiálu,
- pôdy.

Nech firma realizuje $n$ aktivít.

$$
\mathbf{a}
=
\begin{pmatrix}
a_1\\
a_2\\
\vdots\\
a_n
\end{pmatrix},
$$

kde $a_i$ predstavuje rozsah $i$-tej aktivity firmy. 

Zisk firmy je v tvare:

```{math}
:label: pi-a
\Pi(\mathbf{a}) = R(\mathbf{a}) - C(\mathbf{a}),
```

kde $\Pi(\mathbf{a})$ je zisková funkcia, $R(\mathbf{a})$ príjmová funkcia a $C(\mathbf{a})$ nákladová funkcia.

Firma sa usiluje výberom aktivít $\mathbf{a}$ maximalizovať ziskovú funkciu [](pi-a). Rieši úlohu na voľný extrém ({ref}`volny_extrem`), ktorej podmienky prvého rádu sú v tvare:

$$
\frac{\partial \Pi(\mathbf a)}
     {\partial a_i}
=
\frac{\partial R(\mathbf a)}
     {\partial a_i}
-
\frac{\partial C(\mathbf a)}
     {\partial a_i}
=
0,
\qquad
\forall i=1,\ldots n.
$$

Označme

$$
MR_i(\mathbf a)
=
\frac{\partial R(\mathbf a)}
     {\partial a_i}
$$

ako hraničný príjem z aktivity $i$ a

$$
MC_i(\mathbf a)
=
\frac{\partial C(\mathbf a)}
     {\partial a_i}
$$

ako hraničný náklad aktivity $i$.

Iný zápis podmienok prvého rádu je:

```{math}
:label: PPR

MR_i(\mathbf{a}) = MC_i(\mathbf{a}),\qquad \forall i=1,\ldots n,
```

kde $MR_i(\mathbf{a})$ sú hraničné príjmy a $MC_i(\mathbf{a})$ hraničné náklady.

Firma zvyšuje rozsah aktivity dovtedy, kým dodatočný príjem z aktivity je vyšší ako dodatočný náklad. V optimálnom riešení sa hraničný príjem a hraničný náklad každej aktivity rovnajú.

Firma v dokonalej konkurencii
: Nedokáže ovplyvniť ceny vstupov ani výstupov. Považuje ich za dané.

Neskôr sa budeme zaoberať správaním firmy v podmienkach nedokonalej konkurencie.

V tejto časti budú aktivity firmy $\mathbf{a}$ vyjadrené vektorom vstupov $\mathbf{x}.$

V nasledujúcej časti budú aktivity firmy vyjadrené objemom výstupu $y$.

Predstavíme tri základné koncepty správania firmy:

1. maximalizáciu zisku výberom vstupov,
2. minimalizáciu nákladov pri danom výstupe,
3. maximalizáciu zisku výberom výstupu.

(max_zisk_vstupy)=

## Maximalizácia zisku výberom vstupov

Funkcia zisku je v tvare

$$
\Pi(x_1,x_2,\ldots,x_n)
=
pf(x_1,x_2,\ldots,x_n)
-
(c_1x_1+c_2x_2+\cdots+c_nx_n).
$$

- Cena výstupu je $p>0$.
- Cena vstupu $i$ je $c_i>0$.
- Množstvo vstupu $i$ je $x_i\ge0$.
- Produkčná funkcia je $f(x_1,x_2,\ldots,x_n)$.

Firma maximalizuje zisk výberom vstupov
$x_1,x_2,\ldots,x_n$.

Správanie firmy maximalizujúcej zisk možno formulovať ako úlohu na voľný extrém ({ref}`volny_extrem`. Podmienky prvého rádu sú

```{math}
:label: PPR-x
p\frac{\partial f(x_1,x_2,\ldots,x_n)}{\partial x_i}
-c_i
=0,
\qquad
\forall i=1,\ldots n.

Parciálna derivácia produkčnej funkcie podľa vstupu je hraničný produkt vstupu {eq}`mp_i`. 
Podmienky prvého rádu potom možno zapísať v tvare

$$
MP_i(x_1,x_2,\ldots,x_n)
=
\frac{c_i}{p},
\qquad
\forall i=1,\ldots n.
$$

Pre každý vstup sa hraničný produkt musí rovnať pomeru ceny vstupu a ceny výstupu.

Ekvivalentne

$$
p\,MP_i(x_1,x_2,\ldots,x_n)
=
c_i,
\qquad
\forall i=1,\ldots n.
$$

Hodnota hraničného produktu vstupu sa musí rovnať cene vstupu.

Sú podmienky prvého rádu postačujúce?

Otázku zodpovieme na prípade jedného vstupu.

Firma výberom vstupu maximalizuje funkciu zisku

$$
\Pi(x)=pf(x)-cx.
$$

Podmienka prvého rádu je

$$
pf'(x)-c=0.
$$

Derivácia produkčnej funkcie je hraničný produkt vstupu, 

$$
f'(x) = MPx.
$$

Racionálna firma zvyšuje vstup dovtedy, kým sa hodnota dodatočného výstupu rovná nákladom na dodatočnú jednotku vstupu:

$$
p\,MPx=c.
$$

Po vydelení cenou výstupu získame

$$
MPx=\frac{c}{p}.
$$

Geometrická interpretácia úlohy je na {numref}`pf_optimal`. Optimálna stratégia firmy sa realizuje na hranici výrobných možností.

Podmienka druhého rádu je

$$
\Pi''(x)=pf''(x)
\le 0.
$$

Keďže

$$
p>0,
$$

platí

$$
\Pi''(x)\le 0
\quad \Leftrightarrow \quad
f''(x)\le 0.
$$

Ak je produkčná funkcia konkávna, riešenie podmienok prvého rádu predstavuje kandidáta na globálne maximum. Pre určenie globálneho maxima je potrebné jeho hodnotu porovnať s hodnotami ziskovej funkcie na hraniciach definičného oboru.

Pri produkčnej funkcii s viacerými vstupmi predstavuje riešenie podmienok prvého rádu kandidáta na globálne maximum ziskovej funkcie. Ak je produkčná funkcia konkávna, toto riešenie je maximom. Pri hľadaní globálneho maxima je však potrebné porovnať hodnotu ziskovej funkcie aj na hraniciach definičného oboru. Ekonomicky konkávna produkčná funkcia znamená, že hraničné produkty vstupov klesajú (pozri tiež {ref}`hranicne_produkty_vstupov`).

Riešením podmienok prvého rádu {eq}`PPR-x` získame {ref}`redukovany_tvar` úlohy firmy maximalizujúcej zisk v tvare

```{math}
:label: dopyt_firmy
\mathbf{x}(\mathbf{c},p)
=
\begin{pmatrix}
x_1(c_1,c_2,\ldots,c_n,p) \\
x_2(c_1,c_2,\ldots,c_n,p) \\
\vdots \\
x_n(c_1,c_2,\ldots,c_n,p)
\end{pmatrix}.
```

Rovnice {eq}`dopyt_firmy` vyjadrujú dopytové funkcie firmy. Dopyt po $i$-tom vstupe závisí od cien vstupov a ceny výstupu.

Dosadením dopytových funkcií do produkčnej funkcie firmy získame ponukovú funkciu firmy

```{math}
:label: ponuka_firmy_x
y
=
f(\mathbf{x})
=
f[\mathbf{x}(\mathbf{c},p)].
```

Firma dopytuje vstupy $\mathbf{x}$, napríklad kapitál, prácu, materiály alebo pôdu.

Firma ponúka výstup $y$.

Správanie firmy je preto úplne opísané dopytovými funkciami {eq}`dopyt_firmy` a ponukovou funkciou {eq}`ponuka_firmy_x`.

Dopyt po vstupoch aj ponuka výstupu závisia od cien vstupov a ceny výstupu $(\mathbf{c},p)$.

### Dopyt po výrobných faktoroch

Z makroekonomického hľadiska uvažujme dva vstupy ako výrobné faktory:

- kapitál $K$,
- prácu $N$.

Označme dopyt po kapitáli $K^d$ a dopyt po práci $N^d$. Pre firmu maximalizujúcu zisk platia podmienky prvého rádu

```{math}
:label: ppr_mpk
MPK(K^d,N^d)=\frac qP,
```

```{math}
:label: ppr_mpn
MPN(K^d,N^d)=\frac \omega P,
```

kde:

- $q$ je cena jednotky kapitálu,
- $\omega$ je mzdová sadzba (cena jednotky práce),
- $P$ je cena výstupu.

Tieto rovnice určujú optimálny dopyt firmy po kapitáli a práci.

Ak ostatné premenné považujeme za konštantné, klesajúce hraničné produkty faktorov implikujú klesajúci dopyt po kapitáli a práci.

````python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (12, 5)

α = 1/3

def MPK(K, N):
    return α * N / K

def MPN(K, N):
    return (1 - α) * K / N

K = np.linspace(1e-3, 2, 500)
N = np.linspace(1e-3, 2, 500)

N0 = 1
K0 = 1

fig, ax = plt.subplots(1, 2)

# Dopyt po kapitáli

ax[0].plot(
    K,
    MPK(K, N0),
    color="navy",
    linewidth=3
)

ax[0].set_title("Dopyt po kapitáli")
ax[0].set_xlabel("$K^d$")
ax[0].set_ylabel("$q/P$")
ax[0].set_ylim(0, 3)

# Dopyt po práci

ax[1].plot(
    N,
    MPN(K0, N),
    color="darkgreen",
    linewidth=3
)

ax[1].set_title("Dopyt po práci")
ax[1].set_xlabel("$N^d$")
ax[1].set_ylabel("$\omega/P$")
ax[1].set_ylim(0, 3)

plt.tight_layout()
plt.show()
````

### Zisk firmy a konštantné výnosy z rozsahu

Za predpokladu konštantných výnosov z rozsahu možno podľa Eulerovej vety rozložiť výstup na príspevky výrobných faktorov {eq}`prispevky_vstupov`.

Zisková funkcia s makroekonomickou produkčnou funkciou je:

$$
\Pi = P \times AF(K,N) - qK - \omega N. 
$$

Dosaďme do nej podmienky prvého rádu {eq}`ppr_mpk`, {eq}`ppr_mpn` za ceny vstupov:

$$
\Pi = P \times Y - P \cdot MPK \cdot K - P \cdot MPN \cdot N. 
$$

Po vyjmutí ceny výstupu $P$ platí

$$
\Pi = P (Y - MPK \cdot K - MPN \cdot N).
$$

Kombináciou s {eq}`prispevky_vstupov` získame

```{math}
:label: pi_0
\Pi = P (Y - Y) = 0.
```

Ekonomické zdôvodnenie tejto vlastnosti vyplýva z predpokladu konštantných výnosov z rozsahu a z podmienok dokonalej konkurencie.

Neskôr budeme uvažovať o rôznych trhových štruktúrach. Základný predpoklad, ktorého sa doteraz držíme je, že firma pôsobí v štruktúre dokonalej konkurencie, v ktorej pôsobí veľký počet identických firiem. Dopyt po výstupe firiem je dokonale elastický, jeho cenu firma nedokáže ovplyvniť, firma môže slobodne vstúpiť aj opustiť trh. V krátkom období môžu firmy dosahovať kladný aj záporný ekonomický zisk. Ak je záporný, firmy odchádzajú z trhu, naopak ak je kladný, na trhu pribúdajú nové firmy. Z dlhodobého hľadiska sa rovnováha ustáli pri nulovom zisku.

Skutočnosť, že zisk sa rovná $0$ neznamená, že vlastník firmy neprofituje z podnikania. V reálnom svete si podľa účtovných štandardov vlastník firmy vypláca odmenu za podnikanie vo forme zisku. V modeli firmy predpokladáme, že podnikanie je činnosť, ktorá vyžaduje prácu aj kapitál. Vlastník firmy získa odmenu vo forme mzdy za svoju prácu a renty za prenájom kapitálu (nehnuteľností, strojov a zariadení, ktoré vlastní).

## Minimalizácia nákladov pri konštantnom objeme výroby

V nasledovnom probléme predpokladáme, že firma má splniť objednávku na konštantný objem výroby $y_0$. Firma sa snaží splniť objednávku s minimálnymi nákladmi. Úlohu možno zapísať v tvare:

$$
\min_{x_1,x_2,\ldots x_n}
\;
C(x_1,x_2,\ldots x_n)
=
c_1x_1+c_2x_2+\cdots c_nx_n,
$$

pri podmienke

$$
f(x_1,x_2,\ldots,x_n) - y_0 \ge 0.
$$

Firma hľadá najlacnejšiu kombináciu vstupov, ktorá umožňuje vyrobiť požadovaný objem produkcie.

$$
\min_{x_1,x_2 \cdots x_n}{C(x_1,x_2 \cdots x_n) = c_1x_1 + c_2x_2 + \cdots c_nx_n},\\
f(x_1,x_2...x_n) \ge y_0.
$$


Podľa podmienok prvého rádu úlohy ({ref}`ppr_vseobecne`) optimálny bod leží na hranici prípustných riešení

$$
f(x_1,x_2,\ldots,x_n) - y_0 = 0
$$

a zároveň hraničné miery substitúcie účelovej a ohraničujúcej funkcie sa rovnajú. Pre ľubovoľnú dvojicu rozhodovacích premenných $x_i$ a $x_j$ preto platí

$$
MRCS(x_i,x_j)
=
MRTS(x_i,x_j),
\qquad
i\neq j,
$$

kde $MRCS(x_i,x_j)$ je hraničná miera substitúcie nákladovej funkcie $C(x_1,x_2,\ldots,x_n)$,

$MRTS(x_i,x_j)$ je hraničná miera substitúcie produkčnej funkcie.

Hraničná miera substitúcie nákladovej funkcie je 

$$
MRCS(x_i,x_j)
=
\frac{\dfrac{\partial C}{\partial x_i}}
     {\dfrac{\partial C}{\partial x_j}}
=
\frac{c_i}{c_j}.
$$

Je to sklon vrstevnice nákladovej funkcie ({ref}`vrstevnica_funkcie`). Vrstevnica nákladovej funkcie sa nazýva **izonákladová funkcia** a vyjadruje všetky kombinácie vstupov, ktoré vedú k rovnakým celkovým nákladom.

Hraničná miera substitúcie produkčnej funkcie je {ref}`MRTS`

$$
MRTS(x_i,x_j) = \frac{\frac{\partial f}{\partial x_i}}
     {\frac{\partial f}{\partial x_j}}
=
\frac{MP_i}{MP_j}\qquad i\neq j.
$$

Je to sklon izokvanty ({ref}`izokvanta`).

Preto podmienky prvého rádu možno zapísať v tvare

```{math}
:label: ppr_min_naklady_obj1
\frac{MP_i(x_1,x_2,\ldots,x_n)}
     {MP_j(x_1,x_2,\ldots,x_n)}
=
\frac{c_i}{c_j},
\qquad
i\neq j
```

a

```{math}
:label: ppr_min_naklady_obj2
f(x_1,x_2,\ldots,x_n) - y_0.
```

Na {numref}`graf_fix_vystup` je grafická interpretácia úlohy v dvojrozmernom rozhodovacom priestore firmy $[x_1,x_2]$. {ref}`mnozina_pripustnych_rieseni` je zobrazená zelenou plochou ohraničenou modrou krivkou, ktorá predstavuje hranicu prípustných riešení

$$
f(x_1,x_2)=y_0.
$$

Opäť platí interpretácia, podľa ktorej množina prípustných riešení predstavuje {ref}`mnozina_vyrob_moznosti` a hranica prípustných riešení predstavuje {ref}`hvm`.

Optimálne rozhodnutie firmy sa nachádza v bode, v ktorom sa niektorá z izonákladových čiar dotýka hranice prípustných riešení.

V optimálnom bode majú izonákladová funkcia a izokvanta rovnaký sklon. Preto sa hraničná miera technickej substitúcie výrobných faktorov rovná pomeru cien vstupov.

Z grafickej interpretácie úlohy s dvoma vstupmi vyplýva, že riešenie podmienok prvého rádu predstavuje kandidáta na lokálne minimum nákladov. Ak sú izokvanty produkčnej funkcie klesajúce a konvexné, tento bod predstavuje lokálne minimum nákladov pri danom objeme výroby. Pri hľadaní globálneho minima je potrebné porovnať náklady v bode určenom podmienkami prvého rádu s nákladmi na hraniciach definičného oboru.

````python
# obr. Minimalizácia nákladov pri konštantnom objeme výroby

# Príklad y = x1^(1/3) x2^(2/3), c1 = 1/3, c2 = 2/3, y0 = 1
# Riešenie x1 = x2 = C = 1.

# izokvanta
def h(y0,x1):
    x2 = y0**(1.5)*x1**(-.5)
    return x2

#izonákladová čiara
def c(C,x1):
    x2 = C*3/2 - 1/2*x1
    return x2

y0 = 1.
C = 1.
x_1 = np.linspace(0.001,3.) #definičný obor funkcií

izokvanta = h(y0,x_1)
izonaklad = c(C,x_1)

fig, ax = plt.subplots()

ax.plot(x_1,izokvanta,label= r"Izokvanta $y_0$")
ax.fill_between(x_1,izokvanta,y2=3, color = 'lightgreen', alpha=0.5)
ax.plot(x_1,izonaklad,label= 'Izonákladová čiara',color="black")
ax.set(xlabel = "$x_1$", ylabel = "$x_2$",ylim = (0,3.), xlim = (0,3.))

#bod
ax.scatter(1,1,color="red",zorder=10)

ax.text(
    1.05,
    1.05,
    r"$(x_1^*,x_2^*)$"
)

# pomocné čiary
ax.axvline(1, linestyle=":", color="gray")
ax.axhline(1, linestyle=":", color="gray")
ax.legend()
plt.show()
````

Riešením podmienok {eq}`ppr_min_naklady_obj1` a {eq}`ppr_min_naklady_obj2` získame {ref}`redukovany_tvar` úlohy firmy minimalizujúcej náklady pri konštantnom objeme výroby v tvare

```{math}
:label: podmieneny_dopyt
\mathbf{x}
=
\mathbf{x}(y,\mathbf{c}).
```

Funkcie {eq}`podmieneny_dopyt` sú **podmienené** dopytové funkcie. 

Zodpovedajúca **nákladová funkcia** je

```{math}
:label: nakladova_funkcia
C(y)=c_1x_1(y,\mathbf{c}) + c_2x_2(y,\mathbf{c}) + \cdots c_nx_n(y,\mathbf{c})
```

Podmienené dopytové funkcie nezodpovedajú dopytovým funkciám v úlohe maximalizácie zisku ({ref}`max_zisk_vstupy`). Dopyt firmy po vstupoch je v tomto prípade **podmienený** objemom výroby $y$.

Objem výroby nie je rozhodovacou (endogénnou) premennou. Je zadaný exogénne a firma hľadá najlacnejšiu kombináciu vstupov potrebnú na jeho dosiahnutie.

Zodpovedajúca nákladová funkcia závisí od objemu výroby $y$. Takéto nákladové funkcie pužijeme v nasledovnej kapitole.

### Ekonomická interpretácia PPR

Podľa PPR {eq}`ppr_min_naklady_obj1` a {eq}`ppr_min_naklady_obj2` sa v optimálnom riešení hraničná miera technickej substitúcie rovná pomeru cien vstupov.

Hraničnú mieru technickej substitúcie

$$
MRTS(x_i,x_j)
=
\frac{MP_i}{MP_j}
$$

interpretujeme ako internú mieru substitúcie medzi vstupmi. Vyjadruje, v akom pomere možno nahradiť vstup $x_j$ vstupom $x_i$ bez zmeny objemu výroby.

Pomer cien vstupov $c_i/c_j$ predstavuje externú, trhovú mieru substitúcie medzi vstupmi. Vyjadruje, v akom pomere možno nahrádzať vstupy bez zmeny celkových nákladov.

V optimálnom riešení sa interná a externá miera substitúcie rovnajú.

- Ak

  $$
  \frac{c_i}{c_j}
  >
  \frac{MP_i}{MP_j},
  $$

  vstup $i$ je relatívne drahý vzhľadom na jeho prínos k produkcii. Firma by mala znížiť využívanie vstupu $i$ a zvýšiť využívanie vstupu $j$.

- Ak

  $$
  \frac{c_i}{c_j}
  <
  \frac{MP_i}{MP_j},
  $$

  vstup $i$ je relatívne lacný vzhľadom na jeho prínos k produkcii. Firma by mala zvýšiť využívanie vstupu $i$ a znížiť využívanie vstupu $j$.

## Maximalizácia zisku výberom výstupu

Ak poznáme nákladovú funkciu {eq}`nakladova_funkcia`, môžeme ziskovú funkciu firmy zapísať v tvare

$$
\Pi(y)=py-C(y).
$$

Správanie firmy maximalizujúcej zisk výberom výstupu možno formulovať ako úlohu na {ref}`volny_extrem`. Podmienka prvého rádu je

```{math}
:label: PPR_y
p=C'(y)=MC(y),
```

kde $MC(y)$ sú hraničné náklady,

$$
MC(y)=C'(y).
$$

(def_MC)=
Hraničné náklady
: výstupný {ref}`multiplikator-label` nákladov.

Podmienka {eq}`PPR_y` je postačujúca, ak

$$
C''(y)
=
MC'(y)
\ge 0.
$$

Teda ak je nákladová funkcia konvexná.

Firma zvyšuje výrobu dovtedy, kým príjem z poslednej vyrobenej jednotky prevyšuje náklady na jej výrobu. V optimálnom riešení sa cena výstupu rovná hraničným nákladom.

## Krátkodobé správanie firmy

Z krátkodobého hľadiska predpokladáme, že firma nemôže zmeniť hodnotu niektorých vstupov. Rozlišujeme:

- **variabilné vstupy**, ktoré môže firma krátkodobo meniť,
- **fixné vstupy**, ktorých objem je v krátkom období daný.

Pre zjednodušenie predpokladajme dva vstupy. Kapitál $K$ je fixný vstup a práca $N$ je variabilný vstup:

$$
\mathbf{x}
=
\begin{pmatrix}
K\\
N
\end{pmatrix}.
$$

Ceny vstupov sú ceny jednotky práce a kapitálu

$$
\mathbf{c}
=
\begin{pmatrix}
q\\
\omega
\end{pmatrix}.
$$

Krátkodobá zisková funkcia je

$$
\Pi
=
pf(K,N)
-
(qK+\omega N),
$$

kde:

- $w$ je mzdová sadzba,
- $q$ je cena jednotky kapitálu.

Keďže vstup $K$ je fixný, náklad

$$
qK
$$

je v krátkom období konštantný. Pri rozhodovaní firmy preto neovplyvňuje optimálnu voľbu práce $N$.

Krátkodobá variabilná zisková funkcia je

$$
\Pi_v
=
pf(K,N)-\omega N.
$$

Krátkodobá nákladová funkcia je

$$
SC(K,N)
=
qK+\omega N.
$$

Krátkodobá variabilná nákladová funkcia je

$$
SVC(N)
=
\omega N.
$$

### Príklad

Odvoďte dopytovú a ponukovú funkciu firmy, ktorej produkčná funkcia je v Cobbovom-Douglasovom tvare

$$
y=x^\alpha,
$$

kde $c$ je cena vstupu a $p$ cena výstupu.

Preukážte, že dopytová a ponuková funkcia sú rastúce v $p$ a klesajúce v $c$.

#### Riešenie

Podľa podmienky prvého rádu {eq}`PPR-x` platí

$$
MP(x)=\frac{c}{p}.
$$

Pre Cobbovu-Douglasovu produkčnú funkciu

$$
y=x^\alpha
$$

je hraničný produkt

$$
MP(x)
=
\frac{\mathrm{d}y}{\mathrm{d}x}
=
\alpha x^{\alpha-1}.
$$

Po dosadení do podmienky prvého rádu získame

$$
\alpha x^{\alpha-1}
=
\frac{c}{p}.
$$

Po úprave

$$
x^{\alpha-1}
=
\frac{c}{\alpha p}.
$$

Optimálny dopyt po vstupe je teda

$$
x(p,c)
=
\left(
\frac{\alpha p}{c}
\right)^{\frac{1}{1-\alpha}}.
$$

Dosadením dopytovej funkcie do produkčnej funkcie získame ponukovú funkciu

$$
y(p,c)
=
\left(
\frac{\alpha p}{c}
\right)^{\frac{\alpha}{1-\alpha}}.
$$

Pretože

$$
0<\alpha<1,
$$

platí

$$
\frac{1}{1-\alpha}>0,
\qquad
\frac{\alpha}{1-\alpha}>0.
$$

Dopytová funkcia aj ponuková funkcia sú preto rastúce v cene výstupu $p$ a klesajúce v cene vstupu $c$.

#### Python

````python
α, p, c, x = symbols(
    'α p c x',
    positive=True,
    real=True
)

f = x**alpha # produkčná funkcia všeobecne

ppr = diff(f,x) - c/p # podmienka prvého rádu

s = solve(ppr,x,dict=True) # rieš podmienku prvého rádu ppr
s = s[0] if type(s) is list else s

print("Dopytová funkcia x(c,p) = ", s[x])
print("Ponuková funkcia y(c,p) =", f.subs(x,s[x]))
````

### Riešenie

Podľa podmienok prvého rádu {eq}`PPR-x` platí

$$
MP_1=\frac{c_1}{p},
$$

$$
MP_2=\frac{c_2}{p}.
$$

Po vydelení oboch rovníc dostaneme

$$
\frac{MP_1}{MP_2}
=
\frac{c_1}{c_2}.
$$

Výsledok sa zhoduje s podmienkou prvého rádu úlohy minimalizácie nákladov {eq}`ppr_min_naklady_obj1`, {eq}`ppr_min_naklady_obj2`.

Pre Cobbovu-Douglasovu produkčnú funkciu sú hraničné produkty

$$
MP_1
=
\alpha x_1^{\alpha-1}x_2^{1-\alpha},
$$

$$
MP_2
=
(1-\alpha)x_1^\alpha x_2^{-\alpha}.
$$

Po dosadení do podmienky prvého rádu získame

$$
\frac{\alpha x_1^{\alpha-1}x_2^{1-\alpha}}
     {(1-\alpha)x_1^\alpha x_2^{-\alpha}}
=
\frac{c_1}{c_2}.
$$

Po úprave

$$
\frac{\alpha}{1-\alpha}
\frac{x_2}{x_1}
=
\frac{c_1}{c_2}.

Optimálny pomer vstupov je preto

$$
\frac{x_2}{x_1}
=
\frac{1-\alpha}{\alpha}
\frac{c_1}{c_2}.
$$

Podmienky maximalizácie zisku teda určujú nielen optimálny rozsah výroby, ale aj nákladovo efektívny pomer vstupov. Ten sa zhoduje s riešením úlohy minimalizácie nákladov pri danom objeme výroby.

### Príklad

Nákladová funkcia firmy je

$$
C(y)=y^2-3y+4,
$$

kde $y$ je objem výroby. Cena jednotky výstupu je

$$
p=3.
$$

Vypočítajte:

1. optimálny objem výroby,
2. náklady,
3. zisk firmy.

#### Riešenie

Firma maximalizuje zisk

$$
\Pi(y)=py-C(y).
$$

Podľa podmienky prvého rádu {eq}`PPR_y` platí

$$
p=MC(y).
$$

Hraničné náklady sú

$$
MC(y)=C'(y)=2y-3.
$$

Po dosadení ceny výstupu získame

$$
3=2y-3.
$$

Odtiaľ

$$
y=3.
$$

Optimálny objem výroby je

$$
y^*=3.
$$

Náklady firmy sú

$$
C(3)=3^2-3\times3+4=4.
$$

Príjmy firmy sú

$$
R(3)=3\times3=9.
$$

Zisk firmy je

$$
\Pi(3)=9-4=5.
$$

Firma preto vyrába

$$
y^*=3,
$$

jej náklady sú

$$
C=4,
$$

a zisk

$$
\Pi=5.
$$

#### Python

````python
from sympy import symbols, diff, solve

y = symbols('y', real=True)

# cena výstupu
p = 3

# nákladová funkcia
C = y**2 - 3*y + 4

# hraničné náklady
MC = diff(C, y)

# PPR: p = MC
riesenie = solve(MC - p, y, dict = True)
riesenie = riesenie[0] if type(riesenie) is list else riesenie 
y_star = riesenie[y]

# náklady
C_star = C.subs(y, y_star)

# príjmy
R_star = p*y_star

# zisk
Pi_star = R_star * C_star

print("Optimálny výstup", y_star)
print("Náklady:", C_star)
print("Zisk:", Pi_star)
````
