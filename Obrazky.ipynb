# Použité obrázky

Nasledovné kódy slúžia na získanie všetkých obrázkov použitých v učebnici. Kód v každej kapitole vytvorí v pracovnom priečinku používateľa podpriečinok *images* s ďalším podpriečinkom kapitoly, v ktorom sa uložia obrázky.

## Matematický aparát predmetu

Názov podpriečinka je "matematika".

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch

Path("images/matematika").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

#############
# dotyčnice #
#############

#Ako zobraziť dotyčnice ku krivke?
#Dotyčnica je priamka y = a + bx
#sklon dotyčnice je prvá derivácia g(x): b = g'(x)
#Dotyčnica musí prechádzať bodom dotyku, v ktorom platí: g(x) = a + bx
#Teda: g(x) = a + g'(x) x
#Takže a = g(x) - g'(x) x

def dotycnica(g,gder,x):
    b = gder(x)
    a = g(x) - gder(x)*x
    return a,b

# funkcia prírastkov (prvých diferencií)
def Δf(f,x):
    Δy = f(x+1) - f(x)
    return Δy

###################
# Obr funkcia_50x #
###################

def funkcia_50x():
    #Obrázok zobrazuje vzťah medzi dĺžkou plota a nákladmi stavby plota. Je to zobrazenie funkcie y = 50x
 
    #Definícia veľkosti všetkých obrázkov
    #plt.rcParams["figure.figsize"] = (10, 6) #set default figure size
    
    #lineárna funkcia
    def f(x):
        y = 50*x
        return y
    
    #definičný obor ľavý panel
    x = np.linspace(0,5)
    
    y = f(x)
    
    #zobraz priebeh funkcie
    fig, ax = plt.subplots(figsize=(12,6))
    ax.plot(x,y, color = 'k', linewidth = 4)
    ax.set(title = "Funkcia $f(x) = 50x$", xlabel = "$x$", ylabel = "$f(x)$")
    
    plt.savefig(
        "images/matematika/funkcia_50x.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()
    
#################
# Obr oparabola #
#################

def oparabola():
    
    # funkcia f(x) = 25 - sqrt(x)
    def f(x):
        y = 25 - x**2
        return y
    
    # funkcia prvej derivácie, pre dotyčnice
    def fder(x):
        y = -2*x
        return y
    
    #definičný obor
    xl = np.linspace(0,5)
    
    xp = np.arange(1,5,1)
    
    
    #dotyčnice
    a1,b1 = dotycnica(f,fder,1)
    a2,b2 = dotycnica(f,fder,4)
    
    #definičné obory dotyčníc
    x1 = np.linspace(0,2,10)
    x2 = np.linspace(3,5,10)
    
    y1 = a1 + b1*x1
    y2 = a2 + b2*x2
    
    yl = f(xl)
    yp = Δf(f,xp)
    
    #zobraz priebeh funkcie a prírastky
    fig, axes = plt.subplots(1,2, figsize=(12,6))
    
    axes[0].plot(xl,yl, color = 'k', linewidth = 4)
    
    #dotyčnice
    axes[0].plot(x1,y1)
    axes[0].plot(x2,y2)
    
    axes[0].set_xlabel("$x$")
    axes[0].set_ylabel("$g(x)$")
    
    axes[0].set_title("Priebeh funkcie")
    
    axes[1].plot(xp,yp, 'o', color = 'k',  linewidth = 4 )
    
    axes[1].set_xlabel("$x$")
    axes[1].set_xticks(range(1,5)) #jednotky na x osi
    axes[1].set_ylabel("$Δg(x) = g(x+1) - g(x)$")
    
    axes[1].set_title("Prírastky")

    plt.tight_layout()
    plt.savefig(
        "images/matematika/oparabola.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

################
# Obr parabola #
################

def parabola():
    #Obrázok: Konvexná funkcia
    # funkcia f(x) = x^2
    
    def f(x):
        y = x**2
        return y
    
    def fder(x):
        y = 2*x
        return y
    
    #definičný obor
    xl = np.linspace(-10,10)
    xp = np.arange(-9,10,1)
    
    yl = f(xl)
    yp = Δf(f,xp)
    
    #dotyčnice
    a1,b1 = dotycnica(f,fder,-8)
    a2,b2 = dotycnica(f,fder,8)
    a3,b3 = dotycnica(f,fder,0)
    
    #definičné obory dotyčníc
    x1 = np.linspace(-10,-6,10)
    x2 = np.linspace(6,10,10)
    x3 = np.linspace(-2,2,10)
    
    y1 = a1 + b1*x1
    y2 = a2 + b2*x2
    y3 = a3 + b3*x3
    
    #zobraz priebeh funkcie a prírastky
    fig, axes = plt.subplots(1,2,figsize=(12,6))
    axes[0].plot(xl,yl, color = 'k', linewidth = 4)
    
    #dotyčnice
    axes[0].plot(x1,y1)
    axes[0].plot(x2,y2)
    axes[0].plot(x3,y3)
    
    axes[0].set_xlabel("$x$")
    axes[0].set_ylabel("$h(x)$")
    
    axes[0].set_title("Priebeh funkcie")
    
    axes[1].plot(xp,yp, 'o', color = 'k',  linewidth = 4 )
    
    axes[1].set_xlabel("$x$")
    axes[1].set_xticks(range(-9,10)) #jednotky na x osi
    axes[1].set_ylabel("$Δh(x) = h(x+1) - h(x)$")
    
    axes[1].set_title("Prírastky")
    plt.tight_layout()
    plt.savefig(
        "images/matematika/parabola.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

#####################
# Obr geo_derivacia #
#####################

def geo_derivacia():

    # Parameters
    a = 1.5
    x = np.linspace(-6, 6, 1000)
    y = a * x
    
    # Create figure
    plt.figure(figsize=(10, 7))
    plt.plot(x, y, 'b-', linewidth=2.5, label=f'$y = ax$')
    
    # Show the angle α at the origin (0,0)
    origin_x, origin_y = 0, 0
    
    # Arc showing the angle α between x-axis and the line
    arc_radius = 0.8
    arc = Arc((origin_x, origin_y), width=2*arc_radius, height=2*arc_radius,
              angle=0, theta1=0, theta2=np.degrees(np.arctan(a)),
              color='darkgreen', linewidth=2.5)
    plt.gca().add_patch(arc)
    
    # Text for the angle
    plt.text(0.25, 0.12, '$α$', fontsize=18, color='darkgreen', fontweight='bold')
    
    # Tangent and derivative labels
    plt.text(-3.5, 3.2, f'$tg(α) = a$', fontsize=14, color='darkred')
    plt.text(-3.5, 2.5, f"$f'(x) = a$", fontsize=14, color='darkred')
    
    # Annotation connecting derivative and tan α
    plt.annotate("f'(x) = a = tg(α)",
                 xy=(0.9, 1.4), xytext=(-3.5, 4.0),
                 fontsize=13, color='navy',
                 arrowprops=dict(facecolor='navy', shrink=0.05, width=1.5, headwidth=9)
                )
    
    # Small segment near origin to emphasize slope
    dx = 0.6
    plt.plot([0, dx], [0, a*dx], 'r-', linewidth=3, 
             label=f'$sklon = Δy/Δx = a$')
    
    # Arrows on axes
    
    plt.axhline(0, color='black', linewidth=0.8)
    plt.axvline(0, color='black', linewidth=0.8)
    
    plt.grid(True, alpha=0.3, linestyle='--')
    plt.title('Lineárna funkcia $y = ax$    →    $f\'(x) = a = tg(α)$', fontsize=16, pad=15)
    plt.xlabel('x', fontsize=13)
    plt.ylabel('y', fontsize=13)
    plt.legend(loc='upper left', fontsize=11)
    ax = plt.gca()
    ax.set_aspect('equal', adjustable='box')
    
    plt.xlim(-9, 9)
    plt.ylim(-6, 6)
    
    plt.savefig(
        "images/matematika/geo_derivacia.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

###############
# Obr sqrt(x) #
###############

def sqrt_x():

    plt.rcdefaults()
    
    #g(x) = sqrt(x)
    def g(x):
        y = np.sqrt(x)
        return y
           
    def gder(x):
        y = 1/2/np.sqrt(x)
        return y
        
    #definičný obor
    x = np.linspace(0,10,200)
    
    y = g(x)
    
    #dotyčnice
    a1,b1 = dotycnica(g,gder,2)
    a2,b2 = dotycnica(g,gder,6)
    
    #definičné obory dotyčníc
    x1 = np.linspace(0.5,3.5,10)
    x2 = np.linspace(4.5,8.5,10)
    
    y1 = a1 + b1*x1
    y2 = a2 + b2*x2
    
    fig, ax = plt.subplots(figsize=(10,6))
    
    ax.plot(x,y, color ='k', linewidth = 2) #priebeh funkcie
    
    ax.plot(x1,y1) #dotyčnica v okolí bodu 2
    ax.plot(x2,y2, color = 'r') #dotyčnica v okolí bodu 6
    
    ax.plot([2,2], [g(2),0], ':', color = 'b')
    ax.plot([2,0], [g(2),g(2)], ':', color = 'b')
    
    ax.plot([6,6], [g(6),0], ':', color = 'r')
    ax.plot([6,0], [g(6),g(6)], ':', color = 'r')
    
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    #komenty
    ax.text(0.5,1.85,"sklon dotyčnice = $i'(x_1)$", color = 'b')
    ax.text(5,2.85,"sklon dotyčnice = $i'(x_2)$", color = 'r')
    
    #osy
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    ax.set_title("Funkcia $i(x)=\sqrt{x}$")
    
    ax.set_xlabel("$x$")
    ax.set_ylabel("$i(x)$")
    
    plt.savefig(
        "images/matematika/sqrt_x.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

#################
# Obr prirastky #
#################

def prirastky():
    # Diskrétna verzia v definičnom obore x = 1,2, ... 7.
    # Diskrétne zobrazenie multiplikátora 

    def g(x):
        y = np.sqrt(x)
        return y
    
    x = np.arange(1,8)
    y = g(x)
    
    #zobrazíme 4 body charakterizujúce zvýšenie x o jednotku z x=2 na x=3 a x=6 na x=7
    body_x = [2,3,6,7]
    body_y = [g(2),g(3),g(6),g(7)]
    farba = ['r','b','y','m']
    
    fig, ax = plt.subplots()
    
    ax.plot(x,y, 'o', color ='k', linewidth = 2) #priebeh funkcie
    
    #body
    for xx,yy,f in zip(body_x, body_y, farba):
        ax.scatter(xx,yy,marker='x', color = f, linewidth = 4)
        ax.plot([xx,xx], [yy,1], ':', color = f)
        ax.plot([xx,1], [yy,yy], ':', color = f)
    
    #osi
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,1,xh-xd,0)
    ax.arrow(1,yd,0,yh-yd)
    
    #obojstranné šípky zobrazujúce nárast y spôsobený jednotkovým nárastom x v oboch bodoch 
    ax.arrow(1.5,g(2),0,g(3)-g(2), head_width = 0.05, head_length = 0.03, length_includes_head = True)
    ax.arrow(1.5,g(2)+0.03,0,-0.03, head_width = 0.05, head_length = 0.03, length_includes_head = True)
    
    ax.arrow(5,g(6),0,g(7)-g(6), head_width = 0.05, head_length = 0.03, length_includes_head = True)
    ax.arrow(5,g(6)+0.03,0,-0.03, head_width = 0.05, head_length = 0.03, length_includes_head = True)
    
    #komenty
    ax.text(1.6,1.55,"$\mu_1$")
    ax.text(5.1,2.53,"$\mu_2$")
    
    ax.set_title("Geometrická interpretácia zmeny $y$ spôsobenej jednotkovým zvýšením $x$")
    
    ax.set_xlabel("$x$")
    ax.set_ylabel("$i(x)$")
    plt.savefig(
        "images/matematika/prirastky.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

###############
# Obr kubicka #
###############

def kubicka():
    #kubická funkcia
    
    plt.rcParams["figure.figsize"] = (10, 6) #set default figure size
    
    #nelineárna funkcia
    def f(x):
        y = 30*x**2 - 1.5*x**3
        return y
    
    def fder(x):
        y = 60*x - 4.5*x**2
        return y
    
    #definičný obor
    x = np.linspace(-6,20,100)
    
    y = f(x)
    
    a1,b1 = dotycnica(f,fder,0)
    a2,b2 = dotycnica(f,fder,40/3)
    
    x1 = np.linspace(-2,2,10)
    x2 = np.linspace(12,15,10)
    
    y1 = a1 + b1*x1
    y2 = a2 + b2*x2
    
    fig,ax = plt.subplots()
    
    ax.plot(x,y, color ='k', linewidth = 2) #priebeh funkcie
    
    #dotyčnice
    ax.plot(x1,y1)
    ax.plot(x2,y2)
    
    #osi
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    ax.axvline(40/3,yd,yh, linestyle = '--',color = 'k')# pomocná čiarkovaná
    
    #komenty
    ax.text(-6,1750,"$\mu(x)<0$")
    ax.text(5,1750,"$\mu(x)>0$")
    ax.text(16,1750,"$\mu(x)<0$")
    
    ax.set_title("Funkcia $j(x) =30x^3 - 1,5x^2$")
    
    ax.set_xlabel("$x$")
    ax.set_ylabel("$j(x)$")
    
    plt.savefig(
        "images/matematika/kubicka.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

##################
# Obr elasticita #
##################

def elasticita():
    # Obrázok Elasticita
    # Zobrazenie elastického neelastického a jednotkovo elastického vzťahu.
    
    α = np.array([[0.5,1,2],
         [-5,-1,-0.2]]) #exponenty funkcií (elastický, jednotkovo elastický, neelastický)
    tit_e = ["elastický", "jednotkovo elastický", "neelastický"]
    
    tit_k = ["Kladný", "Záporny"]
    
    tit_o = np.array([["a)","b)","c)"],
             ["d)","e)","f)",]])
    
    x = np.array([[np.linspace(0,9),np.linspace(0,3),np.linspace(0,np.sqrt(3))],
         [np.linspace(1,10),np.linspace(1,10),np.linspace(1,10)]])#definičný obor funkcií
    
    fig, axes = plt.subplots(2,3, figsize=(16,10))
    
    for i in range(2) :
        for j in range(3):
            y = x[i,j]**α[i,j] #funkčný vzťah
            axes[i,j].plot(x[i,j],y) #zobraz
            axes[i,j].set(title = tit_o[i,j]+" "+tit_k[i]+" "+tit_e[j]+" vzťah", xlabel = "$x$",) # názov, osy

    plt.savefig(
        "images/matematika/elasticita.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

#################
# Obr. inflexny #
#################

def inflexny():

    def f(x):
        y = 30*x**2 - 1.5*x**3
        return y
    
    def fder(x):
        y = 60*x - 4.5*x**2
        return y
    
    #definičný obor
    x = np.linspace(-6,20,100)
    
    y = f(x)
    
    #dotyčnica
    a,b = dotycnica(f,fder,20/3)
    
    
    #definičné obory dotyčníc
    x1 = np.linspace(3,11,10)
    
    y1 = a + b*x1
    
    fig,ax = plt.subplots()
    
    ax.plot(x,y, color ='k', linewidth = 2) #priebeh funkcie
    ax.plot(x1,y1)#dotyčnica v inflexnom bode
    
    #osi
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.axvline(20/3,yd,yh, linestyle = '--',color = 'k')# pomocná čiarkovaná
    
    #komenty
    ax.text(-6,1750,"konvexný priebeh")
    ax.text(16,1750,"konkávny priebeh")
    
    ax.set_title("Inflexný bod")
    
    ax.set_xlabel("$x$")
    ax.set_ylabel("$f(x)$")
    plt.savefig(
            "images/matematika/inflexny.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

###########################
# Obr vrstevnice_linearne #
###########################

def vrstevnice_linearne():

    #Obrázok Vrstevnice funkcie f(x_1,x_2) = 0,5x_1 + x_2 = c_i
    
    #Izofunkcia
    def h(y_i, x_1):
        x_2 = y_i - 0.5*x_1
        return x_2
    
    #definičný obor ľavý panel
    x_max = np.array([10,20,30])
    
    y = np.array([5,10,15])
    
    #zobraz priebeh funkcie
    fig, ax = plt.subplots()
    i = 1
    for y_i, m in zip (y,x_max):
        x_1 = np.linspace(0,m)
        x_2 = h(y_i, x_1)
        ax.plot(x_1,x_2,linewidth = 2, label=f'$c_{i}$')
        i = i + 1
    ax.legend()
    ax.set_title("Vrstevnice lineárnej funkcie")
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    
    #osi
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    plt.savefig(
            "images/matematika/vrstevnice_linearne.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

#############################
# Obr vrstevnice_nelinearne #
#############################

def vrstevnice_nelinearne():
    #Obrázok zobrazuje nelineárne vrstevnice.
    
    #Nelineárne izočiary
    #ľavá
    def hl(yi,x1):
        x2 = yi**(1.5)*x1**(-.5)
        return x2
    
    #pravá
    def hp(yi,x1):
        x2 = yi+(100-x1)**(1/2)
        return x2
    
    # Kvôli dotyčniciam vyjadrime funkcie prvých derivácií
    
    def hl_der(yi,x1):
        x2 = -0.5*yi**(1.5)*x1**(-1.5)
        return x2
    
    def hp_der(yi,x1):
        x2 = -.5*(100-x1)**(-1/2)
        return x2
    
    # Keďže uvažujeme funkcie 2 premenných potrebujeme predefinovať funkciu dotyčnice.
    def dotycnica2(g,gder,y,x):
        b = gder(y,x)
        a = g(y,x) - gder(y,x)*x
        return a,b
    
    body_x1 = (20,80) #body dotyčníc (x-ova súradnica)
    
    x_1 = np.linspace(0.001,100) #definičný obor funkcií
    
    y = np.array([8,10,12]) #3 izočiary
    
    #zobraz priebehy funkcií
    fig, axes = plt.subplots(1,2, figsize=(12,6))
    i = 1
    for y_i in y:
        x_2l = hl(y_i, x_1)
        x_2p = hp(y_i, x_1)
        axes[0].plot(x_1,x_2l,linewidth = 2, label=f'$c_{i}$')
        axes[1].plot(x_1,x_2p,linewidth = 2, label=f'$c_{i}$')
        axes[0].legend()
        axes[1].legend()
        axes[0].set(xlabel = "$x_1$", ylabel = "$x_2$",ylim = (0,11), xlim = (1,100))
        axes[1].set(xlabel = "$x_1$", ylabel = "$x_2$")
        i = i + 1
    
    #Zobraz dotyčnice
    yi = 10
    
    for bod in body_x1:
        x1 = np.linspace(bod-5,bod+5,10)
        al, bl = dotycnica2(hl,hl_der,yi,bod)
        ap, bp = dotycnica2(hp,hp_der,yi,bod)
        yl = al + bl*x1
        yp = ap + bp*x1
        axes[0].plot(x1,yl,color='k')
        axes[1].plot(x1,yp,color='k')
        
    #komenty
    axes[0].text(17,6.5,"$A$", color = 'k')
    axes[0].text(78,3.1,"$B$", color = 'k')
    axes[1].text(18,18.2,"$C$", color = 'k')
    axes[1].text(78,13.7,"$D$", color = 'k')
    
    fig.suptitle("Nelineárne vrstevnice")

    plt.savefig(
            "images/matematika/vrstevnice_nelinearne.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

##################
# mapa_vrstevnic #
##################

def mapa_vrstevnic():
    # mapa vrstevníc
    
    # pôvodná dvojrozmerná funkcia
    def f(x1,x2):
        y = x1**(1/3)*x2**(2/3)
        return y
    
    x1 = np.linspace(0.0001, 20, 100) # os x
    x2 = np.linspace(0.0001, 20, 100).reshape((100, 1)) # vytvorí stĺpcový vektor; na zmenu tvaru matice slúži .reshape
    
    fig, ax = plt.subplots()
    
    # farebné obrysy
    contours = ax.contourf(x1, x2.flatten(), f(x1, x2)) # https://matplotlib.org/stable/api/_as_gen/matplotlib.axes.Axes.contourf.html
    # .flatten() vytvorí zo stĺpcových vektorov naspäť riadkový, pri f(x1,x2) však potrebujeme kombinovať riadkový (x1) so stĺpcovým (x2). 
    fig.colorbar(contours) #colorbar v pravej časti obrázku
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    ax.set_title("Mapa vrstevníc")

    plt.savefig(
            "images/matematika/mapa_vrstevnic.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

##########################
# Obr pripustne_riesenia #
##########################

def pripustne_riesenia():
    
    def hp(yi,x1):
        x2 = yi+(100-x1)**(1/2)
        return x2    
    
    x_1 = np.linspace(0,100,500)
    
    gm = 0
    
    x_2 = hp(gm,x_1)
    
    fig, ax = plt.subplots()
    ax.plot(x_1,x_2, color = 'k', label = 'Hranica prípustných riešení')
    ax.fill_between(x_1,x_2, color = 'g')
    
    #body
    body_x = [60,50,70]
    body_y = [5,hp(gm,50),8]
    farba = ['r','b','y']
    legenda = ['$g(x_1^r, x_2^r) < 0$', '$g(x_1^b, x_2^b) = 0$', '$g(x_1^y, x_2^y) > 0$']
    
    for xx,yy,f,leg in zip(body_x, body_y, farba, legenda):
        ax.scatter(xx,yy,marker='x', color = f, label = leg)
        ax.plot([xx,xx], [yy,0], ':', color = f)
        ax.plot([xx,0], [yy,yy], ':', color = f)
    
    ax.set_xlim(0,130)
    ax.set_ylim(0,13)
    
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    #osy
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    ax.legend(loc='upper right')
    
    ax.set_title('Množina prípustných riešení')

    plt.savefig(
            "images/matematika/pripustne_riesenia.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

###########################
# Obr graf_viazany_extrem #
###########################

def viazany_extrem():

    #Obr. Grafické riešenie úlohy na viazaný extrém
    
    #Najskôr vypočítajme optimálne riešenie, aby sme vedeli, 
    #ktorým bodom bude prechádzať optimálna izočiara účelovej funkcie. 
    #V tomto bode tiež nakreslíme dotyčnicu.
    
    from scipy.optimize import fsolve #nástroj, pomocou ktorého vyriešime sústavu dvoch nelineárnych rovníc

    def hp(yi,x1):
        x2 = yi+(100-x1)**(1/2)
        return x2    
    def hl(yi,x1):
        x2 = yi**(1.5)*x1**(-.5)
        return x2    # účelová funkcia
    def hl_der(yi,x1):
        x2 = -0.5*yi**(1.5)*x1**(-1.5)
        return x2
    def dotycnica2(g,gder,y,x):
        b = gder(y,x)
        a = g(y,x) - gder(y,x)*x
        return a,b
    
    def f(x1,x2):
        y = x1**(1/3)*x2**(2/3)
        return y
    
    # funkcia v ohraničení
    def g(x1,x2):
        g = x2 - np.sqrt(100 - x1)
        return g
    
    # hraničná miera substitúcie účelovej funkcie
    def mrsf(x1,x2):
        my = 1/2*x2/x1
        return my
    
    # hraničná miera substitúcie funkcie v ohraničení
    def mrsg(x1,x2):
        mg = 1/2/np.sqrt(100 - x1)
        return mg
    
    #podmienky prvého rádu
    def ppr(endog):
        x1, x2 = endog
        return(mrsf(x1,x2) - mrsg(x1,x2), #hraničné miery subtitúcie sa rovnajú
               g(x1,x2)) # ohraničenie sa spĺňa ako rovnosť
    
    x1, x2 = fsolve(ppr, (50,np.sqrt(50))) #optimálne riešenie
    
    f_max = f(x1,x2) #optimálna funkčná hodnota účelovej funkcie
    
    x_1 = np.linspace(1e-15,100,500)
    
    f_hodnoty = (10.5,f_max,16.5)
    legenda = ("$f_1$","$f_{max}$","$f_2$") 
    
    x_2 = hp(0,x_1)
    
    fig,ax = plt.subplots()
    
    ax.plot(x_1,x_2, color = 'k', label = 'Hranica prípustných riešení') #ohraničenie
    ax.fill_between(x_1,x_2, color = 'g', label = "Množina prípustných riešení") #množina prípustných riešení
    
    #vrstevnice účelovej funkcie
    
    ax.set_xlim(0,100)
    ax.set_ylim(0,10)
    
    for ff,leg in zip(f_hodnoty, legenda):
        x_2 = hl(ff,x_1)
        ax.plot(x_1,x_2, label = leg)
    
    #dotyčnica oboch funkcií v optimálnom bode
    # Dotyčnica v bode optima je pre obe funkcie rovnaká, preto nezáleží, z akej funkcie budeme počítať jej konštanty.
    
    a, b = dotycnica2(hl, hl_der, f_max, 50)
    x_1 = np.linspace(30,70) #definícia dotyčnice
    x_2 = a + b*x_1
    ax.plot(x_1,x_2, label = "$MRS_f = MRS_g$")
    
    
    #osy
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    ax.legend()
    ax.set_title("Grafické riešenie úlohy na viazaný extrém")
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")

    plt.savefig(
            "images/matematika/graf_viazany_extrem.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

##########################
# Obr alternativ_viazany #
##########################

def alternativ_viazany():

    # Obrázok: Alternatívna formulácia úlohy na viazaný extrém
    
    
    # Cobb-Douglas vrstevnica
    def f_iso(c, x1):
        return c**(1.5) * x1**(-0.5)
    
    # hranica prípustných riešení
    def g_iso(x1):
        return np.sqrt(100 - x1)
    
    # derivácie
    def df_iso(c, x1):
        return -0.5 * c**(1.5) * x1**(-1.5)
    
    def dg_iso(x1):
        return -0.5 * (100 - x1)**(-0.5)
    
    # optimálny bod (skutočný bod dotyku)
    xE = 50
    yE = np.sqrt(50)
    
    # príslušná hodnota Cobb-Douglasovej funkcie
    c = xE**(1/3) * yE**(2/3)
    
    # dotyčnica
    b = df_iso(c, xE)
    a = yE - b*xE
    
    x = np.linspace(1,99.999,500)
    
    fig, ax = plt.subplots(figsize=(9,6))
    
    # hranica prípustných riešení
    fhran = f_iso(c, x)
    ax.plot(x, fhran,
            color='k',
            linewidth=2,
            label=r'$f(x_1,x_2)=c$')
    
    # prípustná oblasť
    ax.fill_between(x, fhran, 12,
                    color='lightgreen',
                    alpha=0.4,
                    label='Množina prípustných riešení')
    
    # minimalizovaná funkcia g
    g = g_iso(x)
    ax.plot(x, g,
            color='red',
            linewidth=2,
            label=r'$g(x_1,x_2)=0$')
    
    # optimálny bod
    ax.scatter(xE, yE,
               color='blue',
               s=70,
               zorder=5)
    
    ax.text(xE+2,
            yE+0.2,
            "E",
            fontsize=12)
    
    # spoločná dotyčnica
    xt = np.linspace(30,70,100)
    yt = a + b*xt
    
    ax.plot(xt, yt,
            '--',
            color='navy',
            linewidth=2,
            label='Spoločná dotyčnica')
    
    # osi
    ax.set_xlim(0,100)
    ax.set_ylim(0,12)
    
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0,
             length_includes_head=True)
    
    ax.arrow(0,yd,0,yh-yd,
             length_includes_head=True)
    
    ax.set_xlabel("$x_1$")
    ax.set_ylabel("$x_2$")
    
    ax.set_title("Alternatívna formulácia úlohy na viazaný extrém")
    
    ax.legend(loc='center left',
              bbox_to_anchor=(1.02,0.5))
    
    plt.tight_layout()

    plt.savefig(
            "images/matematika/alternativ_viazany.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

#############
# Obr model #
#############

def model():

    from matplotlib.patches import Rectangle
    
    fig, ax = plt.subplots(figsize=(8, 3))
    
    # model
    obdlznik = Rectangle((0.4, 0.3), 0.2, 0.4,
                         fill=False,
                         linewidth=2)
    ax.add_patch(obdlznik)
    
    # vstupná šípka
    ax.arrow(0.1, 0.5, 0.3, 0,
             length_includes_head=True,
             head_width=0.04,
             head_length=0.03,
             color='black')
    
    # výstupná šípka
    ax.arrow(0.6, 0.5, 0.3, 0,
             length_includes_head=True,
             head_width=0.04,
             head_length=0.03,
             color='black')
    
    # popisy
    ax.text(0.5, 0.5, "Model",
            ha='center', va='center',
            fontsize=12)
    
    ax.text(0.08, 0.58,
            "Exogénne\npremenné",
            ha='center')
    
    ax.text(0.92, 0.58,
            "Endogénne\npremenné",
            ha='center')
    
    # odstránenie osí
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    
    plt.savefig(
            "images/matematika/model.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

#########################
# Obr efektivna_hranica #
#########################

def efektivna_hranica():
  
    # alternatívy
    body = {
        "A": (8, 3),
        "B": (10, 4),
        "C": (10, 2),
        "D": (7, 1)
    }
    
    fig, ax = plt.subplots(figsize=(8, 6))
    
    # všetky body
    for nazov, (vynos, riziko) in body.items():
        ax.scatter(riziko, vynos, s=120)
    
        ax.annotate(
            nazov,
            (riziko, vynos),
            xytext=(5, 5),
            textcoords="offset points"
        )
    
    # efektívna hranica
    ef_riziko = [1, 2]
    ef_vynos = [7, 10]
    
    ax.plot(
        ef_riziko,
        ef_vynos,
        linewidth=2,
        color="black",
        label="Efektívna hranica"
    )
    
    ax.set_xlabel("Riziko")
    ax.set_ylabel("Výnos")
    ax.set_title("Dominancia a efektívnosť")
    ax.legend()

    plt.savefig(
            "images/matematika/efektivna_hranica.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()


funkcia_50x()
oparabola()
parabola()
geo_derivacia()
sqrt_x()
prirastky()
kubicka()
elasticita()
inflexny()
vrstevnice_nelinearne()
mapa_vrstevnic()
pripustne_riesenia()
viazany_extrem()
alternativ_viazany()
model()
efektivna_hranica()
```

## Výroba firmy

Názov podpriečinka je *produk*.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch

Path("images/produk").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

#Definícia veľkosti všetkých obrázkov
plt.rcParams["figure.figsize"] = (10, 6) #set default figure size

#definičný obor produkčnej funkcie a podiel vstupu
x = np.linspace(0,150,200)

#produkčná funkcia
def f(x):
    y = x**α
    return y

# dotyčnica
def dotycnica(g,gder,x):
    b = gder(x)
    a = g(x) - gder(x)*x
    return a,b

# funkcia prvej derivácie
def fder(x):
    y = α*x**(α-1)
    return y

α = 0.5
    
y = f(x)

def nakresli_osi(ax):
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()

    ax.arrow(xd,0,xh-xd,0,
             length_includes_head=True)

    ax.arrow(0,yd,0,yh-yd,
             length_includes_head=True)

############################
# Obr mnozina_vyr_moznosti #
############################

def mnozina_vyr_moznosti():
       
    #zobraz množinu výrobných možností
    fig, ax = plt.subplots()
    ax.fill_between(x,y, color = 'g')
    ax.set_title("Množina výrobných možností")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.savefig(
            "images/produk/mnozina_vyr_moznosti.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

########################
# Obr vyrobne_moznosti #
########################

def vyrobne_moznosti():

    #Obrázok Porovnanie niektorých výrobných možností
    #Predchádzajúci obrázok rozšírime o zobrazenia niektorých konkrétnych výrobných možností a tiež o zvýraznenie hlavných osí 
    #(pre lepšiu grafickú interpretovarľnosť)
    
    #zobrazíme 4 body, ktoré sú charakterizované súradnicami x, y a farbou
    
    body_x = [100,25,100,50]
    body_y = [5,5,10,8]
    farba = ['r','b','y','m']
    
    fig, ax = plt.subplots()
    
    #množina výrobných možností
    ax.fill_between(x,y, color = 'g')
    
    #body
    for xx,yy,f in zip(body_x, body_y, farba):
        ax.scatter(xx,yy,marker='x', color = f)
        ax.plot([xx,xx], [yy,0], ':', color = f)
        ax.plot([xx,0], [yy,yy], ':', color = f)
    
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    #osy
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    ax.set_title("Efektívne, neefektívne a neprípustné výrobné možnosti")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.savefig(
            "images/produk/vyrobne_moznosti.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

##################################
# Obr hranica_vyrobnych_moznosti #
##################################

def hranica_vyrobnych_moznosti():

    #Obrázok: Množina a hranica výrobných možností
    #Prvý obrázok rozšírime o hranicu výrobných možností
    
    fig, ax = plt.subplots()
    
    #množina výrobných možností
    ax.fill_between(x,y, color = 'g')
    
    #hranica výrobných možností
    ax.plot(x,y, color = 'k', linewidth = 4)

    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    ax.set_title("Množina a hranica výrobných možností")
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.savefig(
            "images/produk/hranica_vyrobnych_moznosti.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

################
# Obr Produkty #
################

def produkty():
    #Obrázok : Priemerný produkt vstupu vyjadruje sklon úsečky spájajúce začiatok súradnicového systému a danej dvojice vstupu a výstupu
    #Z predchádzajúceho obrázka zobrazíme iba produkčnú funkciu a pridáme spojnicu
    
    x_bod = (40,90)
    
    fig, ax = plt.subplots()
    
    x_prespojnicu = (np.linspace(0,x_bod[0],50),np.linspace(0,x_bod[1],50))
    
    #produkčná funkcia
    ax.plot(x,y, color = 'k', linewidth = 4)
    
    # zobraz spojnice
    for bod,d in zip(x_bod,x_prespojnicu):
        AP=f(bod)/bod
        y_prespojnicu = AP*d
        ax.plot(d,y_prespojnicu)
    
    # zobraz dotyčnice
    for bod in x_bod:
        x_predotycnicu = np.linspace(bod-30,bod+30,10)
        a, b = dotycnica(f,fder,bod)
        y_predotycnicu = a + b*x_predotycnicu
        ax.plot(x_predotycnicu,y_predotycnicu)
    
    ax.text(58,5.5,"sklon = $APx$")
    ax.text(38,6.6,"$A$")
    ax.text(90,9.9,"$B$")

    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.savefig(
            "images/produk/produkty.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

##################
# Obr pf_optimal #
##################

def pf_optimal():

    #Obrázok: Optimálny výber firmy maximalizujúcej zisk
    #Produkčná funkcia a pridáme dotyčnicu 
    
    fig, ax = plt.subplots()
    
    a,b = dotycnica(f,fder,40) 
    x_predotycnicu = np.linspace(10,70,50)
    
    y_predotycnicu = a + b*x_predotycnicu
    
    #produkčná funkcia
    ax.plot(x,y, color = 'k', linewidth = 4)
    ax.plot(x_predotycnicu,y_predotycnicu)
    
    # optimálny bod
    x_opt = 40
    y_opt = f(x_opt)
    
    ax.scatter(x_opt, y_opt,
               color='red',
               s=80,
               zorder=5)
    
    ax.plot([x_opt,x_opt],
            [0,y_opt],
            ':',
            color='red')
    
    ax.plot([0,x_opt],
            [y_opt,y_opt],
            ':',
            color='red')
    
    ax.text(x_opt + 3,
            y_opt - 0.3,
            r"$[x^*,y^*]$")
    
    ax.text(22,7,"sklon = $c/p$")
    
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()

    ax.arrow(xd,0,xh-xd,0)
    ax.arrow(0,yd,0,yh-yd)
    
    plt.xlabel("$x$")
    plt.ylabel("$y$")

    plt.savefig(
            "images/produk/pf_optimal.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

#################
# Obr izokvanty #
#################

def izokvanty():

    #Obrázok zobrazuje izokvanty.
    
    def h(yi,x1):
        x2 = yi**(1.5)*x1**(-.5)
        return x2
    
    def h_der(yi,x1):
        x2 = -0.5*yi**(1.5)*x1**(-1.5)
        return x2
    
    # Keďže uvažujeme funkcie 2 premenných potrebujeme predefinovať funkciu dotyčnice.
    def dotycnica2(g,gder,y,x):
        b = gder(y,x)
        a = g(y,x) - gder(y,x)*x
        return a,b
    
    body_x1 = (20,50,80) #body dotyčníc (x-ova súradnica)
    
    x_1 = np.linspace(0.001,100) #definičný obor funkcií
    
    y = np.array([8,10,12]) #3 izočiary
    
    #zobraz priebehy funkcií
    fig, ax = plt.subplots(figsize = (8,8))
    i = 1
    for y_i in y:
        x_2 = h(y_i, x_1)
        ax.plot(x_1,x_2,linewidth = 2, label=f'$y_{i}$')
        ax.legend()
        ax.set(xlabel = "$x_1$", ylabel = "$x_2$",ylim = (0,11), xlim = (1,100))
        i = i + 1
    
    #Zobraz dotyčnice
    yi = 10
    
    for bod in body_x1:
        x1 = np.linspace(bod-5,bod+5,10)
        a, b = dotycnica2(h,h_der,yi,bod)
        y = a + b*x1
        ax.plot(x1,y,color='k')
            
    #komenty
    ax.text(17,6.5,"$A$", color = 'k')
    ax.text(78,3.1,"$B$", color = 'k')
    ax.text(48,4.1,"$C$", color = 'k')
    
    plt.savefig(
            "images/produk/izokvanty.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

######################
#Obr graf_extenzivna #
######################

def graf_extenzivna():

    def nakresli_osi(ax):
        xd, xh = ax.get_xlim()
        yd, yh = ax.get_ylim()
    
        ax.arrow(xd,0,xh-xd,0,
                 length_includes_head=True)
    
        ax.arrow(0,yd,0,yh-yd,
                 length_includes_head=True)
    
    
    # --------------------------------------------------
    # Produkčná funkcia
    # --------------------------------------------------
    
    α = 0.5
    
    def f(x):
        return x**α
    
    def fder(x):
        return α*x**(α-1)
    
    def dotycnica(g, gder, x):
        b = gder(x)
        a = g(x) - b*x
        return a, b
    
    # --------------------------------------------------
    # Definičný obor
    # --------------------------------------------------
    
    x = np.linspace(1e-15, 150, 500)
    
    # body pre ilustráciu
    K1 = 40
    K2 = 90
    
    # --------------------------------------------------
    # hodnoty produkcie
    # --------------------------------------------------
    
    Y = f(x)
    
    # hraničné produkty
    MP = fder(x)
    
    # priemerné produkty
    AP = f(x)/x
    
    # --------------------------------------------------
    # obrázok
    # --------------------------------------------------
    
    fig, ax = plt.subplots(
        3, 2,
        figsize=(14, 12),
        sharex='col'
    )
    
    # ==================================================
    # 1. riadok: produkčné funkcie
    # ==================================================
    
    for j, (nazov, symbol) in enumerate([
        ("Kapitál", "K"),
        ("Práca", "N")]):
    
        g = ax[0,j]
    
        g.plot(x, Y, color='black', linewidth=3)
    
        for bod, meno in zip([K1, K2], ["1","2"]):
    
            # bod na produkčnej funkcii
            g.scatter(
                bod,
                f(bod),
                color='red',
                zorder=10
            )
    
            g.text(
                bod+2,
                f(bod)+0.8,
                rf"${symbol}_{meno}$"
            )
    
            # spojnice (AP)
            g.plot(
                [0,bod],
                [0,f(bod)],
                color='blue',
                linestyle='--'
            )
    
            # dotyčnice (MP)
            a,b = dotycnica(f,fder,bod)
    
            xx = np.linspace(
                max(1,bod-25),
                bod+25,
                50
            )
    
            yy = a + b*xx
    
            g.plot(
                xx,
                yy,
                color='green'
            )
    
            # pomocná zvislá čiara
            g.axvline(
                bod,
                linestyle=':',
                color='gray'
            )
    
        g.set_title(
            rf"$Y({symbol})$"
        )
    
        g.set_ylabel("$Y$")
    
        nakresli_osi(g)
    
    
    # ==================================================
    # 2. riadok: hraničné produkty
    # ==================================================
    
    for j, (nazov, symbol) in enumerate([
        ("Kapitál", "K"),
        ("Práca", "N")]):
    
        g = ax[1,j]
    
        g.plot(
            x,
            MP,
            color='green',
            linewidth=3
        )
    
        for bod, meno in zip([K1,K2],["1","2"]):
    
            g.scatter(
                bod,
                fder(bod),
                color='green',
                zorder=10
            )
    
            g.text(
                bod+2,
                fder(bod)+0.005,
                rf"$MP{symbol}_{meno}$"
            )
    
            g.axvline(
                bod,
                linestyle=':',
                color='gray'
            )
    
        g.set_title(
            rf"$MP{symbol}$"
        )
    
        g.set_ylabel("MP")
    
    
    
    # ==================================================
    # 3. riadok: priemerné produkty
    # ==================================================
    
    for j, nazov in enumerate(["K","N"]):
    
        g = ax[2,j]
    
        g.plot(
            x,
            AP,
            color='blue',
            linewidth=3
        )
    
        for bod, meno in zip([K1,K2],["1","2"]):
    
            ap_bod = f(bod)/bod
    
            g.scatter(
                bod,
                ap_bod,
                color='blue',
                zorder=10
            )
    
            g.text(
                bod+2,
                ap_bod+0.005,
                rf"$AP{nazov}_{meno}$"
            )
    
            g.axvline(
                bod,
                linestyle=':',
                color='gray'
            )
    
        g.set_title(
            rf"$AP{nazov}$"
        )
    
        g.set_ylabel("AP")
        g.set_xlabel(f"${nazov}$")
    
    # --------------------------------------------------
    
    ax[1,0].set_ylim(0,0.2)
    ax[1,1].set_ylim(0,0.2)
        
    ax[2,0].set_ylim(0,0.35)
    ax[2,1].set_ylim(0,0.35)
    
    plt.tight_layout()

    plt.savefig(
            "images/produk/graf_extenzivna.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

#######################
# Obr graf_intenzivna #
#######################

def graf_intenzivna():
    
    # --------------------------------------------------
    # Produkčná funkcia pracujúceho
    # --------------------------------------------------
    
    A = 1
    α = 0.5
    
    def f(k):
        return A*k**α
    
    # --------------------------------------------------
    # Definičný obor
    # --------------------------------------------------
    
    k = np.linspace(1e-15,150,500)
    
    k1 = 40
    k2 = 90
    
    # --------------------------------------------------
    # hodnoty funkcií
    # --------------------------------------------------
    
    y = f(k)
    
    mpk = fder(k)
    
    apk = f(k)/k
    
    # --------------------------------------------------
    # obrázok
    # --------------------------------------------------
    
    fig, ax = plt.subplots(
        3,1,
        figsize=(10,12),
        sharex=True
    )
    
    # ==================================================
    # Produkčná funkcia
    # ==================================================
    
    g = ax[0]
    
    g.plot(
        k,
        y,
        color='black',
        linewidth=3
    )
    
    for bod, meno in zip([k1,k2],["1","2"]):
    
        # bod
        g.scatter(
            bod,
            f(bod),
            color='red',
            zorder=10
        )
    
        g.text(
            bod+2,
            f(bod)+0.25,
            rf"$k_{meno}$"
        )
    
        # spojnica (APK)
        g.plot(
            [0,bod],
            [0,f(bod)],
            color='blue',
            linestyle='--'
        )
    
        # dotyčnica (MPK)
        a,b = dotycnica(f,fder,bod)
    
        xx = np.linspace(
            max(1,bod-25),
            bod+25,
            50
        )
    
        yy = a + b*xx
    
        g.plot(
            xx,
            yy,
            color='green'
        )
    
        g.axvline(
            bod,
            linestyle=':',
            color='gray'
        )
    
    nakresli_osi(g)
    g.set_title(r"$y=A f(k)$")
    g.set_ylabel("$y$")
    
    # ==================================================
    # MPK
    # ==================================================
    
    g = ax[1]
    
    g.plot(
        k,
        mpk,
        color='green',
        linewidth=3
    )
    
    for bod, meno in zip([k1,k2],["1","2"]):
    
        g.scatter(
            bod,
            fder(bod),
            color='green'
        )
    
        g.text(
            bod+2,
            fder(bod),
            rf"$mpk_{meno}$"
        )
    
        g.axvline(
            bod,
            linestyle=':',
            color='gray'
        )
    
    g.set_title(r"$mpk$")
    g.set_ylabel("$mpk$")
    g.set_ylim(0,0.2)
    
    # ==================================================
    # APK
    # ==================================================
    
    g = ax[2]
    
    g.plot(
        k,
        apk,
        color='blue',
        linewidth=3
    )
    
    for bod, meno in zip([k1,k2],["1","2"]):
    
        ap = f(bod)/bod
    
        g.scatter(
            bod,
            ap,
            color='blue'
        )
    
        g.text(
            bod+2,
            ap,
            rf"$apk_{meno}$"
        )
    
        g.axvline(
            bod,
            linestyle=':',
            color='gray'
        )
    
    g.set_title(r"$apk$")
    g.set_ylabel("$apk$")
    g.set_xlabel("$k$")
    g.set_ylim(0,0.35)
    
    plt.tight_layout()
    
    plt.savefig(
            "images/produk/graf_intenzivna.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

####################
# Obr intenzivna_a #
####################

def intenzivna_a():

    # --------------------------------------------------
    # Parametre
    # --------------------------------------------------
    
    α = 0.5
    
    A1 = 1
    A2 = 1.5
    
    # --------------------------------------------------
    # Funkcie
    # --------------------------------------------------
    
    def f(k,A):
        return A*k**α
    
    def mpk(k,A):
        return A*α*k**(α-1)
    
    def apk(k,A):
        return f(k,A)/k
    
    # --------------------------------------------------
    # Dáta
    # --------------------------------------------------
    
    k = np.linspace(1e-15,150,500)
    
    k1 = 40
    k2 = 90
    
    # --------------------------------------------------
    # Obrázok
    # --------------------------------------------------
    
    fig, ax = plt.subplots(
        3,1,
        figsize=(10,12),
        sharex=True
    )
    
    # ==================================================
    # Produkčná funkcia
    # ==================================================
    
    ax[0].plot(k,f(k,A1),label=r"$A_1f(k)$",linewidth=3)
    ax[0].plot(k,f(k,A2),label=r"$A_2f(k)$",linewidth=3)
    
    for bod in [k1,k2]:
    
        ax[0].axvline(
            bod,
            linestyle=":",
            color="gray"
        )
    
        ax[0].scatter(
            bod,
            f(bod,A1)
        )
    
        ax[0].scatter(
            bod,
            f(bod,A2)
        )
    
    nakresli_osi(ax[0])
    ax[0].set_title("Intenzívna produkčná funkcia")
    ax[0].set_ylabel("$y$")
    ax[0].legend()
    
    # ==================================================
    # MPK
    # ==================================================
    
    ax[1].plot(
        k,
        mpk(k,A1),
        linewidth=3,
        label=r"$mpk_1$"
    )
    
    ax[1].plot(
        k,
        mpk(k,A2),
        linewidth=3,
        label=r"$mpk_2$"
    )
    
    for bod in [k1,k2]:
    
        ax[1].axvline(
            bod,
            linestyle=":",
            color="gray"
        )
    
        ax[1].scatter(
            bod,
            mpk(bod,A1)
        )
    
        ax[1].scatter(
            bod,
            mpk(bod,A2)
        )
    
    ax[1].set_title("Hraničný produkt kapitálu")
    ax[1].set_ylabel("$mpk$")
    ax[1].set_ylim(0,0.2)
    ax[1].legend()
    
    # ==================================================
    # APK
    # ==================================================
    
    ax[2].plot(
        k,
        apk(k,A1),
        linewidth=3,
        label=r"$apk_1$"
    )
    
    ax[2].plot(
        k,
        apk(k,A2),
        linewidth=3,
        label=r"$apk_2$"
    )
    
    for bod in [k1,k2]:
    
        ax[2].axvline(
            bod,
            linestyle=":",
            color="gray"
        )
    
        ax[2].scatter(
            bod,
            apk(bod,A1)
        )
    
        ax[2].scatter(
            bod,
            apk(bod,A2)
        )
    
    ax[2].set_title("Priemerný produkt kapitálu")
    ax[2].set_ylabel("$apk$")
    ax[2].set_xlabel("$k$")
    ax[2].set_ylim(0,0.35)
    ax[2].legend()
    
    plt.tight_layout()
    plt.savefig(
            "images/produk/intenzivna_a.png",
            dpi=300,
            bbox_inches="tight"
        )
    plt.close()

####################
# Obr intenzivna_a #
####################

def intenzivna_a():

    # --------------------------------------------------
    # Parametre
    # --------------------------------------------------

    A1 = 1
    A2 = 1.5

    α = 0.5

    # --------------------------------------------------
    # Funkcie
    # --------------------------------------------------

    def f(k,A):
        return A*k**α

    def mpk(k,A):
        return A*α*k**(α-1)

    def apk(k,A):
        return f(k,A)/k

    # --------------------------------------------------
    # Dáta
    # --------------------------------------------------

    k = np.linspace(1e-15,80,500)

    k1 = 40

    # --------------------------------------------------
    # Obrázok
    # --------------------------------------------------

    fig, ax = plt.subplots(
        3,1,
        figsize=(10,12),
        sharex=True
    )

    # ==================================================
    # Produkčná funkcia
    # ==================================================

    g = ax[0]

    g.plot(k,f(k,A1),color="black",linewidth=3)
    g.plot(k,f(k,A2),color="black",linewidth=3)

    g.axvline(k1,linestyle=":",color="gray")

    g.scatter(k1,f(k1,A1),color="red")
    g.scatter(k1,f(k1,A2),color="red")

    g.text(k1+2,f(k1,A1)-0.8,r"$y_1$")
    g.text(k1+2,f(k1,A2)-0.8,r"$y_2$")

    # spojnice (APK)

    g.plot(
        [0,k1],
        [0,f(k1,A1)],
        "--",
        color="tab:blue",
        linewidth=2
    )

    g.plot(
        [0,k1],
        [0,f(k1,A2)],
        "--",
        color="navy",
        linewidth=2
    )

    # dotyčnice (MPK)

    x_dot = np.linspace(20,60,50)

    a,b = dotycnica(
        lambda x: f(x,A1),
        lambda x: mpk(x,A1),
        k1
    )

    g.plot(
        x_dot,
        a+b*x_dot,
        color="limegreen",
        linewidth=2
    )

    a,b = dotycnica(
        lambda x: f(x,A2),
        lambda x: mpk(x,A2),
        k1
    )

    g.plot(
        x_dot,
        a+b*x_dot,
        color="darkgreen",
        linewidth=2
    )

    # šípka technologického pokroku

    k_sipka = 60

    g.annotate(
        "",
        xy=(k_sipka,f(k_sipka,A2)),
        xytext=(k_sipka,f(k_sipka,A1)),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=2
        )
    )

    g.text(
        k_sipka+2,
        (f(k_sipka,A1)+f(k_sipka,A2))/2,
        r"$A_2>A_1$"
    )

    nakresli_osi(g)

    g.set_title("Vplyv zvýšenia celkovej produktivity faktorov")
    g.set_ylabel("$y$")

    # ==================================================
    # MPK
    # ==================================================

    g = ax[1]

    g.plot(
        k,
        mpk(k,A1),
        color="limegreen",
        linewidth=3
    )

    g.plot(
        k,
        mpk(k,A2),
        color="darkgreen",
        linewidth=3
    )

    g.axvline(
        k1,
        linestyle=":",
        color="gray"
    )

    g.scatter(
        k1,
        mpk(k1,A1),
        color="limegreen"
    )

    g.scatter(
        k1,
        mpk(k1,A2),
        color="darkgreen"
    )

    g.text(
        k1+2,
        mpk(k1,A1)-0.017,
        r"$mpk_1$"
    )

    g.text(
        k1+2,
        mpk(k1,A2)+0.005,
        r"$mpk_2$"
    )

    g.set_ylabel("$mpk$")
    g.set_ylim(0,0.2)

    # ==================================================
    # APK
    # ==================================================

    g = ax[2]

    g.plot(
        k,
        apk(k,A1),
        color="tab:blue",
        linewidth=3
    )

    g.plot(
        k,
        apk(k,A2),
        color="navy",
        linewidth=3
    )

    g.axvline(
        k1,
        linestyle=":",
        color="gray"
    )

    g.scatter(
        k1,
        apk(k1,A1),
        color="tab:blue"
    )

    g.scatter(
        k1,
        apk(k1,A2),
        color="navy"
    )

    g.text(
        k1+2,
        apk(k1,A1)-0.027,
        r"$apk_1$"
    )

    g.text(
        k1+2,
        apk(k1,A2)+0.005,
        r"$apk_2$"
    )

    g.set_ylabel("$apk$")
    g.set_xlabel("$k$")
    g.set_ylim(0,0.35)

    plt.tight_layout()

    plt.savefig(
        "images/produk/intenzivna_a.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

###################
# Obr extenzivna_a #
###################

def extenzivna_a():

    # --------------------------------------------------
    # Parametre
    # --------------------------------------------------

    A1 = 1
    A2 = 1.5

    α = 0.5

    # --------------------------------------------------
    # Funkcie
    # --------------------------------------------------

    def f(x,A):
        return A*x**α

    def mp(x,A):
        return A*α*x**(α-1)

    def ap(x,A):
        return f(x,A)/x

    # --------------------------------------------------
    # Dáta
    # --------------------------------------------------

    x = np.linspace(1e-15,80,500)

    K1 = 40

    # --------------------------------------------------
    # Obrázok
    # --------------------------------------------------

    fig, ax = plt.subplots(
        3,2,
        figsize=(14,12),
        sharex='col'
    )

    # ==================================================
    # Produkčné funkcie
    # ==================================================

    for j,nazov in enumerate(["K","N"]):

        g = ax[0,j]

        g.plot(x,f(x,A1),color="black",linewidth=3)
        g.plot(x,f(x,A2),color="black",linewidth=3)

        g.axvline(K1,linestyle=":",color="gray")

        g.scatter(K1,f(K1,A1),color="red")
        g.scatter(K1,f(K1,A2),color="red")

        g.text(K1+2,f(K1,A1)-0.8,r"$Y_1$")
        g.text(K1+2,f(K1,A2)-0.8,r"$Y_2$")

        # spojnice (AP)

        g.plot(
            [0,K1],
            [0,f(K1,A1)],
            "--",
            color="tab:blue",
            linewidth=2
        )

        g.plot(
            [0,K1],
            [0,f(K1,A2)],
            "--",
            color="navy",
            linewidth=2
        )

        # dotyčnice (MP)

        xx = np.linspace(20,60,50)

        a,b = dotycnica(
            lambda z: f(z,A1),
            lambda z: mp(z,A1),
            K1
        )

        g.plot(
            xx,
            a+b*xx,
            color="limegreen",
            linewidth=2
        )

        a,b = dotycnica(
            lambda z: f(z,A2),
            lambda z: mp(z,A2),
            K1
        )

        g.plot(
            xx,
            a+b*xx,
            color="darkgreen",
            linewidth=2
        )

        # šípka technologického pokroku

        x_s = 60

        g.annotate(
            "",
            xy=(x_s,f(x_s,A2)),
            xytext=(x_s,f(x_s,A1)),
            arrowprops=dict(
                arrowstyle="->",
                linewidth=2
            )
        )

        g.text(
            x_s+2,
            (f(x_s,A1)+f(x_s,A2))/2,
            r"$A_2>A_1$"
        )

        nakresli_osi(g)

        g.set_title(rf"$Y({nazov})$")
        g.set_ylabel("$Y$")

    # ==================================================
    # MPK a MPN
    # ==================================================

    for j,nazov in enumerate(["K","N"]):

        g = ax[1,j]

        g.plot(
            x,
            mp(x,A1),
            color="limegreen",
            linewidth=3
        )

        g.plot(
            x,
            mp(x,A2),
            color="darkgreen",
            linewidth=3
        )

        g.axvline(
            K1,
            linestyle=":",
            color="gray"
        )

        g.scatter(
            K1,
            mp(K1,A1),
            color="limegreen"
        )

        g.scatter(
            K1,
            mp(K1,A2),
            color="darkgreen"
        )

        g.text(
            K1+2,
            mp(K1,A1)-0.017,
            rf"$MP{nazov}_1$"
        )

        g.text(
            K1+2,
            mp(K1,A2)+0.005,
            rf"$MP{nazov}_2$"
        )

        g.set_title(rf"$MP{nazov}$")
        g.set_ylabel("MP")
        g.set_ylim(0,0.2)

    # ==================================================
    # APK a APN
    # ==================================================

    for j,nazov in enumerate(["K","N"]):

        g = ax[2,j]

        g.plot(
            x,
            ap(x,A1),
            color="tab:blue",
            linewidth=3
        )

        g.plot(
            x,
            ap(x,A2),
            color="navy",
            linewidth=3
        )

        g.axvline(
            K1,
            linestyle=":",
            color="gray"
        )

        g.scatter(
            K1,
            ap(K1,A1),
            color="tab:blue"
        )

        g.scatter(
            K1,
            ap(K1,A2),
            color="navy"
        )

        g.text(
            K1+2,
            ap(K1,A1)-0.027,
            rf"$AP{nazov}_1$"
        )

        g.text(
            K1+2,
            ap(K1,A2)+0.005,
            rf"$AP{nazov}_2$"
        )

        g.set_title(rf"$AP{nazov}$")
        g.set_ylabel("AP")
        g.set_xlabel(f"${nazov}$")
        g.set_ylim(0,0.35)

    plt.tight_layout()

    plt.savefig(
        "images/produk/extenzivna_a.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

###################
# Obr intenzivna_k #
###################

def intenzivna_k():

    # --------------------------------------------------
    # Parametre
    # --------------------------------------------------

    A = 1
    α = 0.5

    # --------------------------------------------------
    # Funkcie
    # --------------------------------------------------

    def f(k):
        return A*k**α

    def mpk(k):
        return A*α*k**(α-1)

    def apk(k):
        return f(k)/k

    # --------------------------------------------------
    # Dáta
    # --------------------------------------------------

    k = np.linspace(1e-15,80,500)

    k1 = 20
    k2 = 50

    # --------------------------------------------------
    # Obrázok
    # --------------------------------------------------

    fig, ax = plt.subplots(
        3,1,
        figsize=(10,12),
        sharex=True
    )

    # ==================================================
    # Produkčná funkcia
    # ==================================================

    g = ax[0]

    g.plot(
        k,
        f(k),
        color="black",
        linewidth=3
    )

    # body

    g.scatter(
        k1,
        f(k1),
        color="red",
        zorder=10
    )

    g.scatter(
        k2,
        f(k2),
        color="red",
        zorder=10
    )

    g.text(
        k1+2,
        f(k1)-0.7,
        r"$y_1$"
    )

    g.text(
        k2+2,
        f(k2)-0.7,
        r"$y_2$"
    )

    # pomocné čiary

    g.axvline(
        k1,
        linestyle=":",
        color="gray"
    )

    g.axvline(
        k2,
        linestyle=":",
        color="gray"
    )

    # --------------------------------------------------
    # spojnice (APK)
    # --------------------------------------------------

    g.plot(
        [0,k1],
        [0,f(k1)],
        "--",
        color="tab:blue",
        linewidth=2
    )

    g.plot(
        [0,k2],
        [0,f(k2)],
        "--",
        color="navy",
        linewidth=2
    )

    # --------------------------------------------------
    # dotyčnice (MPK)
    # --------------------------------------------------

    x_dot1 = np.linspace(
        max(1e-15,k1-15),
        k1+25,
        50
    )

    a,b = dotycnica(
        f,
        mpk,
        k1
    )

    g.plot(
        x_dot1,
        a+b*x_dot1,
        color="limegreen",
        linewidth=2
    )

    x_dot2 = np.linspace(
        k2-20,
        k2+20,
        50
    )

    a,b = dotycnica(
        f,
        mpk,
        k2
    )

    g.plot(
        x_dot2,
        a+b*x_dot2,
        color="darkgreen",
        linewidth=2
    )

    # --------------------------------------------------
    # šípka zvýšenia kapitálu
    # --------------------------------------------------

    g.annotate(
        "",
        xy=(k2,0),
        xytext=(k1,0),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=2
        )
    )

    g.text(
        (k1+k2)/2 - 10,
        0.35,
        r"$K_1/N \rightarrow K_2/N$"
    )

    nakresli_osi(g)

    g.set_title("Vplyv zvýšenia intenzívneho kapitálu")
    g.set_ylabel("$y$")

    # ==================================================
    # MPK
    # ==================================================

    g = ax[1]

    g.plot(
        k,
        mpk(k),
        color="green",
        linewidth=3
    )

    g.axvline(
        k1,
        linestyle=":",
        color="gray"
    )

    g.axvline(
        k2,
        linestyle=":",
        color="gray"
    )

    g.scatter(
        k1,
        mpk(k1),
        color="green",
        zorder=10
    )

    g.scatter(
        k2,
        mpk(k2),
        color="green",
        zorder=10
    )

    g.text(
        k1+2,
        mpk(k1)-0.020,
        r"$mpk_1$"
    )

    g.text(
        k2+2,
        mpk(k2)+0.004,
        r"$mpk_2$"
    )

    g.set_ylabel("$mpk$")
    g.set_ylim(0,0.2)

    # ==================================================
    # APK
    # ==================================================

    g = ax[2]

    g.plot(
        k,
        apk(k),
        color="blue",
        linewidth=3
    )

    g.axvline(
        k1,
        linestyle=":",
        color="gray"
    )

    g.axvline(
        k2,
        linestyle=":",
        color="gray"
    )

    g.scatter(
        k1,
        apk(k1),
        color="blue",
        zorder=10
    )

    g.scatter(
        k2,
        apk(k2),
        color="blue",
        zorder=10
    )

    g.text(
        k1+2,
        apk(k1)-0.035,
        r"$apk_1$"
    )

    g.text(
        k2+2,
        apk(k2)+0.004,
        r"$apk_2$"
    )

    g.set_ylabel("$apk$")
    g.set_xlabel("$k$")
    g.set_ylim(0,0.35)

    plt.tight_layout()

    plt.savefig(
        "images/produk/intenzivna_k.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

###################
# Obr extenzivna_k #
###################

def extenzivna_k():

    # --------------------------------------------------
    # Parametre
    # --------------------------------------------------

    A = 1
    α = 0.5

    # --------------------------------------------------
    # Funkcie
    # --------------------------------------------------

    def f(x):
        return A*x**α

    def mp(x):
        return A*α*x**(α-1)

    def ap(x):
        return f(x)/x

    # --------------------------------------------------
    # Dáta
    # --------------------------------------------------

    x = np.linspace(1e-15,80,500)

    K1 = 20
    K2 = 50

    # koeficient posunu produktov práce
    φ = 1.5

    # --------------------------------------------------
    # Obrázok
    # --------------------------------------------------

    fig, ax = plt.subplots(
        3,2,
        figsize=(14,12),
        sharex='col'
    )

    # ==================================================
    # Y(K)
    # ==================================================

    g = ax[0,0]

    g.plot(x,f(x),color="black",linewidth=3)

    g.scatter(K1,f(K1),color="red",zorder=10)
    g.scatter(K2,f(K2),color="red",zorder=10)

    g.text(K1+2,f(K1)-0.7,r"$Y_1$")
    g.text(K2+2,f(K2)-0.7,r"$Y_2$")

    g.axvline(K1,linestyle=":",color="gray")
    g.axvline(K2,linestyle=":",color="gray")

    # spojnice

    g.plot(
        [0,K1],
        [0,f(K1)],
        "--",
        color="tab:blue",
        linewidth=2
    )

    g.plot(
        [0,K2],
        [0,f(K2)],
        "--",
        color="navy",
        linewidth=2
    )

    # dotyčnice

    xx1 = np.linspace(
        max(1e-15,K1-15),
        K1+25,
        50
    )

    a,b = dotycnica(f,mp,K1)

    g.plot(
        xx1,
        a+b*xx1,
        color="limegreen",
        linewidth=2
    )

    xx2 = np.linspace(
        K2-20,
        K2+20,
        50
    )

    a,b = dotycnica(f,mp,K2)

    g.plot(
        xx2,
        a+b*xx2,
        color="darkgreen",
        linewidth=2
    )

    # šípka

    g.annotate(
        "",
        xy=(K2,0),
        xytext=(K1,0),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=2
        )
    )

    g.text(
        (K1+K2)/2-8,
        0.35,
        r"$K_1\rightarrow K_2$"
    )

    nakresli_osi(g)

    g.set_title(r"$Y(K)$")
    g.set_ylabel("$Y$")

    # ==================================================
    # Y(N)
    # ==================================================

    g = ax[0,1]

    g.plot(x,f(x),color="black",linewidth=3)
    g.plot(x,φ*f(x),color="black",linewidth=3)

    g.axvline(K1,linestyle=":",color="gray")

    g.scatter(K1,f(K1),color="red")
    g.scatter(K1,φ*f(K1),color="red")

    g.text(K1+2,f(K1)-0.7,r"$Y_1$")
    g.text(K1+2,φ*f(K1)-0.7,r"$Y_2$")

    g.annotate(
        "",
        xy=(60,φ*f(60)),
        xytext=(60,f(60)),
        arrowprops=dict(
            arrowstyle="->",
            linewidth=2
        )
    )

    g.text(
        62,
        (f(60)+φ*f(60))/2,
        r"$K_2>K_1$"
    )

    nakresli_osi(g)

    g.set_title(r"$Y(N)$")
    g.set_ylabel("$Y$")

    # ==================================================
    # MPK
    # ==================================================

    g = ax[1,0]

    g.plot(x,mp(x),color="green",linewidth=3)

    g.axvline(K1,linestyle=":",color="gray")
    g.axvline(K2,linestyle=":",color="gray")

    g.scatter(K1,mp(K1),color="green")
    g.scatter(K2,mp(K2),color="green")

    g.text(K1+2,mp(K1)-0.020,r"$MPK_1$")
    g.text(K2+2,mp(K2)+0.004,r"$MPK_2$")

    g.set_title(r"$MPK$")
    g.set_ylabel("MP")
    g.set_ylim(0,0.2)

    # ==================================================
    # MPN
    # ==================================================

    g = ax[1,1]

    g.plot(x,mp(x),color="limegreen",linewidth=3)
    g.plot(x,φ*mp(x),color="darkgreen",linewidth=3)

    g.axvline(K1,linestyle=":",color="gray")

    g.scatter(K1,mp(K1),color="limegreen")
    g.scatter(K1,φ*mp(K1),color="darkgreen")

    g.text(K1+2,mp(K1)-0.020,r"$MPN_1$")
    g.text(K1+2,φ*mp(K1)+0.004,r"$MPN_2$")

    g.set_title(r"$MPN$")
    g.set_ylabel("MP")
    g.set_ylim(0,0.3)

    # ==================================================
    # APK
    # ==================================================

    g = ax[2,0]

    g.plot(x,ap(x),color="blue",linewidth=3)

    g.axvline(K1,linestyle=":",color="gray")
    g.axvline(K2,linestyle=":",color="gray")

    g.scatter(K1,ap(K1),color="blue")
    g.scatter(K2,ap(K2),color="blue")

    g.text(K1+2,ap(K1)-0.035,r"$APK_1$")
    g.text(K2+2,ap(K2)+0.004,r"$APK_2$")

    g.set_title(r"$APK$")
    g.set_ylabel("AP")
    g.set_xlabel("$K$")
    g.set_ylim(0,0.35)

    # ==================================================
    # APN
    # ==================================================

    g = ax[2,1]

    g.plot(x,ap(x),color="tab:blue",linewidth=3)
    g.plot(x,φ*ap(x),color="navy",linewidth=3)

    g.axvline(K1,linestyle=":",color="gray")

    g.scatter(K1,ap(K1),color="tab:blue")
    g.scatter(K1,φ*ap(K1),color="navy")

    g.text(K1+2,ap(K1)-0.035,r"$APN_1$")
    g.text(K1+2,φ*ap(K1)+0.004,r"$APN_2$")

    g.set_title(r"$APN$")
    g.set_ylabel("AP")
    g.set_xlabel("$N$")
    g.set_ylim(0,0.5)

    plt.tight_layout()

    plt.savefig(
        "images/produk/extenzivna_k.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

mnozina_vyr_moznosti()
vyrobne_moznosti()
hranica_vyrobnych_moznosti()
produkty()
pf_optimal()
izokvanty()
graf_extenzivna()
graf_intenzivna()
extenzivna_a()
intenzivna_a()
extenzivna_k()
intenzivna_k()
```

## Správanie firmy

Názov podpriečinka je *firma*.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch

Path("images/firma").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

def nakresli_osi(ax):
    xd, xh = ax.get_xlim()
    yd, yh = ax.get_ylim()
    
    ax.arrow(xd,0,xh-xd,0,
             length_includes_head=True)

    ax.arrow(0,yd,0,yh-yd,
             length_includes_head=True)


#####################
# Obr dopyt_faktory
#####################

def dopyt_faktory():

    # --------------------------------------------------
    # Parametre
    # --------------------------------------------------

    α = 1/3

    # --------------------------------------------------
    # Funkcie
    # --------------------------------------------------

    def MPK(K, N):
        return α * N / K

    def MPN(K, N):
        return (1 - α) * K / N

    # --------------------------------------------------
    # Dáta
    # --------------------------------------------------

    K = np.linspace(1e-3, 2, 500)
    N = np.linspace(1e-3, 2, 500)

    N0 = 1
    K0 = 1

    # --------------------------------------------------
    # Obrázok
    # --------------------------------------------------

    fig, ax = plt.subplots(
        1, 2,
        figsize=(12, 5)
    )

    # Dopyt po kapitáli

    ax[0].plot(
        K,
        MPK(K, N0),
        color="navy",
        linewidth=3
    )

    ax[0].set_title("Dopyt po kapitáli")
    ax[0].set_xlabel("$K^d$")
    ax[0].set_ylabel("$q/P$")
    ax[0].set_ylim(0, 3)

    # Dopyt po práci

    ax[1].plot(
        N,
        MPN(K0, N),
        color="darkgreen",
        linewidth=3
    )

    ax[1].set_title("Dopyt po práci")
    ax[1].set_xlabel("$N^d$")
    ax[1].set_ylabel("$\omega/P$")
    ax[1].set_ylim(0, 3)

    plt.tight_layout()

    plt.savefig(
        "images/firma/dopyt_faktory.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

#################
# Obr fix_vystup
#################

def graf_fix_vystup():

    # izokvanta
    def h(y0,x1):
        x2 = y0**(1.5)*x1**(-.5)
        return x2

    # izonákladová čiara
    def c(C,x1):
        x2 = C*3/2 - 1/2*x1
        return x2

    y0 = 1.
    C = 1.
    x_1 = np.linspace(0.001,3.) # definičný obor funkcií

    izokvanta = h(y0,x_1)
    izonaklad = c(C,x_1)

    fig, ax = plt.subplots()

    ax.plot(x_1,izokvanta,label=r"Izokvanta $y_0$")
    ax.fill_between(
        x_1,
        izokvanta,
        y2=3,
        color='lightgreen',
        alpha=0.5
    )

    ax.plot(
        x_1,
        izonaklad,
        label='Izonákladová čiara',
        color="black"
    )

    ax.set(
        xlabel="$x_1$",
        ylabel="$x_2$",
        ylim=(0,3.),
        xlim=(0,3.)
    )

    # bod
    ax.scatter(1,1,color="red",zorder=10)

    ax.text(
        1.05,
        1.05,
        r"$(x_1^*,x_2^*)$"
    )

    # pomocné čiary
    ax.axvline(1, linestyle=":", color="gray")
    ax.axhline(1, linestyle=":", color="gray")

    ax.legend()

    plt.savefig(
        "images/firma/graf_fix_vystup.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

graf_fix_vystup()
dopyt_faktory()
```

## Náklady firmy

Názov podpriečnka je *naklad*.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch

Path("images/naklad").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

##########################
# Obr kratkodobe_naklady #
##########################

def kratkodobe_naklady():

    # Priemerné, priemerné variabilné a hraničné náklady

    def hn(y):
        MC = 0.126*y**2 - 1.8*y + 11
        return MC

    def pvn(y):
        AVC = 0.042*y**2 - 0.9*y + 11
        return AVC

    def pn(y):
        AC = 0.042*y**2 - 0.9*y + 11 + 7/y
        return AC

    y = np.linspace(1e-12, 20, 100)

    naklady = (
        hn(y),
        pvn(y),
        pn(y)
    )

    popisky = (
        "SMC",
        "SAVC",
        "SAC"
    )

    fig, ax = plt.subplots(figsize=[10, 8])

    for n, p in zip(naklady, popisky):
        ax.plot(y, n, label=p)

    ax.set_xlabel("$y$")
    ax.set_ylabel("náklady na jednotku výstupu")

    ax.set_ylim([4, 12])

    ax.legend()

    plt.savefig(
        "images/naklad/kratkodobe_naklady.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
##########################
# Obr. priemerne_naklady #
##########################

def priemerne_naklady():

    y = np.linspace(1, 20, 500)

    # body dotyku
    y1 = 5
    y2 = 9
    y3 = 15

    # dlhodobé priemerné náklady
    LAC = 1 + 12/y + 0.15*y

    # krátkodobé priemerné náklady
    SAC1 = LAC + 0.03*(y-y1)**2
    SAC2 = LAC + 0.02*(y-y2)**2
    SAC3 = LAC + 0.03*(y-y3)**2

    fig, ax = plt.subplots(figsize=(10, 8))

    ax.plot(
        y,
        SAC1,
        "--",
        label=r"$SAC(y,K_1)$"
    )

    ax.plot(
        y,
        SAC2,
        "--",
        label=r"$SAC(y,K_2)$"
    )

    ax.plot(
        y,
        SAC3,
        "--",
        label=r"$SAC(y,K_3)$"
    )

    ax.plot(
        y,
        LAC,
        color="black",
        linewidth=3,
        label="LAC"
    )

    for yp in [y1, y2, y3]:

        lac_p = 1 + 12/yp + 0.15*yp

        ax.scatter(
            yp,
            lac_p,
            color="black",
            zorder=10
        )

        ax.axvline(
            yp,
            linestyle=":",
            color="gray",
            alpha=0.6
        )

    ax.set_xticks([y1, y2, y3])

    ax.set_xticklabels([
        r"$y(K_1)$",
        r"$y(K_2)$",
        r"$y(K_3)$"
    ])

    ax.set(
        xlabel="$y$",
        ylabel="náklady na jednotku výstupu",
        ylim=(3, 8)
    )

    ax.legend()

    plt.savefig(
        "images/naklad/priemerne_naklady.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
###########################
# Obr. vztahy_naklady
###########################

def vztahy_naklady():

    y = np.linspace(1, 20, 500)

    # body dotyku
    y1 = 5
    y2 = 9
    y3 = 15

    # priemerné náklady
    LAC = 1 + 12/y + 0.15*y

    SAC1 = LAC + 0.03*(y-y1)**2
    SAC2 = LAC + 0.02*(y-y2)**2
    SAC3 = LAC + 0.03*(y-y3)**2

    # celkové náklady
    LC  = y * LAC
    SC1 = y * SAC1
    SC2 = y * SAC2
    SC3 = y * SAC3

    # hraničné náklady
    LMC  = np.gradient(LC, y)
    SMC1 = np.gradient(SC1, y)
    SMC2 = np.gradient(SC2, y)
    SMC3 = np.gradient(SC3, y)

    fig, ax = plt.subplots(figsize=(11, 8))

    # =========================
    # PRIEMERNÉ NÁKLADY
    # =========================

    ax.plot(
        y,
        SAC1,
        "--",
        color="tab:blue",
        label=r"$SAC(y,K_1)$"
    )

    ax.plot(
        y,
        SAC2,
        "--",
        color="tab:orange",
        label=r"$SAC(y,K_2)$"
    )

    ax.plot(
        y,
        SAC3,
        "--",
        color="tab:green",
        label=r"$SAC(y,K_3)$"
    )

    ax.plot(
        y,
        LAC,
        color="black",
        linewidth=3,
        label="LAC"
    )

    # =========================
    # HRANIČNÉ NÁKLADY
    # =========================

    w = 3

    mask1 = (y > y1 - w) & (y < y1 + w)
    mask2 = (y > y2 - w) & (y < y2 + w)
    mask3 = (y > y3 - w) & (y < y3 + w)

    ax.plot(
        y[mask1],
        SMC1[mask1],
        ":",
        color="tab:blue",
        linewidth=2.5,
        label=r"$SMC(y,K_1)$"
    )

    ax.plot(
        y[mask2],
        SMC2[mask2],
        ":",
        color="tab:orange",
        linewidth=2.5,
        label=r"$SMC(y,K_2)$"
    )

    ax.plot(
        y[mask3],
        SMC3[mask3],
        ":",
        color="tab:green",
        linewidth=2.5,
        label=r"$SMC(y,K_3)$"
    )

    ax.plot(
        y,
        LMC,
        "-.",
        color="dimgray",
        linewidth=2.5,
        label="LMC"
    )

    # =========================
    # BODY DOTYKU A PRIESEČNÍKY
    # =========================

    for yp in [y1, y2, y3]:

        lac_p = np.interp(yp, y, LAC)
        lmc_p = np.interp(yp, y, LMC)

        # SAC = LAC
        ax.scatter(
            yp,
            lac_p,
            color="black",
            s=60,
            zorder=30
        )

        # SMC = LMC
        ax.scatter(
            yp,
            lmc_p,
            color="black",
            s=60,
            zorder=30
        )

        ax.axvline(
            yp,
            linestyle=":",
            color="gray",
            alpha=0.4
        )

    # =========================
    # OSI
    # =========================

    ax.set_xticks([y1, y2, y3])

    ax.set_xticklabels(
        [
            r"$y(K_1)$",
            r"$y(K_2)$",
            r"$y(K_3)$"
        ]
    )

    ax.set(
        xlabel="$y$",
        ylabel="náklady na jednotku výstupu",
        ylim=(1, 8)
    )

    ax.legend()

    plt.savefig(
        "images/naklad/vztahy_naklady.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

###################
# Obr. graf_zisk  #
###################

def graf_zisk():

    import numpy as np
    import matplotlib.pyplot as plt
    from matplotlib.patches import Rectangle

    # Hraničné náklady
    def hn(y):
        return 0.126 * y**2 - 1.8 * y + 11

    # Priemerné variabilné náklady
    def pvn(y):
        return 0.042 * y**2 - 0.9 * y + 11

    # Priemerné náklady
    def pn(y):
        return 0.042 * y**2 - 0.9 * y + 11 + 7 / y

    # Definičný obor
    y = np.linspace(0.5, 20, 5000)

    # Funkčné hodnoty
    SMC = hn(y)
    SAVC = pvn(y)
    SAC = pn(y)

    # Minimum SAC (nulový ekonomický zisk)
    idx_min = np.argmin(SAC)
    y_zero = y[idx_min]
    p_zero = SAC[idx_min]

    # Tri cenové hladiny
    p_profit = p_zero + 1.2
    p_loss = p_zero - 1.2

    prices = [p_profit, p_zero, p_loss]
    titles = ["Kladný zisk", "Nulový zisk", "Strata"]

    # Obrázok s troma panelmi
    fig, axes = plt.subplots(
        1, 3,
        figsize=(18, 6),
        sharey=True
    )

    for ax, p, title in zip(axes, prices, titles):

        # Nákladové krivky
        ax.plot(y, SMC, label="SMC", linewidth=2)
        ax.plot(y, SAVC, label="SAVC", linewidth=2)
        ax.plot(y, SAC, label="SAC", linewidth=2)

        # Nulový zisk
        if title == "Nulový zisk":

            y_star = y_zero
            AC_star = p_zero

        else:

            # Pravý priesečník SMC = p
            crossings = np.where(
                np.diff(np.sign(SMC - p))
            )[0]

            idx = crossings[-1]

            x1 = y[idx]
            x2 = y[idx + 1]

            mc1 = SMC[idx]
            mc2 = SMC[idx + 1]

            # Presnejší priesečník
            y_star = x1 + (p - mc1) * (x2 - x1) / (mc2 - mc1)

            AC_star = pn(y_star)

        # Cenová čiara
        ax.axhline(
            p,
            color="black",
            linestyle="--",
            linewidth=1.8,
            label="$p$"
        )

        # y*
        ax.vlines(
            y_star,
            ymin=0,
            ymax=max(p, AC_star),
            colors="gray",
            linestyles=":",
            linewidth=1.5
        )

        # AC(y*)
        ax.hlines(
            AC_star,
            xmin=0,
            xmax=y_star,
            colors="gray",
            linestyles=":",
            linewidth=1.2
        )

        # Ekonomický zisk / strata
        profit = p - AC_star

        if profit > 0.02:

            rect = Rectangle(
                (0, AC_star),
                y_star,
                profit,
                facecolor="green",
                edgecolor="green",
                alpha=0.30
            )

            ax.add_patch(rect)

            ax.text(
                y_star / 2,
                (p + AC_star) / 2,
                "zisk",
                ha="center",
                va="center"
            )

        elif profit < -0.02:

            rect = Rectangle(
                (0, p),
                y_star,
                AC_star - p,
                facecolor="red",
                edgecolor="red",
                alpha=0.30
            )

            ax.add_patch(rect)

            ax.text(
                y_star / 2,
                (p + AC_star) / 2,
                "strata",
                ha="center",
                va="center"
            )

        # Bod AC(y*)
        ax.plot(
            y_star,
            AC_star,
            "ko",
            markersize=5
        )

        ax.annotate(
            r"$y^*$",
            xy=(y_star, 4.1),
            xytext=(6, 0),
            textcoords="offset points",
            fontsize=11
        )

        # Označenie ceny
        ax.text(
            0.3,
            p + 0.05,
            r"$p$",
            fontsize=11
        )

        ax.set_title(title)
        ax.set_xlabel(r"$y$")
        ax.set_xlim(0, 20)
        ax.set_ylim(4, 12)

    axes[0].set_ylabel(
        "náklady na jednotku výstupu"
    )

    axes[0].legend(
        loc="upper right"
    )

    plt.tight_layout()

    plt.savefig(
        "images/naklad/graf_zisk.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

####################
# Obr. graf_ponuka #
####################

def graf_ponuka():

    import numpy as np
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(
        1, 2,
        figsize=(14, 6)
    )

    # ==================================================
    # KRÁTKODOBÁ PONUKA
    # ==================================================

    def hn(y):
        return 0.126*y**2 - 1.8*y + 11

    def pvn(y):
        return 0.042*y**2 - 0.9*y + 11

    def pn(y):
        return 0.042*y**2 - 0.9*y + 11 + 7/y

    y = np.linspace(1e-12, 20, 500)

    SMC = hn(y)
    SAVC = pvn(y)
    SAC = pn(y)

    ax[0].plot(y, SMC, label="SMC")
    ax[0].plot(y, SAVC, label="SAVC")
    ax[0].plot(y, SAC, label="SAC")

    # minimum SAVC

    idx_savc = np.argmin(SAVC)

    y_s = y[idx_savc]
    savc_min = SAVC[idx_savc]

    # krátkodobá ponuka

    mask = y >= y_s

    ax[0].plot(
        y[mask],
        SMC[mask],
        color="red",
        linewidth=4,
        label="krátkodobá ponuka"
    )

    ax[0].scatter(
        y_s,
        savc_min,
        color="black",
        zorder=10
    )

    ax[0].text(
        y_s + 0.3,
        savc_min + 0.15,
        r"$\min SAVC$"
    )

    ax[0].axvline(
        y_s,
        linestyle=":",
        color="gray"
    )

    ax[0].set_title("Krátkodobá ponuka firmy")

    ax[0].set(
        xlabel="$y$",
        ylabel="náklady na jednotku výstupu",
        ylim=(4,12)
    )

    ax[0].legend()

    # ==================================================
    # DLHODOBÁ PONUKA
    # ==================================================

    y = np.linspace(1,20,500)

    LAC = 1 + 12/y + 0.15*y

    LC = y * LAC

    LMC = np.gradient(LC,y)

    ax[1].plot(
        y,
        LAC,
        color="black",
        linewidth=3,
        label="LAC"
    )

    ax[1].plot(
        y,
        LMC,
        color="tab:blue",
        linewidth=2.5,
        label="LMC"
    )

    # minimum LAC

    idx_lac = np.argmin(LAC)

    y_l = y[idx_lac]
    lac_min = LAC[idx_lac]

    # dlhodobá ponuka

    mask = y >= y_l

    ax[1].plot(
        y[mask],
        LMC[mask],
        color="red",
        linewidth=4,
        label="dlhodobá ponuka"
    )

    ax[1].scatter(
        y_l,
        lac_min,
        color="black",
        zorder=10
    )

    ax[1].text(
        y_l - 1.8,
        lac_min + 0.2,
        r"$\min LAC$"
    )

    ax[1].axvline(
        y_l,
        linestyle=":",
        color="gray"
    )

    ax[1].set_title("Dlhodobá ponuka firmy")

    ax[1].set(
        xlabel="$y$",
        ylabel="náklady na jednotku výstupu",
        ylim=(1,8)
    )

    ax[1].legend()

    plt.tight_layout()

    plt.savefig(
        "images/naklad/graf_ponuka.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
#####################
# Obr graf_priklad2 #
#####################

def graf_priklad2():

    import numpy as np
    import matplotlib.pyplot as plt

    # interval produkcie

    y = np.linspace(0.2, 25, 2000)

    # dlhodobé priemerné náklady

    LAC = 3 / (2**(2/3)) * y**(1/3)

    # rôzne úrovne kapitálu

    K_values = [1, 2, 3, 4, 6, 8, 10]

    plt.figure(figsize=(10, 6))

    # rodina SAC

    for K in K_values:

        SAC = K**2 / y + y / K

        plt.plot(
            y,
            SAC,
            lw=1.5,
            alpha=0.8,
            label=fr"$SAC,\;K={K}$"
        )

        # bod dotyku SAC a LAC

        y_star = np.sqrt(2 * K**3)
        sac_star = K**2 / y_star + y_star / K

        plt.plot(
            y_star,
            sac_star,
            "ko",
            ms=4
        )

    # LAC

    plt.plot(
        y,
        LAC,
        color="navy",
        lw=3,
        label=r"$LAC$"
    )

    plt.xlabel(r"Produkcia $y$")
    plt.ylabel("Priemerné náklady")
    plt.title("LAC ako spodný obal rodiny SAC")

    plt.xlim(0, 25)
    plt.ylim(0, 15)

    plt.grid(alpha=0.3)

    plt.legend(
        loc="center left",
        bbox_to_anchor=(1.02, 0.5)
    )

    plt.tight_layout()

    plt.savefig(
        "images/naklad/graf_priklad2.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

kratkodobe_naklady()
priemerne_naklady()
vztahy_naklady()
graf_zisk()
graf_ponuka()
graf_priklad2()
```

## Správanie spotrebiteľa

Názov podpriečnka je *spotrebitel*.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch
from itertools import product

Path("images/spotrebitel").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

#################################################
# Obr graf_konvexnost_spotrebitelskych_moznosti #
#################################################

def graf_konvexnost_spotrebitelskych_moznosti():

    import numpy as np
    import matplotlib.pyplot as plt

    fig, ax = plt.subplots(1, 2, figsize=(10, 4))

    # --------------------------------------------------
    # Konvexná množina
    # --------------------------------------------------

    theta = np.linspace(0, 2*np.pi, 400)

    x = 2 + 2*np.cos(theta)
    y = 2 + 1.2*np.sin(theta)

    ax[0].fill(x, y, color="lightblue", alpha=0.6)

    xa = np.array([1.0, 2.0])
    xb = np.array([3.2, 2.4])

    ax[0].plot(
        [xa[0], xb[0]],
        [xa[1], xb[1]],
        color="black",
        lw=2
    )

    ax[0].plot(*xa, "ko")
    ax[0].plot(*xb, "ko")

    ax[0].text(xa[0]-0.25, xa[1]+0.15, r"$\mathbf{x}^a$")
    ax[0].text(xb[0]+0.05, xb[1]+0.15, r"$\mathbf{x}^b$")

    ax[0].set_title("Konvexná množina")

    ax[0].set_xlim(-0.5, 4.5)
    ax[0].set_ylim(0, 4)

    ax[0].set_xlabel(r"$x_1$")
    ax[0].set_ylabel(r"$x_2$")

    # --------------------------------------------------
    # Nekonvexná množina
    # --------------------------------------------------

    t = np.linspace(0, 2*np.pi, 400)

    r = 1.5 + 0.7*np.cos(2*t)

    x = 2 + r*np.cos(t)
    y = 2 + r*np.sin(t)

    ax[1].fill(x, y, color="salmon", alpha=0.6)

    xa = np.array([0.8, 1.1])
    xb = np.array([3.2, 1.1])

    ax[1].plot(
        [xa[0], xb[0]],
        [xa[1], xb[1]],
        color="black",
        lw=2
    )

    ax[1].plot(*xa, "ko")
    ax[1].plot(*xb, "ko")

    ax[1].text(xa[0]-0.25, xa[1]+0.15, r"$\mathbf{x}^a$")
    ax[1].text(xb[0]+0.05, xb[1]+0.15, r"$\mathbf{x}^b$")

    ax[1].set_title("Nekonvexná množina")

    ax[1].set_xlim(-0.5, 4.5)
    ax[1].set_ylim(0, 4)

    ax[1].set_xlabel(r"$x_1$")

    plt.tight_layout()

    plt.savefig(
        "images/spotrebitel/graf_konvexnost_spotrebitelskych_moznosti.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
############
# Obr kose #
############
    
def graf_kose():

    import numpy as np
    import matplotlib.pyplot as plt

    plt.rcParams["figure.figsize"] = (10, 6)

    def ik(x1, u=10):
        return u**(3/2) / x1**(1/2)

    ucka = (5, 10, 15, 4.8)

    x_1 = np.linspace(1e-12, 20, 100)

    body_x = [3, 5, 2, 7, 3.1]
    body_y = [
        ik(3, u=5),
        ik(5),
        ik(2),
        ik(7, u=15),
        ik(3.1, u=4.8)
    ]

    legendy = [
        r'$\mathbf{x}^a$',
        r'$\mathbf{x}^b$',
        r'$\mathbf{x}^c$',
        r'$\mathbf{x}^d$',
        r'$\mathbf{x}^e$'
    ]

    farby = ['r', 'b', 'y', 'g', 'm']

    fig, ax = plt.subplots()

    for uu in ucka:
        x_2 = ik(x_1, u=uu)
        ax.plot(x_1, x_2)

    for xx, yy, f, leg in zip(body_x, body_y, farby, legendy):

        ax.scatter(
            xx,
            yy,
            marker='x',
            color=f,
            label=leg
        )

        ax.plot(
            [xx, xx],
            [yy, 0],
            ':',
            color=f
        )

        ax.plot(
            [xx, 0],
            [yy, yy],
            ':',
            color=f
        )

    ax.set(
        xlabel=r'$x_1$',
        ylabel=r'$x_2$',
        xlim=(0, 20),
        ylim=(0, 25)
    )

    ax.legend()

    plt.tight_layout()

    plt.savefig(
        "images/spotrebitel/graf_kose.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
#################
# Obr graf_mapa #
#################

def graf_mapa():

    # Cobb-Douglasova funkcia užitočnosti
    def u(x1, x2):
        return x1**(1/3) * x2**(2/3)

    x1 = np.linspace(0.0001, 20, 100)
    x2 = np.linspace(0.0001, 20, 100).reshape((100, 1))

    fig, ax = plt.subplots(figsize=(8, 6))

    contours = ax.contourf(
        x1,
        x2.flatten(),
        u(x1, x2)
    )

    fig.colorbar(
        contours,
        ax=ax,
        label="Úroveň užitočnosti"
    )

    ax.set_xlabel(r"$x_1$")
    ax.set_ylabel(r"$x_2$")

    plt.tight_layout()

    plt.savefig(
        "images/spotrebitel/graf_mapa.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
#################
# graf_rozpocet #
#################
    
def graf_rozpocet():

    def ir(x1, M, p1, p2):
        return M / p2 - p1 / p2 * x1

    p_1 = 5
    p_2 = 10
    M = 40

    x_1 = np.linspace(0, M / p_1, 100)
    x_2 = ir(x_1, M, p_1, p_2)

    fig, ax = plt.subplots(figsize=(8, 6))

    ax.plot(
        x_1,
        x_2,
        color='k'
    )

    ax.fill_between(
        x_1,
        x_2,
        color='lightgreen'
    )

    ax.set(
        xlabel=r'$x_1$',
        ylabel=r'$x_2$',
        xlim=(0, M / p_1 + 0.3),
        ylim=(0, M / p_2 + 0.3)
    )

    ax.text(
        M / p_1,
        0.1,
        r'$\frac{M}{p_1}$',
        ha='center'
    )

    ax.text(
        0.05,
        M / p_2 + 0.1,
        r'$\frac{M}{p_2}$',
        va='center'
    )

    plt.tight_layout()

    plt.savefig(
        "images/spotrebitel/graf_rozpocet.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()
    
#######################
# graf_rozpocet_zmeny #
#######################

def graf_rozpocet_zmeny():



    def ir(x1, M, p1, p2):
        return M / p2 - p1 / p2 * x1

    def zobraz(ax, Mka, p1ky, p2ky):

        x_1 = np.linspace(0, 50, 500)

        for (M, p1, p2) in product(Mka, p1ky, p2ky):

            x_2 = ir(x_1, M, p1, p2)

            ax.plot(
                x_1,
                x_2,
                label=rf"$M={M},\;p_1={p1},\;p_2={p2}$"
            )

        ax.set_xlabel(r"$x_1$")
        ax.set_ylabel(r"$x_2$")

        ax.set_xlim(0, 40 / 3)
        ax.set_ylim(0, 8)

        ax.legend(
            loc="upper right",
            frameon=True
        )

    fig, axes = plt.subplots(
        3,
        1,
        figsize=(8, 16)
    )

    # zmena príjmu
    set1 = (
        [30., 40., 50.],
        [5.],
        [10.]
    )

    # zmena ceny prvého statku
    set2 = (
        [40.],
        [3., 5., 8.],
        [10.]
    )

    # zmena ceny druhého statku
    set3 = (
        [40.],
        [5.],
        [5., 10., 15.]
    )

    for ax, exo in zip(
        axes,
        (set1, set2, set3)
    ):

        Mka, p1ky, p2ky = exo

        zobraz(
            ax,
            Mka,
            p1ky,
            p2ky
        )

    axes[0].set_title("Zmena príjmu")
    axes[1].set_title("Zmena ceny prvého statku")
    axes[2].set_title("Zmena ceny druhého statku")

    plt.tight_layout()

    plt.savefig(
        "images/spotrebitel/graf_rozpocet_zmeny.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

def graf_spotrebitel():

    def ik(x1, u):
        return u**(3/2) / x1**(1/2)

    def ir(x1, M, p1, p2):
        return M / p2 - p1 / p2 * x1

    p_1 = 5
    p_2 = 10
    M = 40

    x1opt = M / (3 * p_1)
    x2opt = 2 * M / (3 * p_2)

    uopt = x1opt**(1/3) * x2opt**(2/3)

    x_1 = np.linspace(1e-12, 8, 500)

    iuzit = ik(x_1, u=uopt)
    irozp = ir(x_1, M, p_1, p_2)

    fig, ax = plt.subplots(figsize=(8, 6))

    ax.plot(
        x_1,
        iuzit,
        label='indiferenčná krivka'
    )

    ax.plot(
        x_1,
        irozp,
        color='g',
        label='hranica spotrebiteľských možností'
    )

    ax.scatter(
        x1opt,
        x2opt,
        color='r',
        zorder=5
    )

    ax.text(
        x1opt + 0.2,
        x2opt + 0.1,
        r'$\mathbf{x}^*$'
    )

    ax.set(
        xlabel=r'$x_1$',
        ylabel=r'$x_2$',
        xlim=(0, 8),
        ylim=(0, 6)
    )

    ax.legend()

    plt.tight_layout()

    plt.savefig(
        "images/spotrebitel/graf_spotrebitel.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

graf_konvexnost_spotrebitelskych_moznosti()
graf_kose()
graf_mapa()
graf_rozpocet()
graf_rozpocet_zmeny()
graf_spotrebitel()
```

## Dopyt spotrebiteľa

Názov podpriečinka je *Porovn*

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch
# from itertools import product

Path("images/porovn").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

plt.rcParams["figure.figsize"] = (10, 6)

# --------------------------------------------------
# funkcie
# --------------------------------------------------


def x_opt(alpha, beta, M, p):
    return alpha/(alpha + beta) * M/p

def uzitocnost(x1, x2, alpha, beta):
    return x1**alpha * x2**beta

def rozpocet(x1, M, p1, p2):
    return M/p2 - p1/p2 * x1

def indiferencna_krivka(x1, u, alpha, beta):
    return (u / x1**alpha)**(1/beta)


#################################
# Obr graf_prijmovy_spotrebitel #
#################################

def graf_prijmovy_spotrebitel():
  
    # --------------------------------------------------
    # parametre
    # --------------------------------------------------

    alpha = 1/3
    beta = 2/3

    p1 = 2
    p2 = 4

    M_values = [600, 900]

    x1 = np.linspace(1e-6, 450, 1000)

    # --------------------------------------------------
    # graf
    # --------------------------------------------------

    fig, ax = plt.subplots()

    opt_x = []
    opt_y = []

    for i, M in enumerate(M_values):

        # optimum
        x1_opt = x_opt(alpha, beta, M, p1)
        x2_opt = x_opt(beta, alpha, M, p2)

        u_opt = uzitocnost(
            x1_opt,
            x2_opt,
            alpha,
            beta
        )

        # rozpočtová priamka
        ax.plot(
            x1,
            rozpocet(x1, M, p1, p2),
            label=rf'$M={M}$'
        )

        # indiferenčná krivka
        ax.plot(
            x1,
            indiferencna_krivka(
                x1,
                u_opt,
                alpha,
                beta
            ),
            color='C' + str(i),
            lw=2
        )

        # optimum
        ax.scatter(
            x1_opt,
            x2_opt,
            color='red',
            zorder=5
        )

        label = 'a' if i == 0 else 'b'

        ax.text(
            x1_opt + 8,
            x2_opt + 4,
            rf'$\mathbf{{x}}^{{*{label}}}$'
        )

        opt_x.append(x1_opt)
        opt_y.append(x2_opt)

    # --------------------------------------------------
    # príjmová cesta
    # --------------------------------------------------

    # príjmová cesta

    xp = np.linspace(0, 450, 200)

    ax.plot(
        xp,
        xp,
        'r--',
        lw=1.5,
        label='príjmová cesta'
    )

    # --------------------------------------------------
    # vzhľad
    # --------------------------------------------------

    ax.set(
        xlabel=r'$x_1$',
        ylabel=r'$x_2$',
        xlim=(0, 450),
        ylim=(0, 225)
    )

    ax.legend()

    ax.set_title('Príjmový efekt')

    plt.tight_layout()

    plt.savefig(
        "images/porovn/graf_prijmovy_spotrebitel.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

###############################
# graf_standardny_spotrebitel #
###############################

def graf_standardny_spotrebitel():
    
    
    def zobraz(ax, M, p1ky, p2ky):

        for (M, p1, p2, f, o) in zip(M, p1ky, p2ky, farby, ozn):
            r = rozpocet(x1,M,p1,p2) # rozpočet
            ax.plot(x1,r, label=rf"$p_1 = {p1},\; p_2 = {p2}$", color = f)
            xx1 = x_opt(𝛼, 𝛽, M, p1)
            xx2 = x_opt(𝛽, 𝛼, M, p2)
            h = uzitocnost(xx1,xx2,𝛼, 𝛽)
            ind = h**(1/𝛽) / x1**(𝛼/𝛽) # indiferenčná krivka
            ax.plot(x1,ind, color = f)
            ax.scatter(xx1,xx2,marker='x', color = f, label = o)

        ax.set_xlabel('$x_1$')
        ax.set_ylabel('$x_2$')

    fig, axes = plt.subplots(1, 2, figsize=(16, 8))

    # Parameters
    𝛼, 𝛽 = 1/3, 2/3

    set1 = ([600.,600.], [2.,4.], [4.,4.],['g','r'], ['$\mathbf{x_a}$','$\mathbf{x_b}$'])
    set2 = ([600.,600.], [2.,2.], [4.,8.],['g','r'], ['$\mathbf{x_a}$','$\mathbf{x_b}$'])

    x1 = np.linspace(1e-12,500,300)

    for (ax, exo) in zip(axes, (set1, set2)):
        M, p1ky, p2ky, farby, ozn = exo
        zobraz(ax, M, p1ky, p2ky)
        ax.plot([0,300],[100,100], ':', color = 'black') # čiary cez pôvodný kôš, horizontálna
        ax.plot([100,100],[0,160], ':', color = 'black') # vertikálna
        ax.set_ylim(0,160)
        ax.set_xlim(0,300)
        ax.legend(loc='upper right', frameon=True)

    plt.savefig(
        "images/porovn/graf_standardny_spotrebitel.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

#################
# graf_hicks_ul #
#################

def graf_hicks_ul():

    def ik(x1, u):
        return u**(3/2) / x1**(1/2)

    def ir(x1, M, p1, p2):
        return M / p2 - p1 / p2 * x1

    M = 40
    p_1 = 5
    p_2 = 10

    x1opt = M/3/p_1
    x2opt = 2*M/3/p_2
    x1opt, x2opt
    uopt = x1opt**(1/3) * x2opt**(2/3)

    x_1 = np.linspace(1e-12,8)

    iuzit = ik(x_1,u=uopt)
    irozp = ir(x_1,M,p_1,p_2)

    fig, ax = plt.subplots()

    ax.plot(x_1,iuzit,color='green',
        label='hranica prípustných riešení')
    ax.plot(x_1,irozp,color='b',
        label='vrstevnica rozpočovej funkcie')
    ax.set_xlim(0,8)
    ax.set_ylim(0,6)

    ax.fill_between(x_1,iuzit,y2=6, color = 'lightgreen', alpha=0.5,
                   label = 'množina prípustných riešení')

    ax.scatter(
        x1opt,
        x2opt,
        color='r',
        zorder=5
    )

    ax.text(
        x1opt + 0.2,
        x2opt + 0.1,
        r'$\mathbf{h}^*$'
    )

    ax.set(
        xlabel=r'$x_1$',
        ylabel=r'$x_2$',
        xlim=(0,8),
        ylim=(0,6)
    )

    ax.legend()    
    
    plt.savefig(
        "images/porovn/graf_hicks_ul.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

##############
# graf_hicks #
##############

def graf_hicks():
    
    def zobraz(ax, M, p1ky, p2ky):

        for (M, p1, p2, f, o) in zip(M, p1ky, p2ky, farby, ozn):
            r = rozpocet(x1,M,p1,p2) # rozpočet
            ax.plot(x1,r, label=rf"$p_1 = {p1},\; p_2 = {p2}$", color = f)
            xx1 = x_opt(𝛼, 𝛽, M, p1)
            xx2 = x_opt(𝛽, 𝛼, M, p2)
            h = uzitocnost(xx1,xx2,𝛼, 𝛽)
            ind = h**(1/𝛽) / x1**(𝛼/𝛽) # indiferenčná krivka
            ax.plot(x1,ind, color = f)
            ax.scatter(xx1,xx2,marker='x', color = f, label = o)

        ax.set_xlabel('$x_1$')
        ax.set_ylabel('$x_2$')    
    
    # Parameters
    α, β = 1/3, 2/3
    
    set1 = ([600.,600.], [2.,4.], [4.,4.],['g','r'], ['$\mathbf{x_a}$','$\mathbf{x_b}$'])
    set2 = ([600.,600.], [2.,2.], [4.,8.],['g','r'], ['$\mathbf{x_a}$','$\mathbf{x_b}$'])    
    
    x1a = x_opt(α, β, 600, 2)
    x2a = x_opt(β, α, 600, 4)

    u0 = uzitocnost(x1a, x2a, α, β)

    Mh1 = (
        u0
        * set1[1][1]**α
        * set1[2][1]**β
        / (α**α * β**β)
    )

    Mh2 = (
        u0
        * set2[1][1]**α
        * set2[2][1]**β
        / (α**α * β**β)
    )

    set1h = ([600.,600.,Mh1], [2.,4.,4.], [4.,4.,4.],['g','r','b'], ['$\mathbf{x_a}$','$\mathbf{x_b}$','$\mathbf{x_s}$'])
    set2h = ([600.,600.,Mh2], [2.,2.,2.], [4.,8.,8.],['g','r','b'], ['$\mathbf{x_a}$','$\mathbf{x_b}$','$\mathbf{x_s}$'])

    fig, axes = plt.subplots(2, 1, figsize=(12, 16))

    x1 = np.linspace(1e-12,500,300)

    for (ax, exo) in zip(axes, (set1h, set2h)):
        Mka, p1ky, p2ky, farby, ozn = exo
        zobraz(ax, Mka, p1ky, p2ky)
        ax.set_xlim(0,500)
        ax.set_ylim(0,200)
        ax.legend(loc='upper right', frameon=True)

    axes[0].set_title("Zmena ceny prvého statku")
    axes[1].set_title("Zmena ceny druhého statku")

    plt.savefig(
        "images/porovn/graf_hicks.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

####################
# Obr graf_slutsky #
####################

def graf_slutsky():

    def zobraz(ax, M, p1ky, p2ky):

        for (M, p1, p2, f, o) in zip(M, p1ky, p2ky, farby, ozn):
            r = rozpocet(x1,M,p1,p2) # rozpočet
            ax.plot(x1,r, label=rf"$p_1 = {p1},\; p_2 = {p2}$", color = f)
            xx1 = x_opt(𝛼, 𝛽, M, p1)
            xx2 = x_opt(𝛽, 𝛼, M, p2)
            h = uzitocnost(xx1,xx2,𝛼, 𝛽)
            ind = h**(1/𝛽) / x1**(𝛼/𝛽) # indiferenčná krivka
            ax.plot(x1,ind, color = f)
            ax.scatter(xx1,xx2,marker='x', color = f, label = o)

        ax.set_xlabel('$x_1$')
        ax.set_ylabel('$x_2$')
    
    # Parameters
    α, β = 1/3, 2/3
    
    set1 = ([600.,600.], [2.,4.], [4.,4.],['g','r'], ['$\mathbf{x_a}$','$\mathbf{x_b}$'])
    set2 = ([600.,600.], [2.,2.], [4.,8.],['g','r'], ['$\mathbf{x_a}$','$\mathbf{x_b}$'])
    
    x1a = x_opt(α, β, set1[0][0], set1[1][0])
    x2a = x_opt(β, α, set1[0][0], set1[2][0])

    Ms1 = x1a*set1[1][1] + x2a*set1[2][1]
    Ms2 = x1a*set2[1][1] + x2a*set2[2][1]

    set1s = ([600.,600.,Ms1], [2.,4.,4.], [4.,4.,4.],['g','r','b'], ['$\mathbf{x_a}$','$\mathbf{x_b}$','$\mathbf{x_s}$'])
    set2s = ([600.,600.,Ms2], [2.,2.,2.], [4.,8.,8.],['g','r','b'], ['$\mathbf{x_a}$','$\mathbf{x_b}$','$\mathbf{x_s}$'])

    fig, axes = plt.subplots(2, 1, figsize=(12, 16))

    x1 = np.linspace(1e-12,500,300)

    for (ax, exo) in zip(axes, (set1s, set2s)):
        Mka, p1ky, p2ky, farby, ozn = exo
        zobraz(ax, Mka, p1ky, p2ky)
        ax.set_xlim(0,500)
        ax.set_ylim(0,200)
        ax.legend(loc='upper right', frameon=True)

    plt.savefig(
        "images/porovn/graf_slutsky.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

graf_prijmovy_spotrebitel()
graf_standardny_spotrebitel()
graf_hicks_ul()
graf_hicks()
graf_slutsky()
```

## Voľný čas a spotreba

Názov podpriečnka je *stati*.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch
# from itertools import product

Path("images/stati").mkdir(exist_ok=True) # skontroluj, či v pracovnom priečinku existuje podpriečinok images, ak nie vytvor

####################
# graf_rozpocet_lC #
####################

def graf_rozpocet_lC():
    
    # parametre
    h = 10
    w = 1

    fig, ax = plt.subplots(1, 2, figsize=(12, 5))

    # =====================================================
    # π - T > 0
    # =====================================================

    A = 3  # A = π - T

    lmax = h + A / w

    # hranica rozpočtových možností
    l1 = np.linspace(0, h, 200)
    C1 = w * (h - l1) + A

    # množina spotrebiteľských možností
    poly_x = [0, lmax, h, h, 0]
    poly_y = [0, 0, 0, A, w * h + A]

    ax[0].fill(poly_x, poly_y, color="lightgreen", alpha=0.6)

    # zelená hranica
    ax[0].plot(l1, C1, color="green", lw=3)
    ax[0].plot([h, h], [0, A], color="green", lw=3)

    # pomocné čiary
    ax[0].axvline(h, ls="--", color="gray")
    ax[0].axhline(A, ls="--", color="gray")

    # bod bez práce
    ax[0].scatter([h], [A], color="black", zorder=10)
    ax[0].annotate(r'$(h,\pi-T)$',
                   xy=(h, A),
                   xytext=(10, 10),
                   textcoords='offset points')

    # sklon
    x_arrow = 5
    y_arrow = w * (h - x_arrow) + A

    ax[0].annotate(r'sklon $=-w$',
                   xy=(x_arrow, y_arrow),
                   xytext=(4.0, y_arrow + 2),
                   arrowprops=dict(arrowstyle='->'))

    ax[0].set_title(r'$\pi-T>0$')
    ax[0].set_xlabel(r'$\ell$')
    ax[0].set_ylabel(r'$C$')

    ax[0].set_xticks([h])
    ax[0].set_xticklabels([r'$h$'])

    ax[0].set_xlim(0, 14)
    ax[0].set_ylim(0, 15)

    # prerušované predĺženie rozpočtovej priamky

    l_ext = np.linspace(h, lmax, 100)
    C_ext = w*(h - l_ext) + A

    ax[0].plot(
        l_ext,
        C_ext,
        color="green",
        lw=2,
        ls="--"
    )

    # označenie extrémov

    ax[0].scatter([0, h, lmax],
                  [w*h + A, A, 0],
                  color="black")

    ax[0].set_xticks([h, lmax])
    ax[0].set_xticklabels([
        r'$h$',
        r'$h+\frac{\pi-T}{w}$'
    ])

    ax[0].set_yticks([A, w*h + A])
    ax[0].set_yticklabels([
        r'$\pi-T$',
        r'$wh+\pi-T$'
    ])

    # =====================================================
    # π - T ≤ 0
    # =====================================================

    A = -1.5

    l0 = h + A/w

    l2 = np.linspace(0, l0, 200)
    C2 = w*(h - l2) + A

    # trojuholník spotrebiteľských možností
    ax[1].fill_between(
        l2,
        0,
        C2,
        color="lightgreen",
        alpha=0.6
    )

    # hranica možností
    ax[1].plot(
        l2,
        C2,
        color="green",
        lw=3
    )

    # zvislé časové ohraničenie
    ax[1].axvline(h,
                  ls="--",
                  color="gray")

    # extrémy
    ax[1].scatter(
        [l0, 0],
        [0, w*(h)+A],
        color="black"
    )

    # sklon
    ax[1].annotate(
        r'sklon $=-w$',
        xy=(5.5, w*(h-5.5)+A),
        xytext=(4.0, y_arrow - 2),
        arrowprops=dict(arrowstyle='->')
    )

    ax[1].set_title(r'$\pi-T\leq0$')

    ax[1].set_xlabel(r'$\ell$')
    # ax[1].set_ylabel(r'$C$')

    ax[1].set_xticks([l0, h])
    ax[1].set_xticklabels([
        r'$h+\frac{\pi-T}{w}$',
        r'$h$'
    ])

    ax[1].set_yticks([w*(h)+A])
    ax[1].set_yticklabels([
        r'$wh+\pi-T$'
    ])

    ax[1].set_xlim(0, 14)
    ax[1].set_ylim(0, 15)

    plt.tight_layout()
    
    plt.savefig(
        "images/stati/graf_rozpocet_lC.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

###################
# graf_optimum_lC #
###################

def graf_optimum_lC():

    # parametre
    h = 24
    w = 4
    
    # optimum
    l_star = 10
    C_star = 80

    fig, ax = plt.subplots(figsize=(12, 7))

    # =====================================================
    # π - T > 0
    # =====================================================

    A = 24  # A = π - T

    lmax = h + A / w

    # hranica rozpočtových možností
    l = np.linspace(0, h, 200)
    C = w * (h - l) + A

    # zelená hranica
    ax.plot(l, C, color="green", lw=3)
    ax.plot([h, h], [0, A], color="green", lw=3)

    # pomocné čiary
    ax.axvline(h, ls="--", color="gray")
    ax.axhline(A, ls="--", color="gray")

    # bod bez práce
    ax.scatter([h], [A], color="black", zorder=10)
    ax.annotate(r'$(h,\pi-T)$',
                   xy=(h, A),
                   xytext=(10, 10),
                   textcoords='offset points')

    # sklon
    x_arrow = 5
    y_arrow = w * (h - x_arrow) + A

    ax.text(
        10.1,
        125,
        r'Podmienka optimálnosti: $MRS(\ell^*,C^*)=w$'
    )

    ax.set_xlabel(r'$\ell$')
    ax.set_ylabel(r'$C$')

    ax.set_xticks([h])
    ax.set_xticklabels([r'$h$'])

    ax.set_xlim(0, 33)
    ax.set_ylim(0, 130)

    # prerušované predĺženie rozpočtovej priamky

    l_ext = np.linspace(h, lmax, 100)
    C_ext = w*(h - l_ext) + A

    ax.plot(
        l_ext,
        C_ext,
        color="green",
        lw=2,
        ls="--"
    )

    # označenie extrémov

    ax.scatter([0, h, lmax],
                  [w*h + A, A, 0],
                  color="black")

    ax.set_xticks([l_star, h, lmax])

    ax.set_xticklabels([
        r'$\ell^*$',
        r'$h$',
        r'$h+\frac{\pi-T}{w}$'
    ])

    ax.set_yticks([C_star, A, w*h + A])

    ax.set_yticklabels([
        r'$C^*$',
        r'$\pi-T$',
        r'$wh+\pi-T$'
    ])

    # indiferenčná krivka prechádzajúca optimom

    l_ic = np.linspace(1, 26, 500)
    C_ic = 80*np.sqrt(10/l_ic)

    ax.plot(
        l_ic,
        C_ic,
        color="blue",
        lw=2,
        label="indiferenčná krivka"
    )

    ax.scatter(
        [l_star],
        [C_star],
        color="red",
        s=70,
        zorder=10
    )

    ax.axvline(
        l_star,
        color="black",
        ls="--",
        lw=1
    )

    ax.axhline(
        C_star,
        color="black",
        ls="--",
        lw=1
    )

    ax.annotate(
        '',
        xy=(h, 10),
        xytext=(l_star, 10),
        arrowprops=dict(
            arrowstyle='<->',
            color='darkorange',
            lw=2
        )
    )

    ax.text(
        (l_star + h)/2,
        12,
        r'$N^{s*}$',
        color='darkorange',
        ha='center'
    )

    plt.tight_layout()
    
    plt.savefig(
        "images/stati/graf_optimum_lC.png",
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()    

############
# graf_h_l #
############

def graf_h_l():
    
    import numpy as np
    import matplotlib.pyplot as plt

    def indifference_curve(l, l0, C0, alpha):
        return C0 * (l0 / l)**(alpha / (1 - alpha))

    alpha = 3/4

    # parametre
    h = 24
    w = 3
    A = 48      # π - T
    l0 = h
    C0 = A

    fig, ax = plt.subplots(figsize=(12, 7))

    # hranica spotrebiteľských možností
    l = np.linspace(0, h, 200)
    C = w * (h - l) + A

    # hlavná časť hranice
    ax.plot(l, C, color="green", lw=3)

    # zvislá časť hranice
    ax.plot([h, h], [0, A], color="green", lw=3)

    # pomocné čiary
    ax.axvline(h, ls="--", color="gray")
    ax.axhline(A, ls="--", color="gray")

    # predĺženie rozpočtovej priamky
    lmax = h + A / w

    l_ext = np.linspace(h, lmax, 100)
    C_ext = w * (h - l_ext) + A

    ax.plot(
        l_ext,
        C_ext,
        color="green",
        lw=2,
        ls="--"
    )

    # extrémy
    ax.scatter(
        [0, h, lmax],
        [w*h + A, A, 0],
        color="black",
        zorder=10
    )

    # označenie bodu bez práce
    ax.annotate(
        r'$[h,\pi-T]$',
        xy=(h, A),
        xytext=(10, 10),
        textcoords='offset points'
    )

    # osi
    ax.set_xlabel(r'$\ell$')
    ax.set_ylabel(r'$C$')

    ax.set_xlim(0, 42)
    ax.set_ylim(0, 130)

    # označenie osí
    ax.set_xticks([h, lmax])
    ax.set_xticklabels([
        r'$\ell^* = h$',
        r'$h+\frac{\pi-T}{w}$'
    ])
    ax.set_yticks([A, w*h + A])

    ax.set_yticklabels([
        r'$C^*=\pi-T$',
        r'$wh+\pi-T$'
    ])

    l_ic = np.linspace(4, 40, 500)

    C_ic = indifference_curve(l_ic, l0, C0, alpha)

    ax.plot(
        l_ic,
        C_ic,
        color="blue",
        lw=2
    )
        
    plt.tight_layout()

    plt.savefig(
        "images/stati/graf_h_l.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()     

def graf_zmena_pi_T():    
    
    # --------------------------------------------------
    # model
    # --------------------------------------------------

    h = 24

    def optimum(alpha, w, A):
        M = w*h + A
        l_star = alpha * M / w
        C_star = (1 - alpha) * M
        return l_star, C_star

    def indifference_curve(l, alpha, l_star, C_star):
        return C_star * (l_star/l)**(alpha/(1-alpha))

    # --------------------------------------------------
    # scenáre
    # --------------------------------------------------

    alpha = 1/3

    scenarios = [
        {"A": 30, "w": 4, "color": "blue"},
        {"A": 10, "w": 4, "color": "red"},
    ]

    # --------------------------------------------------
    # graf
    # --------------------------------------------------

    fig, ax = plt.subplots(figsize=(12,7))

    for s in scenarios:

        A = s["A"]
        w = s["w"]
        color = s["color"]

        # optimum
        l_star, C_star = optimum(alpha, w, A)

        # hranica možností
        l = np.linspace(0, h, 300)
        C = w*(h-l) + A

        ax.plot(l, C, color=color, lw=2)

        # vodorovná časť pri π-T
        ax.plot([h, h], [0, A], color=color, lw=2)

        # vodorovná pomocná čiara π-T
        ax.axhline(A, color=color, ls="--", alpha=0.5)

        # indiferenčná krivka
        l_ic = np.linspace(1, h, 500)
        C_ic = indifference_curve(
            l_ic,
            alpha,
            l_star,
            C_star
        )

        ax.plot(
            l_ic,
            C_ic,
            color=color,
            ls=":"
        )

        # optimum
        ax.scatter(
            [l_star],
            [C_star],
            color=color,
            s=70,
            zorder=10
        )
        # vodiace čiary k optimu

        ax.plot(
            [0, l_star],
            [C_star, C_star],
            color=color,
            ls="--",
            lw=1
        )

        ax.plot(
            [l_star, l_star],
            [0, C_star],
            color=color,
            ls="--",
            lw=1
        )

    # spoločná zvislá čiara h
    ax.axvline(h, ls="--", color="black")

    ax.set_xticks([])
    ax.set_yticks([])

    # modré optimum
    l_star_blue, C_star_blue = optimum(alpha, 4, 30)

    ax.text(
        l_star_blue,
        -4,
        r'$\ell^*_1$',
        color='blue',
        ha='center'
    )

    ax.text(
        -2,
        C_star_blue,
        r'$C^*_1$',
        color='blue',
        va='center'
    )

    # červené optimum
    l_star_red, C_star_red = optimum(alpha, 4, 10)

    ax.text(
        l_star_red,
        -4,
        r'$\ell^*_2$',
        color='red',
        ha='center'
    )

    ax.text(
        -2,
        C_star_red,
        r'$C^*_2$',
        color='red',
        va='center'
    )

    # π - T
    ax.text(
        -2,
        30,
        r'$(\pi-T)_1$',
        color='blue',
        va='center'
    )

    ax.text(
        -2,
        10,
        r'$(\pi-T)_2$',
        color='red',
        va='center'
    )

    ax.text(
        h,
        -4,
        r'$h$',
        ha='center'
    )

    ax.set_xlabel(r'$\ell$')
    ax.set_ylabel(r'$C$')

    ax.set_xlim(0, 26)
    ax.set_ylim(0, 110)

    plt.tight_layout()  
    
    plt.savefig(
        "images/stati/graf_zmena_pi_T.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

###################
# graf_zmena_w_lC #
###################

def graf_zmena_w_lC():
    
    import numpy as np
    import matplotlib.pyplot as plt

    # ==================================================
    # Parametre modelu
    # ==================================================

    h = 24
    alpha = 1/3

    scenarios = [
        {"A": 24, "w": 3, "color": "blue"},
        {"A": 24, "w": 10, "color": "red"},
    ]

    # ==================================================
    # Funkcie
    # ==================================================

    def budget(l, w, A):
        """
        Hranica spotrebných možností

        C = w(h-l) + A
        """
        return w * (h - l) + A


    def optimum(alpha, w, A):
        """
        Cobb-Douglas

        U(C,l)=C^(1-alpha) l^alpha
        """

        M = w * h + A

        l_star = alpha * M / w
        C_star = (1 - alpha) * M

        return l_star, C_star


    def indifference_curve(l, alpha, l_star, C_star):
        """
        Indiferenčná krivka pre úroveň užitočnosti
        prechádzajúcu optimom.
        """

        U = C_star**(1-alpha) * l_star**alpha

        return (U / (l**alpha))**(1/(1-alpha))


    # ==================================================
    # Graf
    # ==================================================

    fig, ax = plt.subplots(figsize=(10, 6))

    l_grid = np.linspace(1e-16, h, 500)

    for i, s in enumerate(scenarios, start=1):

        A = s["A"]
        w = s["w"]
        color = s["color"]

        # optimum
        l_star, C_star = optimum(alpha, w, A)

        # hranica možností
        ax.plot(
            l_grid,
            budget(l_grid, w, A),
            color=color,
            lw=2
        )

        l_ext = np.linspace(h, h+8, 100)

        ax.plot(l_ext,
            budget(l_ext,w,A),
            color=color,
            lw=1.5,
            ls="--")

        # indiferenčná krivka
        ax.plot(
            l_grid,
            indifference_curve(
                l_grid,
                alpha,
                l_star,
                C_star
            ),
            color=color,
            ls=":"
        )

        # optimum
        ax.scatter(
            l_star,
            C_star,
            color=color,
            s=60,
            zorder=5
        )

        # pomocné čiary
        ax.plot(
            [0, l_star],
            [C_star, C_star],
            ls="--",
            lw=1,
            color=color,
            alpha=0.7
        )

        ax.plot(
            [l_star, l_star],
            [0, C_star],
            ls="--",
            lw=1,
            color=color,
            alpha=0.7
        )

        # označenia
        ax.text(
            -3,
            C_star,
            rf"$C_{i}$",
            color=color,
            va="center"
        )

        ax.text(
            l_star,
            -8,
            rf"$\ell_{i}$",
            color=color,
            ha="center"
        )

    # ==================================================
    # π-T
    # ==================================================

    A = scenarios[0]["A"]

    ax.axhline(
        A,
        linestyle="--",
        color="black",
        alpha=0.7
    )

    ax.text(
        -3,
        A,
        r"$\pi-T$",
        va="center"
    )

    # ==================================================
    # h
    # ==================================================

    ax.axvline(
        h,
        linestyle="--",
        color="black",
        alpha=0.7
    )

    ax.plot(
        [h, h],
        [0, A],
        color="blue",
        lw=2
    )

    ax.text(
        h,
        -8,
        r"$h$",
        ha="center"
    )

    # ==================================================
    # Úprava osí
    # ==================================================

    C_max = max(
        budget(0, s["w"], s["A"])
        for s in scenarios
    )

    ax.set_xlim(0, h + 8)
    ax.set_ylim(0, 300)

    ax.set_xticks([])
    ax.set_yticks([])

    ax.set_xlabel(r"$\ell$")
    ax.set_ylabel(r"$C$")

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    plt.tight_layout()
    
    plt.savefig(
        "images/stati/graf_zmena_w_lC.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()    

###################
# graf_rozklad_lC #
###################
    
def graf_rozklad_lC():

    # ==================================================
    # Parametre modelu
    # ==================================================

    h = 24
    alpha = 1/3

    scenarios = [
        {"A": 24, "w": 3, "color": "blue"},
        {"A": 24, "w": 10, "color": "red"},
    ]

    # ==================================================
    # Funkcie
    # ==================================================

    def budget(l, w, A):
        """
        Hranica spotrebných možností

        C = w(h-l) + A
        """
        return w * (h - l) + A


    def optimum(alpha, w, A):
        """
        Cobb-Douglas

        U(C,l)=C^(1-alpha) l^alpha
        """

        M = w * h + A

        l_star = alpha * M / w
        C_star = (1 - alpha) * M

        return l_star, C_star


    def indifference_curve(l, alpha, l_star, C_star):
        """
        Indiferenčná krivka pre úroveň užitočnosti
        prechádzajúcu optimom.
        """

        U = C_star**(1-alpha) * l_star**alpha

        return (U / (l**alpha))**(1/(1-alpha))


    # ==================================================
    # Graf
    # ==================================================

    fig, ax = plt.subplots(figsize=(10, 6))

    l_grid = np.linspace(1e-16, h, 500)

    for i, s in enumerate(scenarios, start=1):

        A = s["A"]
        w = s["w"]
        color = s["color"]

        # optimum
        l_star, C_star = optimum(alpha, w, A)

        # hranica možností
        ax.plot(
            l_grid,
            budget(l_grid, w, A),
            color=color,
            lw=2
        )

        l_ext = np.linspace(h, h+8, 100)

        ax.plot(l_ext,
            budget(l_ext,w,A),
            color=color,
            lw=1.5,
            ls="--")

        # indiferenčná krivka
        ax.plot(
            l_grid,
            indifference_curve(
                l_grid,
                alpha,
                l_star,
                C_star
            ),
            color=color,
            ls=":"
        )

        # optimum
        ax.scatter(
            l_star,
            C_star,
            color=color,
            s=60,
            zorder=5
        )

        # pomocné čiary
        ax.plot(
            [0, l_star],
            [C_star, C_star],
            ls="--",
            lw=1,
            color=color,
            alpha=0.7
        )

        ax.plot(
            [l_star, l_star],
            [0, C_star],
            ls="--",
            lw=1,
            color=color,
            alpha=0.7
        )

        # označenia
        ax.text(
            -3,
            C_star,
            rf"$C_{i}$",
            color=color,
            va="center"
        )

        ax.text(
            l_star,
            -8,
            rf"$\ell_{i}$",
            color=color,
            ha="center"
        )

    # ==================================================
    # π-T
    # ==================================================

    A = scenarios[0]["A"]

    ax.axhline(
        A,
        linestyle="--",
        color="black",
        alpha=0.7
    )

    ax.text(
        -3,
        A,
        r"$\pi-T$",
        va="center"
    )

    # ==================================================
    # h
    # ==================================================

    ax.axvline(
        h,
        linestyle="--",
        color="black",
        alpha=0.7
    )

    ax.plot(
        [h, h],
        [0, A],
        color="blue",
        lw=2
    )

    ax.text(
        h,
        -8,
        r"$h$",
        ha="center"
    )

    # ==================================================
    # Úprava osí
    # ==================================================

    C_max = max(
        budget(0, s["w"], s["A"])
        for s in scenarios
    )

    ax.set_xlim(0, h + 8)
    ax.set_ylim(0, 300)

    ax.set_xticks([])
    ax.set_yticks([])

    ax.set_xlabel(r"$\ell$")
    ax.set_ylabel(r"$C$")

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    # ==================================================
    # Hicksova kompenzácia
    # ==================================================

    # pôvodný scenár
    A0 = scenarios[0]["A"]
    w0 = scenarios[0]["w"]

    l0, C0 = optimum(alpha, w0, A0)

    # pôvodná užitočnosť
    U0 = C0**(1-alpha) * l0**alpha

    # nová mzda
    w1 = scenarios[1]["w"]

    # Hicksovo optimum

    ls = U0 / (
        (w1 * (1-alpha)/alpha)**(1-alpha)
    )

    Cs = (
        w1
        * (1-alpha)/alpha
        * ls
    )

    # kompenzovaný nepracovný príjem

    As = Cs - w1*(h-ls)

    # ==================================================
    # Hicksova hranica
    # ==================================================

    ax.plot(
        l_grid,
        budget(l_grid, w1, As),
        color="green",
        lw=2
    )

    ax.scatter(
        ls,
        Cs,
        color="green",
        s=60,
        zorder=6
    )

    ax.plot(
        [0, ls],
        [Cs, Cs],
        color="green",
        ls="--",
        lw=1
    )

    ax.plot(
        [ls, ls],
        [0, Cs],
        color="green",
        ls="--",
        lw=1
    )

    ax.text(
        ls,
        -8,
        r'$\ell_s$',
        color='green',
        ha='center'
    )

    ax.text(
        -3,
        Cs,
        r'$C_s$',
        color='green',
        va='center'
    ) 

    plt.tight_layout()

    plt.savefig(
        "images/stati/graf_rozklad_lC.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()
    
#####################
# graf_ponuka_prace #
#####################
    
def graf_ponuka_prace():
    
    # =========================================
    # Parametre
    # =========================================

    h = 24
    alpha = 1/3

    A1 = 24
    A2 = 48

    w_min = 2
    w_max = 5

    w_grid = np.linspace(w_min, w_max, 500)

    # =========================================
    # Funkcie
    # =========================================

    def leisure(alpha, h, A, w):
        """
        Optimálny voľný čas
        """
        return alpha*h + alpha*A/w


    def consumption(alpha, h, A, w):
        """
        Optimálna spotreba
        """
        return (1-alpha)*(w*h + A)


    def labor_supply(alpha, h, A, w):
        """
        Ponuka práce
        """
        l = leisure(alpha, h, A, w)
        return h - l


    # =========================================
    # Výpočty
    # =========================================

    N1 = labor_supply(alpha, h, A1, w_grid)
    N2 = labor_supply(alpha, h, A2, w_grid)

    # =========================================
    # Graf
    # =========================================

    fig, ax = plt.subplots(figsize=(10,6))

    ax.plot(
        N1,
        w_grid,
        color="blue",
        lw=2,
        label=rf"$\pi-T={A1}$"
    )

    ax.plot(
        N2,
        w_grid,
        color="red",
        lw=2,
        label=rf"$\pi-T={A2}$"
    )

    ax.set_xlabel(r"$N^s$")
    ax.set_ylabel(r"$w$")

    ax.set_xticks([])
    ax.set_yticks([])

    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    #ax.legend()
    ax.set_title("Ponuka práce sa zvýšením $\pi-T$ posunie smerom doprava")

    plt.tight_layout()
    
    plt.savefig(
        "images/stati/graf_ponuka_prace.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

graf_rozpocet_lC()
graf_optimum_lC()
graf_h_l()
graf_zmena_pi_T()
graf_zmena_w_lC()
graf_rozklad_lC()
graf_ponuka_prace()
```

## Spotrebiteľské rozhodovanie medzi súčasnosťou a budúcnosťou

Priečinok: `images/dynam`

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch
# from itertools import product

Path("images/dynam").mkdir(exist_ok=True)

########################
# Funkcie a nastavenia #
########################

plt.rcParams["figure.figsize"] = (8,6)
plt.rcParams["figure.dpi"] = 120

plt.rcParams["font.size"] = 12

plt.rcParams["lines.linewidth"] = 2

plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

def v(c):
    return np.log(c)


def uzitocnost(c, cp, beta=None, r=None):

    if beta is None:
        beta = 1/(1+r)

    return v(c) + beta*v(cp)

def optimum(y, yp, r, t=0, tp=0, beta=None):

    if beta is None:
        beta = 1/(1+r)

    we = y - t + (yp - tp)/(1+r)

    c = we/(1+beta)

    cp = beta*(1+r)*c

    u = uzitocnost(c, cp, beta)

    return {
        "we": we,
        "c": c,
        "cp": cp,
        "u": u,
        "beta": beta
    }

def ekonomicke_osy(ax,xlabel,ylabel):

    ax.set_xticks([])
    ax.set_yticks([])

    ax.set_xlim(left=0)
    ax.set_ylim(bottom=0)

    ax.margins(0)

    ax.text(
        1.00,
        -0.02,
        xlabel,
        transform=ax.transAxes,
        ha="right",
        va="top"
    )

    ax.text(
        -0.02,
        1.00,
        ylabel,
        transform=ax.transAxes,
        ha="right",
        va="top"
    )

    return ax

def nakresli_rozpocet_intertemporal(
    ax,
    r,
    y,
    yp,
    t=0,
    tp=0,
    color="green",
    linestyle="-"
):

    y1 = y - t
    y2 = yp - tp

    c_max = y1 + y2/(1+r)
    cp_max = (1+r)*y1 + y2

    c = np.linspace(0, c_max, 200)

    cp = cp_max - (1+r)*c

    ax.plot(
        c,
        cp,
        color=color,
        linestyle=linestyle
    )

    return {
        "y1": y1,
        "y2": y2,
        "c_max": c_max,
        "cp_max": cp_max
    }

def nakresli_indiferencnu_intertemporal(
    ax,
    u,
    beta,
    c_min=1,
    c_max=100,
    color="blue",
    linestyle="-"
):

    c = np.linspace(c_min, c_max, 500)

    cp = np.exp((u - np.log(c))/beta)

    ax.plot(
        c,
        cp,
        color=color,
        linestyle=linestyle
    )

    return {"u": u}


def nakresli_bod(
    ax,
    x,
    y,
    nazov="E",
    vodiace_ciary=False,
    dx=5,
    dy=5,
    color="black"
):

    ax.scatter(
        [x],
        [y],
        color=color,
        zorder=10
    )

    ax.annotate(
        nazov,
        xy=(x,y),
        xytext=(dx,dy),
        textcoords="offset points"
    )

    if vodiace_ciary:

        ax.plot(
            [0,x],
            [y,y],
            "--",
            color="gray",
            linewidth=1
        )

        ax.plot(
            [x,x],
            [0,y],
            "--",
            color="gray",
            linewidth=1
        )

    return {"x":x,"y":y}

def vyznac_na_osi(
    ax,
    x=None,
    y=None,
    popis_x=None,
    popis_y=None,
    color="black"
):
    """
    Označí hodnoty na osiach.

    Príklady:
        vyznac_na_osi(
            ax,
            x=y1,
            y=y2,
            popis_x=r"$y-t$",
            popis_y=r"$y'-t'$"
        )

        vyznac_na_osi(
            ax,
            x=c_star,
            y=cp_star,
            popis_x=r"$c^*$",
            popis_y=r"$c'^*$"
        )
    """

    if x is not None and popis_x is not None:

        ax.annotate(
            popis_x,
            xy=(x, 0),
            xytext=(0, -15),
            textcoords="offset points",
            ha="center",
            color=color
        )

    if y is not None and popis_y is not None:

        ax.annotate(
            popis_y,
            xy=(0, y),
            xytext=(-10, 0),
            textcoords="offset points",
            ha="right",
            va="center",
            color=color
        )

def nakresli_sipku(
    ax,
    x1,y1,
    x2,y2,
    color="black"
):

    ax.annotate(
        "",
        xy=(x2,y2),
        xytext=(x1,y1),
        arrowprops=dict(
            arrowstyle="->",
            color=color
        )
    )

##########################
# graf_dynamicka_mnozina #
##########################

def graf_dynamicka_mnozina():

    fig, ax = plt.subplots()
    
    rozpocet = nakresli_rozpocet_intertemporal(
        ax,
        r=0.1,
        y=120,
        yp=100,
    ) 
   
    nakresli_bod(
        ax,
        120,
        100,
        nazov="E",
        vodiace_ciary=True,
    )
    
    vyznac_na_osi(
        ax,
        x=120,
        y=100,
        popis_x=r"$y-t$",
        popis_y=r"$y'-t'$",
    )
    
    vyznac_na_osi(
        ax,
        x=rozpocet["c_max"],
        y=rozpocet["cp_max"],
        popis_x=r"$we$",
        popis_y=r"$(1+r)we$",
    )
    
    ax.fill(
        [0, 0, rozpocet["c_max"]],
        [0, rozpocet["cp_max"], 0],
        color="lightgreen",
        alpha=0.3
    )
    
    ax.annotate(
        "veriteľ",
        xy=(50,180),
        xytext=(20,220),
        arrowprops={"arrowstyle":"-"}
    )
    
    ax.annotate(
        "dlžník",
        xy=(160,60),
        xytext=(190,120),
        arrowprops={"arrowstyle":"-"}
    )
    
    ekonomicke_osy(ax,r"$c$",r"$c'$")
    
    ax.set_xlim(0, rozpocet["c_max"]+50)
    ax.set_ylim(0, rozpocet["cp_max"]+20 )
    
    ax.set_xlim(0, rozpocet["c_max"]+50)
    ax.set_ylim(0, rozpocet["cp_max"]+20 )

    plt.savefig(
        "images/dynam/graf_dynamicka_mnozina.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()
    
################
# graf_dyn_kos #
################

def graf_dyn_kos():
    
    fig, axes = plt.subplots(1,2,figsize=(12,6))
    opt1=optimum(300,100,0.05)
    opt2=optimum(100,300,0.05)
    rozpocet1 = nakresli_rozpocet_intertemporal(
        axes[0],
        r=0.05,
        y=300,
        yp=100)
    rozpocet2 = nakresli_rozpocet_intertemporal(
        axes[1] ,
        r=0.05,
        y=100,
        yp=300)
    nakresli_indiferencnu_intertemporal(
        axes[0],
        u=opt1["u"] ,
        beta=opt1["beta"] ,
        c_min=100 ,
        c_max=rozpocet2["c_max"])
    nakresli_indiferencnu_intertemporal(
        axes[1],
        u=opt2["u"] ,
        beta=opt2["beta"] ,
        c_min=100 ,
        c_max=rozpocet2["c_max"])
    nakresli_bod(
        axes[0] ,
        opt1["c"] ,
        opt1["cp"],
        nazov=None,
        vodiace_ciary=True)
    nakresli_bod(
        axes[1] ,
        opt2["c"] ,
        opt2["cp"],
        nazov=None,
        vodiace_ciary=True)
    nakresli_bod(
        axes[0] ,
        300 ,
        100,
        nazov="E",
        vodiace_ciary=True)
    nakresli_bod(
        axes[1] ,
        100 ,
        300,
        nazov="E",
        vodiace_ciary=True)
    vyznac_na_osi(
        axes[0],
        x=300,
        y=100,
        popis_x=r"$y-t$",
        popis_y=r"$y'-t'$",
    )
    vyznac_na_osi(
        axes[1],
        x=100,
        y=300,
        popis_x=r"$y-t$",
        popis_y=r"$y'-t'$",
    )
    vyznac_na_osi(
        axes[0],
        x=opt1["c"] ,
        y=opt1["cp"],
        popis_x=r"$c^*$",
        popis_y=r"$c'^*$",
    )
    vyznac_na_osi(
        axes[1],
        x=opt2["c"] ,
        y=opt2["cp"],
        popis_x=r"$c^*$",
        popis_y=r"$c'^*$",
    )

    y_arrow = 20

    axes[0].annotate(
        "",
        xy=(300, y_arrow),
        xytext=(opt1["c"], y_arrow),
        arrowprops=dict(arrowstyle="<->")
    )

    axes[0].annotate(
        r"$s^*$",
        xy=((300 + opt1["c"])/2, 0),
        xytext=(0, y_arrow),
        textcoords="offset points",
        ha="center"
    )

    axes[1].annotate(
        "",
        xy=(100, y_arrow),
        xytext=(opt2["c"], y_arrow),
        arrowprops=dict(arrowstyle="<->")
    )

    axes[1].annotate(
        r"$-s^*$",
        xy=((100 + opt2["c"])/2,0),
        xytext=(0, y_arrow),
        textcoords="offset points",
        ha="center"
    )
    
    axes[0].set_title("Veriteľ")
    axes[1].set_title("Dlžník")

    for ax in axes:
        ekonomicke_osy(ax,r"$c$",r"$c'$")

    plt.savefig(
        "images/dynam/graf_dyn_kos.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

    
###################
# graf_dyn_prijem #
###################

def graf_dyn_prijem():

    fig, axes = plt.subplots(3,1, figsize=(8,18))
    # pôvodné optimá
    opt1=optimum(300,100,0.05)
    opt2=optimum(100,300,0.05)
    opt3=optimum(200,200,0.05)
    # nové optimá
    opt12=optimum(400,100,0.05)
    opt22=optimum(100,400,0.05)
    opt32=optimum(300,300,0.05)
    # medzioptimum
    m_opt=optimum(300,200,0.05)
    # pôvodné hranice
    rozpocet1 = nakresli_rozpocet_intertemporal(
        axes[0],
        r=0.05,
        y=300,
        yp=100,
        color="blue")
    rozpocet2 = nakresli_rozpocet_intertemporal(
        axes[1] ,
        r=0.05,
        y=100,
        yp=300,
        color="blue")
    rozpocet3 = nakresli_rozpocet_intertemporal(
        axes[2] ,
        r=0.05,
        y=200,
        yp=200,
        color="blue")
    # nové hranice
    rozpocet12 = nakresli_rozpocet_intertemporal(
        axes[0],
        r=0.05,
        y=400,
        yp=100,
        color="red")
    rozpocet22 = nakresli_rozpocet_intertemporal(
        axes[1] ,
        r=0.05,
        y=100,
        yp=400,
        color="red")
    rozpocet32 = nakresli_rozpocet_intertemporal(
        axes[2] ,
        r=0.05,
        y=300,
        yp=300,
        color="red")
    # medzihranica
    m_rozpocet = nakresli_rozpocet_intertemporal(
        axes[2] ,
        r=0.05,
        y=300,
        yp=200,
        color="orange",
        linestyle="--")
    # pôvodné indiferenčné krivky
    nakresli_indiferencnu_intertemporal(
        axes[0],
        u=opt1["u"] ,
        beta=opt1["beta"] ,
        c_min=83 ,
        c_max=360)
    nakresli_indiferencnu_intertemporal(
        axes[1],
        u=opt2["u"] ,
        beta=opt2["beta"] ,
        c_min=120 ,
        c_max=rozpocet22["c_max"])
    nakresli_indiferencnu_intertemporal(
        axes[2],
        u=opt3["u"] ,
        beta=opt3["beta"] ,
        c_min=70 ,
        c_max=rozpocet32["c_max"])
    # nové indiferenčné krivky
    nakresli_indiferencnu_intertemporal(
        axes[0],
        u=opt12["u"] ,
        beta=opt12["beta"] ,
        c_min=130 ,
        c_max=rozpocet12["c_max"],
        color="red")
    nakresli_indiferencnu_intertemporal(
        axes[1],
        u=opt22["u"] ,
        beta=opt22["beta"] ,
        c_min=125 ,
        c_max=rozpocet22["c_max"],
        color="red")
    nakresli_indiferencnu_intertemporal(
        axes[2],
        u=opt32["u"] ,
        beta=opt32["beta"] ,
        c_min=155 ,
        c_max=rozpocet32["c_max"],
        color="red")
    # medzi indiferenčná krivka
    nakresli_indiferencnu_intertemporal(
        axes[2],
        u=m_opt["u"] ,
        beta=m_opt["beta"] ,
        c_min=110 ,
        c_max=rozpocet32["c_max"],
        color="orange",
        linestyle="--")
    # pôvodné optimum body
    nakresli_bod(
        axes[0] ,
        opt1["c"] ,
        opt1["cp"],
        nazov=None,
        vodiace_ciary=True,
        color="blue")
    nakresli_bod(
        axes[1] ,
        opt2["c"] ,
        opt2["cp"],
        nazov=None,
        vodiace_ciary=True,
        color="blue")
    nakresli_bod(
        axes[2] ,
        opt3["c"] ,
        opt3["cp"],
        nazov=None,
        vodiace_ciary=True,
        color="blue")
    vyznac_na_osi(
        axes[0],
        x=opt1["c"] ,
        y=opt1["cp"],
        popis_x=r"$c_1^*$",
        popis_y=r"$c_1'^*$",
        color="blue")
    vyznac_na_osi(
        axes[1],
        x=opt2["c"] ,
        y=opt2["cp"],
        popis_x=r"$c_1^*$",
        popis_y=r"$c_1'^*$",
        color="blue")
    vyznac_na_osi(
        axes[2],
        x=opt3["c"] ,
        y=opt3["cp"],
        popis_x=r"$c_1^*=(y-t)_1$",
        popis_y=r"$c_1'^*=(y'-t')_1$",
        color="blue")
    #nové optimum
    nakresli_bod(
        axes[0] ,
        opt12["c"] ,
        opt12["cp"],
        nazov=None,
        vodiace_ciary=True,
        color = "red")
    nakresli_bod(
        axes[1] ,
        opt22["c"] ,
        opt22["cp"],
        nazov=None,
        vodiace_ciary=True,
        color = "red")
    nakresli_bod(
        axes[2] ,
        opt32["c"] ,
        opt32["cp"],
        nazov=None,
        vodiace_ciary=True,
        color = "red")
    vyznac_na_osi(
        axes[0],
        x=opt12["c"] ,
        y=opt12["cp"],
        popis_x=r"$c_2^*$",
        popis_y=r"$c_2'^*$",
        color="red")
    vyznac_na_osi(
        axes[1],
        x=opt22["c"] ,
        y=opt22["cp"],
        popis_x=r"$c_2^*$",
        popis_y=r"$c_2'^*$",
        color="red")
    vyznac_na_osi(
        axes[2],
        x=opt32["c"] ,
        y=opt32["cp"],
        popis_x=r"$c_2^*=(y-t)_2$",
        popis_y=r"$c_2'^*=(y'-t')_2$",
        color="red")
    # medzi optimum bod
    nakresli_bod(
        axes[2] ,
        m_opt["c"] ,
        m_opt["cp"],
        nazov=None,
        vodiace_ciary=True,
        color = "orange")

    # E2
    nakresli_bod(
        axes[0] ,
        400 ,
        100,
        nazov="$E_2$",
        vodiace_ciary=True,
        dy=3,
        color="red")
    nakresli_bod(
        axes[1] ,
        100 ,
        400,
        nazov="$E_2$",
        vodiace_ciary=True,
        color="red")
    vyznac_na_osi(
        axes[0],
        x=400,
        y=100,
        popis_x=r"$(y-t)_2$",
        popis_y=r"$y'-t'$",
        color="red")
    vyznac_na_osi(
        axes[1],
        x=100,
        y=400,
        popis_x=r"$y-t$",
        popis_y=r"$(y'-t')_2$",
        color="red")
    # E1
    nakresli_bod(
        axes[0] ,
        300 ,
        100,
        nazov="$E_1$",
        vodiace_ciary=True,
        dy=3,
        color="blue")
    nakresli_bod(
        axes[1] ,
        100 ,
        300,
        nazov="$E_1$",
        vodiace_ciary=True,
        color="blue")
    vyznac_na_osi(
        axes[0],
        x=300,
        y=100,
        popis_x=r"$(y-t)_1$",
        popis_y=r"$y'-t'$",
        color="blue")
    vyznac_na_osi(
        axes[1],
        x=100,
        y=300,
        popis_x=r"$y-t$",
        popis_y=r"$(y'-t')_1$",
        color="blue")
    # medzi E
    nakresli_bod(
        axes[2] ,
        300 ,
        200,
        nazov=None,
        vodiace_ciary=True,
        color="orange")
    
    axes[0].plot(
        [0, rozpocet12["c_max"]],
        [0, rozpocet12["c_max"]],
        color="black",
        linestyle=":",
        linewidth=1.5
    )

    axes[1].plot(
        [0, rozpocet22["c_max"]],
        [0, rozpocet22["c_max"]],
        color="black",
        linestyle=":",
        linewidth=1.5
    )

    axes[2].plot(
        [0, rozpocet32["c_max"]],
        [0, rozpocet32["c_max"]],
        color="black",
        linestyle=":",
        linewidth=1.5
    )

    for ax in axes:
        ekonomicke_osy(ax,r"$c$",r"$c'$")
        
    axes[0].set_title("Vplyv dočasného zvýšenia súčasného príjmu")
    axes[1].set_title("Vplyv očakávaného zvýšenia budúceho príjmu")
    axes[2].set_title("Vplyv trvalého zvýšenia príjmu")

    plt.savefig(
        "images/dynam/graf_dyn_prijem.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

################
# graf_zmena_r #
################

def graf_zmena_r():

    def hicksov_bod(u, r, beta):

        c = np.exp(
            (u - beta*np.log(beta*(1+r)))
            /(1+beta)
        )

        cp = beta*(1+r)*c

        return {
            "c": c,
            "cp": cp
        }

    def hicksov_rozpocet(
        ax,
        h,
        r,
        cp_max=None,
        color="orange",
        linestyle="--"
    ):

        we = h["c"] + h["cp"]/(1+r)

        if cp_max is None:
            cp_max = (1+r)*we

        c_start = ( (1+r)*we - cp_max )/(1+r)

        c = np.linspace(c_start, we, 200)

        cp = (1+r)*we - (1+r)*c
        ax.plot(
        c,
        cp,
        color=color,
        linestyle=linestyle
        )

        return {
            "c_max": we,
            "cp_max": (1+r)*we
        }

    beta = 1/(1+0.03)

    pridely = [
        (300, 100),   # veriteľ
        (100, 300)    # dlžník
    ]

    uroky = [0.001, 0.9]

    farby = {
        uroky[0]: "blue",
        uroky[1]: "red"
    }

    fig, axes = plt.subplots(2, 2, figsize=(12,12))

    for j, (y, yp) in enumerate(pridely):
        ax_hore = axes[0, j]
        ax_dole = axes[1, j]
        cp_max=(1+uroky[1])*y + yp

        for ax in [ax_hore, ax_dole]:

            c_max = 0
            for i, r in enumerate(uroky, start=1):
                opt = optimum(y,yp,r,beta=beta)      
                rozpocet = nakresli_rozpocet_intertemporal(ax,r=r,y=y,yp=yp,color=farby[r])
                cmx = rozpocet["c_max"]
                c_max = cmx if cmx>c_max else c_max
                c_min = np.exp(opt["u"] - beta*np.log(cp_max))
                nakresli_indiferencnu_intertemporal(ax,u=opt["u"],beta=beta,c_min=c_min,c_max=c_max,color=farby[r])
                nakresli_bod(ax,opt["c"],opt["cp"],color=farby[r],nazov=None,vodiace_ciary=True)
                vyznac_na_osi(ax,opt["c"],opt["cp"],color=farby[r],popis_x=fr"$c_{i}^*$",popis_y=fr"$c_{i}'^*$")

                # Hicksov rozklad iba v dolnom paneli



            nakresli_bod(ax,y,yp,nazov=None,vodiace_ciary=True)
            vyznac_na_osi(ax,y,yp,popis_x=r"$y-t$",popis_y=r"$y'-t'$")        

            ekonomicke_osy(ax,r"$c$",r"$c'$")

        u_green = optimum(y,yp,uroky[0],beta=beta)["u"]

        h = hicksov_bod(
                u_green,
                uroky[1],
                beta
            )

        hicksov_rozpocet(
            ax_dole,
            h,
            uroky[1],
            cp_max=rozpocet["cp_max"],
            color="orange",
            linestyle="--"
        )

    nakresli_bod(
                axes[1,0],
                h["c"],
                h["cp"],
                nazov=None,
                vodiace_ciary=True,
                color="orange"
            )

    nakresli_bod(
                axes[1,1],
                h["c"],
                h["cp"],
                nazov="SE",
                vodiace_ciary=True,
                color="orange"
            )

    vyznac_na_osi(
                axes[1,0],
                h["c"],
                h["cp"],
                color="orange",
                popis_x=r"$c^{SE}$",
                popis_y=r"$c'^{SE}$"
            )

    vyznac_na_osi(
                axes[1,1],
                h["c"],
                h["cp"],
                color="orange",
                popis_x=None,
                popis_y=r"$c'^{SE}$"
            )

    axes[0,0].set_title("Vplyv zvýšenia úrokovej miery na veriteľa")
    axes[0,1].set_title("Vplyv zvýšenia úrokovej miery na dlžníka")
    axes[1,0].set_title("Rozklad zvýšenia úrokovej miery podľa Hicksa: veriteľ")
    axes[1,1].set_title("dlžník")

    plt.savefig(
        "images/dynam/graf_zmena_r.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()

graf_dynamicka_mnozina()
graf_dyn_kos()
graf_dyn_prijem()
graf_zmena_r()
```

## Teória všeobecnej rovnováhy

Pracovný priečinok je `rovnovaha`.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from matplotlib.patches import Arc, FancyArrowPatch
# from itertools import product

Path("images/rovnovaha").mkdir(exist_ok=True)

plt.rcParams["figure.figsize"] = (10, 10)

def podiel_cien(𝛼A,𝛼B,y1A,y2A,y1B,y2B): # podiel cien riešenie v tvare p1/p2 (sklon ir)
        return (𝛼A*y2A + 𝛼B*y2B) / ((1-𝛼A)*y1A + (1-𝛼B)*y1B)

def x(e1,e2,y,cena): # Marshallove dopytove funkice
        return e1*y/(e1+e2)/cena

def ir(p,y1,y2):
        return p*y1 + y2 - p*x1

# zadanie: 10 jednotiek statku 1 a 10 statku 2
𝛼A = 1/2 # elasticita vo funkcii užitočnosti spotrebiteľa A
𝛼B = 3/4 # elasticita vo funkcii užitočnosti spotrebiteľa B

####################
# hranicny_diagram #
####################

def hranicny_diagram():

    y1A = 6.
    y2A = 5.
    y1B = 10.-y1A
    y2B = 10.-y2A

    p = podiel_cien(𝛼A,𝛼B,y1A,y2A,y1B,y2B) # výpočet podielu cien

    yA = p*y1A+y2A # peňažne vyjadrený príjem spotrebiteľa A
    yB = p*y1B+y2B # peňažne vyjadrený príjem spotrebiteľa B

    # Spotrebiteľské koše oboch spotrebiteľov
    x1A = x(𝛼A,1-𝛼A,yA,p)
    x2A = x(1-𝛼A,𝛼A,yA,1)
    x1B = x(𝛼B,1-𝛼B,yB,p)
    x2B = x(1-𝛼B,𝛼B,yB,1)

    # hodnoty užitočností
    uA = x1A**𝛼A*x2A**(1-𝛼A)
    uB = x1B**𝛼B*x2B**(1-𝛼B)

    prem = np.linspace(1e-12,10-1e-12,100) # definičný obor indiferenčných kriviek a hranice spotrebiteľských možností

    fig, ax = plt.subplots()

    ax.text(10.4,5,'druhý statok', rotation=270)
    ax.text(5,10.5,'prvý statok', rotation=180)

    for i in range(0,11,2):
        ax.text(10.2,10-i,i)
        ax.text(10-i,10.2,i)
        ax.plot([9.95,10.],[10-i,10-i],color='black', linewidth=.7)
        ax.plot([10-i,10-i],[9.95,10.],color='black', linewidth=.7)

    # zobraz hranice možností oboch spotrebiteľov
    ax.plot(prem, p*y1A + y2A - p*prem, label = 'Hranica spotrebiteľských možností\n je pre oboch spotrebiteľov rovnaká')

    # zobraz indiferenčnú krivku A
    ax.plot(prem, uA**(1/(1-𝛼A)) / prem**(𝛼A/(1-𝛼A)), label = 'Indiferenčná krivka spotrebiteľa $A$')

    # zobraz indiferenčnú krivku B
    ax.plot(prem, 10 - uB**(1/(1-𝛼B)) / (10-prem)**(𝛼B/(1-𝛼B)), label = 'indiferenčná krivka spotrebiteľa $B$')

    # príjem:
    ax.scatter(y1A,y2A,marker='x',label='Prídelové koše')
    ax.plot([y1A,y1A],[0,10],':',color='black')
    ax.plot([0,10],[y2A,y2A],':',color='black')

    # spotrebiteľské koše:
    ax.scatter(x1A,x2A,marker='x',label='Spotrebiteľské koše')
    ax.plot([x1A,x1A],[0,10],':',color='black')
    ax.plot([0,10],[x2A,x2A],':',color='black')

    ax.legend()

    ax.set(xlabel = 'prvý statok', ylabel = 'druhý statok',ylim = (0,10), xlim=(0,10))

    plt.savefig(
        "images/rovnovaha/hranicny_diagram.png",
        dpi=300,
        bbox_inches="tight"
    )
    plt.close()    
    
##########################
# graf_pridely_rovnovahy #
##########################

def graf_pridely_rovnovahy():

    def zobraz_rovnovahu(y1A,y2A,text_Y,text_X):
        y1B = 10 - y1A
        y2B = 10 - y2A

        p = podiel_cien(𝛼A,𝛼B,y1A,y2A,y1B,y2B) # výpočet podielu cien

        yA = p*y1A+y2A # peňažne vyjadrený príjem spotrebiteľa A
        yB = p*y1B+y2B # peňažne vyjadrený príjem spotrebiteľa B

        # Spotrebiteľské koše oboch spotrebiteľov
        x1A = x(𝛼A,1-𝛼A,yA,p)
        x2A = x(1-𝛼A,𝛼A,yA,1)
        x1B = x(𝛼B,1-𝛼B,yB,p)
        x2B = x(1-𝛼B,𝛼B,yB,1)

        # hodnoty užitočností
        uA = x1A**𝛼A*x2A**(1-𝛼A)
        uB = x1B**𝛼B*x2B**(1-𝛼B)

        # zobraz hranice možností oboch spotrebiteľov 
        ax.plot(prem, p*y1A + y2A - p*prem, color="green",linestyle="--")

        def_o = np.linspace(max(1e-12,x1A-3.), min(10-1e-12,x1A+3),100)

        # zobraz indiferenčnú krivku A
        ax.plot(def_o, uA**(1/(1-𝛼A)) / def_o**(𝛼A/(1-𝛼A)), color="blue",linestyle="--")

        # zobraz indiferenčnú krivku B
        ax.plot(def_o, 10 - uB**(1/(1-𝛼B)) / (10-def_o)**(𝛼B/(1-𝛼B)), color="orange",linestyle="--")

        # príjem:
        ax.scatter(y1A,y2A,marker='x',label=text_Y)    

        # spotreba:
        ax.scatter(x1A,x2A,marker='x')
        ax.text(x1A+0.1,x2A-0.1,text_X)

    fig, ax = plt.subplots()

    y_bar = 10

    xA1_cc = np.linspace(0.01,9.99,500)
    xA2_cc = 15*xA1_cc/(5+xA1_cc) # kontraktačná krivka pre hodnoty  prvého obrázka

    prem = np.linspace(1e-12,10-1e-12,100) # definičný obor hranice spotrebiteľských možností

    ax.text(10.4,5,'druhý statok', rotation=270)
    ax.text(5,10.5,'prvý statok', rotation=180)

    ax.plot(xA1_cc,xA2_cc,label="Kontraktačná krivka",
             color='red')

    for i in range(0,11,2):
        ax.text(10.2,10-i,i)
        ax.text(10-i,10.2,i)
        ax.plot([9.95,10.],[10-i,10-i],color='black', linewidth=.7)
        ax.plot([10-i,10-i],[9.95,10.],color='black', linewidth=.7)

    zobraz_rovnovahu(2,2,r'$\mathbf{{Y^1}}$',r'$\mathbf{{X^1}}$')
    zobraz_rovnovahu(6,5,r'$\mathbf{{Y^2}}$',r'$\mathbf{{X^2}}$')
    zobraz_rovnovahu(8,8,r'$\mathbf{{Y^3}}$',r'$\mathbf{{X^3}}$')

    ax.legend(loc='lower right', framealpha=0.3)

    ax.set(xlabel = 'prvý statok', ylabel = 'druhý statok',ylim = (0,10), xlim=(0,10))
    plt.savefig(
        "images/rovnovaha/graf_pridely_rovnovahy.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()    

hranicny_diagram()
graf_pridely_rovnovahy()
```

## Statický model všeobecnej rovnováhy

Grafy sa vygenerujú v podpriečinku `static_model`.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from scipy.optimize import fsolve

Path("images/static_model").mkdir(exist_ok=True)


plt.rcParams["figure.figsize"] = (8,6)
plt.rcParams["figure.dpi"] = 120

plt.rcParams["font.size"] = 12

plt.rcParams["lines.linewidth"] = 2

plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

def deep_parameters():
    alpha = 1/3 # podiel kapitálu na HDP
    phi = 1/2 # elasticita užitočnosti podľa voľného času
    g_share = 0.2 # podiel vládnych nákupov na HDP
    h = 1 + phi*(1-g_share)/((1-phi)*(1-alpha)) # voľný čas kalibrovaný tak, aby Y=N=A=1
    return alpha, phi, g_share, h

def deep_parameters_linear():
    alpha = 0 # podiel kapitálu na HDP
    phi = 1/2 # elasticita užitočnosti podľa voľného času
    g_share = 0.6 # podiel vládnych nákupov na HDP
    h = 1 + phi*(1-g_share)/((1-phi)*(1-alpha)) # voľný čas kalibrovaný tak, aby Y=N=A=1
    return alpha, phi, g_share, h

alpha, phi, g_share, h = deep_parameters()
A=1.
G=g_share
pi=alpha
T=g_share
pi_T = pi-T

def volny_cas(w, phi=phi, h=h, pi_T=pi_T):
    """
    Optimálny voľný čas
    """
    return phi*h + phi* pi_T/w


def spotreba(w, phi=phi, h=h, pi_T=pi_T):
    """
    Optimálna spotreba
    """
    return (1-phi)*(w*h +  pi_T)


def ponuka_prace(w, phi=phi, h=h, pi_T=pi_T):
    """
    Ponuka práce
    """
    l = volny_cas(w, phi, h, pi_T=pi_T)
    return h - l

def produkcna_funkcia(N,A=A,alpha=alpha):
    """
    Produkčná funkcia
    """
    return A*N**(1-alpha)

def MPN(N,A=A,alpha=alpha):
    """
    Hraničný produkt práce
    """
    return (1-alpha)*A*N**(-alpha)

def ekonomicke_osi(ax, xlabel, ylabel,
                   xmin=0, ymin=0):

    ax.set_xticks([])
    ax.set_yticks([])

    ax.set_xlim(left=xmin)
    ax.set_ylim(bottom=ymin)

    ax.margins(0)

    ax.text(
        1.02, -0.02, xlabel,
        transform=ax.transAxes,
        ha="right",
        va="top"
    )

    ax.text(
        -0.02, 1.02, ylabel,
        transform=ax.transAxes,
        ha="right",
        va="top"
    )

    return ax


def nakresli_ponuku_prace(ax, phi=phi, h=h, pi_T=pi_T,
                          color="green", linestyle="-",label = "$N^s$"
                         ):
   
    
    w_grid = np.linspace(0.2,1.,500)
    N = ponuka_prace(w_grid)
    ax.plot(N,w_grid,
        color=color,linestyle=linestyle,label=label
    )

def nakresli_dopyt_po_praci(ax, A=A, alpha=alpha,
                            color="blue", linestyle="-",label=r"$N^d$"):
    N_grid = np.linspace(0.7, 1.3, 500)
    w = MPN(N_grid, A, alpha)

    ax.plot(
        N_grid, w,
        color=color,
        linestyle=linestyle,
        label=label
    )

def nakresli_bod(ax,x,y,nazov=None,vodiace_ciary=False,dx=5,dy=5,
                 color="black"
                ):

    ax.scatter(
        [x],
        [y],
        color=color,
        zorder=10
    )

    ax.annotate(
        nazov,
        xy=(x,y),
        xytext=(dx,dy),
        textcoords="offset points"
    )

    if vodiace_ciary:

        ax.plot(
            [0,x],
            [y,y],
            "--",
            color="gray",
            linewidth=1
        )

        ax.plot(
            [x,x],
            [0,y],
            "--",
            color="gray",
            linewidth=1
        )

def vyznac_na_osi(ax,x=None,y=None,popis_x=None,popis_y=None,color="black"):
    """
    Označí hodnoty na osiach.  
    """

    if x is not None and popis_x is not None:

        ax.annotate(popis_x,xy=(x, 0),xytext=(0, -15),textcoords="offset points",ha="center",color=color
        )

    if y is not None and popis_y is not None:

        ax.annotate(popis_y,
            xy=(0, y),
            xytext=(-10, 0),
            textcoords="offset points",
            ha="right",
            va="center",
            color=color
        )

def nakresli_produkcnu_f(ax,A=A,alpha=alpha,h=h,
                            color="green", linestyle="-"):
    N_grid = np.linspace(0,h,500)
    Y = produkcna_funkcia(N_grid,A=A,alpha=alpha)
    ax.plot(N_grid, Y,color=color,linestyle=linestyle)

#Ako zobraziť dotyčnice ku krivke?
#Dotyčnica je priamka y = a + bx
#sklon dotyčnice je prvá derivácia g(x): b = g'(x)
#Dotyčnica musí prechádzať bodom dotyku, v ktorom platí: g(x) = a + bx
#Teda: g(x) = a + g'(x) x
#Takže a = g(x) - g'(x) x

def dotycnica(g,gder,x):
    """
    Príklad: dotycnica(produkcna_funkcia,MPN,0.5)

    """
    b = gder(x) #sklon dotyčnice
    a = g(x) - gder(x)*x #intercept dotyčnice
    return a,b

def nakresli_dotycnicu(ax,g,gder,bod,rozsah,
                      color="black",linestyle="-",label=None):
    a,b = dotycnica(g,gder,bod)
    x = np.linspace(bod-rozsah/2,bod+rozsah/2,500)
    y=a+b*x
    ax.plot(x,y,color=color,linestyle=linestyle,label=label)


def ekonomicke_osi_HVM(ax, xlabel, ylabel,G=g_share):

    ax.set_xticks([])
    ax.set_yticks([])

    # schovaj hornú a pravú os
    ax.spines["top"].set_visible(False)
    ax.spines["right"].set_visible(False)

    # x-os presuň na y=0
    ax.spines["bottom"].set_position(("data", 0))

    # y-os nech zostane v x=0
    ax.spines["left"].set_position(("data", 0))

    # názov osi x na koniec osi
    ax.annotate(xlabel,xy=(ax.get_xlim()[1], 0),xytext=(6, -4),
                textcoords="offset points",ha="left",va="top")
    ax.set_ylim(bottom=-G-0.05)
    ax.set_xlim(left=0)
    # názov osi y hore
    ax.text(-0.02,1.00,ylabel,transform=ax.transAxes,ha="right",va="top")
    

    return ax

def HVM(l,A=A,alpha=alpha,h=h,G=g_share):
    return A*(h-l)**(1-alpha) - G
# táto funkca poslúži pre dotyčnicu v prostrednom 
def HVM0(l,A=A,alpha=alpha,h=h): # HVM s nulovými vládnymi nákupmi
    return HVM(l,A=A,alpha=alpha,h=h,G=0)
def HVM_der(l,A=A,alpha=alpha,h=h):
    return -MPN(h-l,A=A,alpha=alpha)

def nakresli_HVM(ax,
                 A=A, alpha=alpha, h=h, G=g_share,
                 l_max=None,
                 color="green",
                 linestyle="-",
                 prerus_pre_C_zaporne=False):

    if l_max is None:
        l_max = h

    l_grid = np.linspace(0, l_max, 500)

    C = HVM(l_grid,A=A,alpha=alpha,h=h,G=G)

    if not prerus_pre_C_zaporne:

        ax.plot(l_grid, C,
            color=color,linestyle=linestyle
        )

    else:

        kladne = C >= 0
        zaporne = C < 0

        ax.plot(l_grid[kladne],C[kladne],
            color=color,linestyle="-"
        )

        ax.plot(
            l_grid[zaporne],C[zaporne],
            color=color,linestyle="--"
        )

def U_lC(l=h-1,C=1-g_share,phi=phi):
    return l**phi * C**(1-phi)

def nakresli_indiferencnu_lC(ax, U, phi=phi, h=h, l_min=0.64, l_max=h,
                             color="blue",linestyle="-",label=None):

    l_grid = np.linspace(l_min, l_max, 500)

    C = (U / l_grid**phi)**(1/(1-phi))

    ax.plot(l_grid,C,
        color=color,linestyle=linestyle,label=label)

def konkurencna_rovnovaha(phi=phi,h=h,G=g_share,A=A,alpha=alpha):

    def vr(endog):
        l,C,Y,N,w = endog
        return(C*phi/(l*(1-phi)) - A*(1-alpha)*(h-l)**(-alpha), #MRS=MRT
              C - A*(h-l)**(1-alpha) + G, #PPF
              Y-C-G,# Y
              N-h+l,
              w-(1-alpha)*A*(h-l)**(-alpha))

    start = (h-1,1-G,1.,1.,1-alpha)
    
    x = fsolve(vr,start)
    return x

def hicks_SE(A,U,
             phi=phi,h=h,alpha=alpha):

    def rovnica(l):

        C = (U/l**phi)**(1/(1-phi))

        return (
            phi*C/(l*(1-phi))
            - (1-alpha)*A*(h-l)**(-alpha)
        )

    l_SE = fsolve(rovnica,h-1)[0]

    C_SE = (U/l_SE**phi)**(1/(1-phi))

    return l_SE,C_SE
#----------------------------------------------------------------------------

###################################
# graf_rovnovaha_decentralizovana #
####################################

def graf_rovnovaha_decentralizovana():

    fig, ax = plt.subplots()
    
    nakresli_ponuku_prace(ax)
    nakresli_dopyt_po_praci(ax)
    ekonomicke_osi(ax,"$N$","$w$")
    vyznac_na_osi(ax,x=1.,y=None,popis_x="$N^{s*}$",popis_y=None)
    
    x_min = 0.65
    x_max = 1. - x_min + 1.
    
    ax.annotate("$w^*$",xy=(x_min, 2/3),
                xytext=(-10, 0),
                textcoords="offset points",
                ha="right",
                va="center",
                color="black"
            )
    
    nakresli_bod(ax,1.,2/3,vodiace_ciary=True)
    
    ### ohraničenia osí
    
    ax.set_xlim(x_min, x_max)
    
    plt.legend()
    plt.tight_layout()
    
    plt.savefig(
        "images/static_model/graf_rovnovaha_decentralizovana.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#########################
# graf_vyrobne_moznosti #
#########################

def graf_vyrobne_moznosti():

    #------------
    # horný graf
    #------------
    
    bod_dotycnice = 0.5
    rozsah_dotycnice = 0.8
    
    fig, axes = plt.subplots(3,1,figsize=(8,18))
    
    nakresli_produkcnu_f(axes[0])
    nakresli_dotycnicu(axes[0],produkcna_funkcia,MPN,bod_dotycnice,rozsah_dotycnice,
                      label = "sklon=$MPN$")
    axes[0].axvline(h,color="gray",linestyle="--",linewidth=1.0) #-- čiara h
    
    #### množina výrobných možostí svetlou zelenou
    N_grid = np.linspace(0,h,500)
    Y = produkcna_funkcia(N_grid)
    axes[0].fill_between(N_grid,0,Y,
                         color="lightgreen",alpha=0.4)
    vyznac_na_osi(axes[0],x=h,y=0.,popis_x="$h$",popis_y=None,color="black")
    
    axes[0].set_title("Produkčná funkcia")
    
    #------------
    #stredný graf
    #------------
    
    nakresli_HVM(axes[1],G=0)
    bod_dotycnice_HVM = h-bod_dotycnice
    nakresli_dotycnicu(axes[1],HVM0,HVM_der,h-bod_dotycnice,rozsah_dotycnice,
                      label = "$-$sklon=$MPN$")
    
    #### množina výrobných možostí svetlou zelenou
    l_grid = np.linspace(0,h,500)
    Y = HVM(l_grid,G=0)
    axes[1].fill_between(l_grid,0,Y,
                         color="lightgreen",alpha=0.4)
    vyznac_na_osi(axes[1],x=h,y=0.,popis_x="$h$",popis_y=None,color="black")
    
    axes[1].set_title("Výstup ako funkcia $\ell$")
    
    #------------
    #spodný graf
    #------------
    
    nakresli_HVM(axes[2],prerus_pre_C_zaporne=True)
    bod_dotycnice_HVM = h-bod_dotycnice
    nakresli_dotycnicu(axes[2],HVM,HVM_der,h-bod_dotycnice,rozsah_dotycnice,
                      label = "$-$sklon=$MPN$")
    
    #### množina výrobných možostí svetlou zelenou
    l_C0 = h - (G/A)**(1/(1-alpha))
    l_grid = np.linspace(0,l_C0,500)
    Y = HVM(l_grid,)
    
    axes[2].fill_between(l_grid,0,Y,
                         color="lightgreen",alpha=0.4)
    
    vyznac_na_osi(axes[2],y=-g_share,popis_y="$-G$")
    nakresli_bod(axes[2],h,-g_share,nazov="$H$",vodiace_ciary=True,dy=-3)
    vyznac_na_osi(axes[2],x=h+0.05,popis_x="$h$",color="black")
      
    ekonomicke_osi(axes[0],"$N$","$Y$")
    ekonomicke_osi(axes[1],"$\ell$","$Y$")
    ekonomicke_osi_HVM(axes[2],"$\ell$","$C$")
    
    axes[2].set_title("Množina a hranica výrobných možností")

    for ax in axes:
        ax.legend()

    plt.savefig(
        "images/static_model/graf_vyrobne_moznosti.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#####################
# graf_rovnovaha_lc #
#####################

def graf_rovnovaha_lc():
    U_opt = U_lC()
    
    fig,ax= plt.subplots()
    
    nakresli_HVM(ax,prerus_pre_C_zaporne=True)
    nakresli_indiferencnu_lC(ax,U_opt)
    
    vyznac_na_osi(ax,y=-g_share,popis_y="$-G$")
    nakresli_bod(ax,h,-g_share,nazov="$H$",vodiace_ciary=True,dy=-3)
    vyznac_na_osi(ax,x=h+0.05,popis_x="$h$",color="black")
    nakresli_dotycnicu(ax,HVM,HVM_der,h-1,2.,
                      label = r"$-\,\mathrm{sklon}\,=w^*$",color="orange")
    vyznac_na_osi(ax,x=h-1,y=1-g_share,popis_x="$l^*$",popis_y="$C^*$")
    nakresli_bod(ax,h-1,1-g_share,vodiace_ciary=True,nazov="$[l^*,C^*]$")
    vyznac_na_osi(ax,x=None,y=alpha-g_share,popis_y="$\pi-T$")
    nakresli_bod(ax,h,alpha-g_share,vodiace_ciary=True,nazov="$[h,\pi-T]$")
    
    # Y* medzi C* a -G
    ax.annotate(
        "",
        xy=(0.05, 1-g_share),
        xytext=(0.05, -g_share),
        arrowprops=dict(arrowstyle="<->",alpha=0.3)
    )
    
    ax.text(
        0.06,
        (1-g_share-g_share)/2,
        "$Y^*$",
        va="center"
    )
    
    # N* medzi l* a h
    
    ax.annotate(
        "",
        xy=(h-1, 0.03),
        xytext=(h, 0.03),
        arrowprops=dict(arrowstyle="<->",alpha=0.3)
    )
    
    ax.text(
        h-0.5,
        0.05,
        "$N^*$",
        ha="center"
    )
    
    ekonomicke_osi_HVM(ax,"$\ell$","$C$")
    
    ax.legend()

    plt.savefig(
        "images/static_model/graf_rovnovaha_lc.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#######################
# graf_rovnovaha_lc_G #
#######################

def graf_rovnovaha_lc_G():

    # zmena G
    
    G_nove = 0.6 # Nová hodnota vládnyc nákupov
    
    U_opt = U_lC()
    
    fig,ax= plt.subplots()
    
    nakresli_HVM(ax,prerus_pre_C_zaporne=True,color="blue")
    nakresli_indiferencnu_lC(ax,U_opt)
    
    vyznac_na_osi(ax,y=-g_share,popis_y="$-G_1$") 
    nakresli_bod(ax,h,-g_share,nazov="$H_1$",vodiace_ciary=True,dy=-3)
    vyznac_na_osi(ax,x=h+0.05,popis_x="$h$",color="black")
    vyznac_na_osi(ax,x=h-1,y=1-g_share,popis_x="$l^*_1$",popis_y="$C^*_1$",color="blue")
    nakresli_bod(ax,h-1,1-g_share,vodiace_ciary=True,nazov="A")
    
    x_G = konkurencna_rovnovaha(G=G_nove)
    
    l_G, C_G = x_G[0], x_G[1]
    
    U_G = U_lC(l=l_G,C=C_G)
    
    nakresli_HVM(ax,G=G_nove,prerus_pre_C_zaporne=True,color="red")
    nakresli_indiferencnu_lC(ax,U_G,l_min=0.4 ,color="red")
    
    vyznac_na_osi(ax,y=-G_nove,popis_y="$-G_2$") 
    nakresli_bod(ax,h,-G_nove,nazov="$H_2$",vodiace_ciary=True,dy=-3)
    vyznac_na_osi(ax,x=l_G,y=C_G,popis_x="$l^*_2$",popis_y="$C^*_2$",color="red")
    nakresli_bod(ax,l_G,C_G,vodiace_ciary=True,nazov="B")
    
    ekonomicke_osi_HVM(ax,"$\ell$","$C$",G=G_nove)

    plt.savefig(
        "images/static_model/graf_rovnovaha_lc_G.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#######################
# graf_rovnovaha_lc_A #
#######################

def graf_rovnovaha_lc_A():

    # zmena A
    
    A_nove = 1.7 # Nová hodnota A
    
    U_opt = U_lC()
    
    fig,axes= plt.subplots(2,1,figsize=(8,12))
    
    for ax in axes:
    
        nakresli_HVM(ax,prerus_pre_C_zaporne=True,color="blue")
        nakresli_indiferencnu_lC(ax,U_opt,l_min=0.37)
        
        vyznac_na_osi(ax,y=-g_share,popis_y="$-G$") 
        nakresli_bod(ax,h,-g_share,vodiace_ciary=True,dy=-3)
        vyznac_na_osi(ax,x=h+0.05,popis_x="$h$",color="black")
        vyznac_na_osi(ax,x=h-1,y=1-g_share,popis_x="$l^*_1$",popis_y="$C^*_1$",color="blue")
        nakresli_bod(ax,h-1,1-g_share,vodiace_ciary=True,nazov="A",dx=-3)
        
        x_A = konkurencna_rovnovaha(A=A_nove)
        
        l_A, C_A = x_A[0], x_A[1]
        
        U_A = U_lC(l=l_A,C=C_A)
        
        nakresli_HVM(ax,A=A_nove,prerus_pre_C_zaporne=True,color="red")
        nakresli_indiferencnu_lC(ax,U_A,l_min=0.7 ,color="red")
        
        vyznac_na_osi(ax,x=l_A,y=C_A,popis_x="$l^*_2$",popis_y="$C^*_2$",color="red")
        nakresli_bod(ax,l_A,C_A,vodiace_ciary=True,nazov="B")
        ekonomicke_osi_HVM(ax,"$\ell$","$C$")
    
    #----------------
    # Hicks
    #-------------
    
    l_SE,C_SE = hicks_SE(A_nove,U_opt)
    
    nakresli_bod(axes[1],l_SE,C_SE,vodiace_ciary=True,nazov="SE")
    
    vyznac_na_osi(axes[1],x=l_SE,y=C_SE,popis_x="$l^{SE}$",popis_y="$C^{SE}$",color="orange")
    
    G_SE = (A_nove*(h-l_SE)**(1-alpha)- C_SE)
    
    nakresli_HVM(axes[1],A=A_nove,G=G_SE,l_max=1.9,color="orange",linestyle="--")

    plt.savefig(
        "images/static_model/graf_rovnovaha_lc_A.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

####################
# graf_sadzba_dane #
####################

def graf_sadzba_dane():

    alpha_l, phi_l, g_share_l, h_l=deep_parameters_linear()
    
    U_opt = U_lC(C=0.4,l=0.4)
    
    fig,ax= plt.subplots()
    
    nakresli_HVM(ax,A=A,alpha=alpha_l,h=h_l,G=g_share_l,prerus_pre_C_zaporne=True,color="blue")
    nakresli_indiferencnu_lC(ax,U_opt,h=h_l,l_min=0.2,l_max=h_l)
    
    vyznac_na_osi(ax,y=-g_share_l,popis_y="$-G_1$") 
    nakresli_bod(ax,h_l,-g_share_l,nazov="$H$",vodiace_ciary=True,dy=-3)
    vyznac_na_osi(ax,x=h_l+0.05,popis_x="$h$",color="black")
    vyznac_na_osi(ax,x=h_l-1,y=1-g_share_l,popis_x="$l^*_1$",popis_y="$C^*_1$",color="blue")
    nakresli_bod(ax,h_l-1,1-g_share_l,vodiace_ciary=True,nazov="A")
    
    tau = 2*g_share_l / A / h_l
    nakresli_HVM(ax,A=A*(1-tau),alpha=alpha_l,h=h_l,G=0,color="red")
    
    l_SE = h_l/2
    C_SE = A*(1-tau)*(h_l-l_SE)
    U_SE =U_lC(l_SE,C_SE)
    
    vyznac_na_osi(ax,x=l_SE,y=C_SE,popis_x="$l^*_2$",popis_y="$C^*_2$",color="red")
    
    nakresli_bod(ax,l_SE,C_SE,vodiace_ciary=True,nazov="B")
    
    nakresli_indiferencnu_lC(ax,U_SE,h=h_l,l_min=0.09,l_max=h_l,color="red")
    
    ekonomicke_osi_HVM(ax,"$\ell$","$C$",G=g_share_l)

    plt.savefig(
        "images/static_model/graf_sadzba_dane.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

####################
# lafferova_krivka #
####################

def lafferova_krivka():

    A_hodnota = 2.
    τ_hodnoty = np.linspace(0.,1.,500)
    T_hodnoty = A_hodnota**2 * (τ_hodnoty - τ_hodnoty**2)
    
    fig, ax = plt.subplots()
    ax.plot(τ_hodnoty,T_hodnoty)
    ekonomicke_osi(ax,"$\\tau$","$T$")
    vyznac_na_osi(ax, x=0.5, popis_x=r"$\tau_k$")
    nakresli_bod(ax,0.5,A_hodnota**2 * (0.5 - 0.5**2),vodiace_ciary=True)

    plt.savefig(
        "images/static_model/lafferova_krivka.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

graf_rovnovaha_decentralizovana()
graf_vyrobne_moznosti()
graf_rovnovaha_lc()
graf_rovnovaha_lc_G()
graf_rovnovaha_lc_A()
graf_sadzba_dane()
lafferova_krivka()
```

## Model prídelovej ekonomiky a nedokonalosti na finančných trhoch

Grafy sa vygenerujú v podpriečinku `finan`.

```python
import matplotlib.pyplot as plt
import numpy as np
from pathlib import Path
from scipy.optimize import fsolve

Path("images/finan").mkdir(exist_ok=True)


plt.rcParams["figure.figsize"] = (8,6)
plt.rcParams["figure.dpi"] = 120

plt.rcParams["font.size"] = 12

plt.rcParams["lines.linewidth"] = 2

plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False

####################
# Potrebné funkcie #
####################

def v(c):
    return np.log(c)


def uzitocnost(c, cp, beta=None, r=None):

    if beta is None:
        beta = 1/(1+r)

    return v(c) + beta*v(cp)

def optimum(y, yp, r, t=0, tp=0, beta=None):

    if beta is None:
        beta = 1/(1+r)

    we = y - t + (yp - tp)/(1+r)

    c = we/(1+beta)

    cp = beta*(1+r)*c

    u = uzitocnost(c, cp, beta)

    return {
        "we": we,
        "c": c,
        "cp": cp,
        "u": u,
        "beta": beta
    }


def ekonomicke_osi(ax, xlabel, ylabel,
                   xmin=0, ymin=0):

    ax.set_xticks([])
    ax.set_yticks([])

    ax.set_xlim(left=xmin)
    ax.set_ylim(bottom=ymin)

    ax.margins(0)

    ax.text(
        1.02, -0.02, xlabel,
        transform=ax.transAxes,
        ha="right",
        va="top"
    )

    ax.text(
        -0.02, 1.02, ylabel,
        transform=ax.transAxes,
        ha="right",
        va="top"
    )

    return ax

def nakresli_rozpocet_intertemporal(
    ax,
    r,
    y,
    yp,
    t=0,
    tp=0,
    color="green",
    linestyle="-"
):

    y1 = y - t
    y2 = yp - tp

    c_max = y1 + y2/(1+r)
    cp_max = (1+r)*y1 + y2

    c = np.linspace(0, c_max, 200)

    cp = cp_max - (1+r)*c

    ax.plot(
        c,
        cp,
        color=color,
        linestyle=linestyle
    )

    return {
        "y1": y1,
        "y2": y2,
        "c_max": c_max,
        "cp_max": cp_max
    }

def nakresli_indiferencnu_intertemporal(
    ax,
    u,
    beta,
    c_min=1,
    c_max=100,
    color="blue",
    linestyle="-"
):

    c = np.linspace(c_min, c_max, 500)

    cp = np.exp((u - np.log(c))/beta)

    ax.plot(
        c,
        cp,
        color=color,
        linestyle=linestyle
    )

    return {"u": u}


def nakresli_bod(
    ax,
    x,
    y,
    nazov=None,
    vodiace_ciary=False,
    dx=5,
    dy=5,
    color="black"
):

    ax.scatter(
        [x],
        [y],
        color=color,
        zorder=10
    )

    ax.annotate(
        nazov,
        xy=(x,y),
        xytext=(dx,dy),
        textcoords="offset points"
    )

    if vodiace_ciary:

        ax.plot(
            [0,x],
            [y,y],
            "--",
            color="gray",
            linewidth=1
        )

        ax.plot(
            [x,x],
            [0,y],
            "--",
            color="gray",
            linewidth=1
        )

    return {"x":x,"y":y}

def vyznac_na_osi(
    ax,
    x=None,
    y=None,
    popis_x=None,
    popis_y=None,
    color="black"
):
    """
    Označí hodnoty na osiach.
   
    """

    if x is not None and popis_x is not None:

        ax.annotate(
            popis_x,
            xy=(x, 0),
            xytext=(0, -15),
            textcoords="offset points",
            ha="center",
            color=color
        )

    if y is not None and popis_y is not None:

        ax.annotate(
            popis_y,
            xy=(0, y),
            xytext=(-10, 0),
            textcoords="offset points",
            ha="right",
            va="center",
            color=color
        )

def nakresli_sipku(
    ax,
    x1,y1,
    x2,y2,
    color="black"
):

    ax.annotate(
        "",
        xy=(x2,y2),
        xytext=(x1,y1),
        arrowprops=dict(
            arrowstyle="->",
            color=color
        )
    )

def nakresli_rozpocet_klin(
    ax,
    r1,
    r2,
    y,
    yp,
    t=0,
    tp=0,
    color="green",
    linestyle="-",
    aux_linestyle="--"
):
    
    y1 = y - t
    y2 = yp - tp

    # maximálny rozsah grafu
    c_max = y1 + y2/(1+r1)
    cp_max = (1+r1)*y1 + y2

    c = np.linspace(0, c_max, 500)

    # obe rozpočtové priamky cez prídelový bod E=(y1,y2)
    cp_r1 = y2 + (1+r1)*(y1 - c)
    cp_r2 = y2 + (1+r2)*(y1 - c)

    lavy = c <= y1
    pravy = c >= y1

    # ľavá časť: r1 plná
    ax.plot(
        c[lavy],
        cp_r1[lavy],
        color=color,
        linestyle=linestyle
    )

    ax.plot(
        c[lavy],
        cp_r2[lavy],
        color=color,
        linestyle=aux_linestyle,
        linewidth=1.0
    )

    # pravá časť: r2 plná
    ax.plot(
        c[pravy],
        cp_r2[pravy],
        color=color,
        linestyle=linestyle
    )

    ax.plot(
        c[pravy],
        cp_r1[pravy],
        color=color,
        linestyle=aux_linestyle,
        linewidth=1.0
    )

    return {
        "E": (y1, y2),
        "c_max": c_max,
        "cp_max": cp_max
    }
def nakresli_rozpocet_kolateral(
    ax,
    r,
    y,
    yp,
    B,
    t=0,
    tp=0,
    color="green"
):

    """
    B je pH/(1+r)
    """

    y1 = y - t
    y2 = yp - tp + B*(1+r)

    c_grid = np.linspace(0, y1 + B, 300)

    cp_grid = (
        (1+r)*y1
        + y2
        - (1+r)*c_grid
    )

    ax.plot(
        c_grid,
        cp_grid,
        color=color
    )

    # bod zlomu G

    c_G = y1 + B

    cp_G = yp-tp

    # vertikála GD

    ax.plot(
        [c_G, c_G],
        [cp_G, 0],
        color=color
    )


    return {
        "E": (y1, y2),
        "G": (c_G, cp_G)
    }

################
# graf_ricardo #
#################

def graf_ricardo():

    fig,ax=plt.subplots()
    
    opt=optimum(100,300,0.05)
    rozpocet = nakresli_rozpocet_intertemporal(ax ,r=0.05,y=100,yp=300)
    nakresli_indiferencnu_intertemporal(ax,u=opt["u"],beta=opt["beta"],c_min=100,c_max=rozpocet["c_max"])
    nakresli_bod(ax,opt["c"],opt["cp"],nazov=None,vodiace_ciary=True)
    nakresli_bod(ax,100,300,nazov=r"$E_2$",vodiace_ciary=True,dx=-0.01,dy=-20)
    nakresli_bod(ax,150,247.5,r"$E_1$",vodiace_ciary=True,dx=-0.01,dy=-20)
    vyznac_na_osi(ax,x=100,y=300,popis_x=r"$y-t_2$",popis_y=r"$y'-t'_2$",color="red")
    vyznac_na_osi(ax,x=150,y=247.5,popis_x=r"$y-t_1$",popis_y=r"$y'-t'_1$",color="blue")
    vyznac_na_osi(ax,x=opt["c"],y=opt["cp"],popis_x=r"$c^*$",popis_y=r"$c'^*$")
    
    ekonomicke_osi(ax,r"$c$",r"$c'$")
    
    plt.savefig(
        "images/finan/graf_ricardo.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#########################
# graf_urokova_prirazka #
#########################

def graf_urokova_prirazka():

    fig,ax=plt.subplots()
    
    y=200
    yp=200
    r1=0.01
    r2=0.9
    we1 = y+yp/(1+r1)
    we2 = y+yp/(1+r2)
    nakresli_rozpocet_klin(ax,r1,r2,y,yp)
    vyznac_na_osi(ax,x=we1,y=0,popis_x=r"$we_1$")
    vyznac_na_osi(ax,x=we2,y=0,popis_x=r"$we_2$")
    vyznac_na_osi(ax,x=y,y=yp,popis_x=r"$y-t$",popis_y=r"$y'-t'$")
    vyznac_na_osi(ax,x=0,y=we1*(1+r1),popis_y=r"$we_1(1+r_1)$")
    vyznac_na_osi(ax,x=0,y=we2*(1+r2),popis_y=r"$we_2(1+r_2)$")
    nakresli_bod(ax,y,yp,"E",vodiace_ciary=True)
    
    ekonomicke_osi(ax,r"$c$",r"$c'$")
    
    plt.savefig(
        "images/finan/graf_urokova_prirazka.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

######################
# graf_ricardo_nedok #
######################

def graf_ricardo_nedok():

    fig,ax=plt.subplots()

    y=200
    yp=200
    r1=0.01
    r2=5.
    beta = 0.3
    u1=uzitocnost(y, yp, beta=beta)
    we1 = y+yp/(1+r1)
    we2 = y+yp/(1+r2)
    vyznac_na_osi(ax,x=we1,y=0,popis_x=r"$we_1$")
    vyznac_na_osi(ax,x=y,y=yp,popis_x=r"$y-t_1$",popis_y=r"$y'-t_1'$",color="blue")
    vyznac_na_osi(ax,x=0,y=we1*(1+r1),popis_y=r"$we_1(1+r_1)$")
    vyznac_na_osi(ax,x=0,y=we2*(1+r2),popis_y=r"$we_2(1+r_2)$")
    nakresli_bod(ax,y,yp,r"$E_1$",vodiace_ciary=True)
    nakresli_indiferencnu_intertemporal(ax,u=u1,beta=beta,c_min=135,c_max=we1)
    otpimalny_skrt=optimum(y,yp,r1,beta=beta)
    # Najprv nakreslíme optimálny škrt vlády
    rozpocet=nakresli_rozpocet_klin(ax,r1,r2,otpimalny_skrt["c"],otpimalny_skrt["cp"],color="orange")
    nakresli_indiferencnu_intertemporal(ax,u=otpimalny_skrt["u"],beta=beta,
                                        c_min=155,c_max=rozpocet["c_max"],color="orange")
    vyznac_na_osi(ax,x=otpimalny_skrt["c"],y=otpimalny_skrt["cp"],
                  popis_x=r"$y-t^*$",popis_y=r"$y'-t'^*$",color="orange")
    nakresli_bod(ax,otpimalny_skrt["c"],otpimalny_skrt["cp"],r"$E^*$",vodiace_ciary=True)
    
    # potom ukážku škrtu
    u2=uzitocnost(y+50,y-50*(1+r1),beta=beta) # bod o 50 vyššia spotreba v 1. o 50*(1+r1) nižšia v 2. období
    nakresli_indiferencnu_intertemporal(ax,u=u2,beta=beta,
                                        c_min=145,c_max=rozpocet["c_max"],color="red")
    nakresli_rozpocet_klin(ax,r1,r2,otpimalny_skrt["c"],otpimalny_skrt["cp"],color="orange")
    nakresli_rozpocet_klin(ax,r1,r2,y+50,yp-50*(1+r1),color="red")
    vyznac_na_osi(ax,x=y+50,y=yp-50*(1+r1),popis_x=r"$y-t_2$",popis_y=r"$y'-t_2'$",color="red")
    nakresli_bod(ax,y+50,yp-50*(1+r1),r"$E_2$",vodiace_ciary=True,dx=-15,dy=-15)
    # nakoniec nakreslíme pôvodný rozpočet
    nakresli_rozpocet_klin(ax,r1,r2,y,yp,color="blue")
    ax.set_ylim(0, we1*(1+r1)+30 )
    ekonomicke_osi(ax,r"$c$",r"$c'$")

    plt.savefig(
        "images/finan/graf_ricardo_nedok.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##########################
# graf_podiel_neplaticov #
##########################

def graf_podiel_neplaticov():

    fig,ax=plt.subplots()
    
    y=100
    yp=300
    r1=0.01
    r2=0.9
    r22=2.5
    we1 = y+yp/(1+r1)
    we2 = y+yp/(1+r2)
    beta=0.4
    opt1=optimum(y,yp,r2,beta=beta)
    opt2=optimum(y,yp,r22,beta=beta)
    nakresli_indiferencnu_intertemporal(ax,u=opt1["u"],beta=beta,
                                        c_max=we1,color="blue")
    nakresli_indiferencnu_intertemporal(ax,u=opt2["u"],beta=beta,
                                        c_max=we1,color="red")
    vyznac_na_osi(ax,x=opt1["c"],y=opt1["cp"],popis_x=r"$c_1$",popis_y=r"$c'_1$",color="blue")
    nakresli_bod(ax,opt1["c"],opt1["cp"],"A",vodiace_ciary=True)
    vyznac_na_osi(ax,x=y,y=yp,popis_x=r"$y-t$",popis_y=r"$y'-t'$")
    vyznac_na_osi(ax,x=opt2["c"],y=opt2["cp"],popis_x=r"$c_2$",popis_y=r"$c'_2$",color="red")
    nakresli_bod(ax,opt2["c"],opt2["cp"],"B",vodiace_ciary=True)
    nakresli_bod(ax,y,yp,"E",vodiace_ciary=True,dx=-2,dy=10)
    nakresli_rozpocet_klin(ax,r1,r22,y,yp,color="red")
    nakresli_rozpocet_klin(ax,r1,r2,y,yp,color="blue")
    ax.set_ylim(0, we1*(1+r1)+30 )
    ekonomicke_osi(ax,r"$c$",r"$c'$")

    plt.savefig(
        "images/finan/graf_podiel_neplaticov.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_hypoteka #
#################

def graf_hypoteka():

    fig,ax=plt.subplots()
    r=0.05
    y=200
    yp=200
    beta=0.4
    # hypotéka
    # B = pH/(1+r), kde
    # p: cena nehnuteľnosti
    # H: objem nehnuteľnosti
    # c <= y+pH/(1+r)
    #Prídelový bod E: [y,ph] = [y,yp+B*(1+r)]
    B1=100
    B2=50
    # pôvodný koš
    nakresli_bod(ax,y+B1,yp,nazov="A",vodiace_ciary=True)
    nakresli_bod(ax,y+B2,yp,nazov="B",vodiace_ciary=True)
    vyznac_na_osi(ax,y+B2,yp,popis_x="$c_2$",popis_y="$c'=y'-t'$",color="red")
    vyznac_na_osi(ax,y+B1,yp,popis_x=r"$c_1$",popis_y="$c'=y'-t'$",color="blue")
    # Hranice spotrebiteľských možností
    nakresli_rozpocet_kolateral(ax,r,y,yp,B1,color="blue")
    nakresli_rozpocet_kolateral(ax,r,y,yp,B2,color="red")
    # Prídelové body E1 a E2
    nakresli_bod(ax,y,yp+B1*(1+r),nazov=r"$E_1$",vodiace_ciary=True,dx=-10,dy=10)
    nakresli_bod(ax,y,yp+B2*(1+r),nazov=r"$E_2$",vodiace_ciary=True)
    vyznac_na_osi(ax,y,yp+B2*(1+r),popis_x="$y-t$",popis_y=r"$y'-t'+(p_2H)$",color="red")
    vyznac_na_osi(ax,y,yp+B1*(1+r),popis_x="$y-t$",popis_y=r"$y'-t'+(p_1H)$",color="blue")
    # indiferenčné krivky
    u1=uzitocnost(y+B1,yp,beta=beta,r=r)
    u2=uzitocnost(y+B2,yp,beta=beta,r=r)
    nakresli_indiferencnu_intertemporal(ax,u=u1,beta=beta,
                                        c_max=y+B1+50,color="blue")
    nakresli_indiferencnu_intertemporal(ax,u=u2,beta=beta,
                                        c_max=y+B1+50,color="red")
    #ax.set_xlim(0, 500)
    ax.set_ylim(0, y+yp+B1/(1+r)+50)
    
    ekonomicke_osi(ax,r"$c$",r"$c'$")

    plt.savefig(
        "images/finan/graf_hypoteka.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#########################
# graf_priebezny_senior #
#########################

def graf_priebezny_senior():

    fig, ax = plt.subplots()
    r=0.05
    y=250
    yp1=100
    # Penzista získa extra príjem b=100: yp2=yp1+b=200
    yp2=200
    # pôvodné optimá
    opt1=optimum(y,yp1,r)
    opt2=optimum(y,yp2,r)
    rozpocet1 = nakresli_rozpocet_intertemporal(ax,r,y,yp1,color="blue")
    rozpocet1 = nakresli_rozpocet_intertemporal(ax,r,y,yp2,color="red")
    nakresli_indiferencnu_intertemporal(ax,u=opt1["u"],beta=opt1["beta"],
                                        c_max=y+yp2/(1+r))
    nakresli_indiferencnu_intertemporal(ax,u=opt2["u"],beta=opt2["beta"],
                                        c_max=y+yp2/(1+r),color="red")
    # pôvodný optimálny bod
    nakresli_bod(ax,opt1["c"],opt1["cp"],nazov="A",vodiace_ciary=True)
    vyznac_na_osi(ax,opt1["c"],opt1["cp"],popis_x=r"$c_1^*$",popis_y=r"$c_1'^*$",color="blue")
    # nový
    nakresli_bod(ax,opt2["c"],opt2["cp"],nazov="B",vodiace_ciary=True)
    vyznac_na_osi(ax,opt2["c"],opt2["cp"],popis_x=r"$c_2^*$",popis_y=r"$c_2'^*$",color="red")
    #E body
    nakresli_bod(ax,y,yp1,nazov=r"$E_1$",vodiace_ciary=True,dx=-0.1,dy=-20)
    nakresli_bod(ax,y,yp2,nazov=r"$E_2$",vodiace_ciary=True)
    vyznac_na_osi(ax,y,yp2,popis_x="$y$",popis_y="$y'+b$",color="red")
    vyznac_na_osi(ax,y,yp1,popis_x="$y$",popis_y="$y'$",color="blue")
    
    ekonomicke_osi(ax,r"$c$",r"$c'$")
    ax.set_ylim(0, y*(1+r)+yp2+10)

    plt.savefig(
        "images/finan/graf_priebezny_senior.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##################
# graf_priebezny #
##################

def graf_priebezny():

    fig, ax = plt.subplots()
    r=0.05
    y1=450
    # Aktívny stratí t=b/(1+r). Pomery nesedia, sú iba ilustračné
    y2=380
    yp1=50
    # Aktívny získa extra príjem b=100: yp2=yp1+b=200
    yp2=200
    # pôvodné optimá
    opt1=optimum(y1,yp1,r)
    opt2=optimum(y2,yp2,r)
    rozpocet1 = nakresli_rozpocet_intertemporal(ax,r,y1,yp1,color="blue")
    rozpocet1 = nakresli_rozpocet_intertemporal(ax,r,y2,yp2,color="red")
    nakresli_indiferencnu_intertemporal(ax,u=opt1["u"],beta=opt1["beta"],
                                        c_max=y2+yp2/(1+r))
    nakresli_indiferencnu_intertemporal(ax,u=opt2["u"],beta=opt2["beta"],
                                        c_max=y2+yp2/(1+r),color="red")
    # pôvodný optimálny bod
    nakresli_bod(ax,opt1["c"],opt1["cp"],nazov="A",vodiace_ciary=True)
    vyznac_na_osi(ax,opt1["c"],opt1["cp"],popis_x=r"$c_1^*$",popis_y=r"$c_1'^*$",color="blue")
    # nový
    nakresli_bod(ax,opt2["c"],opt2["cp"],nazov="B",vodiace_ciary=True)
    vyznac_na_osi(ax,opt2["c"],opt2["cp"],popis_x=r"$c_2^*$",popis_y=r"$c_2'^*$",color="red")
    #E body
    nakresli_bod(ax,y1,yp1,nazov=r"$E_1$",vodiace_ciary=True)
    nakresli_bod(ax,y2,yp2,nazov=r"$E_2$",vodiace_ciary=True,dx=6,dy=-5)
    vyznac_na_osi(ax,y1,yp1,popis_x="$y$",popis_y="$y'$",color="blue")
    vyznac_na_osi(ax,y2,yp2,popis_x="$y-b/(1+r)$",popis_y="$y'+b$",color="red")
    
    ekonomicke_osi(ax,r"$c$",r"$c'$")
    ax.set_ylim(0, y2*(1+r)+yp2+10)

    plt.savefig(
        "images/finan/graf_priebezny.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

################
# graf_fondovy #
################

def graf_fondovy():

    fig, ax = plt.subplots()
    r=0.05
    y=150
    yp=250
    #vynútená spotreba
    cn = 100
    cnp = yp+(y-cn)*(1+r)
    opt=optimum(y,yp,r)
    un=uzitocnost(cn,cnp,r=r,beta=opt["beta"])
    rozpocet = nakresli_rozpocet_intertemporal(ax,r,y,yp,color="blue")
    nakresli_indiferencnu_intertemporal(ax,u=opt["u"],beta=opt["beta"],
                                        c_max=y+yp/(1+r))
    nakresli_indiferencnu_intertemporal(ax,u=un,beta=opt["beta"],
                                        c_max=y+yp/(1+r),color="red")
    # pôvodný optimálny bod
    nakresli_bod(ax,opt["c"],opt["cp"],nazov="A",vodiace_ciary=True)
    vyznac_na_osi(ax,opt["c"],opt["cp"],popis_x=r"$c_1^*$",popis_y=r"$c_1'^*$",color="blue")
    # nový
    nakresli_bod(ax,cn,cnp,nazov="B",vodiace_ciary=True)
    vyznac_na_osi(ax,cn,cnp,popis_x=r"$c_2^*$",popis_y=r"$c_2'^*$",color="red")
    #E body
    nakresli_bod(ax,y,yp,nazov="$E$",vodiace_ciary=True,dx=8)
    vyznac_na_osi(ax,y,yp,popis_x="$y$",popis_y="$y'$",color="blue")
    
    ekonomicke_osi(ax,r"$c$",r"$c'$")
    ax.set_ylim(0, y*(1+r)+yp+10)

    plt.savefig(
        "images/finan/graf_fondovy.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

graf_ricardo()
graf_urokova_prirazka()
graf_ricardo_nedok()
graf_podiel_neplaticov()
graf_hypoteka()
graf_priebezny_senior()
graf_priebezny()
graf_fondovy()
```

## Úvod do teórie rastu

Grafy sa vygenerujú v podpriečinku `uvod_rast`.

```python
import matplotlib.pyplot as plt
import numpy as np
import grafy
from pathlib import Path

Path("images/uvod_rast").mkdir(exist_ok=True)

grafy.set_theory_style()

k0=0.3
A = 2
s = 0.3
alpha = 0.3
delta = 0.4
n = 0

k_us, y_us, i_us, c_us = grafy.us(A, alpha, n, s, delta)

###########
# graf_45 #
###########

def graf_45():
    
    fig,ax=grafy.plot_cobweb(
        lambda k: grafy.phi(k,A,alpha,n,s,delta),
        0,
        2.3,
        k0,
        var="k",
        steady_state=k_us,
        figsize=(6,6),
        filename="uvod_rast/graf_45"
    )

################
# graf_y_solow #
################

N0=1
k,index,*_=grafy.rady(k0,N0,A,alpha=alpha,n=n,s=s,delta=delta,T=10)
def graf_k_solow():
    fig,ax,df=grafy.plot_model_path(
        k,
        labels=[r"$k_t$"],
        ylabel=r"$k_t$",
        steady_states={
            r"$k^*$": k_us
        },
        filename="uvod_rast/graf_k_solow"
        )

########################
# grafy_zvysnych_radov #
########################
    
y,i,c=grafy.rovnice(k, A=A, alpha=alpha, s=s)
    
def grafy_zvysnych_radov():

    fig,ax,df=grafy.plot_model_path(
        y,
        labels=[r"$y_t$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y^*$": y_us
        },
        filename="uvod_rast/graf_y_solow"
    )
    
    fig,ax,df=grafy.plot_model_path(
        c,
        labels=[r"$c_t$"],
        ylabel=r"$c_t$",
        steady_states={
            r"$c^*$": c_us
        },
        filename="uvod_rast/graf_c_solow"
    )
    
    fig,ax,df=grafy.plot_model_path(
        i,
        labels=[r"$i_t$"],
        ylabel=r"$i_t$",
        steady_states={
            r"$i^*$": i_us
        },
        filename="uvod_rast/graf_i_solow"
    )

###########
# graf_us #
###########

def graf_us():

    grafy.set_theory_style()

    A = 1.
    alpha = 1/2
    s = 0.4
    n = 0.02
    delta = 0.2

    k_us,y_us,i_us,*_=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)

    fig, ax = plt.subplots()

    grafy.nakresli_funkciu(
        ax,
        lambda k: A * grafy.f(k, alpha),
        0,
        5,
        nazov=r"$y=Af(k)$"
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: s * A * grafy.f(k, alpha),
        0,
        5,
        nazov=r"$i=sAf(k)$"
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: (n + delta) * k,
        0,
        5,
        nazov=r"$(n+\delta)k$"
    )

    grafy.vyznac_na_osi(ax, k_us, y_us, r"$k^*$", popis_y=r"$y^*$")
    grafy.vyznac_na_osi(ax, k_us, i_us, popis_y=r"$i^*=sy^*$")
    grafy.nakresli_bod(ax, k_us, y_us, nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(ax, k_us, i_us, nazov=None, vodiace_ciary=True)

    x_arrow = 0.07 * ax.get_xlim()[1]

    grafy.nakresli_dvojsipku(
        ax,
        0.1,
        i_us,
        0.1,
        y_us,
        nazov=r"$c^*$"
    )

    grafy.ekonomicke_osi(ax,xlabel=f"$k$",ylabel=None)
    ax.legend()
    
    plt.savefig(
        "images/uvod_rast/graf_us.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()
    
##############
# graf_zlate #
##############

def graf_zlate():

    grafy.set_theory_style()

    A = 1.
    alpha = 1/2
    n = 0.02
    delta = 0.2

    #množina parametrov miery sporenia  
    ska=[0.4,0.5,0.6]
    slabels=[r"$i=s_1Af(k)$",r"$i=s^*Af(k)$",r"$i=s_2Af(k)$"]
    clabels=[r"$c^*_1$",r"$c^{**}$",r"$c^*_2$"]

    k_zlate,y_zlate,i_zlate,*_=grafy.us(A=A,alpha=alpha,n=n,s=ska[1],delta=delta)

    fig, ax = plt.subplots()

    grafy.nakresli_funkciu(
        ax,
        lambda k: A * grafy.f(k, alpha),
        0,
        xmax=k_zlate*2,
        nazov=r"$y=Af(k)$",linewidth=1.
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: (n + delta) * k,
        0,
        xmax=k_zlate*2,
        nazov=r"$(n+\delta)k$",linewidth=1.
    )


    for s, slabel, clabel in zip(ska, slabels, clabels):
        us=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)

        grafy.nakresli_funkciu(
        ax,
        lambda k: s * A * grafy.f(k, alpha),
        0,
        xmax=k_zlate*2,
        nazov=slabel,
        linewidth=1.
        )
        grafy.nakresli_bod(ax, us[0], us[2], nazov=None, vodiace_ciary=False)
        grafy.nakresli_bod(ax, us[0], us[1], nazov=None, vodiace_ciary=False)

        grafy.nakresli_dvojsipku(
        ax,
        us[0],
        us[2],
        us[0],
        us[1],
        nazov=clabel,
        )

    grafy.ekonomicke_osi(ax,xlabel=f"$k$",ylabel=None)
    grafy.nakresli_dotycnicu(ax,lambda k: A * grafy.f(k, alpha),lambda k: A*alpha*k**(alpha-1),k_zlate,2
                             ,label=f"Sklon = $n+\delta$",linewidth=1.)

    grafy.vyznac_na_osi(ax, k_zlate, i_zlate, popis_x=r"$k^{**}$", popis_y=r"$i^{**}=sy^{**}$")
    grafy.nakresli_bod(ax, k_zlate, i_zlate, nazov=None, vodiace_ciary=True)

    ax.legend()

    plt.savefig(
        "images/uvod_rast/graf_zlate.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()
    
def grafy_extenzivne():
    grafy.set_data_style()

    A = 1.
    s = 1/3
    alpha = 1/3
    delta = 0.1
    n = 0.02

    k_us, y_us, i_us, c_us = grafy.us(A, alpha, n, s, delta)

    k0=k_us
    N0=1000
    k,index,N,*_=grafy.rady(k0,N0,A,alpha=alpha,n=n,s=s,delta=delta,T=50)
    y,i,c=grafy.rovnice(k,A,alpha,s)
    K=k*N
    Y=y*N
    I=i*N
    C=c*N

    fig,ax,df=grafy.plot_model_path(
        K,ylabel=r"$K^*_t$",filename="uvod_rast/graf_Ke_solow"
    )
    
    fig,ax,df=grafy.plot_model_path(
        Y,ylabel=r"$Y^*_t$",filename="uvod_rast/graf_Ye_solow"
    )

    fig,ax,df=grafy.plot_model_path(
        I,ylabel=r"$I^*_t$",filename="uvod_rast/graf_Ie_solow"
    )
    
    fig,ax,df=grafy.plot_model_path(
        C,ylabel=r"$C^*_t$",filename="uvod_rast/graf_Ce_solow"
    )

graf_45()
graf_k_solow()
grafy_zvysnych_radov()
graf_us()
graf_zlate()
grafy_extenzivne()
```

## Konvergencia a ustálený rast

Grafy sa vygenerujú v podpriečinku `rast`.

```python
import matplotlib.pyplot as plt
import numpy as np
import grafy
from pathlib import Path

Path("images/rast").mkdir(exist_ok=True)

grafy.set_theory_style()

A = 1.
alpha = 1/3
s = 1/3
n = 0.02
delta = 0.1

#################
# graf_zmena_As #
#################

def graf_zmena_As():

    A2=1.35
    k_us,y_us,i_us,*_=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)
    k_us2,y_us2,i_us2,*_=grafy.us(A=A2,alpha=alpha,n=n,s=s,delta=delta)

    fig, ax = plt.subplots()

    grafy.nakresli_funkciu(
        ax,
        lambda k: s * A * grafy.f(k, alpha),
        0,
        k_us*2,
        nazov=r"$i=s_1A_1f(k)$",
        color="blue"
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: s * A2 * grafy.f(k, alpha),
        0,
        k_us*2,
        nazov=r"$i=s_2A_2f(k)$",
        color="red"
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: (n + delta) * k,
        0,
        k_us*2,
        nazov=r"$(n+\delta)k$",
        color="orange"
    )

    grafy.vyznac_na_osi(ax, k_us, i_us, popis_x=r"$k^*_1$", popis_y=r"$i^*_1=s_1y^*_1$",color="blue")
    grafy.nakresli_bod(ax, k_us, i_us, nazov=None, vodiace_ciary=True)

    grafy.vyznac_na_osi(ax, k_us2, i_us2, popis_x=r"$k^*_2$", popis_y=r"$i^*_2=s_2y^*_2$",color="red")
    grafy.nakresli_bod(ax, k_us2, i_us2, nazov=None, vodiace_ciary=True)

    grafy.ekonomicke_osi(ax,xlabel=f"$k$",ylabel=None)
    ax.legend()
    
    plt.savefig(
        "images/rast/graf_zmena_As.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#####################
# graf_zmena_ndelta #
#####################
   
    
def graf_zmena_ndelta():    
 
    delta2=0.15

    k_us,y_us,i_us,*_=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)
    k_us2,y_us2,i_us2,*_=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta2)

    x_max=1.1*k_us

    fig, ax = plt.subplots()

    grafy.nakresli_funkciu(
        ax,
        lambda k: s * A * grafy.f(k, alpha),
        0,
        x_max,
        nazov=r"$i=sAf(k)$",
        color="orange"
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: (n + delta2) * k,
        0,
        x_max,
        nazov=r"$(n_2+\delta_2)k$",
        color="red"
    )

    grafy.nakresli_funkciu(
        ax,
        lambda k: (n + delta) * k,
        0,
        x_max,
        nazov=r"$(n_1+\delta_1)k$",
        color="blue"
    )

    grafy.vyznac_na_osi(ax, k_us, i_us, popis_x=r"$k^*_1$", popis_y=r"$i^*_1=sy^*_1$",color="blue")
    grafy.nakresli_bod(ax, k_us, i_us, nazov=None, vodiace_ciary=True)

    grafy.vyznac_na_osi(ax, k_us2, i_us2, popis_x=r"$k^*_2$", popis_y=r"$i^*_2=sy^*_2$",color="red")
    grafy.nakresli_bod(ax, k_us2, i_us2, nazov=None, vodiace_ciary=True)

    grafy.ekonomicke_osi(ax,xlabel=f"$k$",ylabel=None)
    ax.set_ylim(top=i_us*1.3)
    ax.legend()
    
    plt.savefig(
        "images/rast/graf_zmena_ndelta.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()
    
#############
# graf_45Ak #
#############

def graf_45Ak():
    
    k0=1.
    A = 1.
    s = 0.4
    alpha = 1.
    delta = 0.049
    n = 0.02
    
    grafy.set_theory_style()

    fig,ax=grafy.plot_cobweb(
        lambda k: grafy.phi(k,A,alpha,n,s,delta),
        0,
        5.,
        k0,
        var="k",
        figsize=(6,6),
        filename="rast/graf_45Ak"
    )    

###############
# graf_rastAk #
###############


def graf_rastAk():
    
    k0=1.
    A = 1.
    s = 0.4
    alpha = 1.
    delta = 0.049
    n = 0.02
    
    N0=1
    k,index,*_=grafy.rady(k0,N0,A,alpha=alpha,n=n,s=s,delta=delta,T=10)
    fig,ax,df=grafy.plot_model_path(
        k,
        labels=[r"$k_t$"],
        ylabel=r"$k_t$",
        filename="rast/graf_rastAk"
    )
    
#################
# grafy_solow_s #
#################

def grafy_solow_zmeny():
    
    # zmeny s
    ## vplyv na k
    
    A = 1.
    alpha = 1/3
    s = 0.2
    n = 0.02
    delta = 0.1

    N0=1
    T0 = 50
    s2=s*1.35
    T=50

    k_us,y_us,i_us,c_us=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)
    k_us2,y_us2,i_us2,c_us2=grafy.us(A=A,alpha=alpha,n=n,s=s2,delta=delta)

    k_pred = np.full(T0, k_us)

    k_po, *_ = grafy.rady(
        k0=k_us,N0=N0,A0=A,alpha=alpha,n=n,s=s2,delta=delta,
        T=T
    )

    series = np.concatenate([
        k_pred,
        k_po
    ])

    fig,ax,df=grafy.plot_model_path(
        series,
        labels=[r"$k_t$"],
        ylabel=r"$k_t$",
        steady_states={
            r"$k_1^*$": (k_us, "blue"),
            r"$k_2^*$": (k_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_s_k"
    )
    
    ## vplyv na y
    
    y_pred,i_pred,c_pred=grafy.rovnice(k=k_pred,A=A,alpha=alpha,s=s)
    y_po,i_po,c_po=grafy.rovnice(k=k_po,A=A,alpha=alpha,s=s2)

    y = np.concatenate([
        y_pred,
        y_po
    ])

    i = np.concatenate([
        i_pred,
        i_po
    ])

    c = np.concatenate([
        c_pred,
        c_po
    ])

    fig,ax,df=grafy.plot_model_path(
        y,
        labels=[r"$y_t$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y_1^*$": (y_us, "blue"),
            r"$y_2^*$": (y_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_s_y"
    )
    
    ## vplyv na i
    
    fig,ax,df=grafy.plot_model_path(
        i,
        labels=[r"$i_t$"],
        ylabel=r"$i_t$",
        steady_states={
            r"$i_1^*$": (i_us, "blue"),
            r"$i_2^*$": (i_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_s_i"
    )
    
    ## vplyv na c
    
    fig,ax,df=grafy.plot_model_path(
        c,
        labels=[r"$c_t$"],
        ylabel=r"$c_t$",
        steady_states={
            r"$c_1^*$": (c_us, "blue"),
            r"$c_2^*$": (c_us2, "red")
        },

        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_s_c"
    )
    
    # zmena A
    
    A2=A*1.15
    
    ## vplyv na k
    
    k_us,y_us,i_us,c_us=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)
    k_us2,y_us2,i_us2,c_us2=grafy.us(A=A2,alpha=alpha,n=n,s=s,delta=delta)

    k_pred = np.full(T0, k_us)

    k_po, *_ = grafy.rady(
        k0=k_us,N0=N0,A0=A2,alpha=alpha,n=n,s=s,delta=delta,
        T=T
    )

    series = np.concatenate([
        k_pred,
        k_po
    ])

    fig,ax,df=grafy.plot_model_path(
        series,
        labels=[r"$k_t$"],
        ylabel=r"$k_t$",
        steady_states={
            r"$k_1^*$": (k_us, "blue"),
            r"$k_2^*$": (k_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_A_k"
    )
    
    ## vplyv na y
    
    y_pred,i_pred,c_pred=grafy.rovnice(k=k_pred,A=A,alpha=alpha,s=s)
    y_po,i_po,c_po=grafy.rovnice(k=k_po,A=A2,alpha=alpha,s=s)

    y = np.concatenate([
        y_pred,
        y_po
    ])

    i = np.concatenate([
        i_pred,
        i_po
    ])

    c = np.concatenate([
        c_pred,
        c_po
    ])

    fig,ax,df=grafy.plot_model_path(
        y,
        labels=[r"$y_t$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y_1^*$": (y_us, "blue"),
            r"$y_2^*$": (y_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_A_y"
    )
    
    ## vplyv na i
        
    fig,ax,df=grafy.plot_model_path(
        i,
        labels=[r"$i_t$"],
        ylabel=r"$i_t$",
        steady_states={
            r"$i_1^*$": (i_us, "blue"),
            r"$i_2^*$": (i_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_A_i"
    )
    
    ## vplyv na c

    fig,ax,df=grafy.plot_model_path(
        c,
        labels=[r"$c_t$"],
        ylabel=r"$c_t$",
        steady_states={
            r"$c_1^*$": (c_us, "blue"),
            r"$c_2^*$": (c_us2, "red")
        },

        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_A_c"
    )
    
    # zmena n alebo delta
    ## vplyv na k
    
    A = 1.
    alpha = 1/3
    s = 0.2
    n = 0.02
    delta = 0.1

    N0=1
    T0 = 50
    delta2=0.12
    T=50

    k_us,y_us,i_us,c_us=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)
    k_us2,y_us2,i_us2,c_us2=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta2)

    k_pred = np.full(T0, k_us)

    k_po, *_ = grafy.rady(
        k0=k_us,N0=N0,A0=A,alpha=alpha,n=n,s=s,delta=delta2,
        T=T
    )

    series = np.concatenate([
        k_pred,
        k_po
    ])

    fig,ax,df=grafy.plot_model_path(
        series,
        labels=[r"$k_t$"],
        ylabel=r"$k_t$",
        steady_states={
            r"$k_1^*$": (k_us, "blue"),
            r"$k_2^*$": (k_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_ndelta_k"
    )
    
    ## vplyv na y
    
    y_pred,i_pred,c_pred=grafy.rovnice(k=k_pred,A=A,alpha=alpha,s=s)
    y_po,i_po,c_po=grafy.rovnice(k=k_po,A=A,alpha=alpha,s=s)

    y = np.concatenate([
        y_pred,
        y_po
    ])

    i = np.concatenate([
        i_pred,
        i_po
    ])

    c = np.concatenate([
        c_pred,
        c_po
    ])

    fig,ax,df=grafy.plot_model_path(
        y,
        labels=[r"$y_t$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y_1^*$": (y_us, "blue"),
            r"$y_2^*$": (y_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_ndelta_y"
    )
    
    ## vplyv na i
    
    fig,ax,df=grafy.plot_model_path(
        i,
        labels=[r"$i_t$"],
        ylabel=r"$i_t$",
        steady_states={
            r"$i_1^*$": (i_us, "blue"),
            r"$i_2^*$": (i_us2, "red")
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_ndelta_i"
    )
    
    ## vplyv na c
    
    fig,ax,df=grafy.plot_model_path(
        c,
        labels=[r"$c_t$"],
        ylabel=r"$c_t$",
        steady_states={
            r"$c_1^*$": (c_us, "blue"),
            r"$c_2^*$": (c_us2, "red")
        },

        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_ndelta_c"
    )
    
    # zmena N0
    ## vplyv na y
    
    A = 1.
    alpha = 1/3
    s = 0.2
    n = 0.02
    delta = 0.1

    N0=1
    T0 = 50
    N02=1.10
    T=50

    k_us,y_us,i_us,c_us=grafy.us(A=A,alpha=alpha,n=n,s=s,delta=delta)

    k_pred = np.full(T0, k_us)

    k_po, *_ = grafy.rady(
        k0=k_us/N02,N0=N02,A0=A,alpha=alpha,n=n,s=s,delta=delta,
        T=T
    )

    series = np.concatenate([
        k_pred,
        k_po
    ])

    fig,ax,df=grafy.plot_model_path(
        series,
        labels=[r"$k_t$"],
        ylabel=r"$k_t$",
        steady_states={
            r"$k^*$": (k_us, "blue"),
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_N_k"
    )
    
    y_pred,i_pred,c_pred=grafy.rovnice(k=k_pred,A=A,alpha=alpha,s=s)
    y_po,i_po,c_po=grafy.rovnice(k=k_po,A=A,alpha=alpha,s=s)

    y = np.concatenate([
        y_pred,
        y_po
    ])

    i = np.concatenate([
        i_pred,
        i_po
    ])

    c = np.concatenate([
        c_pred,
        c_po
    ])

    fig,ax,df=grafy.plot_model_path(
        y,
        labels=[r"$y_t$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y^*$": (y_us, "blue"),
        },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_N_y"
    )
    
    fig,ax,df=grafy.plot_model_path(
        i,
        labels=[r"$i_t$"],
        ylabel=r"$i_t$",
        steady_states={
            r"$i^*$": (i_us, "blue"),
            },
        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_N_i"
    )
    
    fig,ax,df=grafy.plot_model_path(
        c,
        labels=[r"$c_t$"],
        ylabel=r"$c_t$",
        steady_states={
            r"$c^*$": (c_us, "blue"),
          },

        shock_period=T,
        series_color="orange",
        filename="rast/graf_solow_N_c"
    )
    
###########
# graf_ur #
###########

def graf_ur():
    k0_us=1.
    g=0.02
    
    index = np.arange(100)

    k_trend = k0_us * (1 + g)**(index / (1 - alpha))
    
    fig,ax,df=grafy.plot_model_path(
        k_trend,
        ylabel=r"$k^*_t$",
        filename="rast/graf_ur"
    )

#####################
# graf_konvergencia #
#####################

def graf_konvergencia():
    import grafy
    import pandas as pd
    
    A=2.
    alpha=0.5
    k01=4
    k02=9
    s=0.12
    delta=0.04
    n=0.02
    
    k_us,y_us,*_=grafy.us(A=A, alpha=alpha, n=n, s=s, delta=delta)
    k1,*_=grafy.rady(k0=k01, N0=40, A0=A, alpha=alpha, n=n, s=s, delta=delta)
    k2,*_=grafy.rady(k0=k02, N0=40, A0=A, alpha=alpha, n=n, s=s, delta=delta)
    y1,*_=grafy.rovnice(k=k1, A=A, alpha=alpha, s=s,)
    y2,*_=grafy.rovnice(k=k2, A=A, alpha=alpha, s=s,)
    
    df = pd.DataFrame({
    r"$y_{1t}$": y1,
    r"$y_{2t}$": y2
    })
    
    fig,ax,df=grafy.plot_model_path(
        df,
        labels=[r"$y_{1t}$", r"$y_{2t}$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y^*$": (y_us),
          },
        series_color=["blue", "red"],
        filename="rast/graf_konvergencia"
    )

################################
# graf_podmienena_konvergencia #
################################

def graf_podmienena_konvergencia():

    import grafy
    import pandas as pd
    
    A1=2.
    A2=2.22
    alpha=0.5
    k01=4
    k02=7
    s=0.12
    delta=0.04
    n=0.02
    
    k1_us,y1_us,*_=grafy.us(A=A1, alpha=alpha, n=n, s=s, delta=delta)
    k2_us,y2_us,*_=grafy.us(A=A2, alpha=alpha, n=n, s=s, delta=delta)
    k1,*_=grafy.rady(k0=k01, N0=40, A0=A1, alpha=alpha, n=n, s=s, delta=delta)
    k2,*_=grafy.rady(k0=k02, N0=40, A0=A2, alpha=alpha, n=n, s=s, delta=delta)
    y1,*_=grafy.rovnice(k=k1, A=A1, alpha=alpha, s=s,)
    y2,*_=grafy.rovnice(k=k2, A=A2, alpha=alpha, s=s,)
    
    df = pd.DataFrame({
    r"$y_{1t}$": y1,
    r"$y_{2t}$": y2
    })
    
    fig,ax,df=grafy.plot_model_path(
        df,
        labels=[r"$y_{1t}$", r"$y_{2t}$"],
        ylabel=r"$y_t$",
        steady_states={
            r"$y_1^*$": (y1_us,"blue"),
            r"$y_2^*$": (y2_us,"red")
          },
        series_color=["blue", "red"],
        filename="rast/graf_podmienena_konvergencia"
    )

graf_zmena_As()
graf_zmena_ndelta()
graf_45Ak()
graf_rastAk()
grafy_solow_zmeny()
graf_ur()
graf_konvergencia()
graf_podmienena_konvergencia()
```

## Model vyhľadávania a trh práce

Aplikáciou nasledovného kódu sa grafy nahrajú do podpriečinka `images/hlad_prac'

```python
import grafy
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

Path("images/hlad_prac").mkdir(exist_ok=True)
psi=1.5 # elasticita benefitu hľadania práce
k=95. # náklady na získanie jednotky práce
b=10 # podpora nezamestnanosti
A=210 # celková produktivita faktorov
e=0.95 # efekt párovania práce s voľnými miestami
d=0.5 # podiel spotrebiteľa na odmene z výstupu
eta=(1/3) # elasticita párovania
j=(e*(1-d)*(A-b)/k)**(1/eta)

def exo():
    psi=1.5 # elasticita benefitu hľadania práce
    k=95. # náklady na získanie jednotky práce
    b=10 # podpora nezamestnanosti
    A=210 # celková produktivita faktorov
    e=0.95 # efekt párovania práce s voľnými miestami
    d=0.5 # podiel spotrebiteľa na odmene z výstupu
    eta=(1/3) # elasticita párovania
    j=(e*(1-d)*(A-b)/k)**(1/eta)
    return psi,k,b,A,e,d,eta,j

def m_1j(j,e): #em(1,j)
    return e*j**(1-eta)
def m_1_j1(j,e): #em(1/j,1) dopyt po práci
    return e*j**(-eta)
def P(Q): 
    return Q**psi
def Q(j,b,e,A): # ponuka práce
    return (b+e*m_1j(j,e)*(A-b))**(1/psi)
grafy.set_theory_style()
# exogénne premenné, parametre a miera nasýtenia j

#########################
# graf_benefit_hladania #
#########################

def graf_benefit_hladania():
    psi,k,b,A,e,d,eta,j=exo()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,P,xmin=0.,xmax=45.)
    _=grafy.ekonomicke_osi(ax,xlabel="$Q$",ylabel="$P$")
    
    plt.savefig(
        "images/hlad_prac/graf_benefit_hladania.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

####################
# graf_benefit_bod #
####################

def graf_benefit_bod():
    psi,k,b,A,e,d,eta,j=exo()
    Pval=b+e*m_1j(j,e)*(A-b)
    Qval=Pval**(1/psi)
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,P,xmin=0.,xmax=45.,nazov="$P(Q)$")
    grafy.nakresli_bod(ax, x=Qval, y=Pval,vodiace_ciary="True", nazov=None, color='black')
    grafy.vyznac_na_osi(ax, x=Qval, y=Pval, popis_x="$Q_1$", popis_y="$b+em(1,j)(w-b)$", color='black')
    ax.legend()
    _=grafy.ekonomicke_osi(ax,xlabel="$Q$",ylabel="$P$")

    plt.savefig(
        "images/hlad_prac/graf_benefit_bod.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###################
# graf_dopyt_hlad #
###################

def graf_dopyt_hlad():
    psi,k,b,A,e,d,eta,j=exo()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,
                           lambda j: m_1_j1(j, e),
                           xmin=0.05,xmax=2.,
                           nazov=r"$em\left( \frac{1}{j},1 \right)$")
    grafy.nakresli_bod(ax, x=j, y=k/(1-d)/(A-b),vodiace_ciary="True", nazov=None, color='black')
    grafy.vyznac_na_osi(ax, x=j, y=k/(1-d)/(A-b), popis_x="$j_1$", popis_y=r"$\frac{k}{A-w}$", color='black')
    ax.legend()
    _=grafy.ekonomicke_osi(ax,xlabel="$j$",ylabel=None)
    plt.savefig(
        "images/hlad_prac/graf_dopyt_hlad.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##################
# rovnovaha_hlad #
##################

def graf_rovnovaha_hlad():
    
    psi,k,b,A,e,d,eta,j=exo()
    fig, axes = plt.subplots(2, 1, figsize=(8, 10))
    grafy.nakresli_funkciu(axes[1],lambda j: m_1_j1(j, e),
                           xmin=0.05,xmax=2.,
                           nazov=r"$em\left( \frac{1}{j},1 \right)$")
    grafy.nakresli_funkciu(axes[0],lambda j: Q(j,b,e,A),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b+em(1,j)d(A-b)$")
    for ax in axes: ax.legend() 
    _=grafy.ekonomicke_osi(axes[0],xlabel="$j$",ylabel="$Q$")
    _=grafy.ekonomicke_osi(axes[1],xlabel="$j$",ylabel=None)
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=j,
        points=[Q(j,b,e,A),k/(1-d)/(A-b)],
        value_label=r"$j^*$",
        axis_labels=[
            r"$Q^*$",
            r"$\frac{k}{(1-d)(A-b)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
    )

    plt.savefig(
        "images/hlad_prac/graf_rovnovaha_hlad.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_hlad_b #
###############

def graf_hlad_b():
    psi,k,b,A,e,d,eta,j=exo()
    b2=5*b
    jb=(e*(1-d)*(A-b2)/k)**(1/eta)
    fig, axes = plt.subplots(2, 1, figsize=(8, 10))
    grafy.nakresli_funkciu(axes[1],lambda j: m_1_j1(j, e),
                           xmin=0.05,xmax=2.,
                           nazov=r"$em\left( \frac{1}{j},1 \right)$")
    grafy.nakresli_funkciu(axes[0],lambda j: Q(j,b,e,A),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b+em(1,j)d(A-b_1)$")
    grafy.nakresli_funkciu(axes[0],lambda jb: Q(jb,b2,e,A),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b_2+em(1,j)d(A-b_2)$",
                          color="red")
    for ax in axes: ax.legend() 
    _=grafy.ekonomicke_osi(axes[0],xlabel="$j$",ylabel="$Q$")
    _=grafy.ekonomicke_osi(axes[1],xlabel="$j$",ylabel=None)
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=j,
        points=[Q(j,b,e,A),k/(1-d)/(A-b)],
        value_label=r"$j^*_1$",
        axis_labels=[
            None,
            r"$\frac{k}{(1-d)(A-b_1)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
    )
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=jb,
        points=[Q(jb,b2,e,A),k/(1-d)/(A-b2)],
        value_label=r"$j^*_2$",
        axis_labels=[
            None,
            r"$\frac{k}{(1-d)(A-b_2)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
    )
    plt.savefig(
        "images/hlad_prac/graf_hlad_b.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_hlad_A #
###############

def graf_hlad_A():
    psi,k,b,A,e,d,eta,j=exo()
    A2=1.23*A
    jA=(e*(1-d)*(A2-b)/k)**(1/eta)
    fig, axes = plt.subplots(2, 1, figsize=(8, 10))
    grafy.nakresli_funkciu(axes[1],lambda j: m_1_j1(j, e),
                           xmin=0.05,xmax=2.,
                           nazov=r"$em\left( \frac{1}{j},1 \right)$")
    grafy.nakresli_funkciu(axes[0],lambda j: Q(j,b,e,A),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b+em(1,j)d(A_1-b)$")
    grafy.nakresli_funkciu(axes[0],lambda jA: Q(jA,b,e,A2),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b+em(1,j)d(A_2-b)$",
                          color="red")
    axes[0].set_ylim(0, Q(jA, b, e, A2) + 13)
    axes[0].legend(loc="upper left")
    axes[1].legend(loc="upper left", bbox_to_anchor=(0.03, 1))
    _=grafy.ekonomicke_osi(axes[0],xlabel="$j$",ylabel="$Q$")
    _=grafy.ekonomicke_osi(axes[1],xlabel="$j$",ylabel=None)
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=j,
        points=[Q(j,b,e,A),k/(1-d)/(A-b)],
        value_label=r"$j^*_1$",
        axis_labels=[
            r"$Q^*_1$",
            r"$\frac{k}{(1-d)(A_1-b)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
        axis_label_dy=[0,3]
    )
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=jA,
        points=[Q(jA,b,e,A2),k/(1-d)/(A2-b)],
        value_label=r"$j^*_2$",
        axis_labels=[
            r"$Q^*_2$",
            r"$\frac{k}{(1-d)(A_2-b)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
        axis_label_dy=[0,-3]
    )

    plt.savefig(
        "images/hlad_prac/graf_hlad_A.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_hlad_e #
###############

def graf_hlad_e():
    psi,k,b,A,e,d,eta,j=exo()
    e2=1.1*e
    je=(e2*(1-d)*(A-b)/k)**(1/eta)
    fig, axes = plt.subplots(2, 1, figsize=(8, 10))
    grafy.nakresli_funkciu(axes[1],lambda j: m_1_j1(j, e),
                           xmin=0.05,xmax=2.,
                           nazov=r"$e_1m\left( \frac{1}{j},1 \right)$")
    grafy.nakresli_funkciu(axes[1],lambda je: m_1_j1(je, e2),
                           xmin=0.05,xmax=2.,
                           nazov=r"$e_2m\left( \frac{1}{j},1 \right)$",
                          color="red")
    grafy.nakresli_funkciu(axes[0],lambda j: Q(j,b,e,A),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b+e_1m(1,j)d(A-b)$")
    grafy.nakresli_funkciu(axes[0],lambda je: Q(je,b,e2,A),
                           xmin=0.05,xmax=2.,
                           nazov="$P(Q)=b+e_2m(1,j)d(A-b)$",
                          color="red")
    axes[0].set_ylim(0, Q(je, b, e2, A) + 11)
    for ax in axes: ax.legend() 
    _=grafy.ekonomicke_osi(axes[0],xlabel="$j$",ylabel="$Q$")
    _=grafy.ekonomicke_osi(axes[1],xlabel="$j$",ylabel=None)
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=j,
        points=[Q(j,b,e,A),k/(1-d)/(A-b)],
        value_label=r"$j^*_1$",
        axis_labels=[
            r"$Q^*_1$",
            r"$\frac{k}{(1-d)(A-b)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
    )
    _=grafy.prepoj_body(
        fig=fig,
        axes=axes,
        value=je,
        points=[Q(je,b,e2,A),k/(1-d)/(A-b)],
        value_label=r"$j^*_2$",
        axis_labels=[
            r"$Q^*_2$",
            r"$\frac{k}{(1-d)(A-b)}$"
        ],
        value_label_shift=[-7,0],
        axis_label_dx=[-7,-7],
    )
    plt.savefig(
        "images/hlad_prac/graf_hlad_e.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##########################
# graf_procyklickost_apn #
##########################

def graf_procyklickost_apn():
    import matplotlib.pyplot as plt
    import grafy
    import numpy as np
    def produkcna1(N):
        return 1**(1/2)*N**(1/2)
    def produkcna2(N):
        return 1.5*1**(1/2)*N**(1/2)
    def apn1(N):
        return produkcna1(N1)/N1*N
    def apn2(N):
        return produkcna2(N2)/N2*N
    N1=0.8
    N2=1.
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax, produkcna1, 0, 1.5, nazov=r"$A_1F(K,N)$", color="blue")
    grafy.nakresli_funkciu(ax, produkcna2, 0, 1.5, nazov=r"$A_2F(K,N)$", color="red")
    grafy.nakresli_bod(ax, N1, produkcna1(N1), nazov=None, vodiace_ciary=True, color='blue')
    grafy.nakresli_bod(ax, N2, produkcna2(N2), nazov=None, vodiace_ciary=True, color='red')
    grafy.vyznac_na_osi(ax, x=N1, y=produkcna1(N1), popis_x="$N_1$", popis_y="$Y_1$", color='blue')
    grafy.vyznac_na_osi(ax, x=N2, y=produkcna2(N2), popis_x="$N_2$", popis_y="$Y_2$", color='red')
    grafy.nakresli_funkciu(ax, apn1, 0, N1, nazov="sklon = $APN_1$", 
                           linewidth=1., linestyle="--", color="blue")
    grafy.nakresli_funkciu(ax, apn2, 0, N2, nazov="sklon = $APN_2$", 
                           linewidth=1., linestyle="--", color="red")
    ax.legend()
    _=grafy.ekonomicke_osi(ax, "$N$","$Y$")
    plt.savefig(
        "images/hlad_prac/graf_procyklickost_apn.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

graf_benefit_hladania()
graf_benefit_bod()
graf_dopyt_hlad()
graf_rovnovaha_hlad()
graf_hlad_b()
graf_hlad_A()
graf_hlad_e()
graf_procyklickost_apn()
```

## Dynamický model zatvorenej výrobnej ekonomiky

Obrázky sa nahrajú do podpriečinka `dge`.

```python
import grafy
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

Path("images/dge").mkdir(exist_ok=True)
a=0.  # intercept Ns
b=1.  # sklon     Ns
c=2.  # intercept Nd
d=-1. # sklon     Nd
e=0.  # intercept Ys
f=1.  # sklon     Ys
g=2.  # intercept Yd
e=-1. # sklon     Yd

#################
# graf_NS_w_DGE #
#################

def graf_NS_w_DGE():
    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Ns(a,b,w):
        return a+b*w
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$")
    grafy.nakresli_funkciu(ax,lambda w:Ns(a2,b,w),xmin-a2,xmax,nazov="$N^s(w,r_2,we)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$N^s$","$w$")
    plt.savefig(
        "images/dge/graf_NS_w_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##################
# graf_NS_we_DGE #
##################

def graf_NS_we_DGE():
    a=0 # intercept Ns
    a2=0.5
    b=1 # sklon Ns
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Ns(a,b,w):
        return a+b*w
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r,we_1)$")
    grafy.nakresli_funkciu(ax,lambda w:Ns(a2,b,w),xmin,xmax,nazov="$N^s(w,r,we_2)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$N^s$","$w$")
    plt.savefig(
        "images/dge/graf_NS_we_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_Cd_DGE #
#################

def graf_Cd_DGE():
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:2-r,xmin,xmax,nazov="$C^d(r,we_1)$")
    grafy.nakresli_funkciu(ax,lambda r:2.5-r,xmin+.5,xmax,nazov="$C^d(r,we_2)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$C^d$","$r$")
    plt.savefig(
        "images/dge/graf_Cd_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##################
# graf_Cd_we_DGE #
##################

# Tento graf neexistuje

def graf_Cd_we_DGE():
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda Y:Y,xmin,xmax,nazov="$C^d(Y,r,we_1)$")
    grafy.nakresli_funkciu(ax,lambda Y:0.5+Y,xmin,xmax,nazov="$C^d(Y,r,we_2)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$Y$","$C^d$")
    plt.savefig(
        "images/dge/graf_Cd_we_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_Nd_w_DGE #
#################

def graf_Nd_w_DGE():
    c=2. 
    d=-1.
    c2=2.5
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Nd(c,d,w):
        return c+d*w
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A_1,K)$")
    grafy.nakresli_funkciu(ax,lambda w:Nd(c2,d,w),xmin,xmax+0.5,nazov="$N^d(w,A_2,K)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$N^d$","$w$")
    plt.savefig(
        "images/dge/graf_Nd_w_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_Nd_K_DGE #
#################

def graf_Nd_K_DGE():
    c=2. 
    d=-1.
    c2=1.5
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Nd(c,d,w):
        return c+d*w
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A,K_1)$")
    grafy.nakresli_funkciu(ax,lambda w:Nd(c2,d,w),xmin,xmax-0.5,nazov="$N^d(w,A,K_2)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$N^d$","$w$")
    plt.savefig(
        "images/dge/graf_Nd_K_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_Id_DGE #
###############

def graf_Id_DGE():
    c=2. 
    d=-1.
    c2=2.5
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:c+d*r,xmin,xmax,nazov="$I^d(r,A'_1,K_1)$")
    grafy.nakresli_funkciu(ax,lambda r:c2+d*r,xmin,xmax+0.5,nazov="$I^d(r,A'_2,K_2)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$I^d$","$r$")
    plt.savefig(
        "images/dge/graf_Id_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_Id_x_DGE #
#################

def graf_Id_x_DGE():
    c=2. 
    d=-1.
    c2=1.5
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:c+d*r,xmin,xmax,nazov="$I^d(r+r_{p1},A,K)$")
    grafy.nakresli_funkciu(ax,lambda r:c2+d*r,xmin,xmax-0.5,nazov="$I^d(r+r_{p2},A,K)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$I^d$","$r$")
    plt.savefig(
        "images/dge/graf_Id_x_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###################
# graf_vystup_DGE #
###################

def graf_vystup_DGE():

    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. 
    d=-1.
    c2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Ns(a,b,w):
        return a+b*w
    def Nd(c,d,w):
        return c+d*w
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 1, figsize=(8, 10))
    grafy.nakresli_funkciu(axes[0],lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r,we)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda N:N**(0.5),0,xmax,nazov="$AF(K,N)$",color="blue")
    #grafy.nakresli_funkciu(axes[0],lambda w:Ns(a2,b,w),xmin-a2,xmax,nazov="$N^s(w,r_2,we)$",
    #color="red")
    #grafy.nakresli_funkciu(axes[0],lambda w:Nd(c2,d,w),xmin,xmax-0.5,nazov="$N^d(w,A,K_2)$",
    #color="red")
    axes[0].legend()
    axes[1].legend()
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$N$","$Y$")
    _=grafy.prepoj_body(fig,axes,1,[1,1],axis_labels=["$w^*$","$Y^*$"],value_label="$N^*$",
                       value_label_shift=[-14,0])
    plt.savefig(
        "images/dge/graf_vystup_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_Ys_DGE #
###############

def graf_Ys_DGE():
    
    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. 
    d=-1.
    c2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Ns(a,b,w):
        return a+b*w
    def Nd(c,d,w):
        return c+d*w
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(3, 1, figsize=(8, 15))
    grafy.nakresli_funkciu(axes[0],lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda N:N**(0.5),0,xmax,nazov="$AF(K,N)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:Ns(a2,b,w),xmin-a2,xmax,nazov="$N^s(w,r_2,we)$",
                           color="red")
    #grafy.nakresli_funkciu(axes[0],lambda w:Nd(c2,d,w),xmin,xmax-0.5,nazov="$N^d(w,A,K_2)$",
    #color="red")
    grafy.nakresli_funkciu(axes[2],lambda w:Ns(a,b,w),xmin,xmax,nazov="$Y^s(r)$",color="blue")
    grafy.nakresli_bod(axes[2], 1, 1, nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[2], 1.25**(0.5), 1.25**(0.5), nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[2], x=1, y=1, popis_x="$Y_1$", popis_y="$r_1$")
    grafy.vyznac_na_osi(axes[2], x=1.25**(0.5), y=1.25**(0.5), popis_x="$Y_2$", popis_y="$r_2$")
    axes[0].legend(loc="upper center")
    axes[1].legend()
    axes[2].legend()
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$N$","$Y$")
    grafy.ekonomicke_osi(axes[2],"$Y$","$r$")
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],1,[1,1],axis_labels=["$w_1$","$Y_1$"],
                        value_label="$N_1$",
                       value_label_shift=[-14,0])
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],1.25,[0.75,1.25**(0.5)],axis_labels=["$w_2$","$Y_2$"],
                        value_label="$N_2$",
                       value_label_shift=[-14,0])
    
    plt.savefig(
        "images/dge/graf_Ys_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_Ys_G_DGE #
#################

def graf_Ys_G_DGE():
    
    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. 
    d=-1.
    c2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Ns(a,b,w):
        return a+b*w
    def Nd(c,d,w):
        return c+d*w

    grafy.set_theory_style()
    fig,axes=plt.subplots(3, 1, figsize=(8, 15))
    grafy.nakresli_funkciu(axes[0],lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r,we_1)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda N:N**(0.5),0,xmax,nazov="$AF(K,N)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:Ns(a2,b,w),xmin-a2,xmax,nazov="$N^s(w,r,we_2)$",
                           color="red")
    #grafy.nakresli_funkciu(axes[0],lambda w:Nd(c2,d,w),xmin,xmax-0.5,nazov="$N^d(w,A,K_2)$",
    #color="red")
    grafy.nakresli_funkciu(axes[2],lambda w:Ns(a,b,w),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[2],lambda w:Ns(1-1.25**0.5,b,w),xmin+1.25**0.5-1,
                       xmax,nazov="$Y^s(r,we_2,A,K)$",
                       color="red")
    grafy.nakresli_bod(axes[2], 1, 1, nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[2], 1.25**(0.5), 1, nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[2], x=1, y=1, popis_x="$Y_1$", popis_y="$r$")
    grafy.vyznac_na_osi(axes[2], x=1.25**(0.5), y=1, popis_x="$Y_2$", popis_y=None)
    axes[0].legend(loc="upper center")
    axes[1].legend()
    axes[2].legend()
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$N$","$Y$")
    grafy.ekonomicke_osi(axes[2],"$Y$","$r$")
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],1,[1,1],axis_labels=["$w_1$","$Y_1$"],
                        value_label="$N_1$",
                       value_label_shift=[-14,0])
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],1.25,[0.75,1.25**(0.5)],axis_labels=["$w_2$","$Y_2$"],
                        value_label="$N_2$",
                       value_label_shift=[-14,0])    
    plt.savefig(
        "images/dge/graf_Ys_G_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#################
# graf_Ys_A_DGE #
#################

def graf_Ys_A_DGE():
    
    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. 
    d=-1.
    c2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def Ns(a,b,w):
        return a+b*w
    def Nd(c,d,w):
        return c+d*w
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(3, 1, figsize=(8, 16))
    grafy.nakresli_funkciu(axes[0],lambda w:Ns(a,b,w),xmin,xmax,nazov="$N^s(w,r,we)$",color="green")
    grafy.nakresli_funkciu(axes[0],lambda w:Nd(c,d,w),xmin,xmax,nazov="$N^d(w,A_1,K)$",color="blue")
    grafy.nakresli_funkciu(axes[1],lambda Y:Y**(0.5),0,xmax,nazov="$A_1F(K,N)$",color="blue")
    grafy.nakresli_funkciu(axes[1],lambda Y:1.5*Y**(0.5),0,xmax,nazov="$A_2F(K,N)$",color="red")
    #grafy.nakresli_funkciu(axes[0],lambda w:Ns(a2,b,w),
    #xmin-a2,xmax,nazov="$N^s(w,r,we_2)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:Nd(c2,d,w),xmin,xmax,nazov="$N^d(w,A_2,K)$",color="red")
    grafy.nakresli_funkciu(axes[2],lambda w:Ns(a,b,w),xmin,xmax,nazov="$Y^s(r,we,A_1,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[2],lambda w:Ns(1-1.5*1.25**(0.5),b,w),xmin+1.5*1.25**(0.5)-1,
                           xmax,nazov="$Y^s(r,we,A_2,K)$",
                           color="red")
    grafy.nakresli_bod(axes[2], 1, 1, nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[2], 1.5*1.25**(0.5), 1, nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[2], x=1, y=1, popis_x="$Y_1$", popis_y="$r$")
    grafy.vyznac_na_osi(axes[2], x=1.5*1.25**(0.5), y=1, popis_x="$Y_2$", popis_y=None)
    axes[0].legend(loc="upper center")
    axes[1].legend()
    axes[2].legend()
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$N$","$Y$")
    grafy.ekonomicke_osi(axes[2],"$Y$","$r$")
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],1,[1,1],
                        axis_labels=["$w_1$","$Y_1$"],value_label="$N_1$",
                       value_label_shift=[-14,0])
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],1.25,[1.25,1.5*1.25**(0.5)],
                        axis_labels=["$w_2$","$Y_2$"],
                        value_label="$N_2$",
                       value_label_shift=[-14,0])

    plt.savefig(
        "images/dge/graf_Ys_A_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_Yd_DGE #
###############

def graf_Yd_DGE():

    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:2-r,xmin,xmax,nazov="$Y^d(r,we_1,A'_1,K_1)$")
    grafy.nakresli_funkciu(ax,lambda r:2.5-r,xmin+0.5,xmax,nazov="$Y^d(r,we_2,A'_2,K_2)$",color="red")
    ax.legend()
    grafy.ekonomicke_osi(ax,"$Y^d$","$r$")

    plt.savefig(
        "images/dge/graf_Yd_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

############
# graf_DGE #
############

def graf_DGE():

    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=1.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r,we)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*$", popis_y="$w^*$")
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*$", popis_y="$r^*$")
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.95)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.95)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_DGE_G #
##############

def graf_DGE_G():

    a=0 # intercept Ns
    a2=-0.25
    a3=-0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we_1)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin-a2,xmax,
                           nazov="$N^s(w,r_1,we_2)$",color="orange")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a3,b,w),xmin-a3,xmax,
                           nazov="$N^s(w,r_2,we_2)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A',K,G_1)$",
                           color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g2,h,r),xmin,xmax+0.5,
                           nazov="$Y^d(r,we_2,A',K,G_2)$",color="indianred")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1.25, 0.75, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 1.5-0.25/2, 1.125, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1.5, 1., nazov="$Y^d(r^*_1)$", vodiace_ciary=True,color="indianred")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*_1$", popis_y="$w^*_1$",color="blue")
    grafy.vyznac_na_osi(axes[0], x=1.25, y=0.75, popis_x="$N^*_2$", popis_y="$w^*_2$",color="red")
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*_1$", popis_y="$r^*_1$",dy_y=-2.5,
                        color="blue")
    grafy.vyznac_na_osi(axes[1], x=1.5-0.25/2, y=1.125, popis_x="$Y^*_2$", popis_y="$r^*_2$",dy_y=2.5,
                    color="red")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.1)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.1)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE_G.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_DGE_K #
##############

def graf_DGE_K():

    a=0 # intercept Ns
    a2=-0.5
    #a3=-0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=1.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin-a2,xmax,
                           nazov="$N^s(w,r_2,we)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K_1)$",color="green")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c2,d,w),xmin,xmax-0.5,nazov="$N^d(w,A,K_2)$",
                           color="indianred")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K_1)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin,xmax-0.5,nazov="$Y^s(r,we,A,K_2)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K_1)$",
                           color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g2,h,r),xmin+0.5,xmax,
                           nazov="$Y^d(r,we,A',K_2)$",color="indianred")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1., 0.5, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 1, 1.5, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*$", popis_y="$w^*_1$",color="blue")
    grafy.vyznac_na_osi(axes[0], x=1, y=0.5, popis_y="$w^*_2$",color="red")
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*$", popis_y="$r^*_1$",color="blue")
    grafy.vyznac_na_osi(axes[1], x=1., y=1.5, popis_y="$r^*_2$",color="red")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE_K.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_DGE_A #
##############

def graf_DGE_A():

    a=0 # intercept Ns
    a2=0.25
    #a3=-0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=2.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin,xmax-a2,
                           nazov="$N^s(w,r_2,we)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A_1,K)$",color="green")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c2,d,w),xmin+0.5,xmax,nazov="$N^d(w,A_2,K)$",
                           color="indianred")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A_1,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin+0.5,xmax,nazov="$Y^s(r,we,A_2,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1.125, 1.25+0.125, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 1.25, 0.75, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*_1$", popis_y="$w^*_1$",color="blue",
                        dx_x=-2.5)
    grafy.vyznac_na_osi(axes[0], x=1.125, y=1.25+0.125, popis_x="$N^*_2$", popis_y="$w^*_2$",
                        color="red",
                       dx_x=2.5)
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*_1$", popis_y="$r^*_1$",color="blue")
    grafy.vyznac_na_osi(axes[1], x=1.25, y=0.75, popis_x="$Y^*_2$", popis_y="$r^*_2$",color="red")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE_A.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_DGE_Ap #
##############

def graf_DGE_Ap():   
    
    a=0 # intercept Ns
    a2=0.25
    a3=-0.25
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=2.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we_1)$",color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin,xmax-a2,
                           nazov="$N^s(w,r_1,we_2)$",color="orange")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a3,b,w),xmin-a3,xmax,
                           nazov="$N^s(w,r_2,we_2)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin,xmax-e2,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A'_1,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g2,h,r),xmin+0.5,xmax,nazov="$Y^d(r,we_2,A'_2,K)$",
                           color="indianred")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1.125, 0.875, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 1.125, 1.375, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*_1$", popis_y="$w^*_1$",color="blue",
                        dx_x=-2.,dy_y=+2.)
    grafy.vyznac_na_osi(axes[0], x=1.125, y=0.875, popis_x="$N^*_2$", popis_y="$w^*_2$",color="red",
                       dx_x=2.,dy_y=-2.)
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*_1$", popis_y="$r^*_1$",color="blue",
                       dx_x=-2.5)
    grafy.vyznac_na_osi(axes[1], x=1.125, y=1.375, popis_x="$Y^*_2$", popis_y="$r^*_2$",color="red",
                       dx_x=2.5)
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE_Ap.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_DGE_rp #
###############

def graf_DGE_rp():   

    a=0 # intercept Ns
    a2=-0.25
    a3=0.25
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=2.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we_1)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin-a2,xmax,
                           nazov="$N^s(w,r_1,we_2)$",color="orange")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a3,b,w),xmin,xmax-a3,
                           nazov="$N^s(w,r_2,we_2)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A',K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g2,h,r),xmin,xmax-0.5,nazov="$Y^d(r,we_2,A',K)$",
                           color="indianred")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1-0.125, 1.125, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 1-0.125, 1-0.25-0.125, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*_1$", popis_y="$w^*_1$",color="blue",
                        dx_x=+2.5,dy_y=-1.5)
    grafy.vyznac_na_osi(axes[0], x=1-0.125, y=1.125, popis_x="$N^*_2$", popis_y="$w^*_2$",color="red",
                       dx_x=-2.5,dy_y=+1.5)
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*_1$", popis_y="$r^*_1$",color="blue",
                       dx_x=2.5)
    grafy.vyznac_na_osi(axes[1], x=1-0.125, y=1-0.375, popis_x="$Y^*_2$", popis_y="$r^*_2$",
                        color="red",
                       dx_x=-2.5)
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")
    
    plt.savefig(
        "images/dge/graf_DGE_rp.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_DGE_e #
##############

def graf_DGE_e():
    
    a=0 # intercept Ns
    a2=0.5
    a3=0.25
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=1.75
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s_1(w,r_1,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin,xmax-a2,
                           nazov="$N^s_2(w,r_1,we)$",color="orange")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a3,b,w),xmin,xmax-a3,
                           nazov="$N^s_2(w,r_2,we)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d_1(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c2,d,w),xmin,xmax-0.25,nazov="$N^d_2(w,A,K)$",
                           color="indianred")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s_1(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin,xmax-e2,nazov="$Y^s_2(r,we,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 0.75, 1., nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 0.75, 1.25, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*_1$", popis_y="$w^*$",color="blue")
    grafy.vyznac_na_osi(axes[0], x=0.75, y=1., popis_x="$N^*_2$", color="red")
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*_1$", popis_y="$r^*_1$",color="blue")
    grafy.vyznac_na_osi(axes[1], x=0.75, y=1.25, popis_x="$Y^*_2$", popis_y="$r^*_2$",color="red")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.23)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.23)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE_e.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_DGE_Ns #
###############

def graf_DGE_Ns():

    a=0 # intercept Ns
    a2=0.5
    a3=0.25
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=1.75
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s_1(w,r_1,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin,xmax-a2,
                           nazov="$N^s_2(w,r_1,we)$",color="orange")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a3,b,w),xmin,xmax-a3,
                           nazov="$N^s_2(w,r_2,we)$",color="red")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s_1(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(e2,f,r),xmin,xmax-e2,nazov="$Y^s_2(r,we,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1-0.125, 1.125, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[1], 1-0.125, 1.125, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$N^*_1$", popis_y="$w^*_1$",color="blue",
                       dx_x=2.,dy_y=-1.5)
    grafy.vyznac_na_osi(axes[0], x=1-0.125, y=1.125, popis_x="$N^*_2$", popis_y="$w^*_2$", color="red",
                       dx_x=-2.,dy_y=1.5)
    grafy.vyznac_na_osi(axes[1], x=1., y=1., popis_x="$Y^*_1$", popis_y="$r^*_1$",color="blue",
                       dx_x=1.5,dy_y=-1.5)
    grafy.vyznac_na_osi(axes[1], x=1-0.125, y=1.125, popis_x="$Y^*_2$", popis_y="$r^*_2$",color="red",
                       dx_x=-1.5,dy_y=1.5)
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    axes[1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.17)
    )
    ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$Y$","$r$")

    plt.savefig(
        "images/dge/graf_DGE_Ns.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

graf_NS_w_DGE()
graf_NS_we_DGE()
graf_Cd_DGE()
graf_Nd_w_DGE()
graf_Nd_K_DGE()
graf_Id_DGE()
graf_Id_x_DGE()
graf_vystup_DGE()
graf_Ys_DGE()
graf_Ys_G_DGE()
graf_Ys_A_DGE()
graf_Yd_DGE()
graf_DGE()
graf_DGE_G()
graf_DGE_K()
graf_DGE_A()
graf_DGE_Ap()
graf_DGE_rp()
graf_DGE_e()
graf_DGE_Ns()
```

## Modely otvorenej ekonomiky

Obrázky sa vygenerujú do priečinka `otvorena`.

```python
import grafy
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

Path("images/otvorena").mkdir(exist_ok=True)

###################
# graf_regulaciaK #
###################

def graf_regulaciaK():

    grafy.set_theory_style()
    y,yp,r,beta=200,200,0.01,0.3
    r_regulacia=1/beta-1
    opt=grafy.optimum(y,yp,r,beta=beta)
    regulacia=grafy.optimum(y,yp,r_regulacia,beta=beta)
    c_max=y+yp/(1+r)
    cp_max=yp+y*(1+r_regulacia)
    fig,ax=plt.subplots()
    
    rozpocet=grafy.nakresli_rozpocet_intertemporal(ax,r,y,yp,color="blue")
    rozpocet_regulacia=grafy.nakresli_rozpocet_intertemporal(
        ax,r_regulacia,y,yp,color="red")
    grafy.nakresli_indiferencnu_intertemporal(ax,opt["u"],beta,c_min=157,c_max=c_max)
    grafy.nakresli_indiferencnu_intertemporal(ax,regulacia["u"],beta,c_min=129,c_max=c_max,
                                              color="red")
    grafy.nakresli_bod(ax,opt["c"],opt["cp"],nazov=None,vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax,regulacia["c"],regulacia["cp"],nazov="E",vodiace_ciary=True,
                       color="red")
    grafy.vyznac_na_osi(ax,x=opt["c"],y=opt["cp"],popis_x="$c^*$",popis_y="$c'^*$",
                        color="blue")
    grafy.vyznac_na_osi(ax,x=regulacia["c"],y=regulacia["cp"],popis_x="$c_r=y-t$",
                        popis_y="$c'_r=y'-t'$",
                        color="red")
    #ax.set_ylim(0, cp_max+50)
    grafy.ekonomicke_osi(ax,"$c$","$c'$")

    plt.savefig(
        "images/otvorena/graf_regulaciaK.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

################
# graf_DGE_otv #
################

def graf_DGE_otv():

    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin+0.5,xmax,nazov="$Y^d(r,we,A',K,X)$",
                           color="indianred")
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 1.25, 1.25, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r$",color="blue")
    grafy.vyznac_na_osi(ax, x=1.25, y=1.25, popis_x="$Y_2$", popis_y="$r^*$",color="red")
    ax.legend(
        loc="center",
        bbox_to_anchor=(0.625, 0.88)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    
    grafy.ekonomicke_osi(ax,"$Y$","$r$")
    
    plt.savefig(
        "images/otvorena/graf_DGE_otv.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_otv_r #
##############

def graf_otv_r():  
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K,X_1)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin+0.5,xmax,nazov="$Y^d(r,we,A',K,X_2)$",
                           color="indianred")
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 1.25, 1.25, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(ax, x=1.25, y=1.25, popis_x="$Y_2$", popis_y="$r_2$",color="red")
    ax.legend(
        loc="center",
        bbox_to_anchor=(0.625, 0.9)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    
    grafy.ekonomicke_osi(ax,"$Y$","$r$")
    
    plt.savefig(
        "images/otvorena/graf_otv_r.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_otv_G #
##############

def graf_otv_G():
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    g3=2.25
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A',K,G_1,X_1)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin+0.5,xmax,
                           nazov="$Y^d(r,we_2,A',K,G_2,X_1)$",color="yellow")
    grafy.nakresli_funkciu(ax,lambda r:fun(g3,h,r),xmin+0.25,xmax,
                           nazov="$Y^d(r,we_2,A',K,G_2,X_2)$",color="indianred")
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 1.25, 1., nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(ax, 1.5, 1., nazov="$Y^d(r^*,X_1)$", vodiace_ciary=True,
                       color="yellow")
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r^*$",dy_y=-2.5,
                        color="blue")
    grafy.vyznac_na_osi(ax, x=1.25, y=1., popis_x="$Y_2$",color="red")
    
    ax.legend(
        loc="center",
        bbox_to_anchor=(0.625, 1.)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(ax,"$Y$","$r$")

    plt.savefig(
        "images/otvorena/graf_otv_G.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_otv_A #
##############

def graf_otv_A():

    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A_1,K)$",
                           color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we,A_2,K)$",
                           color="red")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K,G,X_1)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin+0.5,xmax,
                           nazov="$Y^d(r,we,A',K,G,X_2)$",color="indianred")
    
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 1.5, 1., nazov=None, vodiace_ciary=True,color="red")
    
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r^*$",dy_y=-2.5,
                        color="blue")
    grafy.vyznac_na_osi(ax, x=1.5, y=1., popis_x="$Y_2$",color="red")
    
    ax.legend(
        loc="center",
        bbox_to_anchor=(0.625, 0.95)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(ax,"$Y$","$r$")
    
    plt.savefig(
        "images/otvorena/graf_otv_A.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_otv_Ap #
###############

def graf_otv_Ap():
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.5
    g3=1.75
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(e2,f,r),xmin,xmax-e2,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A'_1,K,G,X_1)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin+0.5,xmax,
                           nazov="$Y^d(r,we_2,A'_2,K,G,X_1)$",
                           color="yellow")
    grafy.nakresli_funkciu(ax,lambda r:fun(g3,h,r),xmin,xmax-0.25,
                           nazov="$Y^d(r,we_2,A'_2,K,G,X_2)$",color="indianred")
    
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 0.75, 1., nazov=None, vodiace_ciary=True,color="red")
    
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r^*$",dy_y=-2.5,
                        color="blue")
    grafy.vyznac_na_osi(ax, x=0.75, y=1., popis_x="$Y_2$",color="red")
    
    ax.legend(
        loc="center",
        bbox_to_anchor=(0.565, 1.05)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(ax,"$Y$","$r$")
    
    plt.savefig(
        "images/otvorena/graf_otv_Ap.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_otv_rp #
###############
    
def graf_otv_rp():
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=1.5
    g3=2.25
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A',K,G,X_1)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin,xmax-0.5,nazov="$Y^d(r,we_2,A',K,G,X_1)$",
                           color="yellow")
    grafy.nakresli_funkciu(ax,lambda r:fun(g3,h,r),xmin+0.25,xmax,
                           nazov="$Y^d(r,we_2,A',K,G,X_2)$",color="indianred")
    
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 1.25, 1., nazov=None, vodiace_ciary=True,color="red")
    
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r^*$",dy_y=-2.5,color="blue")
    grafy.vyznac_na_osi(ax, x=1.25, y=1., popis_x="$Y_2$",color="red")
    
    ax.legend(
        loc="center",
        bbox_to_anchor=(0.565, 0.94)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(ax,"$Y$","$r$")

    plt.savefig(
        "images/otvorena/graf_otv_rp.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_otv_N #
##############

def graf_otv_N():

    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=1.5
    g3=2.25
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s_1(r,we,A,K)$",color="blue")
    grafy.nakresli_funkciu(ax,lambda r:fun(e2,f,r),xmin,xmax-e2,nazov="$Y^s_2(r,we,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(ax,lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K,G,X_1)$",
                           color="green")
    grafy.nakresli_funkciu(ax,lambda r:fun(g2,h,r),xmin,xmax-0.5,nazov="$Y^d(r,we,A',K,G,X_2)$",
                           color="indianred")
    
    grafy.nakresli_bod(ax, 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, 0.5, 1., nazov=None, vodiace_ciary=True,color="red")
    
    grafy.vyznac_na_osi(ax, x=1., y=1., popis_x="$Y_1$", popis_y="$r^*$",dy_y=-2.5,color="blue")
    grafy.vyznac_na_osi(ax, x=0.5, y=1., popis_x="$Y_2$",color="red")
    
    ax.legend(
        loc="center left",
        bbox_to_anchor=(0.635, 0.5)
    )
    #ymin, ymax = axes[0].get_ylim()
    #ax.set_ylim(ymin, ymax*1.1)
    grafy.ekonomicke_osi(ax,"$Y$","$r$")
    
    plt.savefig(
        "images/otvorena/graf_otv_N.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()   

graf_regulaciaK()
graf_DGE_otv()
graf_otv_r()
graf_otv_G()
graf_otv_A()
graf_otv_Ap()
graf_otv_rp()
graf_otv_N()
```

## Peniaze, banky a menová politika

Obrázky sa vygenerujú v podpriečinku `peniaze`.

```python
import grafy
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path

Path("images/peniaze").mkdir(exist_ok=True)

#############
# grafy_MdL #
#############

def grafy_MdL():

    grafy.set_theory_style()
    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda P:P,0,2)
    grafy.ekonomicke_osi(ax,"$M^d$","$P$")

    plt.savefig(
        "images/peniaze/graf_MdL.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda P:P,0,2,nazov="$M^d_1=PL(Y_1,r_1)$",color="blue")
    grafy.nakresli_funkciu(ax,lambda P:0.5*P,0,2,nazov="$M^d_2=PL(Y_2,r_2)$",color="red")
    grafy.ekonomicke_osi(ax,"$M^d$","$P$")
    ax.legend()

    plt.savefig(
        "images/peniaze/graf_MdL_posun.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#############
# graf_Mtrh #
#############

def graf_Mtrh():

    fig,ax=plt.subplots()
    grafy.nakresli_bod(ax,1,1,nazov=None,vodiace_ciary=True)
    grafy.nakresli_funkciu(ax,lambda P:P,0,2,nazov="$M^d=PL(Y,r)$",color="green")
    grafy.nakresli_zvislu_ciaru(ax,1,nazov="$M^*=M^s$",color="blue")
    grafy.vyznac_na_osi(ax,x=1,y=1,popis_x="$M^*$",popis_y="$P^*$")
    grafy.ekonomicke_osi(ax,"$M$","$P$")
    ax.legend()
    
    plt.savefig(
        "images/peniaze/graf_Mtrh.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###################
# graf_rovnovahaM #
###################

def graf_rovnovahaM():
    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=1.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 2, figsize=(8, 8)
                         )
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r^*,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0,0], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[0,1], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[0,0], x=1., y=1., popis_x="$N^*$", popis_y="$w^*$")
    grafy.vyznac_na_osi(axes[0,1], x=1., y=1., popis_x="$Y^*$", popis_y="$r^*$")
    axes[0,0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.95)
    )
    axes[0,1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.95)
    )
    grafy.ekonomicke_osi(axes[0,0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[0,1],"$Y$","$r$")
    grafy.nakresli_bod(axes[1,0],1,1,nazov=None,vodiace_ciary=True)
    grafy.nakresli_funkciu(axes[1,0],lambda P:P,0,2,nazov="$M^d=PL(Y^*,r^*)$",color="green")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1,nazov="$M^*=M^s$",color="blue")
    grafy.vyznac_na_osi(axes[1,0],x=1,y=1,popis_x="$M^*$",popis_y="$P^*$")
    axes[1,0].legend(
        loc="lower right",
        bbox_to_anchor=(1.2, 0.1)
    )
    grafy.ekonomicke_osi(axes[1,0],"$M$","$P$")
    #ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    axes[1,1].axis('off')
        
    plt.savefig(
        "images/peniaze/graf_rovnovahaM.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

################
# graf_Mponuka #
################

def graf_Mponuka():

    a=0 # intercept Ns
    a2=-0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=1.5
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 2, figsize=(8, 8)
                         )
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0,0], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[0,1], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[0,0], x=1., y=1., popis_x="$N_1$", popis_y="$w_1$")
    grafy.vyznac_na_osi(axes[0,1], x=1., y=1., popis_x="$Y_1$", popis_y="$r_1$")
    axes[0,0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.95)
    )
    axes[0,1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.95)
    )
    grafy.ekonomicke_osi(axes[0,0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[0,1],"$Y$","$r$")
    grafy.nakresli_bod(axes[1,0],1,1,nazov=None,vodiace_ciary=True,color="blue")
    grafy.nakresli_funkciu(axes[1,0],lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1,nazov="$M_1$",color="blue")
    grafy.vyznac_na_osi(axes[1,0],x=1,y=1,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(axes[1,0],1.25,1.25,nazov=None,vodiace_ciary=True,color="red")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1.25,nazov="$M_2$",color="red")
    grafy.vyznac_na_osi(axes[1,0],x=1.25,y=1.25,popis_x="$M_2$",popis_y="$P_2$",color="red")
    axes[1,0].legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.1)
    )
    grafy.ekonomicke_osi(axes[1,0],"$M$","$P$")
    #ymin, ymax = axes[0].get_ylim()
    #axes[0].set_ylim(ymin, ymax*1.1)
    #axes[1].set_ylim(ymin, ymax*1.1)
    axes[1,1].axis('off')

    plt.savefig(
        "images/peniaze/graf_Mponuka.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

###############
# graf_Mdopyt #
###############

def graf_Mdopyt():

    fig,ax=plt.subplots()
    grafy.nakresli_funkciu(ax,lambda P:P,0,2,nazov="$PL_1(Y,r)$",color="green")
    grafy.nakresli_bod(ax,1,1,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(ax,x=1.,y=0.75,popis_y="$P_2$",color="red")
    grafy.nakresli_zvislu_ciaru(ax,1,nazov="$M$",color="blue")
    grafy.vyznac_na_osi(ax,x=1,y=1,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(ax,1.,0.75,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.nakresli_funkciu(ax,lambda P:0.75*P,0,2,nazov="$PL_2(Y,r)$",color="indianred")
    
    ax.legend(
    )
    grafy.ekonomicke_osi(ax,"$M$","$P$")

    plt.savefig(
        "images/peniaze/graf_Mdopyt.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##################
# graf_likvidita #
##################

def graf_likvidita():

    fig,ax=plt.subplots()
    grafy.nakresli_bod(ax,1,1,nazov=None,vodiace_ciary="horizontal")
    grafy.nakresli_funkciu(ax,lambda P:P,0,2,nazov="$PL(Y,0)$",color="green")
    grafy.nakresli_zvislu_ciaru(ax,1,nazov="$M+B$",color="blue")
    grafy.vyznac_na_osi(ax,x=1,y=1,popis_x="$M+B$",popis_y="$P^*$")
    grafy.ekonomicke_osi(ax,"Likvidita","$P$")
    ax.legend()

    plt.savefig(
        "images/peniaze/graf_likvidita.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

grafy_MdL()
graf_Mtrh()
graf_rovnovahaM()
graf_Mponuka()
graf_Mdopyt()
graf_likvidita()
```

## Modely hospodárskych cyklov s pružnými cenami a mzdami

Podpriečinok, v ktorom sa vygenerujú obrázky sa nazýva `pruzne`.

```python
import grafy
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path
from scipy.optimize import brentq

Path("images/pruzne").mkdir(exist_ok=True)

############
# graf_RBC #
############

def graf_RBC():

    a=0 # intercept Ns
    a2=0.25
    a3=0.5
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    c2=2.75
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.75
    e3=-0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.25
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 2, figsize=(8, 8)
                         )
    # ponuka práce
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we_1)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a2,b,w),xmin,xmax-a2,
                           nazov="$N^s(w,r_1,we_2)$",
                           color="orange")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a3,b,w),xmin,xmax-a3,
                           nazov="$N^s(w,r_2,we_2)$",
                           color="red")
    
    # dopyt po práci
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A_1,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(c2,d,w),xmin+0.75,xmax,
                           nazov="$N^d(w,A_2,K)$",
                           color="indianred")
    
    # ponuka výstupu
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A_1,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e2,f,r),xmin-e2,xmax,
                           nazov="$Y^s(r,we_1,A_2,K)$",
                           color="orange")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e3,f,r),xmin-e3,xmax,
                           nazov="$Y^s(r,we_2,A_2,K)$",
                           color="red")
    
    # dopyt po výstupe
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(g,h,r),xmin,xmax,
                           nazov="$Y^d(r,we_1,A'_1,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(g2,h,r),xmin+0.25,xmax
                           ,nazov="$Y^d(r,we_2,A'_2,K)$",
                           color="indianred")
    
    # pôvodný bod, trh práce a výstupu
    grafy.nakresli_bod(axes[0,0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0,1], 1., 1., nazov=None, vodiace_ciary=True, color="blue")
    grafy.vyznac_na_osi(axes[0,0], x=1., y=1., popis_x="$N_1$", popis_y="$w_1$",color="blue",
                       dx_x=-1.5)
    grafy.vyznac_na_osi(axes[0,1], x=1., y=1., popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    
    # nový bod, trh práce
    grafy.nakresli_bod(axes[0,0], 1.125, 1.625, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0,0], x=1.125, y=1.625, popis_x="$N_2$", popis_y="$w_2$",
                        color="red",
                       dx_x=1.5)
    
    # nový bod, trh výstupu
    grafy.nakresli_bod(axes[0,1], 1.375, 0.875, nazov=None, vodiace_ciary=True, color="red")
    grafy.vyznac_na_osi(axes[0,1], x=1.375, y=0.875, popis_x="$Y_2$", popis_y="$r_2$",
                        color="red")
    
    axes[0,0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.24)
    )
    axes[0,1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.24)
    )
    
    grafy.ekonomicke_osi(axes[0,0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[0,1],"$Y$","$r$")
    
    # trh peňazí
    # pred zmenou
    grafy.nakresli_bod(axes[1,0],1,1,nazov=None,vodiace_ciary=True,color="blue")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1,nazov="$M_1$",color="blue")
    grafy.nakresli_funkciu(axes[1,0],lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.vyznac_na_osi(axes[1,0],x=1,y=1,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    
    # po zmene
    grafy.nakresli_funkciu(axes[1,0],lambda P:0.75*P,0,2,nazov="$PL(Y_2,r_2)$",
                           color="indianred")
    grafy.nakresli_bod(axes[1,0],1.,0.75,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.vyznac_na_osi(axes[1,0],x=1.,y=0.75,popis_x=None,popis_y="$P_2$",color="red")
    axes[1,0].legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.1)
    )
    grafy.ekonomicke_osi(axes[1,0],"$M$","$P$")
    axes[1,1].axis('off')
    
    plt.savefig(
        "images/pruzne/graf_RBC.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############
# graf_RBC_M #
##############

def graf_RBC_M():
    
    fig,ax=plt.subplots()
    grafy.nakresli_zvislu_ciaru(ax,1,nazov="$M_1$",color="blue")
    grafy.nakresli_zvislu_ciaru(ax,1/0.75,nazov="$M_2$",color="red")
    grafy.nakresli_funkciu(ax,lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.nakresli_bod(ax,1,1,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(ax,x=1/.75,y=1.,popis_x="$M_2$",color="red")
    
    grafy.vyznac_na_osi(ax,x=1,y=1,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(ax,1/.75,1,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.nakresli_funkciu(ax,lambda P:0.75*P,0,2,nazov="$PL(Y_2,r_2)$",color="indianred")
    
    ax.legend(
    )
    grafy.ekonomicke_osi(ax,"$M$","$P$")

    plt.savefig(
        "images/pruzne/graf_RBC_M.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

#######################
# graf_pf_koordinacia #
#######################

def graf_pf_koordinacia():

    grafy.set_theory_style()
    
    beta=2. # pracovná elasticita výstupu
    
    fig, axes = plt.subplots(3,1,figsize=(8,16))
    # produkčná funkcia
    grafy.nakresli_funkciu(axes[0], lambda N: N**beta,0,1.8)
    # body
    grafy.nakresli_bod(axes[0],1,1,nazov=None,color="blue")
    grafy.nakresli_bod(axes[0],1.5,1.5**beta,nazov=None,color="red")
    # dotycnice
    grafy.nakresli_dotycnicu(axes[0],
                             lambda N: N**beta,
                            lambda N: beta*N**(beta-1),
                            1.,0.4,linewidth=1.,
                            color="blue")
    grafy.nakresli_dotycnicu(axes[0],
                             lambda N: N**beta,
                            lambda N: beta*N**(beta-1),
                            1.5,0.4,linewidth=1.,
                            color="red")
    # spojnice
    grafy.nakresli_funkciu(axes[0], lambda N: N,0,1,linewidth=1.,linestyle="--",color="blue")
    grafy.nakresli_funkciu(axes[0], lambda N: 1.5**beta/1.5*N,0,1.5,linewidth=1.,
                           linestyle="--",color="red")
    # MPN
    grafy.nakresli_funkciu(axes[1], lambda N: beta*N**(beta-1),0,1.8)
    # APN
    grafy.nakresli_funkciu(axes[2], lambda N: N**(beta-1),0,1.8)
    grafy.ekonomicke_osi(axes[0],"$N$","$Y$")
    grafy.ekonomicke_osi(axes[1],"$N$","$MPN$")
    grafy.ekonomicke_osi(axes[2],"$N$","$APN$")
    


    plt.savefig(
        "images/pruzne/graf_pf_koordinacia.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##########################
# graf_praca_koordinacia #
##########################

def graf_praca_koordinacia():

    a=0. # intercept Ns
    b=1 # sklon Ns
    
    d=2.
    c=1-d
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    xmin2=(d-rozpatie/2)/d
    xmax2=(d+rozpatie/2)/d
    
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    
    grafy.nakresli_funkciu(ax,lambda w:fun(a,b,w),xmin-a,xmax,nazov="$N^s(w,r,we)$",
                           color="blue")
    grafy.nakresli_funkciu(ax,lambda w:fun(c,d,w),xmin2,xmax2,nazov="$N^d(w,A,K)$",
                           color="green")
   # grafy.nakresli_bod(ax,1,1,nazov=None,vodiace_ciary=True)
   # grafy.vyznac_na_osi(ax,x=1.,y=1.,popis_x="$N^*$",popis_y="$w^*$")
    
    ax.legend()
    grafy.ekonomicke_osi(ax,"$N$","$w$")

    plt.savefig(
        "images/pruzne/graf_praca_koordinacia.png",
        dpi=300,
        bbox_inches="tight"
    )    
    plt.close()

##############################
# graf_koordinacia_rovnovaha #
##############################

def graf_koordinacia_Ys():

    a1=-0.25 # intercept Ns
    a2=0.25
    b=1 # sklon N
    e=2 # intercept Ys
    
    d=2.
    c=1-d
    
    x_star1=(c-a1)/(b-d)
    x_star2=(c-a2)/(b-d)
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    xmin2=(d-rozpatie/2)/d
    xmax2=(d+rozpatie/2)/d
    
    beta=2.
    
    j = 0.7   # zakrivenie Ys, vyššia hodnota -> krivka je strmšia a viac zahnutá
    k = -0.1   # horizontálny posun, vyššia hodnota -> doprava
    l = 0.1  # vertikálny posun, vyššia hodnota -> nahor
    
    xmin_Ys = 0.29
    
    def fun(a,b,x):
        return a+b*x
    
    def Ys(j,k,l,Y):
        return j/(Y-k)+l
    
    korene_Ys = lambda Y: fun(e,-1,Y) - Ys(j,k,l,Y)
    
    Y2 = brentq(korene_Ys, 0.3, 0.8)
    Y1 = brentq(korene_Ys, 0.8, 1.5)
    
    r1=fun(e,-1,Y1)
    r2=fun(e,-1,Y2)
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(3,1,figsize=(8,16))
    
    # krivky ponuky práce
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a2,b,w),xmin,xmax-a2,nazov="$N^s(w,r_1,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda w:fun(a1,b,w),xmin-a1,xmax,nazov="$N^s(w,r_2,we)$",
                           color="red")
    
    # dopyt po práci
    grafy.nakresli_funkciu(axes[0],lambda w:fun(c,d,w),xmin2,xmax2,nazov="$N^d(w,A,K)$",
                           color="green")
    axes[0].legend(
        loc="lower right",
        bbox_to_anchor=(0.3, 0.45)
    )    
    
    # produkčná funkcia
    grafy.nakresli_funkciu(axes[1], lambda N: N**beta,0,xmax,nazov="$AF(K,N)$")
    axes[1].legend()

    # ponuka výstupu
    
    grafy.nakresli_funkciu(axes[2], lambda Y: Ys(j,k,l,Y) ,xmin_Ys,xmax,
                           nazov="$Y^s(r,we,A,K)$",
                          color="blue")
    grafy.nakresli_funkciu(axes[2], lambda Y: fun(e,-1,Y),xmin,xmax,nazov="$Y^d(w,A,K)$",
                           color="green")
    grafy.nakresli_bod(axes[2], Y1, r1, nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[2], x=Y1, y=r1, popis_x="$Y_1$", popis_y="$r_1$")
    grafy.vyznac_na_osi(axes[2], x=Y2, y=r2, popis_x="$Y_2$", popis_y="$r_2$")
    grafy.nakresli_bod(axes[2], Y2, r2, nazov=None, vodiace_ciary=True)
    axes[2].legend()

    grafy.ekonomicke_osi(axes[0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[1],"$N$","$Y$")
    grafy.ekonomicke_osi(axes[2],"$Y$","$r$")
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],x_star1,[fun(a1,b,x_star1),x_star1**beta],
                        axis_labels=["$w_2$","$Y_2$"],value_label="$N_2$",
                       value_label_shift=[-14,0])
    _=grafy.prepoj_body(fig,[axes[0],axes[1]],x_star2,[fun(a2,b,x_star2),x_star2**beta],
                        axis_labels=["$w_1$","$Y_1$"],value_label="$N_1$",
                       value_label_shift=[-14,0])
    
    plt.savefig(
    "images/pruzne/graf_koordinacia_Ys.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

##############################
# graf_koordinacia_rovnovaha #
##############################

def graf_koordinacia_rovnovaha():

    a2=-0.25 # intercept Ns
    a1=0.25
    b=1 # sklon N
    e=2 # intercept Ys
    
    d=2.
    c=1-d
    
    x_star1=(c-a1)/(b-d)
    x_star2=(c-a2)/(b-d)
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    xmin2=(d-rozpatie/2)/d
    xmax2=(d+rozpatie/2)/d
    
    beta=2.
    
    j = 0.7   # zakrivenie Ys, vyššia hodnota -> krivka je strmšia a viac zahnutá
    k = -0.1   # horizontálny posun, vyššia hodnota -> doprava
    l = 0.1  # vertikálny posun, vyššia hodnota -> nahor
    
    xmin_Ys = 0.29
    
    def fun(a,b,x):
        return a+b*x
    
    def Ys(j,k,l,Y):
        return j/(Y-k)+l
    
    korene_Ys = lambda Y: fun(e,-1,Y) - Ys(j,k,l,Y)
    
    Y2 = brentq(korene_Ys, 0.3, 0.8)
    Y1 = brentq(korene_Ys, 0.8, 1.5)
    
    r1=fun(e,-1,Y1)
    r2=fun(e,-1,Y2)
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 2, figsize=(8, 8)
                         )
    # krivky ponuky práce
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a1,b,w),xmin,xmax-a1,nazov="$N^s(w,r_1,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a2,b,w),xmin-a1,xmax,nazov="$N^s(w,r_2,we)$",
                           color="red")
    
    # dopyt po práci
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(c,d,w),xmin2,xmax2,nazov="$N^d(w,A,K)$",
                           color="green")
    grafy.nakresli_bod(axes[0,0], x_star1, fun(a1,b,x_star1), nazov=None, vodiace_ciary=True,
                      color="blue")
    grafy.vyznac_na_osi(axes[0,0], x=x_star1, y=fun(a1,b,x_star1), 
                        popis_x="$N_1$", popis_y="$w_1$",color="blue")
    grafy.nakresli_bod(axes[0,0], x_star2, fun(a2,b,x_star2), nazov=None, vodiace_ciary=True,
                      color="red")
    grafy.vyznac_na_osi(axes[0,0], x=x_star2, y=fun(a2,b,x_star2), 
                        popis_x="$N_2$", popis_y="$w_2$",color="red")
    axes[0,0].legend(
        loc="lower right",
        bbox_to_anchor=(0.65, 0.78)
    )    
    
    # ponuka výstupu
    
    grafy.nakresli_funkciu(axes[0,1], lambda Y: Ys(j,k,l,Y) ,xmin_Ys,xmax,
                           nazov="$Y^s(r,we,A,K)$",
                          color="blue")
    grafy.nakresli_funkciu(axes[0,1], lambda Y: fun(e,-1,Y),xmin,xmax,nazov="$Y^d(w,A,K)$",
                           color="green")
    grafy.nakresli_bod(axes[0,1], Y1, r1, nazov=None, vodiace_ciary=True,color="blue")
    grafy.vyznac_na_osi(axes[0,1], x=Y1, y=r1, popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(axes[0,1], x=Y2, y=r2, popis_x="$Y_2$", popis_y="$r_2$",color="red")
    grafy.nakresli_bod(axes[0,1], Y2, r2, nazov=None, vodiace_ciary=True,color="red")
    axes[0,1].legend()
    axes[1,1].axis('off')
    
    # trh peňazí
    # 2
    grafy.nakresli_bod(axes[1,0],1,1,nazov=None,vodiace_ciary=True,color="red")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1,nazov="$M^s$",color="blue")
    grafy.nakresli_funkciu(axes[1,0],lambda P:P,0,2,nazov="$PL(Y_2,r_2)$",color="indianred")
    grafy.vyznac_na_osi(axes[1,0],x=1,y=1,popis_y="$P_2$",color="red")
    
    #1
    grafy.nakresli_funkciu(axes[1,0],lambda P:0.75*P,0,2,nazov="$PL(Y_1,r_1)$",
                           color="green")
    grafy.nakresli_bod(axes[1,0],1.,0.75,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(axes[1,0],x=1.,y=0.75,popis_x="$M$",popis_y="$P_1$",color="blue")
    axes[1,0].legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.1)
    )
    
    grafy.ekonomicke_osi(axes[0,0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[0,1],"$Y$","$r$")
    grafy.ekonomicke_osi(axes[1,0],None,"$P$")
    
    plt.savefig(
    "images/pruzne/graf_koordinacia_rovnovaha.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

#################################
# graf_koordinacia_procyklickeM #
#################################

def graf_koordinacia_procyklickeM():

    fig,ax=plt.subplots()
    
    #1
    grafy.nakresli_zvislu_ciaru(ax,1.25,nazov="$M^s_1$",color="blue")
    grafy.nakresli_funkciu(ax,lambda P:0.7*P,0,2,nazov="$PL(Y_1,r_1)$",
                           color="green")
    grafy.nakresli_bod(ax,1.25,0.7*1.25,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(ax,x=1.25,y=0.7*1.25,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    
    # 2
    grafy.nakresli_bod(ax,1,1,nazov=None,vodiace_ciary=True,color="red")
    grafy.nakresli_zvislu_ciaru(ax,1,nazov="$M^s_2$",color="red")
    
    grafy.nakresli_funkciu(ax,lambda P:P,0,2,nazov="$PL(Y_2,r_2)$",color="indianred")
    grafy.vyznac_na_osi(ax,x=1,y=1,popis_x="$M_2$",popis_y="$P_2$",color="red")
    
    
    ax.legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.1)
    )
    
    grafy.ekonomicke_osi(ax,"$M$","$P$")
    
    plt.savefig(
    "images/pruzne/graf_koordinacia_procyklickeM.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

######################
# graf_koordinacia_G #
######################

def graf_koordinacia_G():

    e=2 # pôvodný intercept Yd
    e2=1.85 # posun Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    
    j = 0.7   # zakrivenie Ys, vyššia hodnota -> krivka je strmšia a viac zahnutá
    k = -0.1   # horizontálny posun, vyššia hodnota -> doprava
    l = 0.1  # vertikálny posun, vyššia hodnota -> nahor
    
    k2=e2 + k - 2*np.sqrt(j) # posun Ys
    
    xmin_Ys = 0.29
    
    def fun(a,b,x):
        return a+b*x
    
    def Ys(j,k,l,Y):
        return j/(Y-k)+l
    
    korene_Ys = lambda Y: fun(e,-1,Y) - Ys(j,k,l,Y)
    
    # pôvodné rovnováhy
    Y2 = brentq(korene_Ys, 0.3, 0.8) 
    Y1 = brentq(korene_Ys, 0.8, 1.5)
    
    r1=fun(e,-1,Y1)
    r2=fun(e,-1,Y2)
    
    # jediná rovnováha
    Ystar = (e - l + k2)/2
    rstar=fun(e2,-1,Ystar)
    
    grafy.set_theory_style()
    fig,ax=plt.subplots()
    
    # pôvodná ponuka výstupu
    grafy.nakresli_funkciu(ax, lambda Y: Ys(j,k,l,Y) ,xmin_Ys,xmax,nazov="$Y^s(r,we_1,A,K)$",
                          color="blue")
    
    # nová ponuka práce
    grafy.nakresli_funkciu(ax, lambda Y: Ys(j,k2,l,Y) ,xmin_Ys+0.2,xmax,
                           nazov="$Y^s(r,we_2,A,K)$",
                          color="red")
    
    # dopyt po práci
    grafy.nakresli_funkciu(ax, lambda Y: fun(e,-1,Y),xmin,xmax,nazov="$Y^d(w,A,K,G_1)$",
                           color="green")
    # nový dopyt po práci
    grafy.nakresli_funkciu(ax, lambda Y: fun(e2,-1,Y),xmin,xmax-0.15,nazov="$Y^d(w,A,K,G_2)$",
                           color="indianred")
    
    grafy.nakresli_bod(ax, Y1, r1, nazov=None, vodiace_ciary=True,color="blue")
    grafy.vyznac_na_osi(ax, x=Y1, y=r1, popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(ax, x=Y2, y=r2, popis_x="$Y_2$", popis_y="$r_2$",color="blue")
    grafy.nakresli_bod(ax, Y2, r2, nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(ax, Ystar, rstar, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(ax, x=Ystar, y=rstar, popis_x="$Y^*$", popis_y="$r^*$",color="red")
    ax.legend()
    
    grafy.ekonomicke_osi(ax,"$Y$","$r$")
    
    plt.savefig(
    "images/pruzne/graf_koordinacia_G.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

graf_RBC()
graf_RBC_M()
graf_pf_koordinacia()
graf_praca_koordinacia()
graf_koordinacia_Ys()
graf_koordinacia_rovnovaha()
graf_koordinacia_procyklickeM()
graf_koordinacia_G()
```

## Súčasné keynesovské modely: nepružné ceny

Obrázky k tejto kapitole sa vygenerujú v podpriečinku `nepruzne`.

```python
import grafy
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

Path("images/nepruzne").mkdir(exist_ok=True)

grafy.set_theory_style

###############
# graf_keynes #
###############

def graf_keynes():

    a=0 # intercept Ns
    b=1 # sklon Ns
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 2, figsize=(8, 8)
                         )
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r^*,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(c,d,w),xmin,xmax,nazov="$N^d(w,A,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(g,h,r),xmin,xmax,
                           nazov="IS: $Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0,0], 0.75, 0.75, nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[0,1], 1., 1., nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[0,1], 0.75, 1.25, nazov=None, vodiace_ciary=True)
    grafy.nakresli_bod(axes[0,1], 1.25, 1.25, nazov=None, vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[0,0], x=0.75, y=0.75, popis_x="$N^*$", popis_y="$w^*$")
    grafy.vyznac_na_osi(axes[0,1], x=1., y=1., popis_x="$Y_m$", popis_y="$r_m$")
    grafy.vyznac_na_osi(axes[0,1], x=0.75, y=1.25, popis_x="$Y^*$", popis_y="$r^*$")
    grafy.vyznac_na_osi(axes[0,1], x=1.25, y=1.25, popis_x="$Y_1$", popis_y=None)
    
    axes[0,0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.98)
    )
    axes[0,1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.98)
    )
    grafy.ekonomicke_osi(axes[0,0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[0,1],"$Y$","$r$")
    grafy.nakresli_bod(axes[1,0],1,1,nazov=None,vodiace_ciary=True)
    grafy.nakresli_funkciu(axes[1,0],lambda P:P,0,2,nazov="$PL(Y^*,r^*)$",color="green")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1,nazov="$M^s$",color="blue")
    grafy.vyznac_na_osi(axes[1,0],x=1,y=1,popis_x="$M^*$",popis_y="$P^*$")
    axes[1,0].legend(
        loc="lower right",
        bbox_to_anchor=(1.05, 0.1)
    )
    grafy.ekonomicke_osi(axes[1,0],"$M$","$P$")
    
    grafy.nakresli_funkciu(axes[1,1],lambda N:N**0.5,0,2,nazov="$AF(K,N)$")
    grafy.nakresli_bod(axes[1,1],0.75,0.75**0.5,nazov=None,vodiace_ciary=True)
    grafy.vyznac_na_osi(axes[1,1],x=0.75,y=0.75**0.5,popis_x="$N^*$",popis_y="$Y^*$")
    grafy.ekonomicke_osi(axes[1,1],"$N$","$Y$")
    axes[1,1].legend()
    
    plt.savefig(
    "images/nepruzne/graf_keynes.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

################
# graf_keynesM #
################

def graf_keynesM():

    a=0 # intercept Ns
    b=1 # sklon Ns
    a2=0.5
    
    c=2. # intercept Nd
    d=-1. # sklon Nd
    
    
    e=0. # intercept Ys
    f=1. # sklon Ys
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(2, 2, figsize=(8, 8)
                         )
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a,b,w),xmin,xmax,nazov="$N^s(w,r_1,we)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,0],lambda w:fun(a2,b,w),xmin,xmax-a2,nazov="$N^s(w,r_2,we)$",
                           color="red")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0,1],lambda r:fun(g,h,r),xmin,xmax,
                           nazov="IS: $Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0,0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0,0], 1.25, 1.75, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[0,1], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0,1], 1.25, 0.75, nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0,0], x=1., y=1., popis_x="$N_1$", popis_y="$w_1$",color="blue")
    grafy.vyznac_na_osi(axes[0,0], x=1.25, y=1.75, popis_x="$N_2$", popis_y="$w_2$",
                        color="red")
    grafy.vyznac_na_osi(axes[0,1], x=1., y=1., popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(axes[0,1], x=1.25, y=0.75, popis_x="$Y_2$", popis_y="$r_2$",
                        color="red")
    
    axes[0,0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.1)
    )
    axes[0,1].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.1)
    )
    grafy.ekonomicke_osi(axes[0,0],"$N$","$w$")
    grafy.ekonomicke_osi(axes[0,1],"$Y$","$r$")
    grafy.nakresli_bod(axes[1,0],1,1,nazov=None,vodiace_ciary=True,color="blue")
    grafy.nakresli_funkciu(axes[1,0],lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.nakresli_funkciu(axes[1,0],lambda P:0.75*P,0,2,nazov="$PL(Y_2,r_2)$",
                           color="indianred")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1,nazov="$M^s_1$",color="blue")
    grafy.nakresli_zvislu_ciaru(axes[1,0],1/0.75,nazov="$M^s_2$",color="red")
    grafy.vyznac_na_osi(axes[1,0],x=1,y=1,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(axes[1,0],1/0.75,1.,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.vyznac_na_osi(axes[1,0],x=1/0.75,y=1.,popis_x="$M_2$",popis_y=None,color="red")
    axes[1,0].legend(
        loc="lower right",
        bbox_to_anchor=(1.18, 0.02)
    )
    grafy.ekonomicke_osi(axes[1,0],"$M$","$P$")
    
    grafy.nakresli_funkciu(axes[1,1],lambda N:N**0.5,0,2,nazov="$AF(K,N)$")
    grafy.nakresli_bod(axes[1,1],1.,1.,nazov=None,vodiace_ciary=True,color="blue")
    grafy.vyznac_na_osi(axes[1,1],x=1.,y=1.,popis_x="$N_1$",popis_y="$Y_1$",color="blue")
    grafy.nakresli_bod(axes[1,1],1.25,1.25**0.5,nazov=None,vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[1,1],x=1.25,y=1.25**0.5,popis_x="$N_2$",popis_y="$Y_2$",
                        color="red")
    grafy.ekonomicke_osi(axes[1,1],"$N$","$Y$")
    axes[1,1].legend()
    
    plt.savefig(
    "images/nepruzne/graf_keynesM.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

######################
# graf_keynes_stabil #
######################

def graf_keynes_stabil():

    e=0. # intercept Ys
    f=1. # sklon Ys
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g,h,r),xmin,xmax,nazov="IS: $Y^d(r,we,A',K)$",
                           color="green")
    grafy.nakresli_bod(axes[0], 0.75, 1.25, nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=0.75, y=1.25, popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$Y_2$", popis_y="$r_2$",color="red")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 0.98)
    )
    grafy.ekonomicke_osi(axes[0],"$Y$","$r$")
    
    l=0.6 # novy sklon dopytu L
    
    grafy.nakresli_funkciu(axes[1],lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda P:l*P,0,2,nazov="$PL(Y_2,r_2)$",
                           color="indianred")
    grafy.nakresli_zvislu_ciaru(axes[1],0.75,nazov="$M^s_1$",color="blue")
    grafy.nakresli_zvislu_ciaru(axes[1],0.75/l,nazov="$M^s_2$",color="red")
    grafy.nakresli_bod(axes[1],0.75,0.75,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(axes[1],x=0.75,y=0.75,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(axes[1],0.75/l,0.75,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.vyznac_na_osi(axes[1],x=0.75/l,y=0.75,popis_x="$M_2$",popis_y=None,color="red")
    grafy.nakresli_bod(axes[1],0.75,0.75*l,nazov="RBC",color="orange",dy=-7)
    axes[1].legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.05)
    )
    axes[1].set_ylim(0, 1.4)
    grafy.ekonomicke_osi(axes[1],"$M$","$P$")

    plt.savefig(
    "images/nepruzne/graf_keynes_stabil.png",
    dpi=300,
    bbox_inches="tight"
)    
    plt.close()

################
# graf_keynesG #
################

def graf_keynesG():

    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.75
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g,h,r),xmin,xmax,
                           nazov="IS: $Y^d(r,we_1,A',K,G_1)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g2,h,r),xmin+0.75,xmax,
                           nazov="IS: $Y^d(r,we_2,A',K,G_2)$",
                           color="indianred")
    grafy.nakresli_bod(axes[0], 0.75, 1.25, nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1.5, 1.25, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, color="orange")
    grafy.vyznac_na_osi(axes[0], x=0.75, y=1.25, popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(axes[0], x=1.5, y=1.25, popis_x="$Y_2$", color="red")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.18)
    )
    grafy.ekonomicke_osi(axes[0],"$Y$","$r$")
    
    l=0.6 # novy sklon dopytu L
    
    grafy.nakresli_funkciu(axes[1],lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda P:l*P,0,2,nazov="$PL(Y_2,r_1)$",
                           color="indianred")
    grafy.nakresli_zvislu_ciaru(axes[1],0.75,nazov="$M^s_1$",color="blue")
    grafy.nakresli_zvislu_ciaru(axes[1],0.75/l,nazov="$M^s_2$",color="red")
    grafy.nakresli_bod(axes[1],0.75,0.75,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(axes[1],x=0.75,y=0.75,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(axes[1],0.75/l,0.75,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.vyznac_na_osi(axes[1],x=0.75/l,y=0.75,popis_x="$M_2$",popis_y=None,color="red")
    grafy.nakresli_bod(axes[1],0.75,0.75*l,nazov=None,color="orange",dy=-7)
    axes[1].legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.05)
    )
    axes[1].set_ylim(0, 1.4)
    grafy.ekonomicke_osi(axes[1],"$M$","$P$")

    plt.savefig(
    "images/nepruzne/graf_keynesG.png",
    dpi=300,
    bbox_inches="tight"
) 
    plt.close()

###########################
# graf_keynes_predpoklady #
###########################

def graf_keynes_predpoklady():

    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.5
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=2.25
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x
    
    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A_1,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e2,f,r),xmin-e2,xmax,
                           nazov="$Y^s(r,we_2,A_2,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g,h,r),xmin,xmax,
                           nazov="IS: $Y^d(r,we_1,A'_1,K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g2,h,r),xmin+0.25,xmax,
                           nazov="IS: $Y^d(r,we_2,A'_2,K)$",
                           color="indianred")
    grafy.nakresli_bod(axes[0], 1., 1, nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1.375, 0.875, nazov=None, vodiace_ciary=True,color="red")
    grafy.nakresli_bod(axes[0], 1.25, 1., nazov=None, vodiace_ciary=True,color="brown")
    
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$Y_1$", popis_y="$r_1$",color="blue")
    grafy.vyznac_na_osi(axes[0], x=1.375, y=0.875, popis_x="$Y_2$", popis_y="$r_2$",
                        color="red")
    grafy.vyznac_na_osi(axes[0], x=1.25, y=1., popis_x="$Y_3$", color="brown")
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.18)
    )
    grafy.ekonomicke_osi(axes[0],"$Y$","$r$")
    
    l=0.6 # novy sklon dopytu L
    
    grafy.nakresli_funkciu(axes[1],lambda P:P,0,2,nazov="$PL(Y_1,r_1)$",color="green")
    grafy.nakresli_funkciu(axes[1],lambda P:l*P,0,2,nazov="$PL(Y_2,r_2)$",
                           color="indianred")
    grafy.nakresli_zvislu_ciaru(axes[1],0.75,nazov="$M^s_1$",color="blue")
    grafy.nakresli_zvislu_ciaru(axes[1],0.75/l,nazov="$M^s_2$",color="red")
    grafy.nakresli_bod(axes[1],0.75,0.75,nazov=None,vodiace_ciary="horizontal",color="blue")
    grafy.vyznac_na_osi(axes[1],x=0.75,y=0.75,popis_x="$M_1$",popis_y="$P_1$",color="blue")
    grafy.nakresli_bod(axes[1],0.75/l,0.75,nazov=None,vodiace_ciary="horizontal",color="red")
    grafy.vyznac_na_osi(axes[1],x=0.75/l,y=0.75,popis_x="$M_2$",popis_y=None,color="red")
    axes[1].legend(
        loc="lower right",
        bbox_to_anchor=(1.15, 0.05)
    )
    axes[1].set_ylim(0, 1.4)
    grafy.ekonomicke_osi(axes[1],"$M$","$P$")

    plt.savefig(
    "images/nepruzne/graf_keynes_predpoklady.png",
    dpi=300,
    bbox_inches="tight"
) 
    plt.close()

##################
# graf_keynes_0r #
##################

def graf_keynes_0r():

    e=0. # intercept Ys
    f=1. # sklon Ys
    e2=-0.25
    
    g=2. # intercept Yd
    h=-1 # sklon Yd
    g2=1.5
    
    rozpatie=1.8
    xmin=1-rozpatie/2
    xmax=1+rozpatie/2
    def fun(a,b,x):
        return a+b*x

    grafy.set_theory_style()
    fig,axes=plt.subplots(1, 2, figsize=(8, 4)
                         )
    
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e,f,r),xmin,xmax,nazov="$Y^s(r,we_1,A,K)$",
                           color="blue")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(e2,f,r),xmin-e2,xmax,nazov="$Y^s(r,we_2,A,K)$",
                           color="red")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g,h,r),xmin,xmax,nazov="$Y^d(r,we_1,A',K)$",
                           color="green")
    grafy.nakresli_funkciu(axes[0],lambda r:fun(g2,h,r),xmin,xmax-0.5,
                           nazov="$Y^d(r,we_2,A',K)$",
                           color="indianred")
    grafy.nakresli_bod(axes[0], 1., 1., nazov=None, vodiace_ciary=True,color="blue")
    grafy.nakresli_bod(axes[0], 1-0.125, 1-0.25-0.125, nazov=None, vodiace_ciary=True,
                       color="orange")
    grafy.vyznac_na_osi(axes[0], x=1., y=1., popis_x="$Y_1$", popis_y="$r_1$",color="blue",
                       dx_x=2.5)
    grafy.vyznac_na_osi(axes[0], x=1-0.125, y=1-0.375, popis_x="$Y_3$", popis_y="$r_2$",
                        color="orange",
                       dx_x=-2.5)
    grafy.nakresli_bod(axes[0], 0.5, 1., nazov=None, vodiace_ciary=True,color="red")
    grafy.vyznac_na_osi(axes[0], x=0.5, y=1., popis_x="$Y_2$",color="red")
    
    
    axes[0].legend(
        loc="center",
        bbox_to_anchor=(0.5, 1.05)
    )
    grafy.ekonomicke_osi(axes[0],"$Y$","$r$")
    
    grafy.nakresli_bod(axes[1],1,1,nazov=None,vodiace_ciary="horizontal")
    grafy.nakresli_funkciu(axes[1],lambda P:P,0,2,nazov="$PL(Y,0)$",color="green")
    grafy.nakresli_zvislu_ciaru(axes[1],1,nazov="$M+B$",color="blue")
    grafy.vyznac_na_osi(axes[1],x=1,y=1,popis_x="$M+B$",popis_y="$P^*$")
    grafy.ekonomicke_osi(axes[1],"Likvidita","$P$")
    axes[1].legend()

    plt.savefig(
    "images/nepruzne/graf_keynes_0r.png",
    dpi=300,
    bbox_inches="tight"
) 
    plt.close()

graf_keynes()
graf_keynesM()
graf_keynes_stabil()
graf_keynesG()
graf_keynes_predpoklady()
graf_keynes_0r()
```
